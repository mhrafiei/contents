## **2. Optimization Problem (Formal Mathematics)**

The formal (standard) mathematical defination of an optimization problem is as follows: Find optimum $\mathbf{x}$ such that all $f_i(\mathbf{x})$ become minimized while all $g_j(\mathbf{x})$ & $h_k(\mathbf{x})$ are satisfied:

\begin{equation}
\begin{aligned}
& \underset{\mathbf{x}}{\text{minimize}}
& & f_i(\mathbf{x}) \quad \text{for } i \in \{1,2,...,I\} \\
& \text{subject to} & &
\begin{cases}
g_j(\mathbf{x}) \leq 0 & \text{for } j \in \{1,2,\ldots,J\} \\
h_k(\mathbf{x}) = 0 & \text{for } k \in \{1,2,\ldots,K\}
\end{cases}
\end{aligned}
\end{equation}

where $f_i(\mathbf{x}): \mathbb{R}^N \to \mathbb{R}$ is the $i^{th}$ objective (fitness or cost) function, $\mathbf{x} = [x_1, x_2, \ldots, x_N]$ include the optimization variables (solution), $g_j(\mathbf{x}): \mathbb{R}^N \to \mathbb{R}$ is the $j^{th}$ inequality constraint, & $h_k(\mathbf{x}): \mathbb{R}^N \to \mathbb{R}$ is the $k^{th}$ equality constraint.

 **Optimization problem types:**
- **Linear** when all $f_i$, $g_j$, & $h_k$ are linear functions otherwise **nonlinear**.
- **Discrete** when all $x_i$ are integers otherwise continuous.
- **Minimization** when all $f_i$ need to be minimized & **maximization** when all $f_i$ need to be maximized.
- **Single-objective** when $I=1$ otherwise **multi-objective**.
- **Constrained** when $K \neq 0$ and/or $J \neq 0$, & **unconstrained** when $K=J=0$.

## **3. Optimization Techniques**


An optimization model, algorithm, or technique describes steps to discover the best (optimum) solution (if exists) to achieve a purpose considering every existing limitation.

**Example:** for continuous $x_1$ & $x_2$:
> * **Minimize** $f(\mathbf{x}) = \sin(x_1) + {x_2}^2$
> * **Subject to:**
>  * One nonlinear Inequality Constraint: $g(\mathbf{x}) = {x_1}^2 + {x_2}^2 - 1 \leq 0$ (the unit circle)
>  * One linear Equality Constraint: $h(\mathbf{x}) = x_1 + x_2 - 1 = 0$ (a line)

> Assuming a feasible solution for this example, we can use Python with the
 <a href="https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.minimize.html">scipy.optimize.minimize</a> library, which provides various optimization algorithms.

```python
#@title Example Solution
'''
Runtime: CPU
'''

import numpy as np
from scipy.optimize import minimize

# Define the objective function
def objective(x):
    return np.sin(x[0]) + x[1]**2

# Define the nonlinear constraint
def constraint1(x):
    return 1 - x[0]**2 - x[1]**2

# Define the linear constraint
def constraint2(x):
    return x[0] + x[1] - 1

# Initial guess
x0 = [0.5, 0.5]

# Define bounds for x1 & x2
b = (-np.inf, np.inf)  # No specific bounds on x1 & x2
bounds = [b, b]

# Define the constraints
con1 = {'type': 'ineq', 'fun': constraint1}
con2 = {'type': 'eq', 'fun': constraint2}
constraints = [con1, con2]

# Solve the problem
solution = minimize(objective, x0, method='SLSQP', bounds=bounds, constraints=constraints)

# Print the solution
print("Solution:", solution.x)
print("Objective function value at solution:", solution.fun)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```

## **4.  ML Optimization Mathematics**


\begin{equation}
\begin{aligned}
& \underset{\mathbf{w}}{\text{minimize}}
& & l_i(\mathbf{w},\mathbf{D}) \quad \text{for } i \in \{1,2,...,I\} \\
& \text{subject to} & &
\begin{cases}
g_j(\mathbf{w},\mathbf{D}) \leq 0 & \text{for } j \in \{1,2,\ldots,J\} \\
h_k(\mathbf{w},\mathbf{D}) = 0 & \text{for } k \in \{1,2,\ldots,K\}
\end{cases}
\end{aligned}
\end{equation}


Here, $l_i(\mathbf{w},\mathbf{D}): \mathbb{R}^N \to \mathbb{R}$ represents the $i^{th}$ loss function, which can also be understood as an objective, fitness, or cost function. The term $\mathbf{w} = [w_1, w_2, \ldots, w_N]$ denotes the trainable parameters in  ML, including weights, biases, & filters (usually, all are referred to as weights) that are subject to optimization. $\mathbf{D}$ signifies the available training data. It is important to note that while the training data $\mathbf{D}$ is known, the parameters $\mathbf{w}$ are initially unknown & must be determined through  ML models using various learning or optimization techniques. Furthermore, it is noteworthy that the set of optimal parameters $\mathbf{w}$ is likely to vary between different training datasets.


### **4.1. Popular Loss Functions**


In  ML, the choice of a loss function is critical as it guides the learning algorithm towards the desired outcome. Here are five of the most important loss functions:

> **Mean Squared Error (MSE):** Commonly used in regression tasks, MSE measures the average of the squares of the errors, i.e., the average squared difference between the estimated values & the actual value. It's particularly useful when large errors are potentially more harmful than smaller ones.
\begin{equation}
\text{MSE}(y, \bar{y}) = \frac{1}{N} \sum_{i=1}^{N} (y_i - \bar{y}_i)^2
\end{equation}
> where $y_i$ are the true values, $\bar{y}_i$ are the predicted values, & $N$ is the number of observations.

> **Cross-Entropy Loss (or Log Loss):** Predominantly used in classification problems, this loss function measures the performance of a classification model whose output is a probability value between 0 & 1. Cross-entropy loss increases as the predicted probability diverges from the actual label, making it effective for models where predictions need to be close to the actual categories.
\begin{equation}
\text{Cross-Entropy}(y, p) = -\sum_{i=1}^{C} y_i \log(p_i)
\end{equation}
>where $y_i$ is the true distribution (usually one-hot encoded labels), $p_i$ is the predicted probability distribution for each class, & $C$ is the number of classes.

> **Binary Cross-Entropy Loss:** A special case of cross-entropy loss, it's used specifically for binary classification tasks. It calculates the loss for each class separately & then takes the average, making it more suitable for models with two classes.
\begin{equation}
\text{Binary Cross-Entropy}(y, p) = -\frac{1}{N} \sum_{i=1}^{N} [y_i \log(p_i) + (1 - y_i) \log(1 - p_i)]
\end{equation}
where $y_i$ & $p_i$ represent the true labels & predicted probabilities, respectively, for each observation in a binary classification, & $N$ is the number of observations.

> **Categorical Cross-Entropy Loss:** It extends binary cross-entropy to multi-class classification problems. It requires the use of one-hot encoding of the class labels, with the model output expected to be a probability distribution.
\begin{equation}
\text{Categorical Cross-Entropy}(y, p) = -\sum_{i=1}^{N} \sum_{j=1}^{C} y_{ij} \log(p_{ij})
\end{equation}
where $y_{ij}$ is the $j^{th}$ element of the one-hot encoded true label for the $i^{th}$ observation, $p_{ij}$ is the corresponding predicted probability, $N$ is the number of observations, & $C$ is the number of classes.

> **Hinge Loss:** Often used for "maximum-margin" classification, most notably for support vector machines (SVMs). It's intended for use with binary classification where the labels are in the set {-1, 1}. Hinge loss is not differentiable, which can be a disadvantage for optimization algorithms that rely on gradients.
\begin{equation}
\text{Hinge Loss}(y, \bar{y}) = \frac{1}{N} \sum_{i=1}^{N} \max(0, 1 - y_i \cdot \bar{y}_i)
\end{equation}
where $y_i$ are the true labels (with values -1 or 1), $\bar{y}_i$ are the predicted values, & $N$ is the number of observations.


Each loss function has its strengths & is chosen based on the specific requirements & nature of the  ML task at hand. For instance, MSE is preferred for regression, while cross-entropy losses are more suited for classification problems.

### **4.2. Potential  ML Constraints**


Constraints in  ML could include:

> **Non-negativity:** Ensuring that certain parameters remain non-negative.
> * Example: Imagine a  ML model designed for resource allocation in a manufacturing process. In this scenario, the non-negativity constraint ensures that the predicted amount of resources (like materials, time, or budget) allocated to each part of the process cannot be negative. This is crucial because, in the real world, you cannot have negative amounts of resources.

> **Normalization:** Keeping the sum of probabilities equal to one or ensuring vector lengths are normalized.
> * Example: In a  ML model used for image classification, the output layer often consists of probabilities indicating the likelihood of the image belonging to each class. The normalization constraint ensures that the sum of these probabilities equals one. This is critical for the model to provide a coherent probabilistic interpretation of its predictions. See <a href="https://arxiv.org/abs/2105.05735">Yoon et al. (2021)</a>.

> **Capacity Constraints:** Limiting the complexity of the model, such as the number of neurons in an NN or the depth of a decision tree.
> * Example: Consider an NN used for predictive maintenance in industrial machinery. The model might employ capacity constraints by limiting the number of layers or neurons to avoid overfitting the training data & ensure generalizability. This ensures the model is complex enough to learn from data but not so complex that it memorizes specific data points. See <a href="https://arxiv.org/abs/2112.01088">Perez et al. (2021)</a>.

> **Domain-specific Constraints:** Constraints that arise from the specific application or data, such as physical boundaries in engineering tasks.
> * Example: In the design of a  ML model for structural health monitoring of buildings, domain-specific constraints might include physical laws & material properties. For instance, the model might incorporate constraints based on the maximum stress a material can withstand or the structure's natural frequencies, ensuring that predictions & analyses are physically plausible. See <a href="https://onepetro.org/REE/article-abstract/22/04/1185/207248/A-Machine-Learning-Methodology-Using-Domain?redirectedFrom=fulltext">Guevara et al. (2019)</a> & <a href="https://arxiv.org/abs/2308.01438">Mohammadshirazi et al. (2023)</a>.

## **5. Popular Learning Techniques**


### **5.1. Gradient Descent**

Gradient Descent is the most basic optimization algorithm. It updates the parameters in the direction that reduces the loss. Gradient descent is simply gradient (slope in 2d) calculation over loss function given all training data, $\mathbf{D}$, in a proper step, called learning rate.

<div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/gradiantdescent_202312041443.png?raw=true" width="35%">
  <br>
  <figcaption>Figure: An Example of Gradient Descent Optimization</figcaption>
</div>

Once the loss function gradient (partial derivatives with respect to all prametes, i.e., $\mathbf{w}$) is calculated, each parameter (i.e., ${w_i}$) is updated as follows:

\begin{equation}
w_{i,{e+1}} = w_{i,{e}} - \gamma \frac{\partial l(\mathbf{w}, \mathbf{D})}{\partial w_{i, {e}}}
\end{equation}

where $w_{i, {e}}$ is the $i^{th}$ parameter in training epoch ${e}$, $\frac{\partial l(\mathbf{w}, \mathbf{D})}{\partial w_{i, {e}}}$ is the partial derivative of loss with respect to $w_{i, {e}}$ using all training data, $\mathbf{D}$, $\gamma$ is the learning rate, & $w_{i,{e+1}}$ is the updated parameter in epoch ${e+1}$. See <a href="https://www.nature.com/articles/nature14539">LeCun et al. (2015)</a>.

**Cons & Pros:**

* Gradient Descent is sensitive to learning rate settings, affecting convergence speed.
* In high-dimensional spaces, its performance decreases due to complexity.
* Prone to getting stuck in local minima, especially in complex problems.
* Limited to differentiable functions and inefficient in parameter scaling.
* Memory-intensive, requiring storage of all training data, problematic for large datasets.



```python
#@title Example of Gradient Descent
'''
Runtime: CPU
'''

import tensorflow as tf

# Sample dataset
datain_tr = tf.constant([[0.1, 0.2], [0.2, 0.2], [0.3, 0.4], [0.4, 0.5]])  # Input data
dataou_tr = tf.constant([[0.1], [0.2], [0.3], [0.4]])  # Output data

# Define the model using the functional API
inputs = tf.keras.Input(shape=(2,))
x = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(1)(x)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compile the model with the Gradient Descent optimizer
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              loss='mean_squared_error')

# Train the model
model.fit(datain_tr, dataou_tr, epochs=10)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```

### **5.2. Stochastic Gradiant Descent**

**Stochastic (Mini-batch) Gradient Descent (SGD):** A variation of gradient descent where the gradient is calculated on a subset (i.e., batch) of training data. Gradient descent is simply gradient (slope in 2d) calculation over loss function given a subset (i.e., batch) of training data, $\mathbf{D}_b$, in a proper step, called learning rate.

<div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/stochasticgradiantdescent_202312041444.png?raw=true" width="35%">
  <br>
  <figcaption>Figure: An Example of Stochastic Gradient Descent Optimization</figcaption>
</div>

Once the loss function gradient (partial derivatives with respect to all prametes, i.e., $\mathbf{w}$) is calculated, each parameter (i.e., ${w_i}$) is updated as follows:

\begin{equation}
w_{i,{t+1}} = w_{i,{t}} - \gamma \frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}
\end{equation}

where $w_{i, {t}}$ is the $i^{th}$ parameter in iteration ${t}$, $\frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}$ is the partial derivative of loss with respect to $w_{i, {t}}$ using the the $b^{th}$ batch of training data, $\mathbf{D}_b$ ($b\in \{1,2,\ldots,B\}$), $\gamma$ is the learning rate, & $w_{i,{t+1}}$ is the updated parameter in iteration ${t+1}$. Note that one epoch is counted one all batches of data is process, one epoch is counted. See <a href="https://www.nature.com/articles/nature14539">LeCun et al. (2015)</a>.

**Cons & Pros:**

* SGD updates parameters using a single data point or small batch, resulting in faster iterations than batch gradient descent.
* SGD, effectively handles streaming data and online learning scenarios.
* Its stochastic nature helps in escaping local minima, increasing the likelihood of finding global minima in non-convex optimization problems.
* SGD leads to significant oscillations and instability in the learning path, requiring careful tuning of the learning rate.
*  The unpredictability of SGD can slow down the convergence to the optimal solution.
* It requires pre-processing steps like normalization for optimal performance.


```python
#@title Example of Stochastic Gradiant Descent
'''
Runtime: CPU
'''

import tensorflow as tf
import numpy as np

# Generating 1024 sample datapoints
datain_tr = np.random.rand(1024, 2)  # 1024 datapoints with 2 features each
dataou_tr = np.random.rand(1024, 1)  # 1024 output values

# Define the model using the functional API
inputs = tf.keras.Input(shape=(2,))
x = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(1)(x)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compile the model with the Stochastic Gradient Descent optimizer
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              loss='mean_squared_error')

# Train the model
model.fit(datain_tr, dataou_tr, epochs=10, batch_size=16)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```

### **5.3. Adaptive Gradient Algorithm (AdaGrad)**

AdaGrad customizes the learning rate for different parameters, applying smaller updates to parameters linked with commonly occurring features & larger updates for those related to rare features. This makes it especially effective for handling sparse data. The central concept involves accumulating squared gradients in the denominator. As each new term is positive, the total sum increases as training progresses, resulting in a progressively decreasing learning rate. While this approach facilitates swift initial progress & rapid adjustment of the learning rate, it can also lead to a significantly reduced learning rate, potentially causing the algorithm to cease learning prematurely.

\begin{equation}
w_{i,{t+1}} = w_{i,{t}} - \frac{\eta}{({G_{t,i} + \epsilon)}^{0.5}} \frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}
\end{equation}

where $G_{t,i} = G_{t-1,i} + [\frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}]^{2}$, $G_{t,i}$ is the sum of the squares of the partial derivatives of loss function with respect to $w_{i, {t}}$ up to iteration $t$, & $\epsilon$ is a small smoothing term to prevent division by zero (often on the order of $1^{-8}$). See <a href="https://www.jmlr.org/papers/volume12/duchi11a/duchi11a.pdf">Duchi et al. (2011)</a>

**Cons & Pros**
* AdaGrad adjusts learning rates based on parameter frequency, effectively increasing rates for infrequent features and decreasing them for frequent ones.
* Particularly efficient for sparse data, due to its adaptive learning rate mechanism.
* Accumulation of squared gradients in AdaGrad results in continually diminishing learning rates.
* This diminishing effect can cause learning to halt prematurely in prolonged training scenarios.
* The aggressive reduction in learning rates makes AdaGrad less suitable for non-convex problems common in deep learning.
* May lead to suboptimal outcomes due to inadequate exploration of complex solution spaces in these scenarios.



```python
#@title Example of Adaptive Gradient Algorithm (AdaGrad)
'''
Runtime: CPU
'''

import tensorflow as tf

# Sample dataset
datain_tr = tf.constant([[0.1, 0.2], [0.2, 0.2], [0.3, 0.4], [0.4, 0.5]])  # Input data
dataou_tr = tf.constant([[0.1], [0.2], [0.3], [0.4]])  # Output data

# Define the model using the functional API
inputs = tf.keras.Input(shape=(2,))
x = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(1)(x)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compile the model with the AdaGrad optimizer
model.compile(optimizer=tf.keras.optimizers.Adagrad(learning_rate=0.01),
              loss='mean_squared_error')

# Train the model
model.fit(datain_tr, dataou_tr, epochs=10)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```

### **5.4. Root Mean Square Propagation (RMSProp)**

RMSProp is an adaptation of AdaGrad designed to rectify its rapidly diminishing learning rate. Rather than accumulating all past squared gradients, RMSProp employs a moving average that prioritizes recent gradients, introducing a recency bias. This average is computed using an exponential decay factor, safeguarding against the reduction of the learning rate to excessively low levels. This modification frequently results in enhanced performance in NNs.

\begin{equation}
w_{i,{t+1}} = w_{i,{t}} - \frac{\eta}{({\nu_{t,i} + \epsilon)}^{1/2}} \frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}
\end{equation}

where $\nu_{t,i} = \beta\nu_{t-1,i} + (1-\beta) [\frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}}]^{2}$, $\nu_{t,i}$ is the decaying average of past squared partial derivatives of loss function with respect to $w_{i, {t}}$ up to iteration $t$, $\beta$ is a hyperparameter that determines the length of the moving average window,  & $\epsilon$ is a small smoothing term to prevent division by zero (often on the order of $1^{-8}$). See <a href="https://arxiv.org/pdf/1609.04747.pdf">Ruder (2017)</a>

**Cons & Pros**
* RMSProp modifies AdaGrad's approach by using a moving average for computing the learning rate, mitigating the issue of rapidly diminishing learning rates.
* Exhibits a recency bias towards recent gradients, making it more responsive and effective in non-stationary problem settings.
* Demonstrates consistent performance in online and non-convex environments, such as training deep neural networks.
* Sensitivity to hyperparameters, particularly the learning rate and decay factor, necessitates careful tuning for optimal performance.
* Despite improvements over AdaGrad, RMSProp may still struggle to achieve optimal convergence for certain complex problems, potentially leading to suboptimal solutions.







```python
#@title Example of Root Mean Square Propagation (RMSProp)
'''
Runtime: CPU
'''
import tensorflow as tf

# Sample dataset
datain_tr = tf.constant([[0.1, 0.2], [0.2, 0.2], [0.3, 0.4], [0.4, 0.5]])  # Input data
dataou_tr = tf.constant([[0.1], [0.2], [0.3], [0.4]])  # Output data

# Define the model using the functional API
inputs = tf.keras.Input(shape=(2,))
x = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(1)(x)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compile the model with the RMSProp optimizer
model.compile(optimizer=tf.keras.optimizers.RMSprop(learning_rate=0.001),
              loss='mean_squared_error')

# Train the model
model.fit(datain_tr, dataou_tr, epochs=10)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```

### **5.5. Adaptive Moment Estimation (Adam)**

Adam integrates elements from both AdaGrad & RMSProp. It employs moving averages of squared gradients for adjusting the learning rate, akin to RMSProp. Additionally, Adam maintains an exponentially decaying average of previous gradients, akin to the concept of "momentum." In the context of optimization algorithms used in  ML, "momentum" is a technique that helps accelerate the algorithm's convergence towards the minimum of the loss function, & can also help to avoid local minima. The concept is inspired by the physical notion of momentum in physics, which is the tendency of a moving object to continue in its direction of motion. In mathematical terms, when learning techniques are augmented with momentum, instead of updating the weights (parameters) solely based on the current gradient, the update also takes into account the previous update. This approach allows Adam to adjust learning rates for each parameter individually. Like RMSProp, it stores an exponentially decaying average of past squared gradients. However, Adam also keeps a similar average of past gradients, functioning as momentum. Essentially, Adam represents a fusion of RMSProp & momentum techniques: it leverages squared gradients for learning rate modulation & utilizes a moving average of the gradient, akin to momentum, rather than the gradient itself:

\begin{equation}
w_{i,{t+1}} = w_{i,{t}} - \frac{\eta}{{{\bar{\nu}_{t,i}}^{1/2} + \epsilon}} \bar{m_t}
\end{equation}

where $\bar{m}_{t,i} = \frac{m_{t,i}}{1-\beta_1^t}$ is called bias-corrected first-moment estimate, $m_{t,i}=\beta_1 m_{t-1,i} + (1-\beta_1)(\frac{\partial l(\mathbf{w}, \mathbf{D}_b}{\partial w_{i, {t}}})$ is called biased firs moment estimate, $\beta_1$ is a hyperparameter called exponential decay rate for the first moment estimates (similar to the momentum term), $\bar{\nu}_{t,i} = \frac{\nu_{t,i}}{1-\beta_2^t}$ is called bias-corrected second raw moment estimate, $\nu_{t,i}=\beta_2 \nu_{t-1,i} + (1-\beta_2)(\frac{\partial l(\mathbf{w}, \mathbf{D}_b)}{\partial w_{i, {t}}})^2$ biased second-moment estimate, & $\beta_2$ is a hyperparameter called the exponential decay rate for the second-moment estimates. See <a href="https://arxiv.org/abs/1412.6980" >Diederik et al. (2014)</a>.

**Cons & Pros**

* Adam optimization algorithm combines aspects of RMSProp and momentum, leading to two main advantages.
* Adapts learning rates for each parameter, ensuring efficient and effective updates in large datasets with sparse gradients.
* Incorporates momentum through moving averages of the gradients, enhancing convergence speed and stability.
* Sensitive to initial learning rate and hyperparameter settings, often requiring fine-tuning for optimal performance.
* In scenarios with noisy or sparse data, Adam may lead to suboptimal convergence due to potentially biased moment estimates.
* Its primary contribution lies in effectively combining momentum with adaptive learning rates for a more efficient optimization process.






```python
#@title Example of Adaptive Moment Estimation (Adam)
'''
Runtime: CPU
'''
import tensorflow as tf

# Sample dataset
datain_tr = tf.constant([[0.1, 0.2], [0.2, 0.2], [0.3, 0.4], [0.4, 0.5]])  # Input data
dataou_tr = tf.constant([[0.1], [0.2], [0.3], [0.4]])  # Output data

# Define the model using the functional API
inputs  = tf.keras.Input(shape=(2,))
x       = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(1)(x)

model   = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compile the model with the Adam optimizer
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='mean_squared_error')

# Train the model
model.fit(datain_tr, dataou_tr, epochs=10)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try: 
    from google.colab import runtime
    import time
    time.sleep(5)
    runtime.unassign()
except ImportError:
    pass
```
