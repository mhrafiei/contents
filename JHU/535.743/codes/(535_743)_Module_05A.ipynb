## **1. Generative AI**


Generative AI refers to a type of AI technology that can generate new content, such as text, images, music, & videos, based on the patterns & data it has learned from. This technology uses ML, particularly those based on neural networks, to understand & replicate complex patterns & distributions in data.

Key characteristics of Generative AI include:

> **Learning from Data**: Generative AI models are trained on large datasets, allowing them to "learn" & understand the underlying structures & nuances of the data. This enables them to produce outputs that are similar in style & content.

> **Creating New Content**: Unlike traditional AI, which primarily focuses on analysis or decision-making, generative AI actively creates new content that didn't exist before, making it useful for a variety of creative & practical applications.

> **Adaptability**: These models can be fine-tuned for specific tasks or styles. For instance, a generative AI trained on classical music can generate new pieces in the style of Bach or Mozart, or an AI trained on images of dogs can generate new, unique dog images.

> **Interactivity**: Generative AI can be interactive, adapting its output based on user feedback or changing requirements, which is particularly useful in fields like gaming, interactive media, & personalized content creation.

Some popular examples of generative AI include:

> **GPT (Generative Pre-trained Transformer)**: Used for generating human-like text based on the input it receives.

> **DALL-E**: Capable of generating detailed images from textual descriptions.

> **Deepfakes**: Video & audio generation that can alter or synthesize human appearances & voices in media.

Generative AI is becoming increasingly integral to many industries, revolutionizing how content is created & customized, though it also raises ethical & regulatory concerns, particularly regarding authenticity, copyright, & misuse.

## **2. Generative Adversarial Networks (GANs)**

[Generative Adversarial Networks (GANs)](https://proceedings.neurips.cc/paper_files/paper/2014/file/5ca3e9b122f61f8f06494c97b1afccf3-Paper.pdf) are a class of AI algorithms used in unsupervised ML. They are implemented by two neural networks (generator & discriminator) contesting each other in a game-theoretic scenario. Ian Goodfellow & his colleagues introduced this approach in 2014, & it has since become a foundational technique in the field of generative AI.

The fundamental concept behind GANs involves two distinct models:

> **Generator**: This network generates new data instances, trying to produce data that is indistinguishable from real-world data. It learns to create fake data by continuously trying to outsmart the discriminator.

> **Discriminator**: This network evaluates data instances to determine whether they are real or generated by the generator. Its job is to accurately classify both real data & fakes generated by the generator.

These two networks undergo an adversarial process, where the generator aims to maximize the errors of the discriminator by creating increasingly convincing fakes, while the discriminator strives to minimize its own errors in distinguishing real from fake.



## **3. GANs Training**

The generator, $G$, produces output from random noise input, & the discriminator, $D$, evaluates this output using a dataset of real, e.g., images, to learn to differentiate. The generator improves its ability to create realistic images, while the discriminator improves its ability to detect fakes. This iterative process continues until the discriminator can no longer easily distinguish real images from fakes, indicating that the generator has learned to produce very realistic data. The generator loss, $\mathbb{L}_D$, & the discriminator loss, $\mathbb{L}_G$, are as follows:

> **Generator Loss**
\begin{equation}
\mathbb{L}_G=-\frac{1}{2} \mathbb{E}_{z \sim p_z(z)}[\log D(G(z))]
\end{equation}

> **Discriminator Loss**
\begin{equation}
\mathbb{L}_D=-\frac{1}{2} \mathbb{E}_{x \sim p_{\text {dati }}(x)}[\log D(x)]-\frac{1}{2} \mathbb{E}_{z \sim p_z(z)}[\log (1-D(G(z)))]
\end{equation}

Here, $ \mathbb{E}_{x \sim p_{\text{data}}(x)}[\log D(x)]$ is the expected log probability that the discriminator correctly classifies real data $x$ as real. Similarly, $ \mathbb{E}_{z \sim p_z(z)}[\log (1 - D(G(z)))] $ is the expected log probability that the discriminator correctly classifies generated data $ G(z) $ as fake. This minimax game aims to find a point where the generator produces data indistinguishable from real data, & the discriminator is at a chance level, theoretically achieving a loss around 0.5 for both real & fake classifications.



### **3.1. GANs Common Issues**

Traditional GAN models suffer from some major problems discussed in this section:

> **Mode collapse:**
Mode collapse is a common problem in the training of GANs where the generator learns to produce a limited variety of outputs. This can occur even if the input noise is varied. In other words, the generator discovers a particular point or a few points in the input space that can deceive the discriminator into thinking they are real data points, & it keeps producing these points over & over again. This is detrimental because the goal of a GAN is typically to produce a diverse range of outputs that accurately reflect the distribution of the training data. Consequences of Mode Collapse include:
* Limited Diversity: The model generates very few types of outputs, which is particularly problematic in applications like image generation where a wide variety of images is desirable.
* Reduced Utility: The ability of the GAN to generalize & perform useful tasks diminishes since it cannot produce diverse results that span the range of the actual data distribution.

> **Unstable Training Dynamics:**
Unstable training dynamics refer to the problems associated with the convergence & learning process of GANs. These include issues where the training process does not lead to a stable equilibrium between the generator & discriminator. Common signs of unstable training include:
* Oscillating Loss: Where the loss values for either the generator or discriminator (or both) do not converge but instead oscillate, indicating that neither network is improving in a stable manner.
* Divergence: Situations where the training leads to the loss going to infinity or collapsing to zero, suggesting that the networks are no longer learning effectively.
* Vanishing Gradients: Particularly in the case of the discriminator outperforming the generator too early in training, the generator may stop learning altogether due to vanishing gradients. This is when gradients passed back to the generator are too small for effective learning.

> **Causes of Unstable Dynamics:**
* Architectural Imbalances: If the discriminator becomes too strong too quickly, it can effectively resolve real vs. fake with high confidence, leading to vanishing gradients for the generator.
* Inappropriate Learning Rates: If the learning rates for the generator & discriminator are not well-matched, one might overpower the other, leading to unstable training.
Inadequate Model Capacity: If either the generator or the discriminator lacks the capacity (due to insufficient layers or nodes), it might not be able to learn the complexity of the data distribution, leading to poor performance.



### **3.2. Addressing GANs Common Issues**

To address GANs' common issues such as combat mode collapse, & unstable training dynamics, several strategies can be employed:

* Alternative Loss Functions: As discussed earlier, using Wasserstein loss or R2 loss can help mitigate these issues by providing more meaningful & stable gradients throughout the training process.
* Regularization Techniques: Techniques like gradient penalty (used in WGAN-GP) enforce a Lipschitz constraint on the discriminator to help stabilize training.
* Architecture Adjustments: Tweaking the architecture, such as adding more layers, changing activation functions, or employing different normalization techniques, can help balance the capabilities of the generator & discriminator.
* Controlled Learning Rates: Adjusting & fine-tuning the learning rates & other hyperparameters during training can prevent one network from overpowering the other.

Addressing these challenges is crucial for ensuring that GANs are able to learn effectively & produce high-quality, diverse outputs ([more](https://proceedings.neurips.cc/paper_files/paper/2016/file/8a3363abe792db2d8761d6403605aeb7-Paper.pdf)).

### **3.3. Wasserstein Loss & R2 Loss**

The Wasserstein loss & R2 loss are two alternative loss functions used in training GANs that aim to address some of the issues inherent in the traditional GAN loss functions, such as mode collapse & unstable training dynamics.



#### 3.3.1. Wasserstein Loss


The Wasserstein loss, also known as Earth Mover's Distance (EMD), measures the cost of optimally transporting mass to transform the generator's distribution into the real data distribution. This loss is used in Wasserstein GANs (WGANs), which modify the standard GAN objective to provide more stable training behavior.

The Wasserstein loss for a GAN can be expressed as:

\begin{equation}
\min _G \max _D \mathbb{E}_{x \sim p_{\text {data }}(x)}[D(x)]-\mathbb{E}_{z \sim p_z(z)}[D(G(z))]
\end{equation}

Here, $D$ is not trained to be a classifier (as in standard GANs) but rather functions as a critic that scores the realness of the samples.

> **Pros**
* Stability of Training: The Wasserstein distance provides useful gradients (almost) everywhere, which helps in stabilizing the training of GANs even when the generator's distribution does not overlap with the real data distribution.
* Improvement in Mode Collapse: It tends to mitigate mode collapse, ensuring that the generator produces diverse samples.
* Interpretable Training Metric: The value of the Wasserstein loss correlates with the quality of generated images, making it a more interpretable measure during training.

> **Cons**
* Computational Cost: Calculating the Wasserstein distance involves solving an optimization problem, which can be computationally expensive.
* Weight Clipping: In its original formulation, WGAN requires weight clipping to enforce a [Lipschitz](https://dash.harvard.edu/bitstream/handle/1/37364596/SARGENT-SENIORTHESIS-2019.pdff) constraint, which can lead to optimization difficulties & capacity underuse ([more](https://ai.stackexchange.com/questions/29904/what-is-lipschitz-constraint-and-why-it-is-enforced-on-discriminator)).

#### 3.3.2. R2 Loss

The R2 loss, used in the Least Squares GAN (LSGAN), aims to penalize samples based on their distance from the decision boundary in a squared error sense. This is intended to address the vanishing gradients problem that can occur with the traditional sigmoid cross-entropy loss used in standard GANs. The R2 loss for the discriminator & generator can be written as:

\begin{equation}
\mathbb{L}_D=\frac{1}{2} \mathbb{E}_{x \sim p_{\text {data }}(x)}\left[(D(x)-1)^2\right]+\frac{1}{2} \mathbb{E}_{z \sim p_z(z)}\left[D(G(z))^2\right]
\end{equation}

\begin{equation}
\mathbb{L}_G=\frac{1}{2} \mathbb{E}_{z \sim p_z(z)}\left[(D(G(z))-1)^2\right]
\end{equation}

> **Pros**
* Stable Gradients: By using a squared loss, it provides stable gradients improving the stability of the GAN training.
* Reduces Mode Collapse: Similar to the Wasserstein loss, it helps in reducing mode collapse.

> **Cons**
* Less Robust to Noisy Labels: Squared error is more sensitive to outliers compared to cross-entropy, which might affect performance when dealing with noisy labels.
* Potential for Oversmoothing: There's a risk that minimizing squared error could lead to oversmoothing of the generated samples, potentially losing some detail.

Both the Wasserstein loss & R2 loss present significant improvements over traditional GAN loss functions, particularly in terms of training stability & reduction of mode collapse. R2 loss is often preferred over Wasserstein loss for training GANs due to its computational simplicity, efficiency, & the stability it offers in training dynamics. It utilizes a straightforward quadratic loss function that is easier to implement & does not require complex hyperparameter tuning, such as enforcing Lipschitz constraints needed with Wasserstein loss. Additionally, R2 loss ensures strong gradients when the generator's outputs are far from realistic, aiding in faster & more reliable convergence. This makes R2 loss particularly attractive in practical applications where ease of implementation, computational efficiency, & stable training are prioritized.



### **3.4. When to Stop Training GANs?**


Determining when a GAN model is sufficiently trained involves assessing several factors to ensure that the model generates high-quality & varied outputs without overfitting. Here are key indicators & methods to evaluate whether a GAN is trained enough:

> **Visual Inspection**
* Quality of Generated Samples: One of the most straightforward methods is visually inspecting the generated samples. In the context of images, you look for realistic textures, correct object proportions, & appropriate colors.
* Diversity of Outputs: Ensure that the GAN produces a variety of outputs. A well-trained GAN should generate diverse results within the scope of the training dataset without repeating the same outputs.

> **Convergence of Discriminator & Generator Loss**
* Stability of Loss Functions: During training, the generator & discriminator have competing goals. The generator tries to fool the discriminator by producing near-realistic outputs, while the discriminator works to distinguish real from fake. A GAN is often considered well-trained when the loss of the discriminator stabilizes around a value that suggests it is unsure if the inputs are real or fake (often around 0.5 in binary classification or 0 in R2 loss).
* Oscillations in Loss: Frequent or high-amplitude oscillations in loss might indicate that the GAN has not yet stabilized & requires more training.

> **Quantitative Metrics**
* [Inception Score](https://proceedings.neurips.cc/paper_files/paper/2016/file/8a3363abe792db2d8761d6403605aeb7-Paper.pdf) (IS): This metric uses a pre-trained Inception model to evaluate how realistic & diverse the generated images are. A higher Inception Score indicates better performance.
* [Fréchet Inception Distance](https://papers.nips.cc/paper_files/paper/2017/file/8a1d694707eb0fefe65871369074926d-Paper.pdf) (FID): FID measures the distance between feature vectors calculated for real & generated images. Lower FID values suggest that the generated images are more similar to the real ones, indicating better model performance.
* Mode Score & Diversity Score: These metrics assess the diversity of the generated samples & their alignment with the real data distribution ([more](https://machinelearningmastery.com/how-to-evaluate-generative-adversarial-networks/)).

> **Stability Over Multiple Runs**
* Consistency Across Training Sessions: A well-trained GAN should demonstrate consistent performance across different training sessions with the same parameters, indicating robustness & stability.

> **Application-Specific Performance**
* Task-Specific Evaluation: Depending on the application, you might have specific performance metrics or goals, such as accuracy in replicating certain features in generated images or the ability to fool expert systems in the domain of application.

> **Avoiding Common Pitfalls**
* Mode Collapse: Ensure that the generator does not produce limited varieties of outputs. Mode collapse is a common issue where the generator starts producing a small set of outputs regardless of input.
* Overfitting: Check that the generator does not simply memorize training examples but instead learns to generate new data based on learned features.

Monitoring these aspects can help determine whether a GAN is adequately trained or if further training or parameter adjustments are necessary. Regular evaluations using a combination of these methods will provide a comprehensive view of the GAN’s performance.

## **4. Diffusion Models**

[Diffusion models](https://colab.research.google.com/drive/1Nu3zXiuPDhNrgSRm-UGLUDVzjKvsO3EE#scrollTo=6VOnXqFWHz0_&line=18&uniqifier=1) are a class of generative models used in ML that have gained significant popularity for their ability to generate high-quality, diverse samples in tasks such as image, audio, & text generation. [These models](https://arxiv.org/pdf/2105.05233.pdf) are based on the concept of gradually adding noise to data & then learning to reverse this process to generate new samples. Here is an introduction to how they work, including their mathematical formulations.

<div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/diffusion_cifar10.png?raw=true" width="75%">
  <br>
  <figcaption>Figure: Example of a diffusion model reversing the noise to generate images similar to the CIFAR10 distribution - Ref: <a href="https://arxiv.org/pdf/2006.11239.pdf">Ho et al. (2020)</a></figcaption>
</div>


Diffusion models are inspired by the physical process of diffusion, where particles spread out from regions of higher concentration to lower concentration until they reach equilibrium. In the context of ML, diffusion models work in two phases: the forward process & the reverse process.
> **Forward Process (Diffusion):** This phase involves gradually adding noise to an original data point until the data becomes pure noise. This process is typically modeled as a [Markov chain](https://builtin.com/machine-learning/markov-chain) where each step adds a small amount of [Gaussian noise](https://en.wikipedia.org/wiki/Gaussian_noise). The forward process can be described mathematically as follows:
\begin{equation}
x_t=\sqrt{\alpha_t} x_{t-1}+\sqrt{1-\alpha_t \epsilon}
\end{equation}
where $x_t$ is the data at time step $t$, $x_{t-1}$ is the data at time step $t-1$, ${\alpha_t}$ is a variance schedule that controls the amount of noise added at each step, typically decreasing as $t$ increases, & $\epsilon$ is a noise term sampled from a normal distribution, $N (0, I)$.
The forward process can also be defined in terms of a variance-preserving transformation over a fixed number of steps, typically parameterized as:
\begin{equation}
q\left(x_t \mid x_0\right)=N\left(x_t ; \sqrt{\bar{\alpha}_t} x_0,\left(1-\bar{\alpha}_t\right) I\right)
\end{equation}
where $\bar{\alpha}_t=\prod_{s=1}^t \alpha_s$


> **Reverse Process (Denoising):** In this phase, the model learns to reverse the noisy data back to its original form. This is done by training a neural network to predict the original data from the noisy data. The reverse process is where the diffusion model generates data by denoising:
\begin{equation}
p_\theta\left(x_{t-1} \mid x_t\right)=N\left(x_{t-1} ; \mu_\theta\left(x_t, t\right), \Sigma_\theta\left(x_t, t\right)\right)
\end{equation}
where $\mu_\theta\left(x_t, t\right)$ & $\Sigma_\theta\left(x_t, t\right)$ are parameterized by the neural network with parameters $\theta$, $\mu_{theta}$ is typically a function predicting the mean of the reverse distribution, & $\Sigma_\theta$ often assumes a fixed or learnable diagonal covariance matrix depending on the model's complexity ([more](https://arxiv.org/pdf/2006.11239.pdf)).


## **5. Multi-Modal Generative Models**

Generative AI in multi-modal models refers to the development & use of AI systems that can generate or synthesize new content by understanding & integrating multiple forms of data, such as text, images, audio, & video. These models are designed to capture & exploit the relationships between different types of data, enabling more complex & nuanced AI applications than those dealing with a single data modality.

> **What are Multi-Modal Models?**
* Multi-modal models are AI systems that process & relate information from two or more distinct modalities (e.g., text & images). They are particularly valuable because they can understand context & content across these modalities, enabling them to perform tasks that involve complex interactions between different forms of data.

> **Applications of Generative AI in Multi-Modal Models**
* Content Creation: These models can generate coherent & contextually appropriate content that combines different modalities. For instance, a model might generate a descriptive caption for a photograph or synthesize a video from a written script.
* Data Augmentation: In scenarios where data in one modality is scarce but abundant in another, generative multi-modal models can be used to generate synthetic data. For example, generating synthetic medical imaging data based on textual disease descriptions to train diagnostic models.
* Enhanced Interaction: Multi-modal generative models can improve human-computer interaction by allowing more natural communication that includes speech, text, & visual elements, enhancing user interfaces & experiences.
* Education & Training: They can create diverse learning materials that integrate text, diagrams, & interactive elements tailored to different learning styles.
* Entertainment & Gaming: In video games & virtual reality, these models can generate dynamic content like dialogues, narratives, & environments that respond to user inputs in multiple modalities.

> **Key Technologies & Techniques**
* Transformers: These are powerful neural network architectures that have been highly successful in modeling language & are increasingly used for other modalities. Transformers can be adapted for multi-modal inputs by modifying their input layers to handle different types of data.
* Variational Autoencoders (VAEs) & Generative Adversarial Networks (GANs): These frameworks are used for generating new data samples. In a multi-modal context, they can be trained to generate data in one modality based on inputs from another, such as generating images from textual descriptions.
* Cross-Modal Learning: This involves training models on paired datasets, like images & their corresponding captions, to learn the relationships between modalities. This training helps the model understand how changes in one modality affect another.
* Attention Mechanisms: These allow models to focus on relevant parts of data within & across modalities, improving the effectiveness of generation tasks by paying attention to the most informative features.

> **Challenges & Research Directions**
* Alignment & Fusion: Determining how best to combine or align different modalities during both training & generation phases is challenging but crucial for performance.
* Scalability & Efficiency: Processing multiple modalities requires significant computational resources, making efficiency improvements a constant area of research.
* Data Scarcity & Bias: Building robust multi-modal datasets is difficult, & biases in data can lead to biased AI behaviors, requiring careful management & mitigation strategies.
* Ethical & Social Implications: As generative multi-modal AI becomes more capable, ensuring these systems are used responsibly & ethically, particularly in sensitive applications, becomes increasingly important.

Generative AI in multi-modal models represents a cutting-edge area of AI research with the potential to significantly enhance the capabilities & applications of AI systems. The interplay between different data types offers a richer understanding of complex environments & tasks, paving the way for innovative applications across various domains.

## **6. Generative AI Applications in Mechanical Engineering**

Generative AI techniques have been gaining traction across various engineering disciplines due to their powerful capabilities in generating realistic data & models. Here are examples of generative AI techniques applications in each of the Mechanical Engineering concentrations in our EP program:

> **Advanced Manufacturing**
 * Custom Tool Design: Generative AI techniques can be used to generate innovative designs for manufacturing tools & components. By training on a dataset of existing tool designs, generative AI techniques can propose new designs that optimize specific manufacturing parameters like strength, weight, or thermal resistance.
 * Quality Control Systems: Generative AI techniques can enhance quality control processes by generating realistic defects & imperfections for training machine vision systems. This allows the systems to better recognize & respond to a wide range of defects in manufactured products.

> **Biomechanical Engineering**
 * Tissue Engineering: In biomechanical engineering, generative AI techniques can be applied to generate complex, 3D tissue structures for simulations & modeling. This aids in the development of tissue implants & in the understanding of tissue growth patterns.
 * Prosthetic Design: Generative AI techniques can help in the design of personalized prosthetics by generating various configurations & forms based on the patient’s unique anatomy, optimizing for factors like comfort, mobility, & aesthetics.

> **Fluid Mechanics & Thermal Science**
 * Flow Simulation Enhancement: Generative AI techniques can be utilized to improve the accuracy of computational fluid dynamics (CFD) simulations by generating high-fidelity synthetic data that can train other predictive models to refine simulation outputs.
 * Heat Transfer Models: They can also assist in developing new models for predicting heat transfer in complex systems, which is crucial for optimizing energy efficiency & designing thermal management systems in engineering applications.

> **Hypersonic Technologies**
 * Material Design: Generative AI techniques can aid in the development of new materials for hypersonic applications by predicting material properties at extreme temperatures & pressures, helping researchers simulate conditions that are otherwise difficult to recreate experimentally.
 * Aerodynamic Shape Optimization: They can generate designs for hypersonic vehicles with optimized aerodynamic shapes that reduce drag & heat build-up during flight, improving performance & safety.

> **Robotics, Dynamics, & Controls**
* Robot Morphology Generation: Generative AI techniques can generate diverse robot morphologies that could be used to optimize robots for specific tasks, environments, or interactions based on a set of input parameters like size, weight, & required flexibility.
* Control System Simulation: By generating data representing various system failures or unexpected environmental interactions, generative AI techniques can improve the robustness of control systems in robotics through enhanced training of autonomous response algorithms.

> **Solids/Mechanics of Materials**
 * Microstructure Optimization: Generative AI techniques can be used to predict & generate optimal microstructures in materials for specific applications, enhancing properties like strength, ductility, & fatigue resistance.
 * Fracture Prediction: They can also contribute to predicting the behavior of materials under stress, including the formation of cracks & fractures, which is crucial for the assessment of material durability & life span.

These examples illustrate the broad applicability of generative AI techniques in addressing complex & varied problems across different fields of Mechanical Engineering.
