## **1. Supervised Learning in General**

> Supervised learning is a cornerstone methodology in ML, characterized by using labeled data to train algorithms. This approach provides the learning algorithm with input-output pairs, where the outputs are known labels or values corresponding to the inputs. The primary goal of supervised learning is for the algorithm to learn a mapping from inputs to outputs, enabling it to make predictions or decisions based on new, unseen inputs. This is achieved through a process of optimization, where the algorithm iteratively adjusts its parameters to minimize the difference between its predicted outputs & actual labels in the training data.

> In the context of supervised learning, two main types of problems are typically addressed: classification & regression. Classification problems are those in which the outputs are categorical labels, & the objective is to correctly identify to which category a new input belongs. This involves training a classifier model on examples of input data paired with categorical labels. On the other hand, regression problems deal with continuous, real-valued outputs. The task here is to predict a numerical value based on the input data, which involves training a regressor model. Both types of problems leverage the inherent patterns in the labeled training data to learn the function that best approximates the relationship between inputs & outputs.

> Deep learning models, which are a subset of ML models, are particularly prevalent in supervised learning tasks due to their ability to learn complex patterns in large datasets. These models are composed of multiple layers, including convolutional layers, pooling layers, recurrent layers, & dense layers, among others. Each layer transforms its input data with the aim of extracting & progressively refining features, facilitating the model's learning of intricate structures in the data. Deep NN classifiers, designed for classification tasks & their counterparts for regression, are distinguished by their depth & the sophistication of their architectures, enabling them to achieve remarkable performance across a wide range of applications. This focus on multi-layered architectures underscores the evolution of ML models towards increasingly complex systems capable of tackling diverse & challenging problems.

### **1.1. Regression Applications in Mechanical Engineering**

Here are examples of Supervised Learning - Regression in various Mechanical Engineering focus areas.

> **Advanced Manufacturing**
- **Example 1**: Predicting the final product weight in a metal casting process based on variables like melting temperature, pouring speed, & cooling rate. This prediction can help adjust process parameters to ensure the desired product weight.
- **Example 2**: Estimating the surface roughness of machined parts from CNC (Computer Numerical Control) parameters such as feed rate, cutting speed, & depth of cut. Accurate predictions enable the optimization of machining parameters for desired surface quality.

> **Biomechanical Engineering**
- **Example 1**: Predicting the contact forces in knee joints during different activities (walking, running, jumping) based on biomechanical data like muscle forces & joint angles. This can aid in the design of more effective knee braces & rehabilitation programs.
- **Example 2**: Estimating the energy expenditure of individuals from wearable sensor data, including heart rate, motion sensors, & environmental factors. This information is crucial for personalized fitness & health management plans.

> **Fluid Mechanics & Thermal Science**
- **Example 1**: Predicting the pressure drop in pipe systems for various fluid types & flow rates using fluid properties (viscosity, density) & pipe characteristics (diameter, length, roughness). Such predictions are vital for the design & operation of efficient fluid transport systems.
- **Example 2**: Estimating the efficiency of heat exchangers based on operational parameters such as inlet temperature, flow rate, & material properties. Optimizing these parameters can significantly enhance heat exchanger performance.

> **Hypersonic Technologies**
- **Example 1**: Predicting the skin friction coefficient on hypersonic vehicles based on flight conditions, including speed, altitude, & angle of attack. Accurate estimations are crucial for minimizing heat stress on the vehicle's surface.
- **Example 2**: Estimating the shock wave angle in front of hypersonic vehicles as a function of Mach number & vehicle geometry. This is critical for aerodynamic shaping & control surface design.

> **Robotics, Dynamics, & Controls**
- **Example 1**: Predicting the end-effector position of a robot arm in real-time based on joint angles & link lengths. This facilitates accurate control & path planning in robotic systems.
- **Example 2**: Estimating the dynamic response of a robotic system to applied loads using models of the system’s mass, damping, & stiffness. This is essential for designing control algorithms that can ensure stability & performance.

> **Solids/Mechanics of Materials**
- **Example 1**: Calculating the stress distribution in a beam under various loading conditions using the beam's material properties & geometric dimensions. This analysis helps in ensuring that structures remain within safe stress limits.
- **Example 2**: Predicting the crack propagation life in materials under cyclic loading by modeling the material fatigue characteristics. This prediction is vital for maintenance scheduling & life extension of structural components.


### **1.2 Classification Applications in Mechanical Engineering**

Here are examples of Supervised Learning - Classification in various Mechanical Engineering focus areas.

> **Advanced Manufacturing**
- **Example 1**: Classifying the quality of products in a manufacturing line as 'Acceptable,' 'Minor defects,' or 'Major defects' based on visual inspection images. This automated classification helps maintain high-quality standards & reduce manual inspection efforts.
- **Example 2**: Identifying the type of tool wear (e.g., flank wear, crater wear) in CNC machining processes through analysis of acoustic emission signals. Timely classification of tool wear can lead to more efficient maintenance schedules & reduce downtime.

> **Biomechanical Engineering**
- **Example 1**: Classifying the risk level of osteoporosis in individuals based on bone density scan images into categories such as 'Normal,' 'Osteopenia,' or 'Osteoporosis.' This assists healthcare professionals in early diagnosis & intervention.
- **Example 2**: Identifying gait abnormalities from motion capture data, classifying them into specific disorders like 'Parkinson's gait,' 'Hemiplegic gait,' or 'Normal gait.' This classification can improve rehabilitation strategies & monitoring.

> **Fluid Mechanics & Thermal Science**

- **Example 2**: Identifying cooling patterns in thermal imaging data of electronics as 'Adequate,' 'Inefficient,' or 'Overcooling.' Efficient classification helps optimize thermal management systems to prevent overheating & ensure longevity.

> **Hypersonic Technologies**
- **Example 1**: Classifying the thermal protection system (TPS) material suitability for different sections of a hypersonic vehicle based on thermal & mechanical properties. This ensures the vehicle's integrity under extreme conditions.
- **Example 2**: Identifying potential areas of aerodynamic heating on hypersonic vehicles using infrared imaging, classifying them into 'Low risk,' 'Moderate risk,' & 'High risk.' This classification aids in focusing cooling & protective measures where they are most needed.

> **Robotics, Dynamics, & Controls**
- **Example 1**: Classifying the intent of human-robot interaction gestures in collaborative tasks, such as 'Start task,' 'Stop task,' 'Increase speed,' or 'Decrease speed.' This enhances the safety & efficiency of human-robot collaboration.
- **Example 2**: Identifying obstacles in the path of autonomous robots using LIDAR or camera sensors, classifying them into types like 'Static,' 'Dynamic,' 'Navigable,' or 'Non-navigable.' Accurate classification is essential for effective path planning & navigation.

> **Solids/Mechanics of Materials**
- **Example 1**: Classifying the failure mode of materials under stress, such as 'Ductile failure,' 'Brittle failure,' 'Yielding,' or 'Fracture.' This information is crucial for selecting the appropriate materials for specific applications.
- **Example 2**: Identify the type of stress concentration in a structural component from strain gauge data, classifying them as 'No concentration, ' 'Mild concentration,' or 'Severe concentration.' This aids in structural optimization & failure prevention.


## **2. Supervised Learning with Randomly Generated Parameters**

A supervised model can be initiated with random parameters. These parameters are then refined through the use of a learning algorithm, such as Adam or Stochastic Gradient Descent (SGD), to ensure the model precisely matches outputs to inputs. In this lecture, we will explore an example of a supervised classification model using the CIFAR-10 dataset. This dataset comprises 60,000 color images of 32x32 pixels across ten categories: airplanes, cars, birds, cats, deer, dogs, frogs, horses, ships, & trucks, with each category having 6,000 images. The CIFAR-10 dataset is segmented into 50,000 images for training & 10,000 for testing. For additional information, please refer to https://en.wikipedia.org/wiki/CIFAR-10.

```python
#@title Example of Supervised Learning with Randomly Generated Parameters - LeNet
'''
Runtime: GPU $$$

tf.keras.datasets.mnist:              https://www.tensorflow.org/api_docs/python/tf/keras/datasets/mnist
tf.keras.utils.to_categorical:        https://www.tensorflow.org/api_docs/python/tf/keras/utils/to_categorical
tf.keras.Input:                       https://www.tensorflow.org/api_docs/python/tf/keras/Input
tf.keras.layers.Conv2D:               https://www.tensorflow.org/api_docs/python/tf/keras/layers/Conv2D
tf.keras.layers.AveragePooling2D:     https://www.tensorflow.org/api_docs/python/tf/keras/layers/AveragePooling2D
tf.keras.layers.Flatten:              https://www.tensorflow.org/api_docs/python/tf/keras/layers/Flatten
tf.keras.layers.Dense:                https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dense
tf.keras.Model:                       https://www.tensorflow.org/api_docs/python/tf/keras/Model
tf.keras.callbacks.EarlyStopping:     https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/EarlyStopping
tf.keras.callbacks.ReduceLROnPlateau: https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/ReduceLROnPlateau
'''
import tensorflow as tf

# Load MNIST dataset
mnist = tf.keras.datasets.mnist
(datain_tr, dataou_tr), (datain_vl, dataou_vl) = mnist.load_data()
datain_tr, datain_vl = datain_tr / 255.0, datain_vl / 255.0  # Normalize data

# Reshape data to add a single channel dimension
datain_tr = datain_tr[..., tf.newaxis]
datain_vl = datain_vl[..., tf.newaxis]

# Convert class vectors to binary class matrices
dataou_tr = tf.keras.utils.to_categorical(dataou_tr, 10)
dataou_vl = tf.keras.utils.to_categorical(dataou_vl, 10)

# Define the input shape
input_shape = datain_tr.shape[1:]

# Create the LeNet model
inputs = tf.keras.Input(shape=input_shape)
x = tf.keras.layers.Conv2D(6, kernel_size=(5, 5), activation='relu')(inputs)
x = tf.keras.layers.AveragePooling2D(pool_size=(2, 2))(x)
x = tf.keras.layers.Conv2D(16, kernel_size=(5, 5), activation='relu')(x)
x = tf.keras.layers.AveragePooling2D(pool_size=(2, 2))(x)
x = tf.keras.layers.Flatten()(x)
x = tf.keras.layers.Dense(120, activation='relu')(x)
x = tf.keras.layers.Dense(84, activation='relu')(x)
outputs = tf.keras.layers.Dense(10, activation='softmax')(x)

# Construct the model
model = tf.keras.Model(inputs=inputs, outputs=outputs, name="LeNet_MNIST_Model")

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_loss',
                                                  patience=3,
                                                  restore_best_weights=True)
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss',
                                                 factor=0.5,
                                                 patience=2,
                                                 min_lr=0.00001)

# Train the model
history = model.fit(datain_tr, dataou_tr,
                    epochs=10,
                    validation_data=(datain_vl, dataou_vl),
                    callbacks=[early_stopping, reduce_lr])

# Evaluate the model
test_loss, test_acc = model.evaluate(datain_vl, dataou_vl, verbose=2)
print(f'\nTest accuracy: {test_acc}, Test loss: {test_loss}')

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

```python
#@title Example of Supervised Learning with Randomly Generated Parameters - MobileNet
'''
Runtime: GPU $$$

tf.keras.datasets.cifar10:              https://www.tensorflow.org/api_docs/python/tf/keras/datasets/cifar10
tf.keras.utils.to_categorical:          https://www.tensorflow.org/api_docs/python/tf/keras/utils/to_categorical
tf.keras.Input:                         https://www.tensorflow.org/api_docs/python/tf/keras/Input
tf.keras.layers.GlobalAveragePooling2D: https://www.tensorflow.org/api_docs/python/tf/keras/layers/GlobalAveragePooling2D
tf.keras.layers.Dense:                  https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dense
tf.keras.layers.Dropout:                https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dropout
tf.keras.Model:                         https://www.tensorflow.org/api_docs/python/tf/keras/Model
tf.keras.callbacks.EarlyStopping:       https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/EarlyStopping
tf.keras.callbacks.ReduceLROnPlateau:   https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/ReduceLROnPlateau
'''
import tensorflow as tf

# Load CIFAR-10 dataset
cifar10 = tf.keras.datasets.cifar10
(datain_tr, dataou_tr), (datain_vl, dataou_vl) = cifar10.load_data()
datain_tr, datain_vl = datain_tr / 255.0, datain_vl / 255.0  # Normalize data

# Convert class vectors to binary class matrices
dataou_tr = tf.keras.utils.to_categorical(dataou_tr, 10)
dataou_vl = tf.keras.utils.to_categorical(dataou_vl, 10)

# Define input shape
input_shape = datain_tr.shape[1:]

# Create the base model from the MobileNet: https://keras.io/api/applications/mobilenet/
base_model = tf.keras.applications.MobileNet(input_shape=input_shape,
                                             include_top=False,
                                             weights=None)  # Use None to initialize weights randomly

# Freeze the base model
base_model.trainable = False

# Create inputs
inputs = tf.keras.Input(shape=input_shape)

# Add the base model
x = base_model(inputs, training=False)

# Add Global Average Pooling layer
x = tf.keras.layers.GlobalAveragePooling2D()(x)

# Add a dense layer
x = tf.keras.layers.Dense(128, activation='relu')(x)

# Add a dropout layer
x = tf.keras.layers.Dropout(0.5)(x)

# Output layer
outputs = tf.keras.layers.Dense(10, activation='softmax')(x)

# Construct the model
model = tf.keras.Model(inputs, outputs)

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_loss',
                                                  patience=10,
                                                  restore_best_weights=True)
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss',
                                                 factor=0.5,
                                                 patience=5,
                                                 min_lr=0.00001)

# Train the model
history = model.fit(datain_tr, dataou_tr,
                    epochs=25,
                    validation_data=(datain_vl, dataou_vl),
                    callbacks=[early_stopping, reduce_lr])

# Evaluate the model
test_loss, test_acc = model.evaluate(datain_vl, dataou_vl, verbose=2)
print(f'\nTest accuracy: {test_acc}, Test loss: {test_loss}')

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```
