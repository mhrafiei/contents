## **1. PyTorch Library**

[PyTorch](https://pytorch.org/) is an open-source machine learning library developed by [Facebook’s AI Research lab (FAIR)](https://ai.meta.com/research/). It is widely used for deep learning applications & offers a flexible & efficient framework for developing & training neural networks.

Here are some key features & aspects of PyTorch:

* **Dynamic Computational Graphs:** Unlike other deep learning frameworks that use static computation graphs (e.g., TensorFlow before version 2.0), PyTorch uses dynamic computation graphs. This means the graph is built on-the-fly as operations are executed, making it more intuitive & easier to debug.
* **Tensor Computation:** PyTorch provides a multi-dimensional array called Tensor, similar to NumPy arrays, but with additional functionalities for GPU acceleration. Tensors are the core data structure used in PyTorch for all computations.
* **Autograd:** PyTorch has an automatic differentiation library called autograd. It records operations performed on tensors to create a computation graph, which is then used to compute gradients automatically. This is essential for training neural networks using backpropagation.
* **Neural Network Module:** PyTorch offers a high-level neural network API called torch.nn. It provides pre-built layers, loss functions, & optimizers, allowing users to build complex neural network architectures easily.
* **GPU Support:** PyTorch supports computations on both CPUs & GPUs, enabling significant acceleration of deep learning tasks. Moving data & models to the GPU is straightforward with PyTorch.
* **Extensibility & Flexibility:** PyTorch is highly extensible, allowing researchers & developers to create custom layers, loss functions, & other components. Its flexible design makes it suitable for both research & production environments.
* **Community & Ecosystem:** PyTorch has a large & active community, with extensive documentation, tutorials, & third-party libraries. This ecosystem includes various tools & libraries for computer vision, natural language processing, reinforcement learning, & more.
* **Deployment:** PyTorch provides tools for deploying models in production, such as TorchScript, which allows models to be serialized & run independently of Python, & integration with frameworks like ONNX (Open Neural Network Exchange) for cross-framework compatibility.

## **2. Basic PyTorch Components & Operations**

In this section, we will learn & practice basic PyTorch tensors & operations.

```python
#@title Basics of tensors
import torch
import numpy as np

# Creating tensors from numpy arrays
a = np.array([10, 2, 3, 4])
b = torch.tensor(a)
c = torch.Tensor(a)

print(f'b dtype: {b.dtype}')
print(f'c dtype: {c.dtype}')
```

```python
#@title Different ways to create tensors
import torch
import numpy as np

x = torch.empty(2, 2)  # Uninitialized tensor
print(x)

x = torch.zeros(2, 2)  # Tensor filled with zeros
print(x)

x = torch.zeros(2, 2, dtype=torch.int64)  # Tensor filled with zeros of type int64
print(x)

x = torch.ones(2, 2)  # Tensor filled with ones
print(x)

x = torch.arange(0, 18, 2).reshape(3, 3)  # Tensor with values from 0 to 18 with a step of 2, reshaped to 3x3
print(x)

x = torch.linspace(0, 18, 19)  # Tensor with 19 evenly spaced values from 0 to 18
print(x)

x = x.type(torch.int32)  # Change tensor type to int32
print(x)
```

```python
#@title Random tensors
import torch
import numpy as np

x = torch.rand(4, 3)  # Tensor with values from a uniform distribution between 0 & 1
print(x)

x = torch.randn(4, 3)  # Tensor with values from a normal distribution
print(x)

x = torch.randint(0, 10, (2, 2))  # Tensor with random integers between 0 & 10
print(x)

# Tensor operations
x = torch.rand_like(c)  # Create a tensor with the same shape as c, filled with random values
print(c)
```

```python
#@title Tensor slicing & reshaping
import torch
import numpy as np

x = torch.randn(4, 3)
print(x)

y = x[:, 1:2] # Slicing
print(y)

y = x.view(6, 2) # Reshaping
print(y)

y = x.shape
print(y)

y = x.view(3, -1)
print(y)
```

```python
#@title Tensor arithmetic
import torch
import numpy as np

# Creating tensors from numpy arrays
a = np.array([10, 2, 3, 4])
print(a)

b = torch.tensor(a)
print(b)

c = torch.Tensor(a)
print(c)

d = b + c  # Element-wise addition
print(d)

d = torch.add(b, c)  # Using torch.add
print(d)

d = b * c  # Element-wise multiplication
print(d)

d = a.dot(b)  # Dot product
print(d)

a = torch.rand(3, 2)
print(a)

b = torch.rand(2, 3)
print(b)

c = torch.mm(a, b)  # Matrix multiplication
print(c)

print(c.norm())  # Norm of the tensor
print(c.numel())  # Number of elements in the tensor
```

```python
#@title Setting the random seed
import torch
import numpy as np

torch.manual_seed(42)

print(torch.rand(2, 3))
```

```python
#@title Autograd: Automatic differentiation
import torch
import numpy as np

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**4 + x**3 + 3 * x**2 + 5 * x + 1
print(y)

y.backward()  # Compute the gradient of y with respect to x
print(x.grad)  # The gradient (dy/dx)

x = torch.rand(2, 3, requires_grad=True)
y = 3 * x + 2
z = 2 * y**2
o = z.mean()
print(z)
print(o)

o.backward()  # Compute the gradient of o with respect to x
print(x.grad)
```
