## **1. Key Python Libraries and Frameworks for Web Development**

Python is a versatile language that powers a wide range of web development frameworks and libraries, making it an excellent choice for building applications in both MLOps and DevOps contexts. For MLOps, Python web frameworks facilitate the development of Application Programming Interfaces (APIs) and web applications that serve machine learning models in production, providing the infrastructure to interact with these models through user interfaces or programmatically. For DevOps, Python libraries enable automation, CI/CD, monitoring, and deployment, allowing developers to streamline and manage complex infrastructure processes.

### **1.1. Flask**

lask is a lightweight, micro-framework widely used in both MLOps and DevOps due to its simplicity and flexibility. In MLOps, Flask is often used to build REST APIs to serve ML models. By defining API endpoints, Flask allows teams to deploy models as web services that can receive data, process it through the model, and return predictions. For DevOps, Flask can be used to create dashboards, monitoring tools, or small web applications that track and visualize metrics, making it useful for lightweight automation interfaces and custom dashboards.

### **1.2. Django**

Django is a more comprehensive framework that includes built-in authentication, database management, and an admin interface. It’s especially beneficial in MLOps when teams need to build robust, data-driven web applications. For example, Django is suitable for managing user access to machine learning model dashboards, tracking experiment results, and handling large amounts of data. In DevOps, Django’s scalability and built-in security features make it suitable for more extensive applications, like CI/CD management interfaces, where DevOps teams may need more control and integration options.

### **1.3. FastAPI**

FastAPI has gained popularity in recent years due to its high-performance capabilities, making it an excellent choice for building APIs that serve machine learning models in real-time. In MLOps, FastAPI’s speed and asynchronous capabilities allow for efficient model serving, making it well-suited for low-latency applications. Additionally, it has strong support for data validation through Pydantic models, ensuring reliable data input handling for ML services. In DevOps, FastAPI is a good fit for creating efficient APIs that can automate CI/CD workflows or handle monitoring requests quickly.

### **1.4. Dash and Streamlit**

Dash and Streamlit are Python libraries specifically designed for building data-driven web applications and dashboards. In MLOps, these tools allow teams to create interactive dashboards for visualizing model performance, data distributions, and results without extensive front-end coding. Streamlit, in particular, is favored for rapid prototyping of ML model interfaces, enabling data scientists to deploy visualizations of model predictions and metrics quickly. In DevOps, Dash and Streamlit can be used to create visualization tools to monitor infrastructure, performance metrics, and deployment pipelines in a user-friendly way.

### **1.5. Tornado**

Tornado is an asynchronous networking library, ideal for handling a high volume of concurrent connections, making it well-suited for real-time model serving in MLOps. For instance, Tornado can serve multiple ML predictions simultaneously without blocking, which is advantageous for applications requiring fast, scalable response times. In DevOps, Tornado can be used in systems that handle real-time logging or monitoring, offering a low-latency solution for tracking system health and performance.

### **1.6. Requests and HTTPx**

While not web frameworks, these libraries are essential in DevOps and MLOps for making HTTP requests to APIs. Requests is often used to interact with REST APIs within CI/CD pipelines, retrieve model outputs, or make calls to various microservices. HTTPx, with its asynchronous support, provides a non-blocking option for making concurrent requests, which is particularly useful in workflows that involve handling data from multiple endpoints simultaneously.

### **1.7. Celery**

Although Celery is primarily a task queue, it’s commonly paired with Flask, Django, or FastAPI to handle background jobs, which are essential in both MLOps and DevOps. For MLOps, Celery can offload data preprocessing, model training, or batch predictions to background tasks, freeing up resources for real-time operations. In DevOps, Celery helps with automating repetitive tasks, scheduling deployments, or managing long-running jobs within pipelines.


## **2. Introduction to Flask**

Flask is a lightweight and flexible web framework for Python, designed to make it easy to develop web applications and APIs. Often referred to as a “micro-framework,” Flask provides only the essentials needed to get a web application running, allowing developers to build projects quickly and scale them as needed. Despite its minimalistic design, Flask is highly extensible, with many plugins and libraries available to add features such as database integration, form handling, authentication, and more.

> **Key Features of Flask**
1.	Simplicity and Flexibility: Flask is easy to set up and use, making it ideal for developers who need to build applications quickly. It’s often chosen for small to medium-sized projects but can also be scaled for larger applications.
2.	Modular Design: Flask follows a modular approach, allowing developers to organize code into reusable components. This makes it easier to manage larger applications as they grow.
3.	RESTful Request Handling: Flask is particularly well-suited for building REST APIs. It enables developers to define routes (URL paths) and specify what actions to perform when a user or client makes a request to those paths.
4.	Jinja2 Templating Engine: Flask uses Jinja2, a templating engine that allows developers to create HTML templates with dynamic content. This is helpful for creating interactive and data-driven web applications.
5.	Extensibility: Flask’s “micro” nature means it does not come bundled with many features by default. However, there are many Flask extensions available to add functionality, such as database connectors, authentication systems, and form handling.

> **Common Use Cases for Flask**
* Building REST APIs: Flask is widely used for building APIs to serve data, interact with databases, or connect to machine learning models. Its simplicity and flexibility make it a popular choice for exposing ML models to applications in MLOps workflows.
* Prototyping and Small Applications: Flask’s minimal setup makes it ideal for quickly prototyping web applications or deploying proof-of-concept applications.
* Microservices: Flask’s small footprint makes it a great choice for microservices architecture, where applications are composed of small, loosely coupled services.
* Serving Machine Learning Models: In MLOps, Flask is commonly used to serve ML models as APIs, allowing other applications to access the model’s predictions. Flask can load a trained model, take input data through API calls, and return predictions in real-time or batch mode.

````python
#@title A Simple Web App

# Initialize a Flask web application.
# Flask is a web framework that allows you to create web applications with Python.
# The `Flask` class is the core of the framework and is used to define routes (URL endpoints) and handle requests.
# - `__name__` tells Flask where to look for templates and static files, which helps organize the application.
# - For more details on setting up a Flask app, see the Flask documentation: https://flask.palletsprojects.com/
app = Flask(__name__)

# Define a route for the home page of the web application.
# - The `@app.route("/")` decorator maps the home URL ("/") to the `home` function.
# - When a user visits this URL, the `home` function is executed.
# - In this case, it returns a simple message: "Hello, World! This is your Flask app running in Google Colab!"
@app.route("/")
def home():
    # Return a basic message to confirm the app is running.
    return "Hello, World! This is your Flask app running in Google Colab!"

# Open an ngrok tunnel on port 5000 (the default port used by Flask).
# - The `ngrok.connect(5000)` function opens a secure public URL (tunnel) to port 5000 on the local machine,
#   allowing external users to access the app at this public URL.
# - This is useful for exposing your app to the internet while running it locally, especially for testing and demos.
# - Each time you run this cell, ngrok provides a new, unique public URL.
public_url = ngrok.connect(5000)

# Print the public URL created by ngrok.
# - This URL can be shared with others to access the application remotely.
# - If you're running the app in Colab, copy this URL to access the app in a new browser tab.
print(" * ngrok tunnel available at:", public_url)

# Run the Flask app on port 5000.
# - The `app.run()` function starts the Flask development server, making the application available locally.
# - By default, Flask runs on port 5000, but you can specify a different port if needed.
# - In Colab, Flask runs on localhost (127.0.0.1), which is accessible only within the Colab environment.
#   However, with the ngrok tunnel, you can access it externally.
app.run(port=5000)
````

### **2.2. Basic HTML**

HTML, or HyperText Markup Language, is the standard language used to create and structure content on the web. It defines the layout, structure, and elements of a webpage. HTML is a markup language, meaning it uses “tags” to enclose different parts of content, which helps web browsers interpret and display the content as intended.

> **Key Aspects of HTML**
1. **Tags and Elements:** HTML uses tags like `<html>`, `<head>`, `<body>`, `<h1>`, `<p>`, etc., to structure different sections of a webpage. Tags are typically paired, with an opening tag (like `<p>`) and a closing tag (like `</p>`), surrounding content to define it as a paragraph, header, link, etc.
><pre><code>&lt;h1&gt;This is a heading&lt;/h1&gt;
&lt;p&gt;This is a paragraph.&lt;/p&gt;
</code></pre>
>2. **Attributes:** HTML tags can have attributes that provide additional information. For instance, the <img> tag, used to insert images, has an src attribute to specify the image source, and an alt attribute to provide alternative text.
><pre><code>&lt;img src=&quot;image.jpg&quot; alt=&quot;A beautiful scenery&quot;&gt;
</code></pre>
> 3. **Document Structure:** A typical HTML document has a structure that includes a <head> section for metadata, styles, and scripts, and a <body> section for content that appears on the page.
<!DOCTYPE html>
><pre><code>&lt;!DOCTYPE html&gt;
&lt;html&gt;
    &lt;head&gt;
        &lt;title&gt;My Webpage&lt;/title&gt;
    &lt;/head&gt;
    &lt;body&gt;
        &lt;h1&gt;Welcome to My Webpage&lt;/h1&gt;
        &lt;p&gt;This is some content on my page.&lt;/p&gt;
    &lt;/body&gt;
&lt;/html&gt;
</code></pre>
> 4. **Hyperlinks:** HTML allows linking between different pages or websites using the `<a>` tag. This is where “HyperText” comes into play—HTML enables the creation of a network of interlinked pages.
><pre><code>&lt;a href=&quot;https://www.example.com&quot;&gt;Visit Example&lt;/a&gt;
</code></pre>

````python
#@title A Simple Flask App with Jinja

# Initialize the Flask web application.
# - Flask allows us to create web applications in Python by defining routes and handling requests.
# - The `Flask` class is initialized with `__name__` to help Flask locate resources, templates, and static files.
app = Flask(__name__)

# Define a simple HTML template using Jinja2 templating.
# - Jinja2 is the templating engine used by Flask to render HTML templates dynamically.
# - This template displays a personalized greeting based on the value of the `name` variable.
# - Flask will replace `{{ name }}` with the actual value of the variable provided.
# - `{% if %} ... {% else %} ... {% endif %}` is Jinja syntax for conditional statements, allowing
#   customization of the HTML output based on conditions.
#   - Jinja documentation: https://jinja.palletsprojects.com/en/3.0.x/templates/
html = """
<html>
    <body>
    <h1>Hello, {{ name }}!</h1>  <!-- Display the name variable passed from Flask -->
    {% if name == "Alice" %}  <!-- If the name is Alice, display a custom message -->
        <p>Welcome, Alice! Nice to see you again.</p>
    {% else %}  <!-- If the name is not Alice, display a generic welcome message -->
        <p>Welcome, Guest!</p>
    {% endif %}
    </body>
</html>
"""

# Define a route for the home page of the application.
# - The `@app.route("/")` decorator maps the home URL ("/") to the `home` function.
# - When a user navigates to this URL, the `home` function executes and returns an HTML response.
@app.route("/")
def home():
    # Define the `name` variable with the value "Alice".
    # This variable is passed to the HTML template, allowing the content to change dynamically.
    # In a more complex app, this value could be dynamically set based on user input or database queries.
    name = "Alice"

    # Render the HTML template using `render_template_string`.
    # - `render_template_string` takes an HTML template string and any variables (like `name`) to pass to the template.
    # - This method allows for inline HTML rendering. For larger applications, Flask typically loads HTML files from
    #   the "templates" folder using `render_template`.
    return render_template_string(html, name=name)

# Open a public ngrok tunnel on port 5000 (Flask's default port).
# - `ngrok.connect(5000)` establishes a secure public URL that exposes the local server on port 5000 to the internet.
# - This is useful for testing and sharing the app with others.
public_url = ngrok.connect(5000)

# Print the public ngrok URL, which can be used to access the application remotely.
# - The printed URL can be copied into a browser to view the app.
print(" * ngrok tunnel available at:", public_url)

# Run the Flask application on port 5000.
# - `app.run()` starts the Flask development server, making the app accessible locally.
# - Since we opened an ngrok tunnel, the app is also accessible online at the ngrok URL.
app.run(port=5000)
````

````python
#@title A Simple Flask App with GET & POST

# Initialize the Flask application.
# Flask provides the tools to create a web server that can respond to different types of requests.
app = Flask(__name__)

# Define a route for the home page, supporting both GET and POST requests.
# - GET request: The default method when the page is accessed, used to retrieve data.
# - POST request: Used when the form is submitted, sending user input (name) to the server.
@app.route("/", methods=["GET", "POST"])
def home():
    # Check if the form was submitted via POST.
    # - `request.method == "POST"` checks if the current request is a POST request.
    # - If it’s a POST request, `request.form.get("name")` retrieves the user's input from the form.
    # - If it’s a GET request (page load without form submission), `name` is set to `None`.
    name = request.form.get("name") if request.method == "POST" else None

    # Render the HTML template, passing the name variable to the template.
    # - `render_template_string(html, name=name)` injects the value of `name` into the template.
    # - If `name` is `None`, the form is displayed. If `name` has a value, a greeting is shown.
    return render_template_string(html, name=name)

# Open a public ngrok tunnel on port 5000 (Flask’s default port).
# - `ngrok.connect(5000)` creates a secure public URL for the application.
# - This allows external access to the app, which is helpful for testing and demos.
public_url = ngrok.connect(5000)

# Print the public URL generated by ngrok.
# - This URL can be accessed in a browser to interact with the app.
print(" * ngrok tunnel available at:", public_url)

# Run the Flask application on port 5000.
# - `app.run()` starts the Flask server, making the app accessible locally and via the ngrok URL.
app.run(port=5000)
````

### **2.7 Basic CSS**

CSS, or Cascading Style Sheets, is a stylesheet language used for describing the presentation of a document written in HTML or XML. It controls the visual layout and design of web pages, allowing developers to change things like colors, fonts, spacing, and layout across a website.

CSS does not use “tags” and “attributes” in the same way HTML does, but it does have similar structures and concepts, like selectors, properties, and values. Let’s break these down to understand their roles in CSS:

> 1. **Selectors (similar to HTML tags)**
* In CSS, selectors are used to target HTML elements that you want to style. This is somewhat similar to how HTML uses tags to define different elements.
* There are different types of selectors in CSS:
    * Element selectors: Target HTML elements directly, like `h1`, `p`, or `div`. For example, styles all `<p>` elements:
        * `p { color: blue; }`
    * Class selectors: Target elements by their class attribute, defined with a dot(`.`) followed by the class name. For example, styles all elements with class="myClass":
        * `.myClass { font-size: 20px; }`
    * ID selectors: Target elements by their id attribute, defined with a hash (`#`) followed by the ID name. For example, styles the element with id="myId":
        * `#myId { background-color: yellow; }`
    * Attribute selectors: Target elements based on an attribute they have, similar to HTML attributes. For example, styles all `<a>` elements with target="_blank
        * `a[target="_blank"] { color: red; }`
	* Pseudo-selectors: Target elements based on their state or position (e.g., :hover, :first-child). For example, changes color when hovered over:
        * `button:hover { background-color: green; }`
> 2. **Properties (similar to HTML attributes)**
* In CSS, properties are similar to HTML attributes, but instead of defining characteristics of elements, they specify the style that should be applied.
* Each CSS rule has one or more properties, like color, font-size, margin, and background.
    * For example, changes the text color:
        * `color: blue;`
    * For example, sets the font size:
        * `font-size: 16px;`
> 3. **Values (similar to HTML attribute values)**
* Values in CSS are the specific settings given to properties, similar to how HTML attributes have specific values.
* Examples:
    * Set the color property to the value "red":
        * `color: red;`
    * Set the font-size property to the value "20px":
        * `font-size: 20px;`




> **Example of CSS Structure:** A complete CSS rule, called a declaration block, includes a selector, properties, and values (`/* */` is used for commenting in a `.css` file):
```
h1 {
    color: blue;          /* Property: color, Value: blue */
    font-size: 24px;      /* Property: font-size, Value: 24px */
    margin: 10px;         /* Property: margin, Value: 10px */
}
```

````python
#@title Helper Pexels API Search Function for One Page For Example 1
# Function to search for photos on a single page from the Pexels API based on a query.
# - `query`: The search term, such as "sunset" or "city".
# - `page_number`: Specifies which page of search results to retrieve.
# - This function returns a dictionary with photo data if successful, or None if an error occurs.
def search_one_page(query: str, page_number: int):
    try:
        # Validate that `page_number` is a positive integer to prevent invalid API requests.
        if page_number < 1:
            raise ValueError("Page number must be a positive integer.")

        # Attempt to fetch photos from the Pexels API.
        # `pexel.search_photos` is a hypothetical function that connects to Pexels' API.
        # Parameters include:
        #   - `query`: Main keyword for image search.
        #   - `page`: Specifies the page of results to fetch.
        #   - `per_page`: Limits the number of results on each page to 80 (maximum allowed by Pexels).
        response = pexel.search_photos(query=query,
                                       orientation='',
                                       size='',
                                       color='',
                                       locale='',
                                       page=page_number,
                                       per_page=80)

        # Ensure that the response contains the expected 'photos' key with image data.
        if 'photos' not in response:
            raise KeyError("The response does not contain 'photos' key.")

        return response  # Return the response containing photos for further processing.

    # Handle various exceptions that may occur during API call or processing.
    except RequestException as e:
        print(f"Network error occurred while fetching photos for query '{query}' on page {page_number}: {e}")
        return None  # Return None if there's a network error.
    except ValueError as e:
        print(f"Invalid value error: {e}")
        return None  # Return None if a value error occurs.
    except KeyError as e:
        print(f"Unexpected response structure: {e}")
        return None  # Return None if the response structure is unexpected.
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return None  # Return None for any other unforeseen errors.
````

````python
#@title Helper Pexels API Search Function for Multi Page For Example 1
# Function to search for photos across multiple pages from Pexels and collect image URLs.
# - `keyword`: The main search term, such as "ocean" or "architecture".
# - `total_pages`: The number of pages to search, defaulting to 3.
# - Returns a random image URL from the collected URLs, or None if no images are found.
def search_pexel_photos(keyword: str, total_pages: int = 3):
    try:
        # Validate that `total_pages` is a positive integer.
        if total_pages < 1:
            raise ValueError("Total pages must be a positive integer.")

        # Gather search results from multiple pages using list comprehension.
        # Calls `search_one_page` for each page in the specified range.
        search_results = [search_one_page(keyword, page_number) for page_number in range(1, total_pages + 1)]

        # Initialize an empty list to store URLs of images.
        urls = []

        # Process each result page obtained from the search.
        for search_result in search_results:
            if search_result is None:  # Skip processing if the page search failed.
                continue

            # Loop over each photo in the 'photos' key of the search result.
            for photo in search_result.get('photos', []):
                # Extract the `src` dictionary, which contains URLs of the image in various resolutions.
                src = photo.get('src')
                if src:
                    # Get the list of resolution keys and choose the highest available resolution.
                    keys = list(src.keys())
                    if keys:
                        url = src.get(keys[-1])  # Access the highest-resolution URL.
                        if url:
                            urls.append(url)  # Append the URL to the list if it exists.

        # If URLs were found, shuffle them and return a random one.
        if urls:
            random.shuffle(urls)  # Introduces randomness by shuffling the URLs.
            return urls[0]  # Return the first URL after shuffling.

        # If no URLs are found after processing all pages, print a message.
        print("No URLs found for the given keyword.")
        return None

    # Handle potential errors during the multi-page search.
    except ValueError as e:
        print(f"Invalid value error: {e}")
        return None  # Return None for invalid parameter errors.
    except Exception as e:
        print(f"An unexpected error occurred during search: {e}")
        return None  # Return None for any other unforeseen errors.
````

````python
#@title Example 2: Text-To-Speech Generation

# Initialize the Flask application
# - This app will provide a simple web interface to input text, convert it to speech, and play or download the resulting audio.
app = Flask(__name__)

# Define the main route for the web application, allowing both GET and POST methods
# - `@app.route("/", methods=["GET", "POST"])` sets up a route for the root URL ("http://<ngrok_url>/").
# - The `methods` argument specifies that this route supports both GET (initial page load) and POST (form submission).
@app.route("/", methods=["GET", "POST"])
def home():
    # Initialize `audio_file_path` to None. This variable will store the path to the generated audio file.
    # - If no audio file is generated, this remains None, and the HTML will not display an audio player.
    audio_file_path = None

    # Check if the form is submitted via POST
    # - `request.method == "POST"` ensures that audio generation only occurs on form submission.
    if request.method == "POST":
        # Retrieve the text entered by the user in the form
        # - `request.form.get("text")` accesses the form field named "text" where the user inputs text for speech conversion.
        text = request.form.get("text")

        # Convert the input text to speech using the Google Text-to-Speech (gTTS) library
        # - `gTTS(text, lang='en')` initializes gTTS with the provided text and sets the language to English ('en').
        # - gTTS is a popular Python library that interfaces with Google's text-to-speech service.
        # - Documentation: https://gtts.readthedocs.io/
        tts = gTTS(text, lang='en')

        # Define the path to save the generated audio file
        # - Here, "static/speech.mp3" refers to a file within the `static` directory, accessible by the web app.
        audio_file_path = "static/speech.mp3"

        # Save the audio file to the specified path
        # - `tts.save(audio_file_path)` writes the audio data to an MP3 file, making it accessible to the HTML template.
        tts.save(audio_file_path)

    # Render the HTML template, passing the `audio_file_path` if an audio file was generated
    # - `render_template_string(html, audio_file_path=audio_file_path if audio_file_path else None)`
    #   injects `audio_file_path` into the HTML template if an audio file was created; otherwise, it passes None.
    # - This conditional handling ensures that the audio player is only shown if an audio file exists.
    return render_template_string(html, audio_file_path=audio_file_path if audio_file_path else None)

# Open an ngrok tunnel to make the app accessible on the internet
# - `ngrok.connect(5000)` opens a secure tunnel on port 5000, the default port for Flask applications.
# - This generates a public URL that can be accessed from any internet-connected device.
public_url = ngrok.connect(5000)

# Print the public URL provided by ngrok, which allows users to access the app externally
print(" * ngrok tunnel available at:", public_url)

# Ensure the `static` directory exists for storing generated audio files
# - `os.makedirs("static", exist_ok=True)` creates the `static` directory if it doesn’t already exist.
# - Flask uses the `static` folder to serve static files like images, CSS, JavaScript, and audio.
os.makedirs("static", exist_ok=True)

# Start the Flask application on port 5000
# - `app.run(port=5000)` launches the Flask development server.
# - With ngrok’s tunnel, this server is accessible via the public URL provided.
app.run(port=5000)
````

````python
#@title Example 3: A Fancier Text-To-Speech Generation

# Initialize the Flask application
# - This app allows users to enter text, convert it to speech, and play/download the generated audio.
app = Flask(__name__)

# Define the main route for the application, supporting both GET and POST requests
# - `@app.route("/", methods=["GET", "POST"])` sets up the root route, enabling the form and audio playback.
@app.route("/", methods=["GET", "POST"])
def home():
    # Initialize `audio_file_path` to store the path to the generated audio file
    # - This variable will hold the path if an audio file is created; otherwise, it remains None.
    audio_file_path = None

    # Initialize `submitted_text` to store the text entered by the user
    # - This variable allows the application to display the user’s input back on the page.
    submitted_text = None

    # Handle form submission with POST request
    # - This block only executes if the form was submitted (method is POST).
    if request.method == "POST":
        # Retrieve the text input from the form
        # - `request.form.get("text")` captures the user’s input from the form field named "text".
        submitted_text = request.form.get("text")

        # Convert the submitted text to speech using gTTS (Google Text-to-Speech)
        # - `gTTS(submitted_text, lang='en')` creates a TTS object with the input text in English.
        # - gTTS is a Python library that sends text to Google’s TTS service and returns an audio response.
        # - Documentation: https://gtts.readthedocs.io/en/latest/
        tts = gTTS(submitted_text, lang='en')

        # Define the path to save the generated audio file in the static folder
        # - "static/speech.mp3" specifies the file location; this folder serves static content like audio files.
        audio_file_path = "static/speech.mp3"

        # Save the audio file to the specified path
        # - `tts.save(audio_file_path)` writes the TTS audio to an MP3 file for playback.
        tts.save(audio_file_path)

    # Render the HTML template, passing any generated audio file and the submitted text for display
    # - `render_template_string(html, audio_file_path=audio_file_path if audio_file_path else None, submitted_text=submitted_text)`
    #   passes both `audio_file_path` and `submitted_text` to the template.
    # - If no audio file was created, `audio_file_path` is set to None to prevent rendering the audio player.
    return render_template_string(html, audio_file_path=audio_file_path if audio_file_path else None, submitted_text=submitted_text)

# Open an ngrok tunnel to allow external access to the app on port 5000
# - `ngrok.connect(5000)` establishes a public URL for the local Flask server.
public_url = ngrok.connect(5000)

# Print the public ngrok URL to the console for easy access
# - This URL allows external devices to access the app from any browser.
print(" * ngrok tunnel available at:", public_url)

# Ensure the `static` directory exists to store audio files
# - `os.makedirs("static", exist_ok=True)` creates the folder if it doesn’t already exist.
# - The static folder is designated by Flask to hold static files like images, audio, etc.
os.makedirs("static", exist_ok=True)

# Run the Flask application on port 5000
# - `app.run(port=5000)` starts the Flask development server, listening for incoming requests on port 5000.
# - Ngrok's tunnel forwards requests to this port.
app.run(port=5000)
````
