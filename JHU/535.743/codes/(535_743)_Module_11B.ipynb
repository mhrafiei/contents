```python
# @title Write to PDF

# Installing the required libraries using pip.
# 'PyPDF2' is for working with PDFs (if needed for future manipulation),
# and 'reportlab' is for generating PDF documents programmatically.
!pip install PyPDF2 reportlab

# Import necessary modules from the ReportLab library:
# 'letter' defines the standard letter-sized page (8.5 x 11 inches).
# 'canvas' is used to draw text, shapes, and other elements on the PDF.
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas

# Step 1: Read the content from a text file
# Using 'with' ensures the file is properly closed after reading.
with open('/content/drive/MyDrive/data_535743/nlp/files/epme.txt', 'r') as f:
    # 'f.read()' reads the entire content of the text file and stores it in the variable 'content'.
    content = f.read()

# Step 2: Define a helper function to wrap the text into multiple lines.
# This function ensures the text fits within a specified maximum width when writing to the PDF.
def wrap_text(c, text, max_width):
    lines = []  # List to store each line of text
    words = text.split()  # Split the entire text into individual words

    current_line = ""  # Temporary storage for the current line being built
    for word in words:
        # Check if adding the current word would exceed the maximum line width
        if c.stringWidth(current_line + " " + word) < max_width:
            # If not, add the word to the current line
            current_line += " " + word
        else:
            # If yes, add the current line to the list of lines and start a new line
            lines.append(current_line.strip())
            current_line = word
    # Add the last line (if any) after the loop
    lines.append(current_line.strip())
    return lines

# Step 3: Define a function to create a new PDF with wrapped text.
def create_pdf(filename, text):
    # Create a new canvas (PDF) with the specified file name and page size (letter).
    c = canvas.Canvas(filename, pagesize=letter)
    width, height = letter  # Get the dimensions of the page

    # Set the font to Helvetica, size 12 for the text in the PDF.
    c.setFont("Helvetica", 12)

    # Define the margins and initial text position.
    margin_left = 100  # Left margin
    margin_right = width - 100  # Right margin
    y_position = height - 100  # Start drawing text 100 points below the top of the page

    # Calculate the maximum width for the text to fit within the margins.
    max_width = margin_right - margin_left

    # Wrap the text into lines that fit within the specified width.
    lines = wrap_text(c, text, max_width)

    # Step 4: Draw each line of text on the PDF.
    for line in lines:
        # Draw the current line of text at the specified position on the page.
        c.drawString(margin_left, y_position, line)
        # Move the y-position down for the next line (14 points line height).
        y_position -= 14

        # If the y-position goes below 40 points (end of the page), start a new page.
        if y_position < 40:
            c.showPage()  # Create a new page
            y_position = height - 100  # Reset y-position for the new page
            c.setFont("Helvetica", 12)  # Reset the font for the new page

    # Step 5: Save the PDF to the file.
    c.save()

# Step 6: Specify the output file name.
filename = "output.pdf"

# Step 7: Call the function to create the PDF with the content from the text file.
create_pdf(filename, content)
```

## **2. Regular Expressions (regex)**

Regular expressions (regex) in Python are a powerful tool for working with strings, allowing you to define search patterns that can match, locate, and manipulate text. Python provides a built-in module called re to work with regular expressions. A regular expression is essentially a sequence of characters that forms a search pattern. This pattern can be used for various operations such as searching for substrings, validating input, or replacing parts of a string. The power of regex comes from the flexibility it provides in defining complex search patterns with just a few symbols.


Different characters, like digits or individual strings, have specific codes representing them. These codes can be used to construct a pattern string. It’s important to note the frequent use of the backslash () in these patterns. To avoid confusion, when defining a regular expression pattern in Python, we use the format `r'mypattern'`.
```python
# @title Example of Regular Expressions

# Import the 're' module, which provides support for regular expressions (regex) in Python.
import re

# A multi-line string (text block) that contains course-related information.
# This is the text where we'll search for the pattern 'AI'.
text = """
This course, i.e., 535.743, offers an applied understanding of artificial intelligence (AI) and machine learning (ML).
535.743 covers topics such as machine learning models, Python essentials, and cloud-based platforms, as well as specialized subjects such as object detection, generative models, AI security, and natural language processing (NLP).
Through a blend of theoretical instruction and hands-on exercises, in 535.743, students will master the algorithms, methodologies, and tools required to solve complex engineering challenges using AI.
Students will develop ML models using TensorFlow and limited PyTorch, object detection techniques such as SSD (Single Shot Detector) and YOLO (You Only Look Once), generative models such as generative adversarial networks (GANs), and various NLP implementations.
They will also learn how to secure AI systems against adversarial attacks and complete exercises on application programming interfaces (APIs), cloud computing, and web development frameworks such as Flask. Emphasis will be placed on real-world applications and state-of-the-art technologies to equip students with the skills required to implement AI solutions effectively and securely in various engineering contexts.
Students need to complete the pre-requisite for this course, 535.742 successfully.
"""

# The pattern we're looking for in the text. In this case, 'AI'.
# This pattern matches the literal string "AI".
pattern = 'AI'

# Search for the first occurrence of the pattern in the text using 're.search()'.
# This function returns a match object if the pattern is found, or 'None' if not.
match_case = re.search(pattern, text)

# Print the match object, which contains details about the found pattern.
# If no match is found, it will print 'None'.
print(match_case)

# Print the start and end positions (span) of the match.
# The 'span()' method returns a tuple containing the start and end indices of the match.
print(match_case.span())

# Print the start position of the match using 'start()'.
# This method returns the index where the match begins in the text.
print(match_case.start())

# Print the end position of the match using 'end()'.
# This method returns the index where the match ends in the text.
print(match_case.end())

# Print the actual string that was matched using 'group()'.
# This method returns the part of the string that matched the pattern.
print(match_case.group())

# Print a separator line of 20 equal signs for better readability in the output.
print("=" * 20)

# Find all occurrences of the pattern in the text using 're.findall()'.
# This function returns a list of all non-overlapping matches of the pattern.
match_cases = re.findall(pattern, text)

# Print the list of all matches found in the text.
print(match_cases)

# Print another separator line.
print("=" * 20)

# Use 're.finditer()' to find all matches of the pattern in the text and return an iterator of match objects.
# This allows us to loop through all matches and examine their positions.
for match_case in re.finditer(pattern, text):
    # For each match, print the start and end positions using 'span()'.
    print(match_case.span())
```

```python
# @title Identifiers for Characters in Patterns

# Import the 're' module, which provides support for working with regular expressions (regex) in Python.
import re

# A multi-line string (text block) containing course-related information.
# This text will be used to search for patterns using regular expressions.
text = """
This course, i.e., 535.743, offers an applied understanding of artificial intelligence (AI) and machine learning (ML).
535.743 covers topics such as machine learning models, Python essentials, and cloud-based platforms, as well as specialized subjects such as object detection, generative models, AI security, and natural language processing (NLP).
Through a blend of theoretical instruction and hands-on exercises, in 535.743, students will master the algorithms, methodologies, and tools required to solve complex engineering challenges using AI.
Students will develop ML models using TensorFlow and limited PyTorch, object detection techniques such as SSD (Single Shot Detector) and YOLO (You Only Look Once), generative models such as generative adversarial networks (GANs), and various NLP implementations.
They will also learn how to secure AI systems against adversarial attacks and complete exercises on application programming interfaces (APIs), cloud computing, and web development frameworks such as Flask. Emphasis will be placed on real-world applications and state-of-the-art technologies to equip students with the skills required to implement AI solutions effectively and securely in various engineering contexts.
Students need to complete the pre-requisite for this course, 535.742 successfully.
"""

# Define the first pattern using regex.
# '\d' matches any digit (0-9), and '.' matches any character.
# '\d\d\d.\d\d\d' matches exactly 3 digits, followed by any character, and then exactly 3 more digits.
pattern = '\d\d\d.\d\d\d'

# Use 're.search()' to find the first occurrence of the pattern in the text.
# This returns a match object if the pattern is found, or 'None' if not.
match_case = re.search(pattern, text)

# Iterate over all matches in the text using 're.finditer()'.
# 'finditer()' returns an iterator yielding match objects for all non-overlapping matches.
for match_case in re.finditer(pattern, text):
    # Get the start and end positions of the current match using 'span()'.
    start, end = match_case.span()
    # Print the start and end positions along with the matched string.
    # The f-string formatting ensures that the output is aligned.
    print(f"Span: ({start:<5}, {end:<5}), Match: {match_case.group()}")

# Print a separator line of 50 equal signs for better readability in the output.
print("=" * 50)

# Define another pattern using curly braces for repetition.
# '\d{3}' matches exactly 3 digits, and '.' matches any character.
# '\d{3}.\d{3}' is equivalent to the previous pattern, but uses {3} to specify 3 repetitions.
pattern = '\d{3}.\d{3}'

# Again, use 're.search()' to find the first occurrence of this new pattern in the text.
match_case = re.search(pattern, text)

# Iterate over all matches in the text using 're.finditer()' for this new pattern.
for match_case in re.finditer(pattern, text):
    # Get the start and end positions of the current match using 'span()'.
    start, end = match_case.span()
    # Print the start and end positions along with the matched string.
    print(f"Span: ({start:<5}, {end:<5}), Match: {match_case.group()}")

# Print another separator line of 50 equal signs.
print("=" * 50)

# Demonstrate exclusion in regex using the caret (^) symbol inside square brackets.
# '[^\d]' matches any character that is not a digit.
# 're.findall()' returns all characters from the first 100 characters of the text that are not digits.
# 'join()' combines these characters into a single string without digits.
print(''.join(re.findall(r'[^\d]', text[:100])))
```

## **3. Review of spaCy Library**

SpaCy is an open-source Python library for advanced natural language processing (NLP). It is designed to help developers build applications that process and understand large amounts of text, providing fast and efficient tools for common NLP tasks such as tokenization, part-of-speech tagging, named entity recognition (NER), dependency parsing, and more.

> **Key Features of spaCy**
* Efficient and Fast: spaCy is built with a focus on performance and efficiency. It is designed to process large volumes of text quickly while maintaining accuracy, making it suitable for production environments.
* Pre-trained Models: spaCy offers pre-trained models in multiple languages that can be used out of the box for tasks like named entity recognition, part-of-speech tagging, and more. These models have been trained on large corpora and provide state-of-the-art accuracy.
* Easy to Use: spaCy has a user-friendly API, making it easy for both beginners and experienced developers to work with complex NLP tasks. It integrates smoothly with deep learning libraries like PyTorch and TensorFlow.
* Tokenization: It breaks down text into individual tokens (words, punctuation, etc.), making it easier to analyze text at a granular level.
* Named Entity Recognition (NER): spaCy can identify and classify named entities in a text, such as people, organizations, locations, dates, and more.
* Part-of-Speech Tagging (POS): It labels each word in a sentence with its corresponding part of speech (noun, verb, adjective, etc.).
* Dependency Parsing: spaCy can analyze the grammatical structure of a sentence, determining the relationships between words (such as subject, object, etc.).
* Word Vectors: spaCy supports word embeddings, which represent words as vectors in a continuous vector space. This helps capture the semantic meaning of words based on their context.
* Customization and Extensions: Developers can add custom pipelines, functions, and annotations to extend spaCy’s capabilities and tailor it to specific needs.
* Text Preprocessing: spaCy is commonly used for preprocessing text data in machine learning tasks, where it prepares text for tasks like sentiment analysis, text classification, or chatbots

## **4. Review of NLTK**

NLTK (Natural Language Toolkit) is a comprehensive Python library designed for working with human language data (text). It is widely used for natural language processing (NLP) tasks and provides easy-to-use interfaces to over 50 corpora and lexical resources, along with tools for text processing, tokenization, stemming, tagging, parsing, and semantic reasoning.

NLTK is especially popular for teaching and research purposes, offering both high-level abstractions for common NLP tasks and low-level control for custom implementations.

> **Key Features of NLTK**
* Text Processing: NLTK provides functions for text processing tasks such as tokenization (splitting text into words or sentences), stemming (reducing words to their base or root form), and lemmatization (normalizing words based on their meaning).
* Corpora and Lexical Resources: NLTK includes interfaces to access large corpora (collections of texts), including the Brown Corpus, Gutenberg Corpus, WordNet (a lexical database), and more. These resources are useful for text analysis and training models.
* Tokenization: NLTK has built-in tools to break down text into smaller components, such as sentences or words. For example, sentence tokenization splits text into individual sentences, while word tokenization divides sentences into words.
* Part-of-Speech (POS) Tagging: It provides the ability to label each word in a sentence with its part of speech (e.g., noun, verb, adjective).
* Named Entity Recognition (NER): NLTK can recognize named entities (such as people, places, and organizations) in text using built-in classifiers.
* Text Classification: You can perform classification tasks, such as determining the sentiment of a text (positive, negative, neutral) or categorizing it into specific classes.
* Parsing and Syntax Trees: NLTK supports syntactic analysis of sentences, enabling you to parse sentences and represent them as syntax trees.
* Stemming and Lemmatization: Stemming reduces words to their base form (e.g., “running” to “run”), while lemmatization is more sophisticated, reducing words based on their meaning (e.g., “better” to “good”).
* Custom Models: NLTK allows you to train custom models using supervised and unsupervised machine learning algorithms for text processing tasks.
* Education-Oriented: NLTK has a strong focus on education and research, with extensive documentation and a large collection of tutorials and books aimed at students and researchers.

```python
# @title VADER module in NLTK
# This code demonstrates sentiment analysis using the VADER module from NLTK and evaluates the model's performance.

# Import the 'nltk' library, which provides tools for natural language processing (NLP), including the VADER sentiment analysis tool.
import nltk

# Download the VADER lexicon, which is a pre-trained model for sentiment analysis.
# VADER (Valence Aware Dictionary and sEntiment Reasoner) is specifically designed to analyze sentiments in social media text.
nltk.download('vader_lexicon')

# Import the SentimentIntensityAnalyzer class from NLTK's VADER module.
# This analyzer will be used to determine the sentiment of a given text.
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Import 'classification_report' from sklearn to evaluate classification performance.
# This function will help us assess the performance of the VADER sentiment analysis.
from sklearn.metrics import classification_report

# Import numpy and pandas for handling arrays and dataframes.
import numpy as np
import pandas as pd

# Define a custom function decorator 'my_divider', which is used to print a divider line after the execution of a function.
# This decorator will be applied to other functions to improve output readability.
def my_divider(func):
    def wrapper(*args, **kwargs):
        # Call the original function with its arguments.
        func(*args, **kwargs)
        # Print a divider line of 75 dashes after the function is called.
        print("-" * 75)
    return wrapper

# Define a function 'dis' (short for display) that prints content.
# This function is wrapped with the 'my_divider' decorator to include a divider after printing.
@my_divider
def dis(content):
    # If the content is a list, print each element in the list.
    if isinstance(content, list):
        for cont in content:
            print(cont)
    else:
        # If the content is not a list, print it directly.
        print(content)

# Initialize the SentimentIntensityAnalyzer from VADER.
# This analyzer provides sentiment scores for texts based on the pre-trained VADER lexicon.
sid = SentimentIntensityAnalyzer()

# Analyze the sentiment of a simple sentence using VADER's 'polarity_scores' method.
# This method returns a dictionary with four sentiment scores: 'neg' (negative), 'neu' (neutral), 'pos' (positive), and 'compound' (overall sentiment score).
text = 'AI is a good subject to learn'
dis(sid.polarity_scores(text))  # Print sentiment scores with a divider line.

# Analyze the sentiment of a more enthusiastic sentence.
text = 'AI is the best and awesome subject to learn EVER!!!'
dis(sid.polarity_scores(text))  # Print sentiment scores with a divider line.

# Analyze the sentiment of a sentence with a cautionary tone about AI.
text = 'AI could potentially pose a significant threat to humanity if not developed responsibly!'
dis(sid.polarity_scores(text))  # Print sentiment scores with a divider line.

# Load a dataset of Amazon reviews from a TSV (tab-separated values) file into a pandas DataFrame.
# The dataset contains customer reviews and their corresponding sentiment labels (e.g., 'positive', 'negative').
df = pd.read_csv('/content/drive/MyDrive/data_535743/nlp/files/amazon_reviews.tsv', sep='\t')

# Print the first few rows of the dataset to inspect it.
dis(df.head())  # Print the first rows with a divider line.

# Print the count of each sentiment label (positive/negative) to understand the class distribution.
dis(df['label'].value_counts())  # Print label distribution with a divider line.

# Apply VADER's sentiment analysis to the first review in the dataset.
# Display the polarity scores, the actual label, and an estimation of the label based on the compound score.
dis([sid.polarity_scores(df.loc[0]['review']),
     f"Actual Label: {df.loc[0]['label']}",
     f"Estimated label (compound): {sid.polarity_scores(df.loc[0]['review'])['compound']} which is negative (negative value)"])

# Convert the real sentiment labels to numerical values: 0 for 'neg' (negative) and 1 for 'pos' (positive).
# This creates a binary classification target for evaluation.
real_scores = df['label'].apply(lambda label: 0 if label.lower().strip() == 'neg' else 1)

# Estimate the sentiment label for each review based on the compound score from VADER's polarity_scores.
# A compound score < 0 is considered negative (0), and a score >= 0 is considered positive (1).
estimated_scores = df['review'].apply(lambda review: 0 if sid.polarity_scores(review)['compound'] < 0 else 1)

# Evaluate the performance of the sentiment analysis by comparing real scores to estimated scores.
# The classification_report shows precision, recall, f1-score, and support for both positive and negative classes.
dis(classification_report(real_scores, estimated_scores))
```

### **5.2. Text Generation**

Text generation in Natural Language Processing (NLP) is the task of automatically creating meaningful, coherent, and contextually relevant sequences of text, such as sentences or paragraphs, using machine learning models. The goal of text generation is to produce human-like text based on a given input (prompt, seed text, or a set of conditions) by predicting the next word or phrase in a sequence, iterating until a complete output is formed.

> **Key Concepts in Text Generation**
* Language Models:
Text generation is typically powered by language models that are trained on large corpora of text. These models learn the statistical properties of language, such as word sequences, grammar, and context. The better the model understands these properties, the more fluent and coherent the generated text will be. Popular models include:
 * n-gram models: Use a fixed-size context window to predict the next word.
 * Recurrent Neural Networks (RNNs): Handle sequential data by using past information to predict future text.
 * Long Short-Term Memory (LSTM): An advanced RNN architecture that helps to capture long-range dependencies.
 * Transformers (like GPT, GPT-3, BERT): The state-of-the-art models for text generation, using self-attention mechanisms to handle longer dependencies and generate more sophisticated text.
* Applications of Text Generation:
 * Chatbots and Conversational Agents: Generating human-like responses in dialogue systems.
 * Creative Writing: Assisting in writing stories, poems, or news articles.
 * Machine Translation: Translating text from one language to another.
 * Code Generation: Writing or completing code snippets in various programming languages.
 * Summarization: Generating concise summaries of longer texts.
 * Automatic Report Generation: Producing reports based on data analysis.

```python
#@title Tfrecords Functions
# Function to serialize input and output numpy arrays to a format compatible with TensorFlow's TFRecord.
# TFRecord is a binary file format used by TensorFlow to store large datasets efficiently.
# This function encodes input and output arrays as byte strings and wraps them in a `tf.train.Example` format.
def serialize_example(input_data, output_data):
    feature = {
        'input': tf.train.Feature(bytes_list=tf.train.BytesList(value=[input_data.tobytes()])),
        'output': tf.train.Feature(bytes_list=tf.train.BytesList(value=[output_data.tobytes()]))
    }
    # Return the serialized example as a byte string, ready to be written to a TFRecord file.
    example_proto = tf.train.Example(features=tf.train.Features(feature=feature))
    return example_proto.SerializeToString()

# Function to write input-output data pairs to a TFRecord file.
# It iterates over the data arrays, serializes each input-output pair, and writes it to the file.
def write_tfrecords(input_data, output_data, filename):
    with tf.io.TFRecordWriter(filename) as writer:
        for i in range(len(input_data)):
            # Serialize the i-th input-output pair and write to file.
            example = serialize_example(input_data[i], output_data[i])
            writer.write(example)

# Function to parse serialized TFRecord examples back into input-output pairs.
# It defines the structure of the data in TFRecords (both input and output as byte strings) and decodes it.
def _parse_function(example_proto):
    # Define the structure of each TFRecord entry.
    feature_description = {
        'input': tf.io.FixedLenFeature([], tf.string),
        'output': tf.io.FixedLenFeature([], tf.string),
    }
    # Parse the example using the provided feature description.
    parsed_example = tf.io.parse_single_example(example_proto, feature_description)

    # Decode the byte strings into numpy arrays of type int16.
    input_data = tf.io.decode_raw(parsed_example['input'], tf.int16)
    output_data = tf.io.decode_raw(parsed_example['output'], tf.int16)

    # Reshape the input and output data to their original shapes.
    input_data = tf.reshape(input_data, (sequence_length-1,))  # Example: Input sequence length of 255.
    output_data = tf.reshape(output_data, (1,))  # Example: Output is a single token.

    return input_data, output_data

# Function to load and prepare a dataset from TFRecord files.
# This function uses TensorFlow's `TFRecordDataset` to read the binary TFRecords and applies the parsing function.
def load_dataset(filenames):
    dataset = tf.data.TFRecordDataset(filenames)
    dataset = dataset.map(_parse_function)  # Apply the parsing function to each example in the dataset.
    return dataset

```

```python
# @title Necessary Functions

# A decorator function that modifies another function by adding a separator line after it runs.
# Decorators are used to modify the behavior of functions without changing their code directly.
def my_divider(func):
    def wrapper(*args, **kwargs):
        # Call the original function with its arguments.
        func(*args, **kwargs)
        # Print a separator line after the function's output.
        print("-" * 75)
    return wrapper

# A display function that prints its content.
# It uses the 'my_divider' decorator to print a separator line after the content is displayed.
@my_divider
def dis(content):
    # If the content is a list, print each element individually.
    if isinstance(content, list):
        for cont in content:
            print(cont)
    else:
        # Otherwise, print the content directly.
        print(content)

# A sampling function for generating words based on model predictions.
# It applies temperature scaling, top-k filtering, and top-p (nucleus) filtering to logits (predicted probabilities).
def sample(logits, temperature=1.0, top_k=0, top_p=0.0):
    # Adjust logits using the temperature, which controls the randomness in predictions.
    logits = np.array(logits) / temperature
    # Top-k filtering: Keep only the top-k highest logits, setting the rest to -infinity.
    if top_k > 0:
        top_k_indices = np.argsort(logits)[-top_k:]
        logits_filtered = np.full_like(logits, -np.inf)
        logits_filtered[top_k_indices] = logits[top_k_indices]
        logits = logits_filtered
    # Top-p (nucleus) filtering: Keep the smallest set of logits whose cumulative probability exceeds top_p.
    if top_p > 0.0:
        sorted_indices = np.argsort(logits)[::-1]
        sorted_logits = logits[sorted_indices]
        cumulative_probs = np.cumsum(np.exp(sorted_logits) / np.sum(np.exp(sorted_logits)))
        cutoff_index = np.where(cumulative_probs > top_p)[0][0]
        logits_filtered = np.full_like(logits, -np.inf)
        logits_filtered[sorted_indices[:cutoff_index + 1]] = logits[sorted_indices[:cutoff_index + 1]]
        logits = logits_filtered
    # Compute probabilities from logits using the softmax function and sample the next word.
    probabilities = np.exp(logits) / np.sum(np.exp(logits))
    return np.random.choice(len(probabilities), p=probabilities)

# Function to generate text using a trained model and a tokenizer.
# The function supports sampling with temperature, top-k, and top-p.
def generate_text(model, tokenizer, sequence_length, initial_text, num_words_to_generate, temperature=1, top_k=10, top_p=0.9):
    generated_words = []  # Store the generated words.
    current_text = deepcopy(initial_text)  # Make a deep copy to avoid modifying the original text.

    # Generate words one at a time in a loop.
    for _ in range(num_words_to_generate):
        # Encode the current text into a sequence of token IDs using the tokenizer.
        encoded_text = tokenizer.texts_to_sequences([current_text])[0]
        # Pad the encoded sequence to the required length for the model.
        padded_encoded = pad_sequences([encoded_text], maxlen=sequence_length, truncating='pre')
        # Predict the logits for the next word using the model.
        logits = model.predict(padded_encoded, verbose=0)[0, :]
        # Use the sample function to select the next word based on logits.
        predicted_word_index = sample(logits, temperature=temperature, top_k=top_k, top_p=top_p)
        # Convert the predicted word index back to a word using the tokenizer's dictionary.
        predicted_word = tokenizer.index_word.get(predicted_word_index, '')
        current_text += ' ' + predicted_word  # Update the current text with the predicted word.
        generated_words.append(predicted_word)  # Add the predicted word to the list.

    # Return the full generated text, starting with the initial input followed by the generated words.
    return initial_text + ' ' + ' '.join(generated_words)

# Function to preprocess text by tokenizing and removing unwanted characters (e.g., punctuation, whitespace).
# Tokenizes text using spaCy and converts it to lowercase, excluding specific characters.
def cleansing(doc):
    return [token.text.lower() for token in doc if token.text not in '\n\n \n\n\n!"-#$%&()--.*+,-/:;<=>?@[\\]^_`{|}~\t\n ']

# Function to read a file and return its content as a string.
def read_file(path):
    with open(path, 'r') as f:
        return f.read()

# Function to create sequences of tokens for training.
# It generates sequences of length `sequence_length` from a list of tokens.
def sequence_maker(tokens, sequence_length, num_tokens):
    return [tokens[i-sequence_length:i] for i in range(sequence_length + 1, num_tokens)]


# Custom learning rate scheduler function to modify the learning rate during training.
# It decreases the learning rate gradually after a few epochs to ensure smooth convergence.
def scheduler(epoch, lr):
    if epoch < 5:
        return lr  # Keep the learning rate constant for the first 5 epochs.
    else:
        return lr * 0.999  # Decay the learning rate by 0.1% after each epoch.

```

```python
#@title Model Building Classes

# Custom embedding class to combine token and positional embeddings for transformer models.
class TokenAndPositionEmbedding(Layer):
    def __init__(self, maxlen, vocab_size, embed_dim, trainable=True, **kwargs):
        """
        Initializes the TokenAndPositionEmbedding layer.

        Args:
        - maxlen: The maximum length of the input sequence (i.e., the number of tokens).
        - vocab_size: The size of the vocabulary (i.e., the number of unique tokens).
        - embed_dim: The dimensionality of the embeddings (i.e., the size of the vector space).
        - trainable: Boolean indicating whether the layer's weights are trainable.
        - kwargs: Additional arguments for the parent Layer class.
        """
        # Call the parent class's initializer, passing in whether the layer is trainable and any other kwargs.
        super(TokenAndPositionEmbedding, self).__init__(trainable=trainable, **kwargs)

        # Embedding layer for tokens (word embeddings). Each token is mapped to an 'embed_dim'-dimensional vector.
        self.token_emb = Embedding(input_dim=vocab_size+1, output_dim=embed_dim)  # input_dim = vocab_size + 1 to account for padding

        # Embedding layer for positions (positional embeddings). Each position in the input sequence has its own embedding.
        self.pos_emb = Embedding(input_dim=maxlen, output_dim=embed_dim)  # input_dim = maxlen, embedding for each position

    # Forward pass: Defines how the input data flows through the layer.
    def call(self, x):
        """
        Combines token embeddings and positional embeddings.

        Args:
        - x: Input tensor, typically containing integer-encoded tokens.

        Returns:
        - A tensor that is the sum of token embeddings and positional embeddings.
        """
        # Get the length of the input sequence.
        maxlen = tf.shape(x)[-1]

        # Create a range of position indices from 0 to maxlen-1.
        positions = tf.range(start=0, limit=maxlen, delta=1)

        # Get the positional embeddings for the positions in the input sequence.
        positions = self.pos_emb(positions)

        # Get the token embeddings for the tokens in the input sequence.
        x = self.token_emb(x)

        # Return the sum of token embeddings and positional embeddings.
        return x + positions


# Transformer block class that implements a single layer of the transformer architecture.
class TransformerBlock(Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, rate=0.33, trainable=True, **kwargs):
        """
        Initializes the TransformerBlock layer, which contains multi-head attention,
        feed-forward network (FFN), and normalization layers.

        Args:
        - embed_dim: The dimensionality of the input embeddings.
        - num_heads: The number of attention heads in the multi-head attention layer.
        - ff_dim: The dimensionality of the feed-forward network's intermediate layer.
        - rate: Dropout rate to apply during training.
        - trainable: Boolean indicating whether the layer's weights are trainable.
        - kwargs: Additional arguments for the parent Layer class.
        """
        # Call the parent class's initializer, passing whether the layer is trainable and any other kwargs.
        super(TransformerBlock, self).__init__(trainable=trainable, **kwargs)

        # Multi-head attention layer: allows the model to attend to different parts of the input sequence simultaneously.
        self.att = MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)

        # Feed-forward network (FFN) consisting of two dense layers:
        # - The first dense layer projects the input to a higher dimension (ff_dim) with ReLU activation.
        # - The second dense layer projects it back to the original dimension (embed_dim).
        self.ffn = tf.keras.Sequential([Dense(ff_dim, activation='relu'), Dense(embed_dim)])

        # Layer normalization applied after the attention and feed-forward layers for training stability.
        self.layernorm1 = LayerNormalization(epsilon=1e-6)
        self.layernorm2 = LayerNormalization(epsilon=1e-6)

        # Dropout layers to prevent overfitting during training.
        self.dropout1 = Dropout(rate)
        self.dropout2 = Dropout(rate)

    # Forward pass: Defines how the input data flows through the layer.
    def call(self, inputs, training=False):
        """
        Processes the input data through the transformer block, including attention,
        feed-forward, layer normalization, and dropout.

        Args:
        - inputs: Input tensor.
        - training: Boolean flag indicating whether the model is in training mode.

        Returns:
        - A tensor that has passed through the transformer block.
        """
        # Apply multi-head attention to the inputs. Attention is computed on the inputs themselves (self-attention).
        attn_output = self.att(inputs, inputs)

        # Apply dropout to the attention output if in training mode.
        attn_output = self.dropout1(attn_output, training=training)

        # Residual connection: Add the attention output back to the original input and normalize.
        out1 = self.layernorm1(inputs + attn_output)

        # Apply the feed-forward network (FFN) to the output from the attention block.
        ffn_output = self.ffn(out1)

        # Apply dropout to the FFN output if in training mode.
        ffn_output = self.dropout2(ffn_output, training=training)

        # Residual connection: Add the FFN output back to the output of the attention block and normalize.
        return self.layernorm2(out1 + ffn_output)

```

```python
#@title Tfrecords Functions
# Function to serialize input and output numpy arrays to a format compatible with TensorFlow's TFRecord.
# TFRecord is a binary file format used by TensorFlow to store large datasets efficiently.
# This function encodes input and output arrays as byte strings and wraps them in a `tf.train.Example` format.
def serialize_example(input_data, output_data):
    feature = {
        'input': tf.train.Feature(bytes_list=tf.train.BytesList(value=[input_data.tobytes()])),
        'output': tf.train.Feature(bytes_list=tf.train.BytesList(value=[output_data.tobytes()]))
    }
    # Return the serialized example as a byte string, ready to be written to a TFRecord file.
    example_proto = tf.train.Example(features=tf.train.Features(feature=feature))
    return example_proto.SerializeToString()

# Function to write input-output data pairs to a TFRecord file.
# It iterates over the data arrays, serializes each input-output pair, and writes it to the file.
def write_tfrecords(input_data, output_data, filename):
    with tf.io.TFRecordWriter(filename) as writer:
        for i in range(len(input_data)):
            # Serialize the i-th input-output pair and write to file.
            example = serialize_example(input_data[i], output_data[i])
            writer.write(example)

# Function to parse serialized TFRecord examples back into input-output pairs.
# It defines the structure of the data in TFRecords (both input and output as byte strings) and decodes it.
def _parse_function(example_proto):
    # Define the structure of each TFRecord entry.
    feature_description = {
        'input': tf.io.FixedLenFeature([], tf.string),
        'output': tf.io.FixedLenFeature([], tf.string),
    }
    # Parse the example using the provided feature description.
    parsed_example = tf.io.parse_single_example(example_proto, feature_description)

    # Decode the byte strings into numpy arrays of type int16.
    input_data = tf.io.decode_raw(parsed_example['input'], tf.int16)
    output_data = tf.io.decode_raw(parsed_example['output'], tf.int16)

    # Reshape the input and output data to their original shapes.
    input_data = tf.reshape(input_data, (sequence_length-1,))  # Example: Input sequence length of 255.
    output_data = tf.reshape(output_data, (1,))  # Example: Output is a single token.

    return input_data, output_data

# Function to load and prepare a dataset from TFRecord files.
# This function uses TensorFlow's `TFRecordDataset` to read the binary TFRecords and applies the parsing function.
def load_dataset(filenames):
    dataset = tf.data.TFRecordDataset(filenames)
    dataset = dataset.map(_parse_function)  # Apply the parsing function to each example in the dataset.
    return dataset

```

```python
# @title Necessary Functions

# A decorator function that modifies another function by adding a separator line after it runs.
# Decorators are used to modify the behavior of functions without changing their code directly.
def my_divider(func):
    def wrapper(*args, **kwargs):
        # Call the original function with its arguments.
        func(*args, **kwargs)
        # Print a separator line after the function's output.
        print("-" * 75)
    return wrapper

# A display function that prints its content.
# It uses the 'my_divider' decorator to print a separator line after the content is displayed.
@my_divider
def dis(content):
    # If the content is a list, print each element individually.
    if isinstance(content, list):
        for cont in content:
            print(cont)
    else:
        # Otherwise, print the content directly.
        print(content)

# A sampling function for generating words based on model predictions.
# It applies temperature scaling, top-k filtering, and top-p (nucleus) filtering to logits (predicted probabilities).
def sample(logits, temperature=1.0, top_k=0, top_p=0.0):
    # Adjust logits using the temperature, which controls the randomness in predictions.
    logits = np.array(logits) / temperature
    # Top-k filtering: Keep only the top-k highest logits, setting the rest to -infinity.
    if top_k > 0:
        top_k_indices = np.argsort(logits)[-top_k:]
        logits_filtered = np.full_like(logits, -np.inf)
        logits_filtered[top_k_indices] = logits[top_k_indices]
        logits = logits_filtered
    # Top-p (nucleus) filtering: Keep the smallest set of logits whose cumulative probability exceeds top_p.
    if top_p > 0.0:
        sorted_indices = np.argsort(logits)[::-1]
        sorted_logits = logits[sorted_indices]
        cumulative_probs = np.cumsum(np.exp(sorted_logits) / np.sum(np.exp(sorted_logits)))
        cutoff_index = np.where(cumulative_probs > top_p)[0][0]
        logits_filtered = np.full_like(logits, -np.inf)
        logits_filtered[sorted_indices[:cutoff_index + 1]] = logits[sorted_indices[:cutoff_index + 1]]
        logits = logits_filtered
    # Compute probabilities from logits using the softmax function and sample the next word.
    probabilities = np.exp(logits) / np.sum(np.exp(logits))
    return np.random.choice(len(probabilities), p=probabilities)

# Function to generate text using a trained model and a tokenizer.
# The function supports sampling with temperature, top-k, and top-p.
def generate_text(model, tokenizer, sequence_length, initial_text, num_words_to_generate, temperature=1, top_k=10, top_p=0.9):
    generated_words = []  # Store the generated words.
    current_text = deepcopy(initial_text)  # Make a deep copy to avoid modifying the original text.

    # Generate words one at a time in a loop.
    for _ in range(num_words_to_generate):
        # Encode the current text into a sequence of token IDs using the tokenizer.
        encoded_text = tokenizer.texts_to_sequences([current_text])[0]
        # Pad the encoded sequence to the required length for the model.
        padded_encoded = pad_sequences([encoded_text], maxlen=sequence_length, truncating='pre')
        # Predict the logits for the next word using the model.
        logits = model.predict(padded_encoded, verbose=0)[0, :]
        # Use the sample function to select the next word based on logits.
        predicted_word_index = sample(logits, temperature=temperature, top_k=top_k, top_p=top_p)
        # Convert the predicted word index back to a word using the tokenizer's dictionary.
        predicted_word = tokenizer.index_word.get(predicted_word_index, '')
        current_text += ' ' + predicted_word  # Update the current text with the predicted word.
        generated_words.append(predicted_word)  # Add the predicted word to the list.

    # Return the full generated text, starting with the initial input followed by the generated words.
    return initial_text + ' ' + ' '.join(generated_words)

# Function to preprocess text by tokenizing and removing unwanted characters (e.g., punctuation, whitespace).
# Tokenizes text using spaCy and converts it to lowercase, excluding specific characters.
def cleansing(doc):
    return [token.text.lower() for token in doc if token.text not in '\n\n \n\n\n!"-#$%&()--.*+,-/:;<=>?@[\\]^_`{|}~\t\n ']

# Function to read a file and return its content as a string.
def read_file(path):
    with open(path, 'r') as f:
        return f.read()

# Function to create sequences of tokens for training.
# It generates sequences of length `sequence_length` from a list of tokens.
def sequence_maker(tokens, sequence_length, num_tokens):
    return [tokens[i-sequence_length:i] for i in range(sequence_length + 1, num_tokens)]


# Custom learning rate scheduler function to modify the learning rate during training.
# It decreases the learning rate gradually after a few epochs to ensure smooth convergence.
def scheduler(epoch, lr):
    if epoch < 5:
        return lr  # Keep the learning rate constant for the first 5 epochs.
    else:
        return lr * 0.999  # Decay the learning rate by 0.1% after each epoch.

```

```python
#@title Model Building Classes

# Custom embedding class to combine token and positional embeddings for transformer models.
class TokenAndPositionEmbedding(Layer):
    def __init__(self, maxlen, vocab_size, embed_dim, trainable=True, **kwargs):
        """
        Initializes the TokenAndPositionEmbedding layer.

        Args:
        - maxlen: The maximum length of the input sequence (i.e., the number of tokens).
        - vocab_size: The size of the vocabulary (i.e., the number of unique tokens).
        - embed_dim: The dimensionality of the embeddings (i.e., the size of the vector space).
        - trainable: Boolean indicating whether the layer's weights are trainable.
        - kwargs: Additional arguments for the parent Layer class.
        """
        # Call the parent class's initializer, passing in whether the layer is trainable and any other kwargs.
        super(TokenAndPositionEmbedding, self).__init__(trainable=trainable, **kwargs)

        # Embedding layer for tokens (word embeddings). Each token is mapped to an 'embed_dim'-dimensional vector.
        self.token_emb = Embedding(input_dim=vocab_size+1, output_dim=embed_dim)  # input_dim = vocab_size + 1 to account for padding

        # Embedding layer for positions (positional embeddings). Each position in the input sequence has its own embedding.
        self.pos_emb = Embedding(input_dim=maxlen, output_dim=embed_dim)  # input_dim = maxlen, embedding for each position

    # Forward pass: Defines how the input data flows through the layer.
    def call(self, x):
        """
        Combines token embeddings and positional embeddings.

        Args:
        - x: Input tensor, typically containing integer-encoded tokens.

        Returns:
        - A tensor that is the sum of token embeddings and positional embeddings.
        """
        # Get the length of the input sequence.
        maxlen = tf.shape(x)[-1]

        # Create a range of position indices from 0 to maxlen-1.
        positions = tf.range(start=0, limit=maxlen, delta=1)

        # Get the positional embeddings for the positions in the input sequence.
        positions = self.pos_emb(positions)

        # Get the token embeddings for the tokens in the input sequence.
        x = self.token_emb(x)

        # Return the sum of token embeddings and positional embeddings.
        return x + positions


# Transformer block class that implements a single layer of the transformer architecture.
class TransformerBlock(Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, rate=0.33, trainable=True, **kwargs):
        """
        Initializes the TransformerBlock layer, which contains multi-head attention,
        feed-forward network (FFN), and normalization layers.

        Args:
        - embed_dim: The dimensionality of the input embeddings.
        - num_heads: The number of attention heads in the multi-head attention layer.
        - ff_dim: The dimensionality of the feed-forward network's intermediate layer.
        - rate: Dropout rate to apply during training.
        - trainable: Boolean indicating whether the layer's weights are trainable.
        - kwargs: Additional arguments for the parent Layer class.
        """
        # Call the parent class's initializer, passing whether the layer is trainable and any other kwargs.
        super(TransformerBlock, self).__init__(trainable=trainable, **kwargs)

        # Multi-head attention layer: allows the model to attend to different parts of the input sequence simultaneously.
        self.att = MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)

        # Feed-forward network (FFN) consisting of two dense layers:
        # - The first dense layer projects the input to a higher dimension (ff_dim) with ReLU activation.
        # - The second dense layer projects it back to the original dimension (embed_dim).
        self.ffn = tf.keras.Sequential([Dense(ff_dim, activation='relu'), Dense(embed_dim)])

        # Layer normalization applied after the attention and feed-forward layers for training stability.
        self.layernorm1 = LayerNormalization(epsilon=1e-6)
        self.layernorm2 = LayerNormalization(epsilon=1e-6)

        # Dropout layers to prevent overfitting during training.
        self.dropout1 = Dropout(rate)
        self.dropout2 = Dropout(rate)

    # Forward pass: Defines how the input data flows through the layer.
    def call(self, inputs, training=False):
        """
        Processes the input data through the transformer block, including attention,
        feed-forward, layer normalization, and dropout.

        Args:
        - inputs: Input tensor.
        - training: Boolean flag indicating whether the model is in training mode.

        Returns:
        - A tensor that has passed through the transformer block.
        """
        # Apply multi-head attention to the inputs. Attention is computed on the inputs themselves (self-attention).
        attn_output = self.att(inputs, inputs)

        # Apply dropout to the attention output if in training mode.
        attn_output = self.dropout1(attn_output, training=training)

        # Residual connection: Add the attention output back to the original input and normalize.
        out1 = self.layernorm1(inputs + attn_output)

        # Apply the feed-forward network (FFN) to the output from the attention block.
        ffn_output = self.ffn(out1)

        # Apply dropout to the FFN output if in training mode.
        ffn_output = self.dropout2(ffn_output, training=training)

        # Residual connection: Add the FFN output back to the output of the attention block and normalize.
        return self.layernorm2(out1 + ffn_output)
```


```python
# @title Model Development & Training - Small Transformer Experiment
# This code develops a Transformer-based text generation model.
# It defines hyperparameters for the Transformer model, sets up the model architecture, and includes training and model-saving components.

# Transformer hyperparameters:
# - `embed_dim`: The size of the word embedding vectors, which represent each token as a 256-dimensional vector.
# - `num_heads`: The number of attention heads in the multi-head self-attention mechanism, set to 2 here.
# - `ff_dim`: The size of the feed-forward network (FFN) inside each transformer block, also set to 256.
# - `num_layers`: The number of stacked transformer layers in the model. More layers allow the model to capture more complex patterns.
embed_dim = 128
num_heads = 2
ff_dim = 128
num_layers = 1

# Define the input layer for the model, specifying the input shape as (sequence_length - 1).
# Each input sequence will have 255 tokens, as the sequence length is 256 and the last token is the target label.
inputs = Input(shape=(sequence_length - 1,))

# Initialize the custom Token and Position Embedding layer.
# This layer combines token embeddings with positional embeddings, which allow the transformer to understand word order in the sequence.
embedding_layer = TokenAndPositionEmbedding(sequence_length - 1, vocabulary_size, embed_dim)
x = embedding_layer(inputs)  # Apply the embedding layer to the input tokens.

# Add the Transformer blocks to the model.
# The model will contain `num_layers` (1) transformer blocks, each consisting of multi-head attention and feed-forward layers.
for i in range(num_layers):
    # For each iteration, a new TransformerBlock is added to the model.
    # The block includes multi-head attention, feed-forward networks (FFN), layer normalization, and residual connections.
    x = TransformerBlock(embed_dim, num_heads, ff_dim)(x)

# Apply global average pooling across the sequence dimension.
# This operation reduces the output sequence into a single vector by averaging all time steps, creating a fixed-size representation.
x = GlobalAveragePooling1D()(x)

# Add the output layer, which produces a probability distribution over the vocabulary.
# The `softmax` activation function ensures that the output is a valid probability distribution.
# The model will predict the next word in the sequence based on the vocabulary size (`vocabulary_size + 1`).
outputs = Dense(vocabulary_size+1, activation="softmax")(x)

# Define the model, specifying the input and output layers.
model = Model(inputs=inputs, outputs=outputs)

# Compile the model:
# - `optimizer=Adam`: Adam optimizer is used with a learning rate of 1e-3. The `clipnorm=1.0` parameter ensures that gradient norms are clipped,
#   preventing exploding gradients during training.
# - `loss='sparse_categorical_crossentropy'`: This loss function is used for multi-class classification tasks with integer labels (token IDs).
# - `metrics=['accuracy']`: The model's performance will be evaluated based on accuracy during training and validation.
optimizer = Adam(learning_rate=1e-3, clipnorm=1.0)
model.compile(optimizer=optimizer, loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Set up a learning rate scheduler callback to adjust the learning rate dynamically during training.
# This helps the model converge more efficiently by lowering the learning rate as training progresses.
lr_callback = LearningRateScheduler(scheduler)

# Set up early stopping to prevent overfitting.
# Training will stop if the validation accuracy (`val_accuracy`) does not improve for 5 consecutive epochs.
# The `restore_best_weights=True` option ensures that the best model (based on validation accuracy) is restored after training.
early_stopping = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)

# Set up a model checkpoint to save the best model during training.
# The model with the lowest loss will be saved to a file named 'best_model_transformer.keras'.
checkpoint = ModelCheckpoint(os.path.join(path_sequence, 'best_model_transformer.keras'), monitor='loss', save_best_only=True)

# Display the model's architecture and parameters using the summary method.
model.summary()

#dfsasdf=asdfasdfasdfasdf

# Define the number of steps per epoch for training.
# `1237877` represents the total number of sequences, and `(1 - validation_rate)` gives the percentage used for training.
# The batch size determines how many training samples are processed in one iteration.
steps_per_epoch = max(1, int(1237877 * (1 - validation_rate) // batch_size))

# Define the number of validation steps per epoch.
# This is calculated similarly to the training steps but for the validation data.
validation_steps = max(1, int(1237877 * validation_rate // batch_size))

# Train the model using the training dataset `dataset_tr` and validate using `dataset_vl`.
# - `steps_per_epoch`: The number of batches of training data processed in each epoch.
# - `validation_steps`: The number of batches of validation data processed in each epoch.
# - `epochs=100`: Train the model for a maximum of 100 epochs (iterations over the entire dataset).
# - `verbose=1`: Displays progress during training.
# - `callbacks`: Includes the learning rate scheduler, early stopping, and model checkpointing.
model.fit(dataset_tr, steps_per_epoch=steps_per_epoch, validation_steps=validation_steps,
          epochs=100, verbose=1, validation_data=dataset_vl,
          callbacks=[lr_callback, early_stopping, checkpoint])


# Save the trained model to a file named 'best_model_transformer.h5' for later use.
model.save(os.path.join(path_sequence,'best_model_transformer.h5'))
```
