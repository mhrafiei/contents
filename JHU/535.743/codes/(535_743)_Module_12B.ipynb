## **1. Introduction to Language Tasks & Models in Hugging Face**

Large Language Models (LLMs) are advanced AI systems designed to understand, generate, & rk with human language at a high level of proficiency. These models, such as [OpenAI’s GPT](https://platform.openai.com/docs/models) (Generative Pre-trained Transformer) series, are based on transformer architectures, a type of deep learning model that relies on self-attention mechanisms to process & predict text. LLMs are trained on diverse & extensive datasets, allowing them to develop a broad understanding of language patterns, context, & even the nuances of different writing styles.

One of the most exciting aspects of LLMs is their versatility. They can be applied to a variety of tasks including text generation, translation, summarization, & sentiment analysis, making them invaluable tools in fields ranging from customer service to creative writing. Their ability to engage in dialogue & generate coherent & contextually relevant responses also makes them ideal for building conversational agents.

Hugging Face, a leading company in the AI space, has significantly contributed to the democratization of LLMs through its platform & products. It offers a variety of free LLMs through its ‘transformers’ library, which is an open-source repository that provides pre-trained models to the public. This allows developers & researchers to utilize state-of-the-art models without the overhead of training them from scratch. The Hugging Face ecosystem supports a collaborative approach, enabling users to easily share their models & benefit from improvements made by others in the community. This open model has accelerated innovation in AI, as developers can build upon each other’s work, adapting & refining models to better suit their specific needs.

In this section, we review major free Hugging Face LLM models for various tasks.

### **1.2. Question Answering**

The following table compares three models specialized for question answering (QA)—BERT, RoBERTa, & ALBERT—based on various performance metrics. Bias & toxicity are concerns for all three models, though ALBERT’s design emphasizes efficiency & mitigation efforts. ### **1.3. Text Classification**

The following table compares three transformer-based models—DistilBERT, XLNet, & T5—across various metrics. DistilBERT, developed by Hugging Face in 2019, is a lighter, distilled version of BERT with 66M parameters, offering faster inference speeds & reduced memory usage while maintaining competitive performance in classification tasks. XLNet, created by Google Brain & Carnegie Mellon University in 2019, uses a permutation-based autoregressive approach, improving bidirectional context understanding without the need for masking, but has slower inference due to its autoregressive nature. It comes in both base & large versions with up to 340M parameters. T5, developed by Google Research in 2020, follows a text-to-text framework, making it highly versatile for different NLP tasks, & is available in sizes ranging from 60M to 11B parameters. T5 achieves state-of-the-art performance on benchmarks like GLUE & SuperGLUE, while offering variable inference speeds depending on model size.

All three models are open-source & widely available on Hugging Face, with strong fine-tuning capabilities across tasks. While memory usage for DistilBERT is minimal (~256MB), XLNet & T5 require significantly more memory, especially for larger models. In terms of bias & ethical considerations, each model exhibits biases from its training data, & mitigation efforts are ongoing. DistilBERT stands out for its efficiency through distillation, XLNet for its innovative training method, & T5 for its adaptable text-to-text framework. Each model has strong transfer learning capabilities & performs well on standard benchmarks, with T5 leading in performance due to its broader flexibility.

### **1.4. Named Entity Recognition (NER)**

The following table compares three popular models for Named Entity Recognition (NER)—SpaCy Transformer Models, Flair NER, & BERT for NER—across various performance metrics. SpaCy Transformer Models, developed by Explosion AI in 2019, integrate transformer-based models like BERT & RoBERTa directly into the SpaCy ecosystem, offering flexible NER capabilities depending on the chosen transformer model (e.g., BERT-base with 110M parameters). Flair NER, created by Zalando Research in 2018, uses a combination of contextual string embeddings & sequence models, offering high-performance NER capabilities with state-of-the-art results on multiple benchmarks. BERT for NER, developed by Google Research in 2018, leverages its bidirectional transformer architecture, excelling in entity recognition tasks with models ranging from 110M to 340M parameters.

All three models demonstrate strong performance on benchmarks like CoNLL-2003 & OntoNotes 5. SpaCy models offer efficient inference thanks to SpaCy’s optimized processing, while Flair provides moderate speed & exceptional performance with its embeddings. BERT for NER provides moderate inference speed but increases significantly in memory usage for larger models. All models can be easily fine-tuned for specific NER tasks & exhibit strong transfer learning capabilities. Bias & ethical considerations are present across all models due to their training data, with ongoing research on bias mitigation.

SpaCy’s unique integration with its NLP pipeline makes it an efficient choice for NER tasks, while Flair NER stands out for its unique combination of embeddings & traditional sequence models. BERT for NER, with its bidirectional transformer architecture, enhances entity recognition accuracy. All three models are open-source & widely available for use via platforms like Hugging Face, with extensive support for fine-tuning & transfer learning across diverse NER tasks.

### **1.5. Summarization**

The following table compares three prominent transformer-based models for text summarization—BART, T5, & PEGASUS—across various metrics. BART, developed by Facebook AI Research (FAIR) in 2019, uses a sequence-to-sequence architecture with a denoising autoencoder objective, making it highly effective in abstractive summarization tasks. T5, created by Google Research in 2020, uses a text-to-text framework that allows it to handle a wide range of NLP tasks, not just summarization. PEGASUS, also by Google Research (2020), is optimized specifically for summarization tasks using its unique gap-sentence generation (GSG) pretraining method.

In terms of parameters, BART’s models range from 139M to 406M parameters, T5 models range from 60M (small) to 11B, & PEGASUS models are around 568M. All three models are capable of high-performance on summarization benchmarks like CNN/DailyMail & XSum. PEGASUS, in particular, excels at summarization with ROUGE scores slightly higher than BART & T5 in certain datasets, but BART leads on CNN/DailyMail with ROUGE-1 at 46.5.

Inference speed varies: BART offers moderate speed, T5’s speed depends heavily on the model size, & PEGASUS is optimized for summarization, with inference time scaling based on the model size. Memory usage follows a similar trend, with smaller T5 models being more lightweight (~240MB), while larger versions of all three models require several gigabytes of memory (up to 35GB for T5-11B).

Bias & toxicity remain concerns across all models due to the nature of the training data, & all three models are easily fine-tunable for specific summarization tasks, supported by libraries like Hugging Face Transformers. Each model also supports strong transfer learning capabilities.

BART stands out for its effectiveness in abstractive summarization, T5 for its flexibility in handling a variety of NLP tasks via its text-to-text framework, & PEGASUS for its targeted optimization for summarization using GSG. All three models are open-source & available on platforms like Hugging Face, making them accessible for further research & practical applications.

### **1.6. Translation**


The following table compares three popular models for machine translation—MarianMT, M2M-100, & T5 for Translation—based on various performance metrics. MarianMT, developed by Helsinki NLP in 2020, is a transformer-based sequence-to-sequence model trained on the OPUS parallel corpus, supporting a wide variety of language pairs, particularly low-resource languages. M2M-100, developed by Facebook AI in 2020, is a multilingual transformer model capable of translating directly between 100+ languages without relying on English as an intermediary. T5 for Translation, developed by Google Research in 2020, is based on the text-to-text framework & is highly adaptable for translation & other NLP tasks.

Regarding parameters, MarianMT models depend on the language pair, with the English-German model having about 75M parameters, while M2M-100 has a 418M base model & a 1.2B large model. T5, on the other hand, ranges from 60M for the small version to 11B for the largest model. All three models are transformer-based, but M2M-100 specializes in multilingual translation, while MarianMT excels in low-resource language pairs, & T5 is versatile across multiple NLP tasks due to its text-to-text design.

On benchmark performance, MarianMT achieves BLEU scores of 38-45 on WMT20 for various language pairs, while M2M-100 demonstrates strong performance across 100+ languages, scoring ~23.2 BLEU for English-German. T5 shows competitive performance on translation tasks when formatted as input-output pairs, along with strong results across other NLP tasks.

In terms of inference speed, MarianMT is efficient, especially for supported language pairs. M2M-100 is slower, particularly for low-resource languages, due to its size. T5’s speed varies significantly with the model size, with smaller models being faster. Memory usage also varies, with MarianMT models taking up ~300MB for specific language pairs, while M2M-100’s 418M model requires ~1.5GB & its 1.2B model uses ~3.5GB. T5’s memory usage ranges from ~240MB for the small model to 35GB for the largest.

All models exhibit biases inherited from their training data, with ongoing research aimed at mitigating these biases. Fine-tuning capabilities are strong across all models, supported by Hugging Face Transformers, making them highly adaptable for specific translation tasks. MarianMT is especially suited for low-resource language pairs, M2M-100 excels in direct multilingual translation, & T5’s text-to-text framework enables flexibility across various NLP applications.

Each model is open-source & available on platforms like Hugging Face, offering strong transfer learning potential for translation & related tasks.

### **1.10. Multi-Modal & Other Specialized Models**

This table compares four models—CLIP, Whisper, DALL·E, & Speech Models—across several metrics. CLIP, developed by OpenAI in 2021, is a transformer-based multi-modal model designed to link images & text for tasks like zero-shot image-text retrieval & classification. It uses around 149M to 428M parameters & is trained on 400M image-text pairs. Whisper, also from OpenAI (2022), is a speech recognition model designed to handle multilingual & multitask transcription with models ranging from 74M to 1.55B parameters. It is trained on 680,000 hours of supervised speech data & achieves strong performance in word error rate (WER). DALL·E, a text-to-image generation model from OpenAI (2021), has about 12B parameters & generates high-quality images from textual descriptions, excelling in creative image generation tasks. Speech Models, from various developers (Google, OpenAI, Hugging Face), include models like Wav2Vec 2.0, which support automatic speech recognition (ASR), text-to-speech (TTS), & speech translation tasks, with parameters ranging from 317M to larger sizes depending on the model.

CLIP is known for its strong performance in zero-shot tasks (e.g., 76.2% accuracy on ImageNet), Whisper excels in speech transcription (e.g., WER of ~2.7% on LibriSpeech), DALL·E leads in text-to-image generation, & speech models like Wav2Vec 2.0 perform well on speech benchmarks with a WER of ~1.8%. All models exhibit biases from their training data, with ongoing efforts to mitigate these issues. They are open-source & available on Hugging Face, with fine-tuning capabilities supported for specific tasks, making them versatile tools for multi-modal, speech, & image generation applications.

```python
#@title Initiate

# Clear the current Python environment to start fresh.
# This removes any existing variables, functions, or states that were previously loaded.
# The '%reset -f' command forcefully resets the current IPython environment without confirmation.
%reset -f

# Install the safetensors library.
# safetensors is a library used to handle model weights & data more efficiently & safely than typical formats.
# We use '!pip install' to install the required Python packages in Google Colab or other Jupyter environments.
!pip install --upgrade safetensors

# Import the os module to interact with the operating system (e.g., file paths, directories).
import os

# Import Google Drive module to access files from Google Drive.
from google.colab import drive

# Mount Google Drive.
# This gives us access to our Google Drive files directly in the Colab environment.
drive.mount('/content/drive')

# Check if the cache directory is already linked to a symbolic link.
# Symbolic links allow us to point one directory to another (in this case, pointing the cache directory to Google Drive).
# This step ensures that heavy files like model weights or cache are stored in Google Drive & not in Colab's temporary storage.
if not os.path.islink('/root/.cache'):
    # If the cache directory is not a symbolic link, remove the existing cache directory.
    !rm -rf /root/.cache/

    # Create a symbolic link between Colab's cache directory & a folder in Google Drive.
    # This prevents losing cached files (like model weights) every time the Colab session resets.
    !ln -s /content/drive/MyDrive/data_535743/nlp/cache /root/.cache
else:
    # If the symbolic link already exists, no action is taken.
    pass

# Import additional modules to support various functionalities.
from IPython.display import clear_output  # Allows clearing the output in Jupyter/Colab cells.
import time  # Provides time-related functions, e.g., for delays.
import sys  # Provides system-related functions, e.g., for output management.
import torch  # PyTorch library for machine learning models.
from huggingface_hub import login  # Allows logging in to Hugging Face Hub to download or upload models.
from transformers import pipeline # Importing the pipeline module from Hugging Face's transformers

from typing import *

# Define the Hugging Face API token to authenticate access.
# Hugging Face provides pre-trained models for natural language processing (NLP), computer vision, & more.
# The token ensures access to private models or advanced API functionalities.
hf_api_token = "YOUR TOKEN!!"  # Replace with your actual Hugging Face API token.

# Log in to Hugging Face using the provided API token.
# This login ensures we can download models or datasets from Hugging Face.
login(token=hf_api_token)

# Define the directory where we want to store offloaded model parts or large files.
offload_dir = "/content/offload"

# Ensure that the directory for offloaded data exists.
# If the directory does not exist, it will be created. This is useful for managing temporary files in Colab.
os.makedirs(offload_dir, exist_ok=True)

# Typing effect function that simulates typing out each character of a word, one by one.
# This is used to make printed text appear as if it's being typed out slowly.
def typing_effect(word, max_width=100, current_length=0, delay=0.01):
    """
    Prints a single word with a typing effect.

    Parameters:
    - word (str): The word to be printed with the typing effect.
    - max_width (int): The maximum width of the line before wrapping.
    - current_length (int): The current length of the printed line.
    - delay (float): Time delay between printing each character, simulating typing.

    Returns:
    - int: The updated current length after printing the word.
    """

    # Check if the current word can fit in the remaining space on the current line.
    # If it can't, move to the next line by printing a newline character ('\n').
    if current_length + len(word) + 1 > max_width:  # +1 for the space after the word
        sys.stdout.write('\n')  # Move to the next line
        current_length = 0  # Reset the length counter to start from the new line

    # Loop through each character of the word to print it out one by one.
    for char in word:
        sys.stdout.write(char)  # Print each character
        sys.stdout.flush()  # Ensure that each character is immediately displayed
        time.sleep(delay)  # Add a small delay to simulate typing
        current_length += 1  # Increment the current length as we print each character

    # After printing the word, add a space character to separate words.
    sys.stdout.write(' ')
    current_length += 1  # Account for the space after the word

    # Return the updated current length of the line.
    return current_length

# Function to display the full text with a typing effect, handling word wrapping & newlines.
def display_text(text, max_width=100, delay=0.01):
    """
    Prints the entire text with a typing effect, handling word wrapping & newlines.

    Parameters:
    - text (str): The text to be printed.
    - max_width (int): The maximum width of each line before wrapping to the next line.
    - delay (float): Time delay between printing each character.
    """

    # Split the input text into words based on spaces.
    words = text.split(' ')
    current_length = 0  # Initialize the current line length counter

    # Loop through each word to print it with a typing effect.
    for word in words:
        if '\n' in word:
            # Handle newlines that might be embedded in the text.
            parts = word.split('\n')
            for i, part in enumerate(parts):
                if i > 0:
                    sys.stdout.write('\n')  # Start a new line
                    current_length = 0  # Reset the current line length
                if part:
                    current_length = typing_effect(part, max_width, current_length, delay=delay)
        else:
            current_length = typing_effect(word, max_width, current_length, delay=delay)

    # Print a final newline after the text is printed.
    sys.stdout.write('\n')
    sys.stdout.flush()

# Clear the previous output from the notebook/cell.
clear_output()

# Display the message "Initiation is complete!" with a typing effect.
display_text("Initiation is complete!", delay=0.1)
```

```python
#@title Load GPT-2 (~30 sec)

# Import the necessary classes from the transformers library:
# - GPT2LMHeadModel: A language model head on top of the GPT-2 architecture used for text generation.
# - GPT2Tokenizer: A tokenizer specifically designed for GPT-2, used to convert text into token sequences & vice versa.
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Define the name of the pre-trained model we want to load. In this case, it's 'gpt2'.
# GPT-2 is a large pre-trained language model designed by OpenAI.
# It's excellent for generating human-like text based on input prompts.
model_name = 'gpt2'

# Load the pre-trained GPT-2 tokenizer:
# The tokenizer converts text into numerical token sequences that can be processed by the GPT-2 model.
# It also converts token sequences back into human-readable text after generation.
# The "from_pretrained" method automatically downloads the tokenizer configuration & vocab files from Hugging Face.
tokenizer = GPT2Tokenizer.from_pretrained(model_name)

# Load the pre-trained GPT-2 model:
# GPT-2 is a transformer-based language model capable of generating high-quality, coherent text.
# It has multiple layers of attention mechanisms that allow it to learn contextual relationships in language.
# The "from_pretrained" method downloads the model weights from Hugging Face.
model = GPT2LMHeadModel.from_pretrained(model_name)

# Clear the current output (if using in a notebook environment).
clear_output()

# Display a success message after loading the model, simulating a typing effect.
display_text("Model loaded successfully!", delay=0.1)
```

```python
#@title GPT-2 Inference

# Define the prompt related to mechanical engineering.
# The prompt guides the GPT-2 model to generate a relevant response based on this initial input.
# In this case, we ask the model to summarize a design report for a high-efficiency axial pump used in hydraulic systems.
prompt = ("Generate a summary for a design report of a high-efficiency axial pump used in hydraulic systems. "
          "Include key design features, materials used, & performance metrics.")

# Encode the prompt using the GPT-2 tokenizer:
# The tokenizer converts the prompt into token IDs (numerical representations) that GPT-2 can process.
# The 'return_tensors' argument ensures the tokens are returned as PyTorch tensors ('pt').
inputs = tokenizer.encode(prompt, return_tensors='pt')

# Generate text using the GPT-2 model:
# The model generates text based on the provided input & certain parameters.
# - 'max_length=200' limits the generated text to 200 tokens.
# - 'num_return_sequences=1' generates one output sequence.
# - 'no_repeat_ngram_size=2' prevents the model from repeating two-word sequences, improving text diversity.
# - 'temperature=1.0' controls the randomness in the model's predictions. A higher temperature leads to more creative outputs.
# - 'top_k=50' limits the sampling to the top 50 most likely next tokens.
# - 'top_p=0.9' applies nucleus sampling, ensuring only the smallest number of tokens whose cumulative probability exceeds 0.9 are considered.
# - 'early_stopping=True' stops generation if the model determines that the text is complete.
outputs = model.generate(inputs, max_length=200, num_return_sequences=1,
                         no_repeat_ngram_size=2,
                         temperature=1.0, top_k=50, top_p=0.9,
                         early_stopping=True)

# Clear the previous output (if using in a notebook environment).
clear_output()

# Decode the generated output:
# The tokenizer converts the tokenized output back into human-readable text.
# 'skip_special_tokens=True' ensures that special tokens like '<|endoftext|>' are not included in the final output.
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Display the generated text using a typing effect or directly print it.
display_text(generated_text)
```

```python
#@title Initiate

# Clear the current Python environment to start fresh.
# This removes any existing variables, functions, or states that were previously loaded.
# The '%reset -f' command forcefully resets the current IPython environment without confirmation.
%reset -f

# Install the safetensors library.
# safetensors is a library used to handle model weights & data more efficiently & safely than typical formats.
# We use '!pip install' to install the required Python packages in Google Colab or other Jupyter environments.
!pip install --upgrade safetensors

# Import the os module to interact with the operating system (e.g., file paths, directories).
import os

# Import Google Drive module to access files from Google Drive.
from google.colab import drive

# Mount Google Drive.
# This gives us access to our Google Drive files directly in the Colab environment.
drive.mount('/content/drive')

# Check if the cache directory is already linked to a symbolic link.
# Symbolic links allow us to point one directory to another (in this case, pointing the cache directory to Google Drive).
# This step ensures that heavy files like model weights or cache are stored in Google Drive & not in Colab's temporary storage.
if not os.path.islink('/root/.cache'):
    # If the cache directory is not a symbolic link, remove the existing cache directory.
    !rm -rf /root/.cache/

    # Create a symbolic link between Colab's cache directory & a folder in Google Drive.
    # This prevents losing cached files (like model weights) every time the Colab session resets.
    !ln -s /content/drive/MyDrive/data_535743/nlp/cache /root/.cache
else:
    # If the symbolic link already exists, no action is taken.
    pass

# Import additional modules to support various functionalities.
from IPython.display import clear_output  # Allows clearing the output in Jupyter/Colab cells.
import time  # Provides time-related functions, e.g., for delays.
import sys  # Provides system-related functions, e.g., for output management.
import torch  # PyTorch library for machine learning models.
from huggingface_hub import login  # Allows logging in to Hugging Face Hub to download or upload models.
from transformers import pipeline # Importing the pipeline module from Hugging Face's transformers

from typing import *

# Define the Hugging Face API token to authenticate access.
# Hugging Face provides pre-trained models for natural language processing (NLP), computer vision, & more.
# The token ensures access to private models or advanced API functionalities.
hf_api_token = "YOUR TOKEN!!"  # Replace with your actual Hugging Face API token.

# Log in to Hugging Face using the provided API token.
# This login ensures we can download models or datasets from Hugging Face.
login(token=hf_api_token)

# Define the directory where we want to store offloaded model parts or large files.
offload_dir = "/content/offload"

# Ensure that the directory for offloaded data exists.
# If the directory does not exist, it will be created. This is useful for managing temporary files in Colab.
os.makedirs(offload_dir, exist_ok=True)

# Typing effect function that simulates typing out each character of a word, one by one.
# This is used to make printed text appear as if it's being typed out slowly.
def typing_effect(word, max_width=100, current_length=0, delay=0.01):
    """
    Prints a single word with a typing effect.

    Parameters:
    - word (str): The word to be printed with the typing effect.
    - max_width (int): The maximum width of the line before wrapping.
    - current_length (int): The current length of the printed line.
    - delay (float): Time delay between printing each character, simulating typing.

    Returns:
    - int: The updated current length after printing the word.
    """

    # Check if the current word can fit in the remaining space on the current line.
    # If it can't, move to the next line by printing a newline character ('\n').
    if current_length + len(word) + 1 > max_width:  # +1 for the space after the word
        sys.stdout.write('\n')  # Move to the next line
        current_length = 0  # Reset the length counter to start from the new line

    # Loop through each character of the word to print it out one by one.
    for char in word:
        sys.stdout.write(char)  # Print each character
        sys.stdout.flush()  # Ensure that each character is immediately displayed
        time.sleep(delay)  # Add a small delay to simulate typing
        current_length += 1  # Increment the current length as we print each character

    # After printing the word, add a space character to separate words.
    sys.stdout.write(' ')
    current_length += 1  # Account for the space after the word

    # Return the updated current length of the line.
    return current_length

# Function to display the full text with a typing effect, handling word wrapping & newlines.
def display_text(text, max_width=100, delay=0.01):
    """
    Prints the entire text with a typing effect, handling word wrapping & newlines.

    Parameters:
    - text (str): The text to be printed.
    - max_width (int): The maximum width of each line before wrapping to the next line.
    - delay (float): Time delay between printing each character.
    """

    # Split the input text into words based on spaces.
    words = text.split(' ')
    current_length = 0  # Initialize the current line length counter

    # Loop through each word to print it with a typing effect.
    for word in words:
        if '\n' in word:
            # Handle newlines that might be embedded in the text.
            parts = word.split('\n')
            for i, part in enumerate(parts):
                if i > 0:
                    sys.stdout.write('\n')  # Start a new line
                    current_length = 0  # Reset the current line length
                if part:
                    current_length = typing_effect(part, max_width, current_length, delay=delay)
        else:
            current_length = typing_effect(word, max_width, current_length, delay=delay)

    # Print a final newline after the text is printed.
    sys.stdout.write('\n')
    sys.stdout.flush()

# Clear the previous output from the notebook/cell.
clear_output()

# Display the message "Initiation is complete!" with a typing effect.
display_text("Initiation is complete!", delay=0.1)
```

```python
#@title Initiate

# Clear the current Python environment to start fresh.
# This removes any existing variables, functions, or states that were previously loaded.
# The '%reset -f' command forcefully resets the current IPython environment without confirmation.
%reset -f

# Install the safetensors library.
# safetensors is a library used to handle model weights & data more efficiently & safely than typical formats.
# We use '!pip install' to install the required Python packages in Google Colab or other Jupyter environments.
!pip install --upgrade safetensors

# Import the os module to interact with the operating system (e.g., file paths, directories).
import os

# Import Google Drive module to access files from Google Drive.
from google.colab import drive

# Mount Google Drive.
# This gives us access to our Google Drive files directly in the Colab environment.
drive.mount('/content/drive')

# Check if the cache directory is already linked to a symbolic link.
# Symbolic links allow us to point one directory to another (in this case, pointing the cache directory to Google Drive).
# This step ensures that heavy files like model weights or cache are stored in Google Drive & not in Colab's temporary storage.
if not os.path.islink('/root/.cache'):
    # If the cache directory is not a symbolic link, remove the existing cache directory.
    !rm -rf /root/.cache/

    # Create a symbolic link between Colab's cache directory & a folder in Google Drive.
    # This prevents losing cached files (like model weights) every time the Colab session resets.
    !ln -s /content/drive/MyDrive/data_535743/nlp/cache /root/.cache
else:
    # If the symbolic link already exists, no action is taken.
    pass

# Import additional modules to support various functionalities.
from IPython.display import clear_output  # Allows clearing the output in Jupyter/Colab cells.
import time  # Provides time-related functions, e.g., for delays.
import sys  # Provides system-related functions, e.g., for output management.
import torch  # PyTorch library for machine learning models.
from huggingface_hub import login  # Allows logging in to Hugging Face Hub to download or upload models.
from transformers import pipeline # Importing the pipeline module from Hugging Face's transformers

from typing import *

# Define the Hugging Face API token to authenticate access.
# Hugging Face provides pre-trained models for natural language processing (NLP), computer vision, & more.
# The token ensures access to private models or advanced API functionalities.
hf_api_token = "YOUR TOKEN!!"  # Replace with your actual Hugging Face API token.

# Log in to Hugging Face using the provided API token.
# This login ensures we can download models or datasets from Hugging Face.
login(token=hf_api_token)

# Define the directory where we want to store offloaded model parts or large files.
offload_dir = "/content/offload"

# Ensure that the directory for offloaded data exists.
# If the directory does not exist, it will be created. This is useful for managing temporary files in Colab.
os.makedirs(offload_dir, exist_ok=True)

# Typing effect function that simulates typing out each character of a word, one by one.
# This is used to make printed text appear as if it's being typed out slowly.
def typing_effect(word, max_width=100, current_length=0, delay=0.01):
    """
    Prints a single word with a typing effect.

    Parameters:
    - word (str): The word to be printed with the typing effect.
    - max_width (int): The maximum width of the line before wrapping.
    - current_length (int): The current length of the printed line.
    - delay (float): Time delay between printing each character, simulating typing.

    Returns:
    - int: The updated current length after printing the word.
    """

    # Check if the current word can fit in the remaining space on the current line.
    # If it can't, move to the next line by printing a newline character ('\n').
    if current_length + len(word) + 1 > max_width:  # +1 for the space after the word
        sys.stdout.write('\n')  # Move to the next line
        current_length = 0  # Reset the length counter to start from the new line

    # Loop through each character of the word to print it out one by one.
    for char in word:
        sys.stdout.write(char)  # Print each character
        sys.stdout.flush()  # Ensure that each character is immediately displayed
        time.sleep(delay)  # Add a small delay to simulate typing
        current_length += 1  # Increment the current length as we print each character

    # After printing the word, add a space character to separate words.
    sys.stdout.write(' ')
    current_length += 1  # Account for the space after the word

    # Return the updated current length of the line.
    return current_length

# Function to display the full text with a typing effect, handling word wrapping & newlines.
def display_text(text, max_width=100, delay=0.01):
    """
    Prints the entire text with a typing effect, handling word wrapping & newlines.

    Parameters:
    - text (str): The text to be printed.
    - max_width (int): The maximum width of each line before wrapping to the next line.
    - delay (float): Time delay between printing each character.
    """

    # Split the input text into words based on spaces.
    words = text.split(' ')
    current_length = 0  # Initialize the current line length counter

    # Loop through each word to print it with a typing effect.
    for word in words:
        if '\n' in word:
            # Handle newlines that might be embedded in the text.
            parts = word.split('\n')
            for i, part in enumerate(parts):
                if i > 0:
                    sys.stdout.write('\n')  # Start a new line
                    current_length = 0  # Reset the current line length
                if part:
                    current_length = typing_effect(part, max_width, current_length, delay=delay)
        else:
            current_length = typing_effect(word, max_width, current_length, delay=delay)

    # Print a final newline after the text is printed.
    sys.stdout.write('\n')
    sys.stdout.flush()

# Clear the previous output from the notebook/cell.
clear_output()

# Display the message "Initiation is complete!" with a typing effect.
display_text("Initiation is complete!", delay=0.1)
```

```python
#@title Generate Chat Response

def generate_chat_response(model, tokenizer, history: list, max_new_tokens: int = 256, max_prompt_length: int = 1024) -> str:
    """
    Generates a response from the conversational model based on the interaction history.

    Parameters:
    - model: The pre-trained language model used to generate responses.
    - tokenizer: The tokenizer associated with the model for encoding & decoding text.
    - history (list): A list of tuples representing the conversation history, where each tuple
                      contains (user_input, assistant_response).
    - max_new_tokens (int): The maximum number of new tokens the model should generate.
    - max_prompt_length (int): The maximum length of the input prompt in tokens before truncation.

    Returns:
    - str: The generated response from the assistant.

    For more information on how LLMs are used for dialogue generation, refer to:
    https://huggingface.co/blog/how-to-generate
    """
    # Initialize an empty string to build the conversation prompt.
    prompt = ''

    # Iterate over the conversation history to construct the prompt.
    for i, (user_input, assistant_response) in enumerate(history):
        if i == 0:
            # For the first interaction, include a system prompt to establish the assistant's role.
            prompt += f"[INST] <<SYS>>\nYou are a helpful assistant.\n<</SYS>>\n\n{user_input.strip()} [/INST] {assistant_response.strip()} "
        else:
            # For subsequent interactions, format user & assistant exchanges without the system prompt.
            prompt += f"[INST] {user_input.strip()} [/INST] {assistant_response.strip()} "

    # Encode the prompt without padding to calculate its token length.
    # This helps determine if the prompt exceeds the maximum allowed length.
    encoded_prompt = tokenizer.encode(prompt, add_special_tokens=False)

    # Check if the prompt length exceeds the `max_prompt_length`.
    if len(encoded_prompt) > max_prompt_length:
        # Calculate the number of tokens to remove to fit within `max_prompt_length`.
        tokens_to_remove = len(encoded_prompt) - max_prompt_length
        # Truncate the prompt from the start to remove the oldest tokens.
        encoded_prompt = encoded_prompt[tokens_to_remove:]
        # Decode the truncated prompt back to a string for input to the model.
        prompt = tokenizer.decode(encoded_prompt, skip_special_tokens=True)

    # Encode the prompt with padding to ensure uniform input length for the model.
    encoding = tokenizer(
        prompt,
        return_tensors='pt',
        truncation=True,
        max_length=max_prompt_length,
        padding='max_length'  # Pad the input to the maximum length.
    )

    # Move the input IDs & attention mask to the model's device (CPU or GPU).
    input_ids = encoding['input_ids'].to(model.device)
    attention_mask = encoding['attention_mask'].to(model.device)

    # Generate a response from the model using the provided input & settings.
    output_ids = model.generate(
        input_ids=input_ids,
        attention_mask=attention_mask,
        max_new_tokens=max_new_tokens,  # Limit the number of generated tokens.
        do_sample=True,  # Enable sampling to make the output less deterministic.
        temperature=0.7,  # Control randomness in the generated output (lower values for more deterministic results).
        top_p=0.9,  # Use nucleus sampling to limit token selection based on cumulative probability.
        repetition_penalty=1.1,  # Penalize repeated tokens to reduce repetitive outputs.
        eos_token_id=tokenizer.eos_token_id,  # Specify the end-of-sequence token ID.
        pad_token_id=tokenizer.pad_token_id  # Use the pad token ID for padding.
    )

    # Decode the generated output from token IDs to text.
    output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

    # Extract & return the assistant's response by removing the prompt portion from the generated text.
    assistant_response = output_text[len(prompt):].strip()

    return assistant_response
```

```python
#@title Chat
def chat(max_history: int = 10) -> None:
    """
    A function to facilitate a simple chat interaction using a pre-trained Llama-2 language model.
    The function runs an interactive loop where the user inputs text, & the model generates responses.
    The conversation continues until the user types 'exit' to end the session.

    Parameters:
    - max_history (int): The maximum number of past user-assistant interactions to keep in memory.
                         This helps maintain context without exceeding memory limits.

    For more details on building conversational agents with LLMs, see:
    https://huggingface.co/blog/transformers-conversational-agents
    """
    # Display a welcome message using a typing effect function (e.g., `display_text`).
    # This sets the context for the user, explaining how to exit the chat session.
    display_text("Welcome to the Llama-2 Chatbot! Type 'exit' to quit.")
    # Print a separator line to create a visual distinction in the chat interface.
    print("-"*100)

    # Initialize an empty list to keep track of the conversation history.
    # The history will store tuples of (user input, assistant response).
    history = []

    # Start an infinite loop to facilitate the continuous chat until the user decides to exit.
    while True:
        # Prompt the user to input their message, displaying "User:" for clarity.
        print('User:\n')
        user_input = input()

        # Check if the user has typed 'exit' (case-insensitive) to end the chat session.
        if user_input.lower() == 'exit':
            # Display a goodbye message from the assistant using a typing effect for interactivity.
            print('\nAssistant:\n')
            display_text("Goodbye!")
            print('')
            break  # Exit the loop, ending the chat session.

        # Add the user's input to the conversation history.
        # The conversation history helps maintain context across multiple turns.
        history.append((user_input, ''))

        # Ensure that the conversation history does not exceed the maximum allowed interactions.
        # This limits the history to the most recent `max_history` entries for efficiency & context management.
        if len(history) > max_history:
            history.pop(0)

        # Try to generate a response from the model based on the current conversation history.
        try:
            # Call a function `generate_chat_response` that uses the pre-trained model & tokenizer
            # to create a response based on the conversation history.
            assistant_response = generate_chat_response(model, tokenizer, history)
        except Exception as e:
            # If an error occurs during response generation, display an error message.
            display_text(f"An error occurred: {e}")
            print("-"*100)
            # Provide a fallback response indicating that an error was encountered.
            assistant_response = "I'm sorry, but I encountered an error while processing your request.\n"

        # Update the most recent entry in the history with the assistant's generated response.
        # This allows the conversation history to include both the user's input & the model's response.
        history[-1] = (user_input, assistant_response)

        # Display the assistant's response using a typing effect for a more interactive experience.
        print('\nAssistant:\n')
        display_text(f"{assistant_response}")
        # Print a separator line to delineate interactions in the chat.
        print("-"*100)
```

```python
#@title Llama Inference
# Set the maximum number of past interactions to keep in memory during the chat.
# This helps maintain context by allowing the model to refer back to previous turns while avoiding excessive memory usage.
# Keeping track of too many past interactions can increase computational cost & lead to context overflow.
max_history = 10  # Maximum number of past interactions to keep

# Call the `chat` function to initiate the conversational loop.
# The `chat` function runs an interactive session where the user inputs text & receives model-generated responses.
# - `max_history=max_history` sets the limit for how many prior interactions are stored in the conversation history.
# This is important for managing the context of the conversation & ensuring that the input sequence size does not become too large.
chat(max_history=max_history)

# Explanation:
# - The `chat` function will continue running until the user types 'exit'.
# - The conversation history allows the model to generate contextually relevant responses by considering past exchanges.
# - For more details on how maintaining conversation history enhances LLMs' performance in generating coherent responses, see:
# https://huggingface.co/blog/transformers-conversational-agents
```

```python
#@title Stable Diffusion & BLIP

# Import the necessary modules for image generation & captioning.
# `StableDiffusionPipeline` is used for generating images from text prompts.
# `BlipProcessor` & `BlipForConditionalGeneration` are used for generating image captions.
# `PIL` (Python Imaging Library) is imported to handle image processing tasks.
from diffusers import StableDiffusionPipeline
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image

# Define the ID for the pre-trained Stable Diffusion model.
# This model is used to generate images from text prompts.
# Stable Diffusion is a popular text-to-image generation model.
model_image_generation_id = "sd-legacy/stable-diffusion-v1-5"

# Define the ID for the pre-trained BLIP model for image captioning.
# BLIP (Bootstrapped Language-Image Pre-training) is used for generating descriptive captions from images.
model_image_captioning_id = "Salesforce/blip-image-captioning-large"

# Load the Stable Diffusion pipeline from the specified model ID.
# The `torch_dtype=torch.float16` argument ensures that the model uses half-precision for better performance & lower memory usage.
# The `.to("cuda")` ensures the model runs on a GPU if available, accelerating the image generation process.
model_image_generation = StableDiffusionPipeline.from_pretrained(model_image_generation_id, torch_dtype=torch.float16).to("cuda")

# Load the BLIP model for conditional image captioning from the specified model ID.
# This model is used for generating captions based on an image input.
# The `.to("cuda")` ensures that the model runs on a GPU for faster inference.
model_image_captioning = BlipForConditionalGeneration.from_pretrained(model_image_captioning_id).to("cuda")

# Load the BLIP processor for preprocessing input images & text before feeding them into the BLIP model.
# The processor handles tasks like tokenizing text & resizing images as required by the model.
processor_image_captioning = BlipProcessor.from_pretrained(model_image_captioning_id)

# Explanation:
# - The Stable Diffusion model is designed for generating high-quality images based on textual prompts.
#   For more information on Stable Diffusion, refer to: https://huggingface.co/blog/stable_diffusion
# - The BLIP model is used for generating captions from images, which helps bridge the gap between visual & language processing.
#   More details on BLIP can be found at: https://huggingface.co/Salesforce/blip-image-captioning-large
```

```python
#@title Generate An Image from A Prompt

# Define the text prompt that will be used to generate an image.
# The prompt should describe the content & details of the image to be generated.
# The more descriptive the prompt, the more specific & detailed the generated image will be.
prompt = "A detailed jet engine with all its components"

# Use the `model_image_generation` pipeline to generate an image based on the provided prompt.
# The `.images[0]` extracts the first image from the list of generated images (usually there is one image generated).
# Stable Diffusion uses the prompt to create a realistic image that matches the given description.
image = model_image_generation(prompt).images[0]

# Save the generated image to a file called "generated_image.png" in the current directory.
# Saving the image allows you to keep a record of generated outputs for future reference or sharing.
image.save("./generated_image.png")

# Display the generated image in the notebook or script output.
# This line allows the image to be displayed directly within a Jupyter Notebook or a similar interactive environment.
image

# Explanation:
# - Image generation using Stable Diffusion leverages latent diffusion techniques to create high-quality, detailed images from textual descriptions.
# - For more on text-to-image generation & how prompt engineering can influence outputs, see:
# https://huggingface.co/docs/diffusers/index
```

```python
#@title Caption the Generated Image

# Define the path to the image file to be captioned.
# The image was previously generated & saved as "generated_image.png".
img_url = './generated_image.png'

# Open the image using PIL (Python Imaging Library) & convert it to RGB format.
# This step ensures that the image is properly formatted for processing by the BLIP model.
raw_image = Image.open(img_url).convert('RGB')

# Conditional image captioning:
# Define an initial prompt or context for the BLIP model to use when generating a caption.
# This prompt helps guide the model to generate a caption that matches the context provided.
text = "The following gadget"

# Preprocess the image & text using the BLIP processor.
# This step tokenizes the text & prepares the image for input to the BLIP model.
# The output tensors are moved to the GPU for faster processing.
inputs = processor_image_captioning(raw_image, text, return_tensors="pt").to("cuda")

# Clear the current output in the Jupyter Notebook for a cleaner display.
clear_output()

# Generate a caption using the BLIP model with the processed inputs.
# The model outputs token IDs that represent the generated text.
out = model_image_captioning.generate(**inputs)

# Display a message to indicate that this is an example of conditional image captioning.
display_text('Conditional image captioning example:')

# Decode the generated token IDs to a readable string, skipping any special tokens.
# This step converts the model's output into a human-readable caption.
display_text(processor_image_captioning.decode(out[0], skip_special_tokens=True))

# Unconditional image captioning:
# Prepare the image input without any accompanying text to let the model generate a caption solely based on the image.
inputs = processor_image_captioning(raw_image, return_tensors="pt").to("cuda")

# Generate an image caption using the BLIP model without the context prompt.
out = model_image_captioning.generate(**inputs)

# Display a separator line for readability in the output.
display_text("\n" + "-" * 50 + "\n")

# Display a message to indicate that this is an example of unconditional image captioning.
display_text('Unconditional image captioning example:')

# Decode & display the unconditional caption generated by the model.
display_text(processor_image_captioning.decode(out[0], skip_special_tokens=True))

# Explanation:
# - Conditional captioning allows the model to incorporate contextual prompts in its output.
# - Unconditional captioning lets the model create a caption solely based on visual input, showcasing its ability to understand images without extra guidance.
# - For more on image captioning with BLIP & other models, visit:
# https://huggingface.co/docs/transformers/main/en/model_doc/blip
```
