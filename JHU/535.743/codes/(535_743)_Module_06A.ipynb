## **1. Super-Resolution Non-ML/AI Methods**

Image enhancement involves techniques used to improve the visual quality of an image. When dealing with non-ML/AI methods, traditional signal processing techniques come into play. Here are some common image super-resolution techniques:

> **Nearest Neighbor Upsampling**
* Description: This is the simplest form of image scaling. Each pixel in the enlarged image is assigned the value of the nearest pixel from the original image.
* Advantages: Computationally efficient, simple to implement.
* Disadvantages: Produces a blocky or pixelated result & lacks smooth transitions.

  <br>
  <figcaption>Figure: Example of Original Image (left) & Nearest Neighbor Upsampling (right)</a></figcaption>
</div>

> **Bilinear Upsampling**
* Description: This method interpolates the value of a new pixel based on a weighted average of the four nearest original pixels. It considers the distance of each pixel from the new pixel.
* Advantages: Smoother than nearest neighbor, balances simplicity & quality.
* Disadvantages: Can still produce blurry edges, does not handle high-frequency details well.

> <div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/535_743/module_06/Img_Bilinear_Upsampling.png?raw=true" width="40%">
  <br>
  <figcaption>Figure: Example of Original Image (left) & Bilinear Upsampling (right)</a></figcaption>
</div>

> **Bicubic Upsampling**
* Description: Bicubic interpolation uses the nearest 16 pixels (4x4 grid) to compute the value of a new pixel. It takes into account the values & the gradients of these pixels.
* Advantages: Produces smoother & more visually pleasing results compared to bilinear.
* Disadvantages: More computationally intensive, may introduce artifacts like ringing.

</div>

> **Lanczos Resampling**
* Description: A high-quality resampling method that uses a sinc function to interpolate the new pixel values, typically involving a wider neighborhood of pixels.
* Advantages: Preserves details better, reduces aliasing & moiré patterns.
* Disadvantages: Computationally expensive, can introduce ringing artifacts.

## **2. Denoising Non-ML/AI Methods**

Denoising aims to remove unwanted noise from an image while preserving important details. Traditional methods include:

> **Gaussian Blurring**
* Description: Applies a Gaussian function to blur the image, averaging pixel values with their neighbors to reduce noise.
* Advantages: Simple & fast, effective at reducing random noise.
* Disadvantages: Can blur important details & edges.

> <div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/535_743/module_06/Img_Gaussian_Blurring.png?raw=true" width="40%">
  <br>
  <figcaption>Figure: Example of Original Image (left) & Gaussian Blurring (right)</a></figcaption>
</div>

> **Median Filtering**
* Description: Replaces each pixel value with the median value of the neighboring pixels within a specified window.
* Advantages: Effective at removing salt-and-pepper noise, preserves edges better than Gaussian blurring.
* Disadvantages: Computationally more intensive than Gaussian blur, may not handle larger noise patterns well.

> <div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/535_743/module_06/Img_Median_Filtering.png?raw=true" width="40%">
  <br>
  <figcaption>Figure: Example of Original Image (left) & Median Filtering (right)</a></figcaption>
</div>

> **Bilateral Filtering**
* Description: A non-linear filter that considers both the spatial proximity & the intensity difference of pixels, preserving edges while reducing noise.
* Advantages: Maintains edges while smoothing out noise.
* Disadvantages: Computationally expensive, sensitive to parameter settings.

> <div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/535_743/module_06/Img_Bilateral_Filtering.png?raw=true" width="40%">
  <br>
  <figcaption>Figure: Example of Original Image (left) & Bilateral Filtering (right)</a></figcaption>
</div>

> **Non-Local Means (NLM)**
* Description: Replaces the pixel value with a weighted average of similar pixels found within a large search window, not limited to immediate neighbors.
* Advantages: Excellent at preserving textures & details while removing noise.
* Disadvantages: Very computationally intensive & complex to implement.

## **3. Edge Enhancement Non-ML/AI Methods**

Edge enhancement techniques are used to make the edges within an image more distinct.

> **Unsharp Masking**
* Description: Enhances edges by subtracting a blurred version of the image from the original image, emphasizing high-frequency components.
* Advantages: Simple to implement, effective for enhancing edges.
* Disadvantages: Can introduce halos & increase noise.

</div>

> **Laplacian Filter**
* Description: Applies a second derivative filter to highlight regions of rapid intensity change, which correspond to edges.
* Advantages: Clearly defines edges.
* Disadvantages: Can enhance noise, may need post-processing to adjust edge strength.

> **High-Pass Filtering**
* Description: Passes high-frequency components (edges) while attenuating low-frequency components (smooth areas).
* Advantages: Sharpens edges effectively.
* Disadvantages: It can make the image look noisy & requires careful parameter selection.

## **4. Histogram Equalization Non-ML/AI Methods**

Histogram equalization enhances contrast by spreading out the most frequent intensity values.

> **Global Histogram Equalization**
* Description: Adjusts the intensity distribution of the entire image to span the full intensity range.
* Advantages: Simple & effective for improving overall contrast.
* Disadvantages: Can over-enhance noise, may not be suitable for images with varying lighting conditions.

> <div align="left">
  <img src="https://github.com/mhrafiei/figures/blob/main/535_743/module_06/Img_Global_Histogram_Equalization.png?raw=true" width="40%">
  <br>
  <figcaption>Figure: Example of Original Image (left) & Global Histogram Equalization (right)</a></figcaption>
</div>

> **Adaptive Histogram Equalization (AHE)**
* Description: Applies histogram equalization to small regions (tiles) of the image, then combines them.
* Advantages: Improves local contrast, better handles varying lighting.
* Disadvantages: Computationally intensive, can produce over-contrasted areas.

## **5. Image Enhancement Non-ML/AI Methods Scripts**

```python
#@title  Non-ML/AI Image Enhancement
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Download images
!wget https://github.com/mhrafiei/figures/blob/main/535_743/module_06/535743_module06_verylowres_noisy.png
!wget https://github.com/mhrafiei/figures/blob/main/535_743/module_06/535743_module06_verylowres_original.png

# Load image paths
image_path_original = '/content/535743_module06_verylowres_original.png'
image_path_noisy = '/content/535743_module06_verylowres_noisy.png'

# Load & convert images to RGB
image_original = cv2.cvtColor(cv2.imread(image_path_original, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)
image_noisy = cv2.cvtColor(cv2.imread(image_path_noisy, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

def show_and_save(before: np.ndarray, after: np.ndarray, method_name: str) -> None:
    """
    Display & save the before & after images.

    Args:
        before (np.ndarray): The original image.
        after (np.ndarray): The processed image.
        method_name (str): The name of the processing method.
    """
    method_name = 'Image_' + '_'.join(method_name.split())
    fig, axs = plt.subplots(1, 2, figsize=(12, 6))
    axs[0].imshow(before)
    axs[0].axis('off')
    axs[1].imshow(after)
    axs[1].axis('off')
    plt.tight_layout(pad=0.05)
    plt.savefig(f'{method_name}.png', bbox_inches='tight', pad_inches=0)
    plt.show()

def nearest_neighbor_upsampling(image: np.ndarray, scale: int = 2) -> np.ndarray:
    """
    Perform nearest neighbor upsampling on the image.

    Args:
        image (np.ndarray): The input image.
        scale (int): The upsampling scale factor.

    Returns:
        np.ndarray: The upsampled image.
    """
    h, w, _ = image.shape
    resized = cv2.resize(image, (w * scale, h * scale), interpolation=cv2.INTER_NEAREST)
    return resized

def bilinear_upsampling(image: np.ndarray, scale: int = 2) -> np.ndarray:
    """
    Perform bilinear upsampling on the image.

    Args:
        image (np.ndarray): The input image.
        scale (int): The upsampling scale factor.

    Returns:
        np.ndarray: The upsampled image.
    """
    h, w, _ = image.shape
    resized = cv2.resize(image, (w * scale, h * scale), interpolation=cv2.INTER_LINEAR)
    return resized

def bicubic_upsampling(image: np.ndarray, scale: int = 2) -> np.ndarray:
    """
    Perform bicubic upsampling on the image.

    Args:
        image (np.ndarray): The input image.
        scale (int): The upsampling scale factor.

    Returns:
        np.ndarray: The upsampled image.
    """
    h, w, _ = image.shape
    resized = cv2.resize(image, (w * scale, h * scale), interpolation=cv2.INTER_CUBIC)
    return resized

def lanczos_resampling(image: np.ndarray, scale: int = 2) -> np.ndarray:
    """
    Perform Lanczos resampling on the image.

    Args:
        image (np.ndarray): The input image.
        scale (int): The resampling scale factor.

    Returns:
        np.ndarray: The resampled image.
    """
    h, w, _ = image.shape
    resized = cv2.resize(image, (w * scale, h * scale), interpolation=cv2.INTER_LANCZOS4)
    return resized

def gaussian_blurring(image: np.ndarray, ksize: tuple[int, int] = (5, 5)) -> np.ndarray:
    """
    Apply Gaussian blurring to the image.

    Args:
        image (np.ndarray): The input image.
        ksize (tuple[int, int]): The kernel size for blurring.

    Returns:
        np.ndarray: The blurred image.
    """
    blurred = cv2.GaussianBlur(image, ksize, 0)
    return blurred

def median_filtering(image: np.ndarray, ksize: int = 5) -> np.ndarray:
    """
    Apply median filtering to the image.

    Args:
        image (np.ndarray): The input image.
        ksize (int): The kernel size for filtering.

    Returns:
        np.ndarray: The filtered image.
    """
    filtered = cv2.medianBlur(image, ksize)
    return filtered

def bilateral_filtering(image: np.ndarray,
                        d: int = 9,
                        sigma_color: int = 75,
                        sigma_space: int = 75) -> np.ndarray:
    """
    Apply bilateral filtering to the image.

    Args:
        image (np.ndarray): The input image.
        d (int): Diameter of each pixel neighborhood.
        sigma_color (int): Filter sigma in the color space.
        sigma_space (int): Filter sigma in the coordinate space.

    Returns:
        np.ndarray: The filtered image.
    """
    filtered = cv2.bilateralFilter(image, d, sigma_color, sigma_space)
    return filtered

def unsharp_masking(image: np.ndarray,
                    sigma: float = 1.0,
                    strength: float = 1.5) -> np.ndarray:
    """
    Apply unsharp masking to the image.

    Args:
        image (np.ndarray): The input image.
        sigma (float): The sigma value for Gaussian blurring.
        strength (float): The strength of the sharpening effect.

    Returns:
        np.ndarray: The sharpened image.
    """
    blurred = cv2.GaussianBlur(image, (0, 0), sigma)
    sharpened = cv2.addWeighted(image, 1 + strength, blurred, -strength, 0)
    return sharpened

def laplacian_filter(image: np.ndarray) -> np.ndarray:
    """
    Apply Laplacian filter to the image.

    Args:
        image (np.ndarray): The input image.

    Returns:
        np.ndarray: The filtered image with edges highlighted.
    """
    laplacian = cv2.Laplacian(image, cv2.CV_64F)
    return cv2.convertScaleAbs(laplacian)

def high_pass_filter(image: np.ndarray) -> np.ndarray:
    """
    Apply high-pass filter to the image.

    Args:
        image (np.ndarray): The input image.

    Returns:
        np.ndarray: The high-pass filtered image.
    """
    blurred = cv2.GaussianBlur(image, (21, 21), 0)
    high_pass = cv2.subtract(image, blurred)
    return high_pass

def global_histogram_equalization(image: np.ndarray) -> np.ndarray:
    """
    Apply global histogram equalization to the image.

    Args:
        image (np.ndarray): The input image.

    Returns:
        np.ndarray: The image with equalized histogram.
    """
    img_yuv = cv2.cvtColor(image, cv2.COLOR_RGB2YUV)
    img_yuv[:, :, 0] = cv2.equalizeHist(img_yuv[:, :, 0])
    equalized = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)
    return equalized

def adaptive_histogram_equalization(image: np.ndarray,
                                    clip_limit: float = 2.0,
                                    tile_grid_size: tuple[int, int] = (8, 8)) -> np.ndarray:
    """
    Apply adaptive histogram equalization to the image.

    Args:
        image (np.ndarray): The input image.
        clip_limit (float): Threshold for contrast limiting.
        tile_grid_size (tuple[int, int]): Size of the grid for the histogram equalization.

    Returns:
        np.ndarray: The image with equalized histogram.
    """
    img_yuv = cv2.cvtColor(image, cv2.COLOR_RGB2YUV)
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    img_yuv[:, :, 0] = clahe.apply(img_yuv[:, :, 0])
    equalized = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)
    return equalized

# List of methods to apply
methods = [
    ("Nearest Neighbor Upsampling", nearest_neighbor_upsampling),
    ("Bilinear Upsampling", bilinear_upsampling),
    ("Bicubic Upsampling", bicubic_upsampling),
    ("Lanczos Resampling", lanczos_resampling),
    ("Gaussian Blurring", gaussian_blurring),
    ("Median Filtering", median_filtering),
    ("Bilateral Filtering", bilateral_filtering),
    ("Unsharp Masking", unsharp_masking),
    ("Laplacian Filter", laplacian_filter),
    ("High-Pass Filtering", high_pass_filter),
    ("Global Histogram Equalization", global_histogram_equalization),
    ("Adaptive Histogram Equalization", adaptive_histogram_equalization)
]

# Apply each method & save the results
for method_name, method in methods:
    if method_name in {"Gaussian Blurring", "Median Filtering", "Bilateral Filtering"}:
        result_image = method(image_noisy)
        show_and_save(image_noisy, result_image, method_name)
    else:
        result_image = method(image_original)
        show_and_save(image_original, result_image, method_name)
```
