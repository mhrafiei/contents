## **2. Object Oriented Programmin, Git, & GitHub**

In this section, we will see simple Object Oriented Programming (OOP) Python examples, & learn how to work with git & GitHub for version, repository, & project management. Here is a list of git commands we are going to use in this section (note that these commands are valid as of July 2024 & potential future changes might occur):

> Shows the username configured in git, useful to verify your identity in Git records:
```sh
git config user.name

> Shows the email configured in git, important for associating commits with your identity:
```sh
git config user.email

> Sets your username for all your Git operations, which labels your commits with your name:
```sh
git config --global user.name "John Doe"

> Sets your email for all your Git operations, used for identifying commit authorship:
```sh
git config --global user.email "doe1234.535743.jhu@gmail.com"

> Displays the current working directory's status, including untracked, modified, or staged files:
```sh
git status

> Initializes a new Git repository in the current directory, creating the .git folder to track changes:
```sh
git init

> Copies a Git repository from a remote source, typically from a URL pointing to a project on GitHub:
```sh
git clone <repository-url>

> Adds specified files to the staging area, preparing them to be included in the next commit:
```sh
git add <file-name>

> Records staged changes to the repository along with a descriptive message about what was changed:
```sh
git commit -m "commit message"

> Sends local branch commits to the remote repository, updating it with your changes:
```sh
git push

> Retrieves changes from a remote repository & tries to merge them into the current branch:
```sh
git pull

> Displays a list of all past commits along with details like author, date, & commit message:
```sh
git log

> Shows the differences between files or commits, useful for seeing what has changed:
```sh
git diff

> Lists all the remote connections set up with other repositories, displaying their URLs:
```sh
git remote -v

> Adds a new remote connection under the given name pointing to the specified URL:
```sh
git remote add <name> <url>

> Adds a 'default' remote named 'origin'. This is typically used to point to the primary repository:
```sh
git remote add origin <url>

> Changes the name of an existing remote. Useful for correcting errors or updating remote names:
```sh
git remote rename <old> <new>

> Deletes the association to a remote repository. Useful when you no longer need to connect to a remote:
```sh
git remote remove <name>

> Creates a new public repository on GitHub from the command line, initiating a repository on your account:
```sh
gh repo create --public <repo-name>

> Lists all repositories under your GitHub profile, helpful for managing or reviewing your projects:
```sh
gh repo list

> Removes the 'origin' remote, often done before reassigning the origin to a new URL:
```sh
git remote remove origin

> Adds a new 'origin' remote with a URL that includes an access token for authentication:
```sh
git remote add origin https: ```sh//token@github.com/mhrafiei/sib

> Pushes your commits to the newly set origin & sets the upstream branch for tracking:
```sh
git push -u origin main

> Displays the remote connections again to confirm changes:
```sh
git remote -v

> Renames the current branch to 'main', often done to establish 'main' as the primary development branch:
```sh
git branch -M main

> Downloads objects & refs from another repository. Useful for syncing local data with remote changes:
```sh
git fetch

> Fetches updates from a specific branch of a specified remote:
```sh
git fetch <remote> <branch>

> Specific fetch from 'origin' targeting a particular branch:
```sh
git fetch origin <branch>

> Checks out the 'main' branch, ensuring you're working on the primary branch:
```sh
git checkout main

> An alternative command to switch branches, modern replacement for 'checkout':
```sh
git switch main

> Lists all local branches, letting you see what branches are available for checkout:
```sh
git branch

> Creates a new branch with the given name, allowing you to diverge from the main line of development:
```sh
git branch <name-new-branch>

> Renames a branch, useful when you need to change the purpose or clarify the role of a branch:
```sh
git branch -m <old-name> <new-name>

> Deletes a branch that is no longer needed, cleaning up your project branches:
```sh
git branch -d <branch-to-delete>

> Switches to another branch, moving your workspace to the state of that branch:
```sh
git switch <name-branch>

> Checks out a branch, changing your working directory to reflect the state of that branch:
```sh
git checkout <branch-name>

> Merges changes from one branch (e.g., 'new_branch') into your current branch (usually 'main'):
```sh
git switch main
git merge new_branch

> Provides a concise view of the commit log, showing each commit as a single line:
```sh
git log --oneline

> Restores files from a specific commit, effectively undoing changes since then:
```sh
git checkout <commit-hash>

> Moves back to the main branch after checking out a specific commit:
```sh
git checkout main

> Restores a specific file to its state at the HEAD commit, useful for undoing recent changes:
```sh
git restore <file-name>

> Restores a file to how it was N commits ago, allowing you to backtrack changes:
```sh
git restore --source HEAD~N <file-name>

> Unstages a file, keeping the changes in your working directory but not ready for commit:
```sh
git restore --staged <file-name>

> Removes files from staging & optionally undoes changes in your working directory:
```sh
git reset <file-path>

> Resets your HEAD pointer to a specific commit, & optionally undoes changes:
```sh
git reset <commit-hash>

> Resets your repository to a specific commit, discarding all changes in the directory:
```sh
git reset <commit-hash> --hard

> Hard resets the repository to one commit prior, losing all changes in the current commit:
```sh
git reset --hard HEAD~1

> Resets the repository to the initial state, erasing all commits & changes:
```sh
git reset --hard

> Stashes your changes, allowing you to save them temporarily without committing:
```sh
git stash push -u

> Alternative syntax to stash, including untracked files:
```sh
git stash push --include-untracked

> Lists all stashed changes, giving you an overview of what has been stashed:
```sh
git stash list

> Applies the most recent stash, restoring the changes to your working directory:
```sh
git stash pop

> Lists all remotes, showing you the connections set up with other repositories:
```sh
git remote

> Lists all remotes with detailed URLs, helping you verify the links:
```sh
git remote -v

> Adds a new remote with a specified name, linking to a repository URL:
```sh
git remote add <name> <url>

> Removes a specified remote, severing the connection to that repository:
```sh
git remote rm <name>

> Setup a .gitignore file to specify patterns of files to ignore, like logs or build directories:
> You manually create or edit the .gitignore file in your project directory.

> Forcefully removes all files & directories in the current directory:
```sh
rm -rf *




# <font color="#418FDE" size="6.5" uppercase>**B: Programming with VSC, Git, GitHub, & Codespaces**</font>
----

In this lecture, you learned to:

* Develop Object Oriented Programming (OOP) scripts in Visual Studio Code (VSC) desktop & web (codespaces).
* Develop & manage repositories using Git & GitHub.


In the next Module (Module 9), we go over "Object Detection, Part 1."
