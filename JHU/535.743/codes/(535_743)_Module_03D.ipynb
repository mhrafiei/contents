## **1. Self-Supervised Learning Concept**

> Self-Supervised Learning (SSL), also termed representation learning, is a burgeoning field in machine learning that addresses the challenge of utilizing unlabeled data. Consider a dataset with ten million images of cats and dogs, of which only 10,000 (0.1%) have labels. Traditional supervised learning approaches would overlook the vast majority of this dataset, focusing only on the labeled portion. However, SSL techniques aim to harness the information latent in the unlabeled data. By doing so, SSL can significantly enhance the learning process, providing a richer, more generalized model that benefits from the full spectrum of available data.

> In the context of SSL, the process begins by designing a pretext task that the model can use to learn from the unlabeled data. This task is structured to enable the model to capture meaningful patterns and features from the data without requiring any labels. For example, the model might learn to predict the next frame in a video sequence or to colorize a grayscale image. Through this task, the model develops a base representation of the data, gaining a fundamental understanding of its structure and variability. This learned representation forms the foundation upon which further learning can be built.

> Once the pretext task is completed, the model, now enriched with knowledge gleaned from the unlabeled data, is fine-tuned on the labeled dataset in a downstream task. This task is typically the primary objective, such as classifying images as cats or dogs. Because the model has already learned a substantial amount about the visual features common to both categories from the pretext task, it require.

> How does one construct a pretext model utilizing both labeled and unlabeled data? There are two main SSL strategies for generating such a model: contrastive learning and generative learning.

## **2. Contrastive Learning**

> Contrastive learning is a technique in machine learning that teaches a model to understand the difference between similar and dissimilar examples. It operates by creating pairs of related data points—known as positive pairs—and pairs of unrelated data points—known as negative pairs. The goal is to adjust the model so that it learns to bring the positive pairs closer together in the representation space while pushing the negative pairs apart. This process relies heavily on data augmentation to generate varied examples of data points that the model can learn from.

> The following figure illustrates several image augmentation techniques, such as cutout, rotation, and adding noise, which are used to create these positive pairs by altering the original images in a controlled manner.

> The augmentation process is crucial to contrastive learning, as it generates the transformed versions of the original data. For instance, by applying the cutout augmentation, parts of an image are masked, creating a new version of the image that retains its essential features while differing in some respects from the original. When this technique is applied across a dataset, it effectively doubles the number of images, as each original image is paired with its augmented counterpart. The model uses these pairs to learn which features are important for identifying the subject of an image, regardless of the presence of alterations or occlusions.

> In this domain, augmentations can include modifications to the amplitude, temporal shifts, or the addition of noise to the original signals. These transformations create diverse scenarios from which the model can learn, enhancing its ability to generalize and recognize patterns despite variations in the signal's presentation. By training with both original and augmented data, the model becomes more robust to changes and can make more accurate predictions or classifications.

```python
#@title Plot Function for Image Augmentation
'''
Runtime: CPU
mt.pyplot: https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.html
'''

# Import the required module for data visualization
import matplotlib as mt

# Define a function to plot an original image and its augmented version side by side
def fun_plot(image_original, image_augmented, general_title):
    # Create a list of titles for the subplots
    titles = ['original', 'augmented']
    # Create a list of the images to be displayed
    images = [image_original, image_augmented]

    # Initialize a figure and a set of subplots with 1 row and 2 columns
    fig, ax = mt.pyplot.subplots(1, 2, figsize=(4, 3))

    # Set the main title of the figure, with the title text capitalized
    fig.suptitle(general_title.title())

    # Adjust the layout of the plots to make sure there is no overlap
    fig.tight_layout()
    # Adjust the spacing between the subplots to minimize white space
    mt.pyplot.subplots_adjust(wspace=0.05, hspace=0.05)

    # Iterate over the indices (0 and 1) of the two images
    for i0 in range(2):
        # Display the ith image in the ith subplot
        ax[i0].imshow(images[i0])
        # Set the title of the ith subplot, with the title text capitalized
        ax[i0].set_title(titles[i0].title())
        # Remove the x-axis ticks by setting them to an empty list
        ax[i0].set_xticks([])
        # Remove the y-axis ticks by setting them to an empty list
        ax[i0].set_yticks([])

    # Display the figure with the two subplots
    mt.pyplot.show()
```

```python
#@title Load and Process the CIFAR-10 Data
'''
Runtime: CPU
'''

# Import TensorFlow
import tensorflow as tf

# Load the CIFAR-10 dataset
(datain_tr, dataou_tr), (datain_te, dataou_te) = tf.keras.datasets.cifar10.load_data()

# Normalize the images to [0, 1]
datain_tr, datain_te = datain_tr / 255.0, datain_te / 255.0

# Print the shape of the normalized training data
print('Shape of datain_tr: {}'.format(datain_tr.shape))
```

```python
#@title Example of Image Augmentations - Random Flip
'''
Runtime: CPU
tf.keras.layers.RandomFlip:         https://www.tensorflow.org/api_docs/python/tf/keras/layers/RandomFlip.
'''

# Import TensorFlow, a machine learning and deep learning library
import tensorflow as tf

# Define an image augmentation function that applies random flips both horizontally and vertically
fun_augment = tf.keras.layers.RandomFlip(mode='horizontal_and_vertical')

# Select the last image from the training dataset as the original image
image_original = datain_tr[-1, ...]
# Apply the defined augmentation function to the original image to get the augmented image
image_augmented = fun_augment(image_original)

# Use the previously defined function to plot the original and augmented images with the title 'Random Flip'
fun_plot(image_original, image_augmented, 'Random Flip')
```

```python
#@title Example of Image Augmentations - Random Zoom
'''
Runtime: CPU
tf.keras.layers.RandomZoom:         https://www.tensorflow.org/api_docs/python/tf/keras/layers/RandomZoom.
'''

# Import TensorFlow, assuming it's already included in the context
import tensorflow as tf

# Define an augmentation function using TensorFlow's Keras API to apply random zoom
# height_factor = 0.75 means the image will be zoomed in by up to 75% randomly
fun_augment = tf.keras.layers.RandomZoom(height_factor = .75)

# Select the first image from the CIFAR-10 training dataset for augmentation
# datain_tr is assumed to be previously loaded and preprocessed CIFAR-10 dataset
image_original = datain_tr[-1,...]

# Apply the random zoom augmentation to the original image
# The augmentation function expects a batch of images, but here it's applied to a single image
image_augmented = fun_augment(image_original)

# Visualize the original and augmented images using a custom plotting function
# 'Random Zoom' is the title for the visualization
fun_plot(image_original, image_augmented, 'Random Zoom')
```

```python
#@title Example of Image Augmentations - Random Crop and Resize
'''
Runtime: CPU
tf.keras.layers.RandomCrop:         https://www.tensorflow.org/api_docs/python/tf/keras/layers/RandomCrop.
tf.keras.layers.Resizing:           https://www.tensorflow.org/api_docs/python/tf/keras/layers/Resizing.
'''

# Import TensorFlow library
import tensorflow as tf

# Define the first augmentation operation: Randomly crop images to 20x20 pixels
fun_augment_01  = tf.keras.layers.RandomCrop(height = 20, width = 20)
# Define the second augmentation operation: Resize the images back to their original dimensions
fun_augment_02  = tf.keras.layers.Resizing(height = datain_tr.shape[1], width = datain_tr.shape[2])

# Combine the two augmentation operations into a sequential model
fun_augment     = tf.keras.Sequential([fun_augment_01, fun_augment_02])

# Select the first image from the dataset as the original image
image_original  = datain_tr[-1, ...]

# Apply the combined augmentations to the original image
image_augmented = fun_augment(image_original)

# Use the previously defined function to plot the original and augmented images
fun_plot(image_original, image_augmented, 'Crop & Resize')
```

```python
#@title Example of Image Augmentations - Random Saturation (Jitter)
'''
Runtime: CPU
tf.image.random_saturation:         https://www.tensorflow.org/api_docs/python/tf/image/random_saturation.
'''

# Use TensorFlow for image processing
import tensorflow as tf

# Define an augmentation function using a lambda expression. This function applies random saturation
# to the input image. The image's pixel values are first scaled to [0, 255] (assuming they were in [0, 1]),
# then the saturation is adjusted, and finally, the image is scaled back to [0, 1].
fun_augment = lambda image: tf.cast(tf.image.random_saturation(image*255,
                                                               lower=1,
                                                               upper=5),
                                    dtype=tf.float32) / 255

# Select the last image from the dataset as the original image
image_original = datain_tr[-1, ...]

# Apply the augmentation function to the original image to create the augmented image
image_augmented = fun_augment(image_original)

# Use the previously defined function to plot the original and augmented images,
# with a title indicating the type of augmentation applied
fun_plot(image_original, image_augmented, 'Random Saturation (Jitter)')
```

```python
#@title Example of Creating Contrastive Training Inputs
'''
Runtime: CPU
tf.keras.layers.RandomRotation:         https://www.tensorflow.org/api_docs/python/tf/keras/layers/RandomRotation.
'''

# Import TensorFlow library
import tensorflow as tf

# Define the augmentation operation: Randomly rotate images within a specified factor
fun_augment = tf.keras.layers.RandomRotation(factor = 0.2)

# Apply the augmentation to the entire training dataset
datain_tr_augmented = fun_augment(datain_tr)

# Plot the first five pairs of original and augmented images for visualization
for i0 in range(5):
    fun_plot(datain_tr[i0, ...], datain_tr_augmented[i0, ...], 'Random Rotation')

# Concatenate the original and augmented datasets to form a larger training set
# This new dataset combines original and modified images for contrastive learning
datain_tr_prx = tf.concat([datain_tr, datain_tr_augmented], axis=0)

# Print the shapes of the original, augmented, and concatenated datasets
print("Shape of original training inputs: {}".format(datain_tr.shape))
print("Shape of augmented training inputs: {}".format(datain_tr_augmented.shape))
print("Shape of pretext inputs: {}".format(datain_tr_prx.shape))
```

```python
#@title Example of Creating Contrastive Training Inputs Using Random Augmentations & TensorFlow Dataset API
'''
Runtime: CPU
tf.image.random_flip_left_right:       https://www.tensorflow.org/api_docs/python/tf/image/random_flip_left_right
tf.keras.layers.RandomRotation:       https://www.tensorflow.org/api_docs/python/tf/keras/layers/RandomRotation.
tf.image.rot90:                       https://www.tensorflow.org/api_docs/python/tf/image/rot90
tf.image.random_crop:                 https://www.tensorflow.org/api_docs/python/tf/image/random_crop
tf.data.Dataset.from_tensor_slices:   https://www.tensorflow.org/api_docs/python/tf/data/Dataset
'''

import tensorflow as tf

# Define a function to apply random augmentations
def fun_random_augment(image, label):
    # Random flip left-right
    image = tf.image.random_flip_left_right(image)
    # Random rotation
    image = tf.image.rot90(image, k=tf.random.uniform(shape=[], minval=0, maxval=4, dtype=tf.int32))
    # Random zoom, cropping to simulate zoom
    image = tf.image.random_crop(image, size=[28, 28, 3])
    return image, label

# Prepare the training dataset using the TensorFlow Dataset API
data_tr = tf.data.Dataset.from_tensor_slices((datain_tr, dataou_tr))
data_tr = data_tr.shuffle(buffer_size=1024).map(fun_random_augment).batch(32).prefetch(tf.data.experimental.AUTOTUNE)

# The testing dataset does not need augmentation but let's prepare it similarly
data_te = tf.data.Dataset.from_tensor_slices((datain_te, dataou_te)).batch(32).prefetch(tf.data.experimental.AUTOTUNE)
```

## **3. Supervised Contrastive Learning**

In Supervised Contrastive Learning, the process involves pseudo-labeling an augmented image as positive, denoted by class 1, while the original counterpart is tagged as negative, represented by class 0. The primary objective of the pretext task is for a model, referred to as the pretext model, to master the differentiation between these pseudo-labels across a substantial input dataset with no original label. Following this, the learned pretext model undergoes a transfer-learning and fine-tuning phase, leveraging a small set of labeled data pertinent to the downstream task to enhance its application and performance in specific tasks.

```python
#@title Example of Creating Pseudo Labels
'''
Runtime: CPU
'''

dataou_tr_prx_positive = tf.ones((datain_tr.shape[0],1))
dataou_tr_prx_negative = tf.zeros((datain_tr_augmented.shape[0],1))

#based on the order of data points in datain_tr_prx
dataou_tr_prx = tf.concat([dataou_tr_prx_negative, dataou_tr_prx_positive], axis = 0)

#binary categorical for SoftMax:
dataou_tr_prx = tf.keras.utils.to_categorical(dataou_tr_prx)

print("Shape of training pretext output: {}".format(dataou_tr_prx.shape))

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

## **4. Unsupervised Contrastive Learning**


In Unsupervised Contrastive Learning, no pseudo-labeling is required, a model is trained to distinguish between similar (positive) and dissimilar (negative) pairs of data points without the use of explicitly provided labels. This method is particularly valuable when dealing with large datasets where manual labeling is impractical or infeasible.

In the context of images, an original image serves as an anchor, and through data augmentation techniques, a positive pair is created. This augmented version, while altered, still retains the core characteristics of the original image, allowing the model to learn that these two are similar. Dissimilar images, or negative pairs, are concurrently presented to the model, which could be other non-augmented images in the batch. The objective is for the model to learn an embedding space where positive pairs are closer together, while negative pairs are further apart, enhancing the model's ability to understand the essence of the data.

The training process involves a contrastive loss function, such as the triplet loss or the hinge loss, which quantifies the distance between positive and negative pairs in the embedding space. By minimizing this loss, the model becomes adept at identifying the defining features of an image that are invariant to the transformations applied during augmentation. Once this pretext task is accomplished, the learned representations can be used to improve performance on downstream tasks, such as classification or object detection, even with limited labeled data. This transfer of knowledge from the self-supervised task to a supervised task is where the model's utility is fully realized, allowing it to make accurate predictions by leveraging the rich, complex representations developed during the self-supervised learning phase.

## **3. Generative Learning**

> In Generative Self-Supervised Learning (SSL), the process begins by leveraging a model developed through unsupervised learning techniques, with or without applying data augmentation strategies. A common approach involves employing a generative adversarial network (GAN), similar to the figure, for a pretext task. This task entails training the GAN to produce synthetic data, such as images of cats and dogs, that closely mimics the distribution observed in the existing dataset. Following this, the discriminator component of the GAN is adapted and refined using a small amount of labeled data for a specific downstream application, thereby enhancing its performance on the targeted task.

> Another strategy involves the utilization of an autoencoder, particularly its encoder component, which has been trained to accurately replicate the distribution of the input data. This encoder is then extracted and similarly fine-tuned with a limited set of labeled data pertinent to the downstream task. This fine-tuning process allows the model to maintain its general capabilities while being optimized for a specific application, thereby leveraging the strengths of generative models in SSL frameworks.
