## **1. High-Level Introduction to Google Cloud**

Google Cloud is a comprehensive suite of cloud computing services offered by Google, designed to empower businesses and developers with robust, scalable, and secure infrastructure. It provides a wide array of solutions, including computing power, data storage, machine learning tools, and big data analytics, all accessible through the same infrastructure that Google uses internally for its global operations. With services such as Google Kubernetes Engine (GKE), BigQuery, and Firebase, Google Cloud enables organizations to build, deploy, and scale applications seamlessly. Known for its commitment to open-source technologies, cutting-edge AI capabilities, and a strong emphasis on security, Google Cloud is a versatile platform that supports industries ranging from startups to enterprise-level corporations.

### **1.1. Google Cloud Free Tier**

As of December 2024, the Google Cloud Free Tier provides new users with an excellent opportunity to explore and experiment with Google Cloud services at no cost for the first 90 days. As part of this offer, users receive $300 in free credits, which can be applied to most Google Cloud products and services. The free tier also includes access to Always Free resources, which provide limited access to specific services even after the trial ends. These include small-scale Compute Engine instances, Cloud Storage, and BigQuery usage, enabling users to continue building and testing projects with no cost beyond the trial period. ### **1.2. Google Cloud Project**

In Google Cloud, a project is a logical container for resources such as virtual machines, databases, storage, and APIs. It provides a way to organize and manage resources, billing, and permissions in a cohesive manner. Each project is assigned a Project ID, a globally unique identifier that is essential for interacting with Google Cloud services and APIs. For example, if you’re building a web application that uses Google Cloud Storage for hosting files and a Cloud SQL database for data management, all these resources would be grouped under a single project. This makes it easier to monitor usage, apply security policies, and manage billing for all components. The Project ID plays a crucial role in API calls and configuration, ensuring requests are routed to the correct project within the vast Google Cloud ecosystem.

### **1.4. IAM & Admin**

Google Cloud Identity and Access Management (IAM) & Admin is a critical component for managing access and security within the Google Cloud environment. IAM enables administrators to define and enforce permissions for users, groups, and service accounts at a highly granular level. By assigning roles to entities, administrators can ensure that users have access only to the resources they need, following the principle of least privilege. Google Cloud offers predefined roles for common tasks and custom roles for specific requirements, ensuring flexibility in access control. The Admin tools complement IAM by providing centralized management of resources, billing accounts, and organizational policies. Features like IAM Policies, Policy Analyzer, and Audit Logs help maintain compliance, monitor activities, and identify potential risks. Together, IAM & Admin ensure that organizations can securely manage access and maintain control over their cloud resources in a scalable, efficient, and compliant manner.

The Google Cloud Service Agent role is a specialized Identity and Access Management (IAM) role that allows Google-managed services to perform necessary operations within your project. Service agents are automatically created and managed by Google for specific services, such as Cloud Pub/Sub, Compute Engine, or Kubernetes Engine, to access and manage the resources required for their functionality. These roles follow the principle of least privilege, granting only the permissions needed for the associated service to operate effectively. For example, the Compute Engine Service Agent role enables Compute Engine to manage instances, disks, and networks within a project. Administrators can view, but not modify, these roles to ensure the seamless operation of managed services while maintaining a secure environment. For more details, visit the [Service Agents Documentation](https://cloud.google.com/iam/docs/service-agents).

Here we go over a number of Service Agent roles in Google Cloud:

1. **roles/pubsub.serviceAgent**: The Cloud Pub/Sub Service Agent role allows the Pub/Sub service to access necessary resources within your project to function correctly. [Service Agents Documentation](https://cloud.google.com/iam/docs/service-agents)

2. **roles/run.serviceAgent**: The Cloud Run Service Agent role enables the Cloud Run service to manage and deploy applications within your project. [Cloud Run IAM Roles](https://cloud.google.com/run/docs/reference/iam/roles)

3. **roles/compute.serviceAgent**: The Compute Engine Service Agent role permits the Compute Engine service to operate and manage virtual machine instances and related resources in your project. [Compute Engine IAM Roles](https://cloud.google.com/compute/docs/access/iam)

4. **roles/editor**: The Editor role provides read and write access to most resources within a project, allowing users to create and modify resources. [IAM Roles Overview](https://cloud.google.com/iam/docs/understanding-roles)

5. **roles/container.serviceAgent**: The Kubernetes Engine Service Agent role allows the Google Kubernetes Engine service to manage Kubernetes clusters and associated resources in your project. [Service Agents Documentation](https://cloud.google.com/iam/docs/service-agents)

6. **roles/owner**: The Owner role grants full control over all resources in a project, including the ability to manage roles and permissions. [IAM Roles Overview](https://cloud.google.com/iam/docs/understanding-roles)

7. **roles/iam.serviceAccountUser**: The Service Account User role allows users to attach service accounts to resources like virtual machines, enabling those resources to act as the service account. [Service Account Permissions](https://cloud.google.com/iam/docs/service-account-permissions)

8. **roles/aiplatform.serviceAgent**: The Vertex AI Service Agent role enables the Vertex AI service to access necessary resources within your project for machine learning operations. [Service Agents Documentation](https://cloud.google.com/iam/docs/service-agents)

9. **roles/storage.admin**: The Storage Admin role provides full control over Cloud Storage resources, including the ability to create, modify, and delete buckets and objects. [Cloud Storage IAM Roles](https://cloud.google.com/storage/docs/access-control/iam-roles)

### **1.5 Compute Engine**

Google Cloud Compute Engine is a robust Infrastructure-as-a-Service (IaaS) offering that provides scalable virtual machines (VMs) for running workloads in the cloud. With Compute Engine, users can create and manage custom VMs tailored to their specific needs, choosing from a variety of machine types, operating systems, and configurations. It supports both Linux and Windows environments and offers options for preemptible VMs for cost-efficient batch processing and GPU-optimized instances for high-performance computing tasks like machine learning and scientific simulations. Compute Engine integrates seamlessly with other Google Cloud services, such as Cloud Storage and BigQuery, enabling efficient data workflows. Advanced features like autoscaling, load balancing, and live migration ensure high availability and reliability for applications. Whether running simple applications or powering complex enterprise workloads, Compute Engine provides the flexibility, performance, and security needed for modern cloud computing.

### **1.6. Kubernetes Engine**

Google Kubernetes Engine (GKE) is a fully managed service designed to simplify the deployment, management, and scaling of containerized applications using Kubernetes, the industry-leading container orchestration platform. GKE offers a seamless environment for developers and DevOps teams to run applications in highly available, scalable, and secure Kubernetes clusters. With automated features such as node management, auto-scaling, and self-healing, GKE reduces the operational burden of managing Kubernetes infrastructure. It also integrates seamlessly with other Google Cloud services, including Cloud Storage, BigQuery, and Cloud Monitoring, providing a unified ecosystem for building and scaling cloud-native applications. GKE supports hybrid and multi-cloud deployments through tools like Anthos, enabling businesses to run workloads across environments with consistency. As a pioneer in Kubernetes development, Google’s GKE ensures top-notch performance, security, and reliability for both small-scale and enterprise-level workloads.

### **1.7. Cloud Storage**

Google Cloud Storage is a highly scalable and secure object storage service designed to handle a wide range of use cases, from simple file storage to complex data archiving and big data analytics. It offers multiple storage classes—Standard, Nearline, Coldline, and Archive—allowing users to optimize costs based on the frequency and speed of access needed. Cloud Storage is ideal for storing unstructured data like images, videos, backups, and logs, as well as for serving static content for websites and applications. It features strong data durability, encryption by default, and seamless integration with other Google Cloud services, such as BigQuery for analytics and Compute Engine for hosting applications. With features like lifecycle management, versioning, and fine-grained access controls through IAM, Cloud Storage ensures cost-efficiency, ease of use, and security for managing data at any scale.

### **1.8. Service Accounts**

A Google Cloud Service Account is a specialized identity used by applications, virtual machines, or other resources to interact securely with Google Cloud APIs and services. Unlike user accounts, service accounts are intended for non-human interactions and provide a secure way to grant permissions to applications without exposing personal credentials. Each service account is associated with a private key that allows it to authenticate and access resources, adhering to the principle of least privilege through role-based access control (RBAC). Administrators can create, manage, and monitor service accounts using Google Cloud IAM, assigning specific roles to control the scope of their access. Service accounts are essential for automating tasks, integrating third-party applications, or running workloads that require API access. By using these accounts, organizations can ensure secure, controlled, and efficient interactions between their cloud services and applications.

### **1.9. Quotas & System Limits**

Google Cloud Quotas and System Limits are essential mechanisms designed to ensure fair usage, maintain system performance, and protect users from unexpected overages. Quotas define the maximum amount of resources a project can use, such as API calls, storage, or virtual machine instances, helping to prevent accidental overuse. These limits are applied at the project, region, or global level, depending on the service. Quotas can often be increased by submitting a request to Google Cloud support, based on specific needs. In addition to quotas, Google Cloud enforces system limits, which are fixed constraints on resources or operations, such as the maximum number of networks per project or database size. These limits ensure the platform’s reliability and scalability while guiding users in architecting their applications. By monitoring usage in the Google Cloud Console and setting alerts, users can effectively manage their quotas and ensure smooth operations within their projects.

### **2.1. Setup**

The following Bash script automates the setup of a Google Cloud project for deploying and managing a Flask application by defining project-specific variables such as PROJECT_ID, SERVICE_ACCOUNT_NAME, and REGION. It sets the active project in Google Cloud CLI, creates a service account, and assigns it multiple IAM roles for accessing services like [Pub/Sub](https://cloud.google.com/pubsub?hl=en), Cloud Run, Compute Engine, and Cloud Storage. The script enables a comprehensive list of Google Cloud APIs, generates a service account key saved locally, and configures environment variables for authentication. It also creates a Cloud Storage bucket, grants permissions to the service account and public users, and sets up [CORS](https://cloud.google.com/storage/docs/using-cors) for the bucket. The script ensures that the necessary infrastructure and permissions are ready for deploying a Flask app on Google Cloud.


```bash
#!/bin/bash

# Define essential variables for Google Cloud setup
# Replace this with your Google Cloud project ID
PROJECT_ID="DEFINE IT HERE"

# Replace with a descriptive name for the service account to be created
SERVICE_ACCOUNT_NAME="DEFINE IT HERE"

# Construct the service account email using the project ID and service account name
SERVICE_ACCOUNT_EMAIL="${SERVICE_ACCOUNT_NAME}@${PROJECT_ID}.iam.gserviceaccount.com"

# Replace with your Gmail address (required for managing project permissions)
USER_EMAIL="DEFINE IT HERE (YOUR CLASS GMAIL)"

# File path where the service account's JSON key will be saved
KEY_FILE_PATH="./key.json"

# Name for the Google Cloud Storage bucket, using the project ID for uniqueness
BUCKET_NAME="${PROJECT_ID}-bucket"

# Google Cloud Storage bucket URL format, prefixed with "gs://"
BUCKET_URL="gs://${BUCKET_NAME}"

# Specify the Google Cloud region where the bucket will be created
REGION="us-central1"

# Define a list of APIs required for the project; each API will be enabled later
APIS=(
    "aiplatform.googleapis.com"            # AI Platform API
    "analyticshub.googleapis.com"          # Analytics Hub API
    "apigateway.googleapis.com"            # API Gateway
    "artifactregistry.googleapis.com"      # Artifact Registry API
    "autoscaling.googleapis.com"           # Autoscaling API
    "bigquery.googleapis.com"              # BigQuery API
    "bigqueryconnection.googleapis.com"    # BigQuery Connection API
    "bigquerydatapolicy.googleapis.com"    # BigQuery Data Policy API
    "bigquerymigration.googleapis.com"     # BigQuery Migration API
    "bigqueryreservation.googleapis.com"   # BigQuery Reservation API
    "bigquerystorage.googleapis.com"       # BigQuery Storage API
    "cloudapis.googleapis.com"             # General Cloud APIs
    "cloudresourcemanager.googleapis.com"  # Cloud Resource Manager API
    "cloudtrace.googleapis.com"            # Cloud Trace API
    "compute.googleapis.com"               # Compute Engine API
    "container.googleapis.com"             # Kubernetes Engine API
    "containerfilesystem.googleapis.com"   # Container File System API
    "containerregistry.googleapis.com"     # Container Registry API
    "dataform.googleapis.com"              # Dataform API
    "dataplex.googleapis.com"              # Dataplex API
    "datastore.googleapis.com"             # Datastore API
    "dns.googleapis.com"                   # Cloud DNS API
    "gkebackup.googleapis.com"             # Kubernetes Backup API
    "iam.googleapis.com"                   # Identity and Access Management API
    "iamcredentials.googleapis.com"        # IAM Credentials API
    "logging.googleapis.com"               # Cloud Logging API
    "monitoring.googleapis.com"            # Cloud Monitoring API
    "networkconnectivity.googleapis.com"   # Network Connectivity API
    "oslogin.googleapis.com"               # OS Login API
    "privilegedaccessmanager.googleapis.com"  # Privileged Access Manager API
    "pubsub.googleapis.com"                # Pub/Sub API
    "run.googleapis.com"                   # Cloud Run API
    "secretmanager.googleapis.com"         # Secret Manager API
    "servicemanagement.googleapis.com"     # Service Management API
    "serviceusage.googleapis.com"          # Service Usage API
    "sql-component.googleapis.com"         # Cloud SQL Component API
    "storage-api.googleapis.com"           # Cloud Storage API
    "storage-component.googleapis.com"     # Cloud Storage Component API
    "storage.googleapis.com"               # Cloud Storage API
    "texttospeech.googleapis.com"          # Text-to-Speech API
)

# Notify the user that the setup process is starting
echo "Starting Google Cloud setup in Cloud Shell..."

# Set the active project for Google Cloud commands
echo "Setting the active project to: $PROJECT_ID"
gcloud config set project $PROJECT_ID

# Create a new service account with the specified name
echo "Creating service account: $SERVICE_ACCOUNT_NAME"
gcloud iam service-accounts create $SERVICE_ACCOUNT_NAME \
    --description="Service account for project operations" \
    --display-name="Service Account for Cloud Setup"

# Define roles to be assigned to the service account
# These roles provide necessary permissions for the service account to operate
ROLES=(
    "roles/pubsub.serviceAgent"           # Permissions for Pub/Sub
    "roles/run.serviceAgent"              # Permissions for Cloud Run
    "roles/compute.serviceAgent"          # Permissions for Compute Engine
    "roles/editor"                        # Full editing capabilities for the project
    "roles/container.serviceAgent"        # Permissions for Kubernetes Engine
    "roles/owner"                         # Project ownership rights
    "roles/iam.serviceAccountUser"        # Allows service account impersonation
    "roles/aiplatform.serviceAgent"       # Permissions for AI Platform
    "roles/storage.admin"                 # Full access to manage storage
)

# Assign each role to the service account
for ROLE in "${ROLES[@]}"; do
    echo "Granting role $ROLE to $SERVICE_ACCOUNT_EMAIL"
    gcloud projects add-iam-policy-binding $PROJECT_ID \
        --member="serviceAccount:$SERVICE_ACCOUNT_EMAIL" \
        --role="$ROLE"
done

# Grant your personal Gmail account permission to manage API services
echo "Granting Service Usage Admin role to your email: $USER_EMAIL"
gcloud projects add-iam-policy-binding $PROJECT_ID \
    --member="user:$USER_EMAIL" \
    --role="roles/serviceusage.serviceUsageAdmin"

# Enable the listed APIs for the project
echo "Enabling required Google Cloud APIs..."
for API in "${APIS[@]}"; do
    echo "Enabling API: $API"
    gcloud services enable "$API"
done

# Create a JSON key for the service account and save it to the specified path
echo "Creating a key for the service account and saving to: $KEY_FILE_PATH"
gcloud iam service-accounts keys create $KEY_FILE_PATH \
    --iam-account=$SERVICE_ACCOUNT_EMAIL

# Optional: Set an environment variable to point to the key file
echo "Setting GOOGLE_APPLICATION_CREDENTIALS to point to the key file..."
export GOOGLE_APPLICATION_CREDENTIALS=$KEY_FILE_PATH

# Create a Google Cloud Storage bucket in the specified region
echo "Creating Cloud Storage bucket: $BUCKET_NAME in region: $REGION"
gcloud storage buckets create $BUCKET_URL \
    --project=$PROJECT_ID \
    --location=$REGION \
    --uniform-bucket-level-access

# Grant the service account permission to access and manage the bucket
echo "Granting the service account access to the bucket..."
gcloud storage buckets add-iam-policy-binding $BUCKET_URL \
    --member="serviceAccount:$SERVICE_ACCOUNT_EMAIL" \
    --role="roles/storage.objectAdmin"

# Allow public access to the bucket for viewing its contents
gcloud storage buckets add-iam-policy-binding $BUCKET_URL \
    --member="allUsers" \
    --role="roles/storage.objectViewer"

# Set CORS (Cross-Origin Resource Sharing) policy for the bucket using a JSON file
echo "Setting CORS policy for the bucket using cors.json..."
gsutil cors set cors.json $BUCKET_URL

# Notify the user that the bucket setup is complete and provide key file details
echo "Bucket $BUCKET_NAME created successfully. Setup is complete!"
echo "Service account key saved at: $KEY_FILE_PATH"
```


### **2.2. Rquired Python Libraries**

The following libraries are required to be installed for the application:

>**Gunicorn:**
Gunicorn (Green Unicorn) is a Python WSGI HTTP server that is commonly used to serve Python web applications. It acts as an interface between your application and the server, ensuring it can handle multiple requests efficiently. Gunicorn is lightweight, easy to configure, and works well with frameworks like Flask and Django. It supports synchronous workers, multiple worker processes, and can scale to handle high-traffic environments by running multiple instances of the application.

>**pexels-api-py:**
The `pexels-api-py` library provides an easy way to interact with the Pexels API, allowing developers to search for and retrieve high-quality photos and videos. It abstracts the complexities of API calls, enabling you to integrate Pexels' resources into Python projects seamlessly. Use cases include searching for images based on keywords, retrieving trending photos, and working with curated collections.

>**Flask:**
Flask is a micro web framework for Python, designed for simplicity and scalability. It is lightweight and minimalistic, making it a popular choice for building APIs and web applications. Flask is extensible, allowing developers to integrate third-party libraries and tools. It includes routing, template rendering, and request/response handling as core features, making it ideal for quick prototyping and production-ready applications.

>**pydub:**
Pydub is an audio manipulation library that simplifies working with audio files in Python. It supports basic operations like slicing, merging, and volume adjustments, as well as format conversions (e.g., MP3 to WAV). Built on FFmpeg, it handles various audio file formats and enables developers to automate audio editing tasks programmatically.

>**Pillow:**
Pillow is a powerful Python Imaging Library (PIL) fork for image manipulation. It enables developers to open, process, and save images in many formats (e.g., JPEG, PNG, BMP). Features include resizing, cropping, drawing, and applying filters. Pillow is a go-to library for image processing tasks in Python.

>**Requests:**
Requests is a user-friendly HTTP library for Python that simplifies sending HTTP/HTTPS requests. It supports GET, POST, PUT, DELETE, and other HTTP methods, as well as handling headers, cookies, and sessions. Requests is widely used for interacting with APIs, downloading data, and web scraping due to its ease of use and robust features.

>**google-cloud-aiplatform:**
The `google-cloud-aiplatform` library is a client for Google Cloud’s AI Platform, enabling developers to manage machine learning workflows. This cors.json configuration file defines Cross-Origin Resource Sharing (CORS) rules for a Google Cloud Storage bucket, specifying how external clients can interact with the bucket’s resources. It allows requests from all origins ("*"), which means any domain can access the bucket, making it suitable for public resources but requiring caution for sensitive data. The allowed HTTP methods are limited to GET, HEAD, and OPTIONS, enabling read-only access and preflight checks without allowing data modification. The response headers exposed to clients are restricted to Content-Type, ensuring they can identify the file type but not access unnecessary information. Finally, the maxAgeSeconds value of 3600 seconds (1 hour) caches the results of preflight requests in the browser, reducing overhead and improving performance for repeated requests. This configuration is designed for secure, efficient, and controlled access to bucket resources.


````python
#@title The "app.py" (Do Not Run This Cell)

# Importing necessary components from Flask, a popular web framework for Python
# Flask documentation: https://flask.palletsprojects.com/
from flask import Flask, request, render_template, jsonify, abort, make_response

# Importing built-in Python libraries for various tasks
import os  # Provides functions for interacting with the operating system (e.g., creating directories)
import uuid  # Generates unique identifiers, useful for creating unique session IDs
import time  # Provides functions for handling time (e.g., timestamps)
import subprocess  # Enables execution of external commands or programs
import sys  # Provides access to system-specific parameters and functions

# Importing the Google Cloud Storage library to interact with cloud storage
# Documentation: https://cloud.google.com/python/docs/reference/storage
from google.cloud import storage

# Importing the requests library to perform HTTP requests
# Requests documentation: https://docs.python-requests.org/en/latest/
import requests

# Uncomment the following line if you have a custom module named 'generate' for specific functionalities
# from generate import *

# Specifying the path to the JSON key file for Google Cloud authentication
# Learn how to set up authentication: https://cloud.google.com/iam/docs/creating-managing-service-account-keys
json_path = './key.json'

# Replace "EFINE IT HERE" with your Google Cloud project ID (find it in the Cloud Console)
project_id = "EFINE IT HERE"

# Construct the bucket name by appending '-bucket' to the project ID
bucket_name = project_id + '-bucket'

# Initializing a Flask application, which will serve as the backend of the web app
# static_folder specifies the folder for serving static files (e.g., CSS, JavaScript)
app = Flask(__name__, static_folder='static')


# Function to upload files to Google Cloud Storage (GCS)
def upload_to_gcs(bucket_name, source_file_name, destination_blob_name):
    """
    Uploads a file to a Google Cloud Storage bucket.

    Args:
        bucket_name: The name of your GCS bucket.
        source_file_name: Path to the local file you want to upload.
        destination_blob_name: The desired name/path for the file in the bucket.

    Example:
        upload_to_gcs('my-bucket', 'local_file.txt', 'folder/file_in_gcs.txt')

    Google Cloud Storage guide: https://cloud.google.com/storage/docs/uploading-objects
    """
    # Create a Google Cloud Storage client object to manage storage interactions
    storage_client = storage.Client()

    # Access the specific bucket in GCS using the provided bucket name
    bucket = storage_client.bucket(bucket_name)

    # Create a blob object (represents a file in GCS) for the target file
    blob = bucket.blob(destination_blob_name)

    # Upload the local file to the specified bucket and blob
    blob.upload_from_filename(source_file_name)


# Function to check if a file exists in a specific GCS bucket
def check_file_availability(bucket_name, session_id, file_name):
    """
    Checks whether a specific file exists in the Google Cloud Storage bucket.

    Args:
        bucket_name: The name of the bucket where the file should exist.
        session_id: A unique session ID used as a folder in the bucket.
        file_name: The name of the file to check for availability.

    Returns:
        True if the file exists, False otherwise.

    Example:
        check_file_availability('my-bucket', 'session123', 'image.jpg')
    """
    # Initialize a Google Cloud Storage client
    storage_client = storage.Client()

    # Get the GCS bucket object
    bucket = storage_client.bucket(bucket_name)

    # Construct the file's path in GCS (e.g., 'session123/image.jpg')
    blob = bucket.blob(f"{session_id}/{file_name}")

    # Check if the file exists in the bucket and return the result
    return blob.exists()


# Define the root (home) route of the web application
@app.route("/", methods=["GET", "POST"])
def home():
    """
    Displays the homepage and handles form submissions for generating content.

    HTTP Methods:
        - GET: Displays the homepage.
        - POST: Processes form input to start a session and triggers content generation.
    """
    # Handle form submission (POST request)
    if request.method == "POST":
        # Generate a unique session ID using UUID and a timestamp
        session_id = uuid.uuid4().hex + str(int(time.time() * 10e10))

        # Retrieve the text (keyword) input from the submitted form and remove extra spaces
        keyword = request.form.get("text").strip()

        # Define the local path for storing session-specific files
        session_path = os.path.join('./static', session_id)

        # Create the session directory if it doesn't already exist
        os.makedirs(session_path, exist_ok=True)

        # Construct a command to execute an external script (e.g., 'generate.py') with arguments
        command = [
            sys.executable,  # Path to the current Python interpreter
            'generate.py',  # Script to be executed
            '--keyword', keyword,  # Pass the keyword input as an argument
            '--json_path', json_path,  # Provide the path to the JSON key file
            '--project_id', project_id,  # Specify the Google Cloud project ID
            '--session_id', session_id,  # Pass the session ID for identification
            '--session_path', session_path,  # Specify the local path for session files
            '--bucket_name', bucket_name,  # Specify the GCS bucket name
        ]

        # Execute the external script as a separate process
        subprocess.Popen(
            command,
            stdout=subprocess.PIPE,  # Capture standard output
            stderr=subprocess.PIPE,  # Capture standard error
            text=True  # Handle output as text (not bytes)
        )

        # Render the homepage with the session details after submission
        return render_template('index.html', session_id=session_id, keyword=keyword, bucket_name=bucket_name)
    else:
        # Handle GET requests by simply rendering the homepage
        return render_template('index.html')


# Define a route to fetch content for a specific session
@app.route('/content/<session_id>')
def content(session_id):
    """
    Dynamically retrieves content (image, audio, text) for a given session ID.

    Args:
        session_id: The unique identifier for the session.

    Returns:
        A JSON response with URLs or text content for available files.
    """
    # Construct URLs for potential files in the GCS bucket
    image_url = f"https://storage.googleapis.com/{bucket_name}/{session_id}/image.jpg"
    audio_url = f"https://storage.googleapis.com/{bucket_name}/{session_id}/audio.mp3"
    text_url = f"https://storage.googleapis.com/{bucket_name}/{session_id}/content.txt"

    # Check if the image file exists
    image_available = check_file_availability(bucket_name, session_id, "image.jpg")

    # Check if the audio file exists
    audio_available = check_file_availability(bucket_name, session_id, "audio.mp3")

    # Initialize text content to None
    text_content = None

    # Check if the text file exists and fetch its content if available
    if check_file_availability(bucket_name, session_id, "content.txt"):
        response = requests.get(text_url)  # Perform an HTTP GET request to fetch the file
        if response.status_code == 200:  # Ensure the request was successful
            text_content = response.text.strip()  # Extract and clean up the text content

    # Return a JSON response with the available content
    return jsonify({
        'image_url': image_url if image_available else None,  # Provide image URL if it exists
        'audio_url': audio_url if audio_available else None,  # Provide audio URL if it exists
        'text_content': text_content  # Provide text content if available
    })


# Run the Flask development server when the script is executed directly
if __name__ == "__main__":
    # Start the server on host 0.0.0.0 (accessible externally) and port 8080
    app.run(host="0.0.0.0", port=8080)
````

````python
#@title The "generate.py" (Do Not Run This Cell)

# Importing the io library for handling input and output operations, such as reading and writing files in memory
# Documentation: https://docs.python.org/3/library/io.html
import io

# Importing the os module to interact with the operating system, including file and directory operations
# Documentation: https://docs.python.org/3/library/os.html
import os

# Importing the random module to generate random numbers and perform random operations like shuffling
# Documentation: https://docs.python.org/3/library/random.html
import random

# Importing the time module to work with time-related functions, such as delays and timestamps
# Documentation: https://docs.python.org/3/library/time.html
import time

# Importing the sys module to interact with the Python interpreter, accessing command-line arguments and system-specific parameters
# Documentation: https://docs.python.org/3/library/sys.html
import sys

# Importing the uuid module to generate universally unique identifiers, useful for creating unique session IDs
# Documentation: https://docs.python.org/3/library/uuid.html
import uuid

# Importing the shutil module for high-level file operations like copying and removing files
# Documentation: https://docs.python.org/3/library/shutil.html
import shutil

# Importing the argparse module to parse command-line arguments, making scripts flexible and user-friendly
# Documentation: https://docs.python.org/3/library/argparse.html
import argparse

# Importing the json module to work with JSON data, enabling parsing and serialization
# Documentation: https://docs.python.org/3/library/json.html
import json

# Importing the requests library to make HTTP requests, useful for interacting with web APIs
# Documentation: https://docs.python-requests.org/en/latest/
import requests

# Importing the base64 module for encoding and decoding data in base64, commonly used for handling binary data in text formats
# Documentation: https://docs.python.org/3/library/base64.html
import base64

# Importing the vertexai library to interact with Google Cloud's Vertex AI for machine learning tasks
# Documentation: https://cloud.google.com/vertex-ai/docs
import vertexai

# Importing specific classes from vertexai.generative_models for working with generative AI models and safety settings
from vertexai.generative_models import GenerativeModel, Part, SafetySetting

# Importing the Pexels API client to search and retrieve stock photos from Pexels
# Documentation: https://github.com/pexels/pexels-api-client
from pexelsapi.pexels import Pexels

# Importing RequestException to handle exceptions that may occur during HTTP requests
from requests.exceptions import RequestException

# Importing PIL (Python Imaging Library) modules for image processing, such as creating and editing images
# Documentation: https://pillow.readthedocs.io/
from PIL import Image, ImageDraw, ImageFont

# Importing the Google Cloud Text-to-Speech client to convert text into spoken audio
# Documentation: https://cloud.google.com/text-to-speech/docs/reference/libraries
from google.cloud import texttospeech

# Importing AudioSegment from pydub for audio manipulation, such as combining audio clips and adding effects
# Documentation: https://github.com/jiaaro/pydub
from pydub import AudioSegment

# Importing the Sine generator from pydub to create sine wave audio tones
from pydub.generators import Sine

# Importing the Google Cloud Storage client to interact with Google Cloud Storage buckets for file storage
# Documentation: https://cloud.google.com/storage/docs/reference/libraries
from google.cloud import storage


# Defining a function to upload files to Google Cloud Storage (GCS)
def upload_to_gcs(bucket_name, source_file_name, destination_blob_name):
    # Initialize the Google Cloud Storage client
    storage_client = storage.Client()

    # Access the specific GCS bucket by its name
    bucket = storage_client.bucket(bucket_name)

    # Create a blob object in the bucket representing the destination file
    blob = bucket.blob(destination_blob_name)

    # Upload the local file to the specified blob in GCS
    # This transfers the file from the local filesystem to the cloud storage bucket
    blob.upload_from_filename(source_file_name)

    # Set cache-control metadata to prevent caching of the uploaded file by browsers
    # This ensures that the latest version is always fetched
    blob.cache_control = 'no-store, no-cache, must-revalidate, max-age=0'

    # Apply the updated metadata to the blob in GCS
    blob.patch()


# Defining a function to generate text using Vertex AI's generative models
# The decorator @response_certainty(3) is commented out but can be used to modify function behavior if defined
#@response_certainty(3)
def generate_text(prompt,
                  project_id,
                  temperature=1,
                  top_p=0.95,
                  model_name="gemini-1.5-flash-002",
                  location="us-central1",
                  stream=True):
    # Configuration dictionary for text generation parameters
    generation_config = {
        "max_output_tokens": 8192,  # Maximum number of tokens to generate
        "temperature": temperature,  # Controls randomness in generation; higher values mean more random
        "top_p": top_p,  # Nucleus sampling parameter to limit the pool of tokens considered
        "response_mime_type": "application/json",  # Specifies the MIME type of the response
    }

    # List of safety settings to control the content generated by the AI model
    # These settings help prevent the generation of harmful or inappropriate content
    safety_settings = [
        SafetySetting(
            category=SafetySetting.HarmCategory.HARM_CATEGORY_HATE_SPEECH,
            threshold=SafetySetting.HarmBlockThreshold.OFF
        ),
        SafetySetting(
            category=SafetySetting.HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT,
            threshold=SafetySetting.HarmBlockThreshold.OFF
        ),
        SafetySetting(
            category=SafetySetting.HarmCategory.HARM_CATEGORY_SEXUALLY_EXPLICIT,
            threshold=SafetySetting.HarmBlockThreshold.OFF
        ),
        SafetySetting(
            category=SafetySetting.HarmCategory.HARM_CATEGORY_HARASSMENT,
            threshold=SafetySetting.HarmBlockThreshold.OFF
        ),
    ]

    # Initialize Vertex AI with the specified Google Cloud project and location
    vertexai.init(project=project_id, location=location)

    # Load the specified generative model by name
    model = GenerativeModel(model_name)

    # Generate content by providing the prompt and configuration settings to the model
    responses = model.generate_content(
        [prompt],  # List of prompts to generate content for
        generation_config=generation_config,  # Pass the generation configuration
        safety_settings=safety_settings,  # Apply safety settings
        stream=stream,  # Whether to stream the responses
    )

    # Combine all streamed responses into a single JSON object and parse it
    return json.loads(''.join([response.text for response in responses]))


# Defining a function to search for photos on Pexels API for a specific page
def search_pexel_one_page(query: str, page_number: int, pexel):
    try:
        # Validate that the page number is a positive integer
        if page_number < 1:
            raise ValueError("Page number must be a positive integer.")

        # Perform the API call to search for photos with the given query and page number
        response = pexel.search_photos(query=query,
                                       orientation='',
                                       size='',
                                       color='',
                                       locale='',
                                       page=page_number,
                                       per_page=80)

        # Check if the response contains the 'photos' key, indicating successful retrieval
        if 'photos' not in response:
            raise KeyError("The response does not contain 'photos' key.")

        # Return the response containing photo data
        return response

    # Handle network-related exceptions that may occur during the API call
    except RequestException as e:
        print(f"Network error occurred while fetching photos for query '{query}' on page {page_number}: {e}")
        return None  # Return None if a network error occurs

    # Handle value-related exceptions, such as invalid page numbers
    except ValueError as e:
        print(f"Invalid value error: {e}")
        return None  # Return None if a value error occurs

    # Handle key-related exceptions, such as missing 'photos' in the response
    except KeyError as e:
        print(f"Unexpected response structure: {e}")
        return None  # Return None if the response structure is unexpected

    # Handle any other unforeseen exceptions
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return None  # Return None for any other unforeseen errors


# Defining a function to search for multiple photos on Pexels across multiple pages
def search_pexel_photos(keyword: str, pexel, total_pages: int = 3, total_images: int = 1):
    try:
        # Validate that the total number of pages is a positive integer
        if total_pages < 1:
            raise ValueError("Total pages must be a positive integer.")

        # Perform searches across the specified number of pages
        search_results = [search_pexel_one_page(keyword, page_number, pexel)
                          for page_number in range(1, total_pages + 1)]
        urls = []  # Initialize a list to store image URLs

        # Process each search result page obtained from the Pexels API
        for search_result in search_results:
            if search_result is None:  # Skip processing if the page search failed
                continue

            # Loop over each photo in the 'photos' key of the search result
            for photo in search_result.get('photos', []):
                # Extract the 'src' dictionary, which contains URLs of the image in various resolutions
                src = photo.get('src')
                if src:
                    # Get the list of resolution keys and choose the second highest available resolution
                    keys = list(src.keys())
                    if keys:
                        url = src.get(keys[-2])  # Access the second highest-resolution URL
                        if url:
                            urls.append(url)  # Append the URL to the list if it exists

        # If URLs were found, shuffle them and return the specified number of random URLs
        if urls:
            random.shuffle(urls)  # Introduce randomness by shuffling the URLs
            if len(urls) < total_images:
                total_images = len(urls)
            return urls[:total_images]  # Return the first 'total_images' URLs after shuffling

        # If no URLs are found after processing all pages, print a message
        print("No URLs found for the given keyword.")
        return None

    # Handle value-related exceptions, such as invalid parameters
    except ValueError as e:
        print(f"Invalid value error: {e}")
        return None  # Return None for invalid parameter errors

    # Handle any other unforeseen exceptions during the multi-page search
    except Exception as e:
        print(f"An unexpected error occurred during search: {e}")
        return None  # Return None for any other unforeseen errors


# Defining a function to convert text to speech and save it as an MP3 file with added silence
def text2speech(text, filename, silence_duration_ms=500):  # 500ms of silence
    """
    Generates speech from text with silence added at the beginning and end, and saves it as an MP3 file.

    Args:
        text (str): The text to convert to speech.
        filename (str): The name of the output MP3 file.
        silence_duration_ms (int): Duration of silence in milliseconds to add before and after the speech.
    """
    # Initialize the Google Cloud Text-to-Speech client
    client = texttospeech.TextToSpeechClient()

    # Create a synthesis input object with the provided text
    input_text = texttospeech.SynthesisInput(text=text)

    # Define the voice parameters, selecting language and specific voice
    voice = texttospeech.VoiceSelectionParams(
        language_code="en-US",  # Language of the voice
        name="en-US-Studio-O",  # Specific voice name
    )

    # Define the audio configuration, including encoding and speaking rate
    audio_config = texttospeech.AudioConfig(
        audio_encoding=texttospeech.AudioEncoding.MP3,  # Output format
        speaking_rate=1  # Normal speaking rate
    )

    # Perform the text-to-speech request with the specified input, voice, and audio configuration
    response = client.synthesize_speech(
        request={"input": input_text, "voice": voice, "audio_config": audio_config}
    )

    # Load the MP3 audio content from the response into an AudioSegment object for manipulation
    speech_audio = AudioSegment.from_file(io.BytesIO(response.audio_content), format="mp3")

    # Generate silence using AudioSegment.silent() with the specified duration
    silence = AudioSegment.silent(duration=silence_duration_ms)

    # Concatenate silence before and after the speech audio to create the final audio segment
    final_audio = silence + speech_audio + silence

    # Export the final audio segment as an MP3 file with the specified filename
    final_audio.export(filename, format="mp3")


# Defining a function to generate content by combining text, image, and audio based on a keyword
def generate_content(keyword, project_id, session_id, bucket_name, session_path, pexel):
    # Create a prompt for generating introductory content relevant to the keyword
    prompt = f"Provide five paragraphs of introductory content relevant to '{keyword}.' \
               Your response is in JSON with the paragraph number as key and the content as value."

    # Step 1: Generate text content using the generate_text function
    df_paragraphs = generate_text(prompt, project_id)
    # Combine all paragraphs into a single string separated by two newlines
    content = '\n\n'.join(list(df_paragraphs.values()))
    # Define the path to save the generated content locally
    content_path = os.path.join(session_path, "content.txt")
    # Write the generated content to the content.txt file
    with open(content_path, 'w') as f:
        f.write(content)

    # Step 2: Search for relevant photos on Pexels based on the keyword
    image_urls = search_pexel_photos(keyword, pexel, total_images=1)
    # Fetch the first image from the retrieved URLs
    response = requests.get(image_urls[0])
    # Define the path to save the downloaded image locally
    image_path = os.path.join(session_path, "image.jpg")
    # Write the image content to the image.jpg file in binary mode
    with open(image_path, "wb") as file:
        file.write(response.content)

    # Step 3: Convert the generated text content to speech and save as an MP3 file
    audio_path = os.path.join(session_path, "audio.mp3")
    # Add silence before and after the content and generate the audio file
    text2speech("    " + content + "    ", audio_path)

    # Step 4: Define the final paths for content, image, and audio to be uploaded to GCS
    content_path_final = os.path.join(f"{session_id}", "content.txt")
    image_path_final = os.path.join(f"{session_id}", "image.jpg")
    audio_path_final = os.path.join(f"{session_id}", "audio.mp3")

    # Upload the content.txt file to GCS
    upload_to_gcs(bucket_name, content_path, content_path_final)
    # Upload the image.jpg file to GCS
    upload_to_gcs(bucket_name, image_path, image_path_final)
    # Upload the audio.mp3 file to GCS
    upload_to_gcs(bucket_name, audio_path, audio_path_final)

    # Pause the script for 5 seconds to ensure all uploads are completed
    time.sleep(5)


# Defining a function to parse command-line arguments using argparse
def get_args():
    # Help information to display when the user requests help
    arg_help = '''
               Some Help Information Here
               '''
    # Initialize the ArgumentParser with program name, epilog, and description
    parser = argparse.ArgumentParser(prog        = 'Generator',
                                     epilog      = '** Version 20241122 **',
                                     description = arg_help)

    # Add argument for session_id with short and long flags, default value, type, and help description
    parser.add_argument('-session_id',
                        '--session_id',
                        default = None,
                        type    = str,
                        help    = 'a random string representing session id')

    # Add argument for project_id with short and long flags, default value, type, and help description
    parser.add_argument('-project_id',
                        '--project_id',
                        default = None,
                        type    = str,
                        help    = 'project id from google cloud')

    # Add argument for json_path with short and long flags, default value, type, and help description
    parser.add_argument('-json_path',
                        '--json_path',
                        default = None,
                        type    = str,
                        help    = 'google cloud json_path')

    # Add argument for keyword with short and long flags, default value, type, and help description
    parser.add_argument('-keyword',
                        '--keyword',
                        default = None,
                        type    = str,
                        help    = 'keyword')

    # Add argument for session_path with short and long flags, default value, type, and help description
    parser.add_argument('-session_path',
                        '--session_path',
                        default = None,
                        type    = str,
                        help    = 'session_path')

    # Add argument for bucket_name with short and long flags, default value, type, and help description
    parser.add_argument('-bucket_name',
                        '--bucket_name',
                        default = None,
                        type    = str,
                        help    = 'bucket_name')

    # Parse and return the arguments provided by the user
    return parser.parse_args()


# Defining the main function to execute the script's primary functionality
def main():
    # Parse command-line arguments and store them in 'args'
    args = get_args()

    # Set the environment variable for Google Cloud credentials using the provided JSON path
    os.environ['GOOGLE_APPLICATION_CREDENTIALS'] = args.json_path

    # Define the Pexels API key (replace 'DEFINE IT HERE' with your actual API key)
    pexel_api_key = 'DEFINE IT HERE'
    # Initialize the Pexels API client with the provided API key
    pexel = Pexels(pexel_api_key)

    # Call the generate_content function with the parsed arguments and Pexels client
    generate_content(args.keyword,
                     args.project_id,
                     args.session_id,
                     args.bucket_name,
                     args.session_path,
                     pexel)


# Entry point of the script
# This ensures that the main function runs only when the script is executed directly, not when imported as a module
if __name__ == '__main__':
    # Execute the main function
    main()
````
