# Comprehensive Guide to Training Models: FSP, PRX, and DWM with Transfer Learning and Fine-Tuning

In this section, we will explore a detailed workflow for training and comparing different neural network models using TensorFlow and Keras. Specifically, we focus on:

- **FSP (Fully Supervised Pretext):** Training a model using fully labeled data.
- **PRX (Pretext):** Utilizing self-supervised learning with unlabeled data through a pretext task.
- **DWM (Downstream Model):** Leveraging the representations learned from PRX for a downstream classification task.

This guide is structured to help students understand the purpose and implementation of each code block. We will explain the key concepts, functions, and training strategies used throughout the workflow.

## 1. **Model Initialization**

### **Base Model Setup**

- **Purpose:** Use a pre-trained model to leverage transfer learning, which can improve training efficiency and model performance.
- **Implementation:**
  - Load a pre-trained model (e.g., DenseNet121) without the top classification layer (`include_top=False`).
  - Set the input shape to match the data (`input_shape=(res, res, 3)`), where `res` is the image resolution.
  - The base model is pre-trained on ImageNet (`weights='imagenet'`).

### **Model Cloning**

- **Purpose:** Create separate instances of the base model for different tasks (FSP and PRX) to ensure fair comparison.
- **Implementation:**
  - Clone the base model to create `model_base_fsp` and `model_base_prx`.
  - Synchronize their weights to ensure both start with identical parameters.

---

## 2. **Model Construction**

### **Fully Supervised Pretext (FSP) Model**

- **Purpose:** Train a model solely on labeled data.
- **Architecture:**
  - **Input Layer:** Accepts images with a specific shape.
  - **Convolutional Layer:** Converts the number of channels to 3 (RGB) using a 1x1 convolution.
  - **Base Model:** Processes the input through the pre-trained base model (`model_base_fsp`).
  - **Flatten Layer:** Flattens the output of the base model.
  - **Batch Normalization:** Normalizes the flattened output for faster convergence.
  - **Dense Layer:** Final classification layer with softmax activation for output classes.

### **Pretext (PRX) Model**

- **Purpose:** Use self-supervised learning to learn representations from both labeled and unlabeled data.
- **Architecture:**
  - Similar to the FSP model, but uses `model_base_prx` as the base model.

---

## 3. **Utility Functions**

### **Model Compilation Function**

- **Purpose:** Compile the model with the appropriate optimizer, loss function, and metrics.
- **Components:**
  - **Optimizer:** Adam optimizer with a specified learning rate.
  - **Loss Function:** Sparse categorical crossentropy for multi-class classification.
  - **Metrics:** Accuracy to monitor during training.

### **Model Summary Function**

- **Purpose:** Display the architecture of the model for verification.
- **Usage:** Call this function after modifying the model to ensure layers are correctly configured.

### **Callbacks Function**

- **Purpose:** Define callbacks to optimize training.
- **Components:**
  - **EarlyStopping:** Stops training when validation loss stops improving.
  - **ReduceLROnPlateau:** Reduces the learning rate when a metric has stopped improving.

### **Model Training Function**

- **Purpose:** Train the model using training and validation datasets.
- **Parameters:**
  - Training and validation datasets.
  - Number of epochs.
  - Callbacks.
  - Steps per epoch.

### **Model Evaluation Function**

- **Purpose:** Evaluate the model on a test dataset to obtain performance metrics.

### **Control Function**

- **Purpose:** Compare weights of base and trained models for debugging and verification.
- **Usage:** Optional, can be used to check if weights are being updated as expected.

---

## 4. **Training Phases**

### **A. FSP Training**

#### **Transfer Learning Phase**

- **Steps:**
  - Freeze the base model layers to retain pre-trained weights.
  - Keep the Batch Normalization layer trainable to adapt to new data distributions.
  - Compile the model with a suitable learning rate.
  - Train the model on the labeled dataset.
  - Evaluate and store the results.

#### **Fine-Tuning Phase**

- **Steps:**
  - Unfreeze the base model layers to allow fine-tuning.
  - Freeze the Batch Normalization layer to stabilize training.
  - Recompile the model with a lower learning rate.
  - Train the model further to fine-tune weights.
  - Evaluate and update the results.

### **B. PRX Training**

#### **Transfer Learning Phase**

- **Steps:**
  - Similar to FSP, but train on both labeled and unlabeled data using a pretext task.
  - Save model weights after training.

#### **Fine-Tuning Phase**

- **Steps:**
  - Unfreeze the base model layers.
  - Correctly set the trainable attribute of the Batch Normalization layer (ensure syntax is correct).
  - Recompile and train the model.
  - Evaluate and compare the performance.

### **C. Selecting the Best PRX Model**

- **Purpose:** Choose between the model after transfer learning and after fine-tuning based on performance.
- **Implementation:** Load the model weights that give better validation accuracy.

---

## 5. **Creating the Downstream Model (DWM)**

- **Purpose:** Use the representations learned from PRX to perform a downstream classification task.
- **Architecture:**
  - Similar to FSP, but utilizes `model_base_prx`.
  - Initialize weights to match those of the FSP model for fair comparison.
  - Transfer weights from PRX to DWM to leverage learned features.

---

## 6. **DWM Training**

### **Transfer Learning Phase**

- **Steps:**
  - Freeze the base model layers to retain pre-trained weights from PRX.
  - Keep the Batch Normalization layer trainable.
  - Compile and train the model on the downstream task.
  - Evaluate and store the results.

### **Fine-Tuning Phase**

- **Steps:**
  - Unfreeze the base model layers for fine-tuning.
  - Correctly set the trainable attribute of the Batch Normalization layer.
  - Recompile with a lower learning rate.
  - Train the model further to adapt to the downstream task.
  - Evaluate and update the results.

---

## 7. **Displaying Results**

- **Purpose:** Compare the performance of FSP, PRX, and DWM models.
- **Implementation:**
  - Print out the accuracies stored in the results dictionary.
  - Display the total time taken for the experiments.

---

## 8. **Key Concepts and Best Practices**

- **Transfer Learning:** Leveraging pre-trained models to improve training efficiency and performance.
- **Fine-Tuning:** Unfreezing some layers and retraining with a lower learning rate to adapt to new tasks.
- **Batch Normalization Layers:** Careful handling of these layers is important when freezing/unfreezing models.
- **Model Evaluation:** Consistent evaluation metrics are crucial for fair comparison.
- **Resource Management:** Deleting unused variables and forcing garbage collection helps manage memory in Colab.

---

## 9. **Common Pitfalls**

- **Incorrect Layer Freezing:** Ensure that you correctly set the `trainable` attribute of layers. Misconfigurations can lead to unintended training behavior.
- **Learning Rate Adjustment:** Use a higher learning rate for training new layers and a lower learning rate for fine-tuning pre-trained layers.
- **Data Consistency:** When comparing models, ensure they are trained and evaluated on the same datasets.

---

## 10. **Conclusion**

This workflow demonstrates how to:

- Utilize pre-trained models for different tasks.

- Leverage self-supervised learning with unlabeled data.
- Compare different training strategies to select the best-performing model.

```python
#@title Data Processing Functions

#Data Scaling Function
def fun_scale_vectorized(chunk):
    """
    Vectorized scaling of the input chunk data to a range between 0 and 1 for all channels.

    Parameters:
    chunk (numpy.ndarray): A 2D array representing a chunk of data across multiple channels.

    Returns:
    numpy.ndarray: The scaled chunk data.
    """
    # Compute min values across the time dimension for each channel
    min_vals = np.min(chunk, axis=0)
    # Explanation:
    # - 'np.min(chunk, axis=0)' calculates the minimum value for each channel across all time steps.
    # - This identifies the lowest value in each channel, which is essential for scaling.
    # - Ensures that scaling is done relative to the data's actual range.
    # Reference: https://numpy.org/doc/stable/reference/generated/numpy.min.html

    # Compute max values across the time dimension for each channel
    max_vals = np.max(chunk, axis=0)
    # Explanation:
    # - 'np.max(chunk, axis=0)' calculates the maximum value for each channel across all time steps.
    # - Identifies the highest value in each channel, necessary for determining the scaling range.
    # - Ensures that scaling accounts for the full range of data in each channel.
    # Learn more about numpy.max: https://numpy.org/doc/stable/reference/generated/numpy.max.html

    # Perform min-max scaling vectorized for all channels
    scaled_chunk = (chunk - min_vals) / (max_vals - min_vals + 1e-8)
    # Explanation:
    # - Min-max scaling transforms data to a fixed range, typically [0, 1].
    # - '(chunk - min_vals)' shifts the data so that the minimum value in each channel becomes 0.
    # - '(max_vals - min_vals)' computes the range of each channel.
    # - Dividing by the range scales the data to [0, 1].
    # - Adding a small epsilon (1e-8) prevents division by zero in case 'max_vals' equals 'min_vals'.
    # - This vectorized operation efficiently scales all channels simultaneously without explicit loops.
    # Learn more about min-max scaling: https://en.wikipedia.org/wiki/Feature_scaling#Rescaling_(min-max_normalization)

    return scaled_chunk
    # Explanation:
    # - Returns the scaled chunk data as a NumPy array.
    # - The output has the same shape as the input, with values scaled between 0 and 1 for each channel.
    # - Ensures that the data is normalized, which can improve model training stability and performance.
    # Reference: https://numpy.org/doc/stable/user/quickstart.html#the-basics

# Define a generator function to create data chunks from CSV files
def generate_chunks(files, label):
    """
    Generates scaled and reshaped data chunks from a list of CSV files.

    Parameters:
    files (list of str): List of file paths to CSV files.
    label (int): Label associated with the data (e.g., 0 for normal, 1 for abnormal).

    Yields:
    tuple: A tuple containing a data chunk as a float32 tensor and its corresponding label as an int32.
    """
    # Iterate over each file in the provided list
    for file in files:
        # Read the CSV file into a pandas DataFrame with float32 dtype to reduce memory usage
        data = pd.read_csv(file, dtype='float32')
        # Explanation:
        # - pd.read_csv reads a CSV file into a DataFrame.
        # - Setting dtype='float32' ensures that all numerical data is stored in 32-bit floating point format,
        #   which conserves memory compared to the default float64.

        # Extract the header (column names), convert them to float32, and reshape to a 2D array with one row
        header = np.array([float(val) for val in data.columns.values], dtype='float32').reshape(1, -1)
        # Explanation:
        # - data.columns.values retrieves the column names from the DataFrame.
        # - The list comprehension converts each column name to a float.
        # - np.array creates a NumPy array from the list, specifying dtype='float32'.
        # - reshape(1, -1) transforms the array into a 2D array with one row and as many columns as needed.

        # Concatenate the header with the data values along the first axis (rows)
        data_array = np.concatenate([header, data.values], axis=0).astype('float32')
        # Explanation:
        # - data.values converts the DataFrame into a NumPy array.
        # - np.concatenate combines the header and data arrays vertically (along rows).
        # - astype('float32') ensures the combined array has float32 data type.

        # Assign the concatenated array to 'signal' for further processing
        signal = data_array
        # Explanation:
        # - 'signal' now contains both the header and the data from the CSV file, ready for chunking and scaling.

        # Initialize the starting index for chunking
        start = 0
        # Explanation:
        # - 'start' marks the beginning of the current chunk within the signal.

        # Loop to create chunks as long as there's enough data left
        while start + window_size <= len(signal):
            # Define the end index of the current chunk
            end = start + window_size
            # Explanation:
            # - 'window_size' determines the number of data points in each chunk.
            # - 'end' marks the end of the current chunk.

            # Extract the chunk of data from 'start' to 'end'
            chunk = signal[start:end, :]
            # Explanation:
            # - Slices the 'signal' array to obtain a subset of data points for the current chunk.
            # - ':' indicates that all channels are included.

            # Vectorized scaling of the chunk across all channels
            chunk = fun_scale_vectorized(chunk)
            # Explanation:
            # - Applies the previously defined scaling function to normalize the chunk's data to [0, 1].

            # Reshape the chunk to a 3D array suitable for machine learning models
            # Assuming window_size is a perfect square, reshape to (sqrt(window_size), sqrt(window_size), num_channels)
            chunk = chunk.reshape(int(np.sqrt(chunk.shape[0])), int(np.sqrt(chunk.shape[0])), num_channels)
            # Explanation:
            # - Reshaping is necessary to format the data correctly for input into machine learning models, such as Convolutional Neural Networks (CNNs).
            # - The code assumes that 'window_size' is a perfect square (e.g., 1024 = 32x32).
            # - 'num_channels' represents the number of data channels (features) per data point.
            # - The resulting shape is (height, width, channels).

            # Yield the processed chunk and its corresponding label
            yield (chunk.astype('float32'), label)
            # Explanation:
            # - 'yield' makes this function a generator, producing one chunk-label pair at a time.
            # - 'chunk.astype('float32')' ensures the data is in float32 format, compatible with TensorFlow.

            # Move the 'start' index forward to the next position, allowing for overlap between chunks
            start += (window_size - overlap)
            # Explanation:
            # - 'overlap' determines how much consecutive chunks overlap with each other.
            # - For example, with window_size=100 and overlap=50, chunks will be [0-100], [50-150], [100-200], etc.
            # - This overlapping can help models learn temporal dependencies better.

# Data Loading Function
def read_dataset_all():
    """
    Loads and prepares the dataset by reading CSV files from 'normal' and 'imbalance' directories,
    determining the number of channels, and creating TensorFlow Datasets for normal and abnormal signals.

    Returns:
        tf.data.Dataset: A concatenated dataset containing both normal and abnormal signal data.
    """

    # Data Loading (Fully On-The-Fly)

    # Use the glob module to retrieve all CSV files from the 'normal' directory
    files_normal = glob.glob('normal/*.csv')  # List of file paths for normal signals
    # Explanation:
    # - 'glob.glob' searches for all files matching the pattern 'normal/*.csv'.
    # - 'files_normal' will contain a list of file paths to all CSV files in the 'normal' directory.
    # - These files represent normal (e.g., non-defective) signal data.
    # Learn more about the glob module: https://docs.python.org/3/library/glob.html

    # Use the glob module to retrieve all CSV files from the 'imbalance' directory
    files_abnormal = glob.glob('imbalance/*.csv')  # List of file paths for abnormal signals
    # Explanation:
    # - Similar to 'files_normal', this line collects all CSV files in the 'imbalance' directory.
    # - These files represent abnormal (e.g., defective) signal data.
    # - Separating normal and abnormal data aids in balanced training and evaluation.
    # Reference: https://docs.python.org/3/library/glob.html

    # Load the first 'normal' CSV file to inspect its structure and determine the number of channels
    sample_data = pd.read_csv(files_normal[0], dtype='float32')  # Read the first CSV file in the 'normal' list
    # Explanation:
    # - Reads the first CSV file from 'files_normal' to understand the data structure.
    # - This is useful for determining parameters like the number of channels.
    # - Ensures that all data chunks are correctly shaped for model input.
    # Learn more about pandas.read_csv: https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html

    # Determine the number of channels (columns) in the dataset
    num_channels = sample_data.shape[1]  # Number of columns corresponds to the number of channels
    # Explanation:
    # - 'sample_data.shape[1]' retrieves the number of columns in the DataFrame, which corresponds to the number of channels or features.
    # - 'num_channels' is used later to correctly reshape data chunks.
    # - Ensures compatibility between data and model architecture.
    # Reference: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.shape.html

    # Create a TensorFlow Dataset for normal signals using the generator function
    dataset_normal = tf.data.Dataset.from_generator(
        lambda: generate_chunks(files_normal, 0),
        output_types=(tf.float32, tf.int32),
        output_shapes=((res, res, num_channels), ())
    )
    # Explanation:
    # - 'tf.data.Dataset.from_generator' creates a TensorFlow Dataset from a generator function.
    # - The lambda function calls 'generate_chunks' with 'files_normal' and label '0' (indicating normal).
    # - 'output_types' specifies the data types of the outputs: 'tf.float32' for data chunks and 'tf.int32' for labels.
    # - 'output_shapes' defines the shape of each output element. Here, each chunk is expected to have shape (32, 32, num_channels), and the label is a scalar ().
    # - This setup facilitates efficient on-the-fly data loading and preprocessing.
    # Learn more about tf.data.Dataset.from_generator: https://www.tensorflow.org/api_docs/python/tf/data/Dataset#from_generator

    # Create a TensorFlow Dataset for abnormal signals using the generator function
    dataset_abnormal = tf.data.Dataset.from_generator(
        lambda: generate_chunks(files_abnormal, 1),
        output_types=(tf.float32, tf.int32),
        output_shapes=((res, res, num_channels), ())
    )
    # Explanation:
    # - Similar to 'dataset_normal', but uses 'files_abnormal' and label '1' (indicating abnormal).
    # - Labels help in distinguishing between normal and abnormal signals during training.
    # - Facilitates binary classification tasks by providing clear target labels.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/data/Dataset#from_generator

    # Concatenate the normal and abnormal datasets into a single dataset
    dataset_all = dataset_normal.concatenate(dataset_abnormal)
    # Explanation:
    # - Combines 'dataset_normal' and 'dataset_abnormal' into one unified dataset.
    # - This allows for mixed batches containing both normal and abnormal data, promoting balanced training.
    # - Ensures that the model learns to differentiate between both classes effectively.
    # - Facilitates easier shuffling and batching when preparing data for training.
    # Learn more about dataset concatenation: https://www.tensorflow.org/api_docs/python/tf/data/Dataset#concatenate

    return dataset_all
    # Explanation:
    # - Returns the concatenated TensorFlow Dataset containing both normal and abnormal signal data.
    # - This dataset can be further processed (e.g., batching, shuffling) before being fed into the model for training and evaluation.
    # - Provides a unified interface for accessing all data samples in a streamlined manner.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/data/Dataset#concatenate

# Function to serialize a single example
def serialize_example(feature, label):
    """
    Serializes a single feature-label pair into a TFRecord Example.

    Parameters:
        feature (tf.Tensor): The feature tensor to serialize.
        label (tf.Tensor): The label tensor to serialize.

    Returns:
        bytes: The serialized Example as a string.
    """
    # Serialize the feature tensor to a string
    feature_serialized = tf.io.serialize_tensor(feature)
    # Explanation:
    # - 'tf.io.serialize_tensor' converts the tensor into a serialized string representation.
    # - This is necessary for storing tensors in TFRecord files, which require binary data.
    # - Ensures that the tensor can be efficiently written to disk and later reconstructed.
    # Learn more about tensor serialization: https://www.tensorflow.org/api_docs/python/tf/io/serialize_tensor

    # Serialize the label tensor to a string
    label_serialized = tf.io.serialize_tensor(label)
    # Explanation:
    # - Similar to 'feature_serialized', this line serializes the label tensor.
    # - Consistent serialization ensures that both features and labels can be stored together in the same Example.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/io/serialize_tensor

    # Create a dictionary mapping feature names to tf.train.Feature
    feature_dict = {
        'feature': tf.train.Feature(bytes_list=tf.train.BytesList(value=[feature_serialized.numpy()])),
        'label': tf.train.Feature(bytes_list=tf.train.BytesList(value=[label_serialized.numpy()])),
    }
    # Explanation:
    # - Constructs a dictionary where each key corresponds to a feature in the Example.
    # - 'tf.train.Feature' is used to define the type of each feature; in this case, both are byte strings.
    # - 'bytes_list' is appropriate for serialized tensors, which are stored as byte strings.
    # - 'feature_serialized.numpy()' converts the serialized tensor from a TensorFlow tensor to a NumPy array, which is then converted to bytes.
    # - This structure is required by the TFRecord format to store complex data types.
    # Learn more about tf.train.Feature: https://www.tensorflow.org/api_docs/python/tf/train/Feature

    # Create a tf.train.Example message using the feature dictionary
    example_proto = tf.train.Example(features=tf.train.Features(feature=feature_dict))
    # Explanation:
    # - 'tf.train.Example' is a protocol buffer used to serialize data for TensorFlow.
    # - It contains a 'Features' message, which holds the actual data in the form of a feature dictionary.
    # - This structure allows for flexible and efficient storage of complex data types.
    # - Essential for creating TFRecord files that can be efficiently read during training.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/train/Example

    # Serialize the Example protocol buffer to a string
    return example_proto.SerializeToString()
    # Explanation:
    # - 'SerializeToString()' converts the Example protocol buffer into a serialized string.
    # - This serialized string can be written directly to a TFRecord file.
    # - Ensures that the Example is stored in a compact binary format suitable for high-performance I/O operations.
    # Learn more about serialization: https://developers.google.com/protocol-buffers/docs/encoding

# Function to process and serialize the dataset
def write_dataset_to_tfrecord(dataset, tfrecord_file):
    """
    Writes a TensorFlow Dataset to a TFRecord file by serializing each example.

    Parameters:
        dataset (tf.data.Dataset): The dataset to serialize and write.
        tfrecord_file (str): The path to the output TFRecord file.
    """
    # Open a TFRecordWriter to write the serialized examples
    with tf.io.TFRecordWriter(tfrecord_file) as writer:
        # Explanation:
        # - 'tf.io.TFRecordWriter' is used to write serialized Example protocol buffers to a TFRecord file.
        # - It handles the low-level file I/O operations, ensuring data is written in the correct format.
        # - Efficient for writing large datasets as it supports streaming data to disk.
        # Learn more about TFRecordWriter: https://www.tensorflow.org/api_docs/python/tf/io/TFRecordWriter

        # Iterate over each feature-label pair in the dataset
        for feature, label in dataset:
            # Serialize the current example
            serialized_example = serialize_example(feature, label)
            # Explanation:
            # - Calls the 'serialize_example' function to convert the feature-label pair into a serialized Example.
            # - Ensures that each example is properly formatted before writing to the TFRecord file.
            # - Maintains consistency in data formatting across all examples.
            # Reference: https://www.tensorflow.org/api_docs/python/tf/io/serialize_tensor

            # Write the serialized example to the TFRecord file
            writer.write(serialized_example)
            # Explanation:
            # - Writes the serialized Example string to the TFRecord file.
            # - Appends each example sequentially, building the complete dataset in binary format.
            # - Enables efficient reading and parsing during training.
            # Learn more about writing TFRecords: https://www.tensorflow.org/tutorials/load_data/tfrecord#writing_tfrecord_files

# Function to parse a single serialized example
def parse_example(serialized_example):
    """
    Parses a single serialized TFRecord Example into feature and label tensors.

    Parameters:
        serialized_example (tf.Tensor): A scalar string Tensor; a single serialized Example.

    Returns:
        tuple: A tuple containing the feature tensor and the label tensor.
    """
    # Define the feature description for parsing
    feature_description = {
        'feature': tf.io.FixedLenFeature([], tf.string),
        'label': tf.io.FixedLenFeature([], tf.string),
    }
    # Explanation:
    # - 'feature_description' specifies the expected structure of the serialized Example.
    # - 'tf.io.FixedLenFeature' is used for features with a fixed length; here, both 'feature' and 'label' are scalar strings.
    # - This description guides TensorFlow on how to parse the serialized data correctly.
    # - Essential for accurate deserialization of TFRecord data.
    # Learn more about feature descriptions: https://www.tensorflow.org/api_docs/python/tf/io/parse_single_example

    # Parse the serialized Example using the feature description
    example = tf.io.parse_single_example(serialized_example, feature_description)
    # Explanation:
    # - 'tf.io.parse_single_example' parses a single serialized Example proto.
    # - Returns a dictionary mapping feature names to TensorFlow tensors.
    # - Converts the binary data back into usable tensors for model input.
    # - Ensures that the data is correctly interpreted according to the feature description.
    # - Facilitates the reconstruction of the original feature and label tensors.
    # Learn more about tf.io.parse_single_example: https://www.tensorflow.org/api_docs/python/tf/io/parse_single_example

    # Deserialize the feature tensor from the parsed Example
    feature = tf.io.parse_tensor(example['feature'], out_type=tf.float32)
    # Explanation:
    # - 'tf.io.parse_tensor' deserializes the byte string back into a TensorFlow tensor of the specified type.
    # - 'out_type=tf.float32' ensures that the feature tensor is converted to float32, matching the original data type.
    # - Maintains consistency between serialized and deserialized data types.
    # - Facilitates the reconstruction of the original feature tensor for model input.
    # Learn more about parse_tensor: https://www.tensorflow.org/api_docs/python/tf/io/parse_tensor

    # Deserialize the label tensor from the parsed Example
    label = tf.io.parse_tensor(example['label'], out_type=tf.int32)
    # Explanation:
    # - Similar to the feature tensor, this line deserializes the label tensor.
    # - 'out_type=tf.int32' ensures that the label tensor is converted to int32, matching the original data type.
    # - Ensures that labels are correctly interpreted for loss calculation and model evaluation.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/io/parse_tensor

    # Explicitly set the shape of the feature tensor
    feature.set_shape([res, res, num_channels])  # e.g., [32, 32, 8]
    # Explanation:
    # - 'feature.set_shape' explicitly defines the shape of the feature tensor.
    # - Ensures that TensorFlow knows the exact dimensions of the tensor, which is necessary for model compatibility.
    # - Helps TensorFlow optimize the computation graph by knowing tensor shapes in advance.
    # - Prevents shape-related errors during model training and inference.
    # Reference: https://www.tensorflow.org/api_docs/python/tf/Tensor#set_shape

    # Explicitly set the shape of the label tensor
    label.set_shape([])  # Scalar label
    # Explanation:
    # - 'label.set_shape' explicitly defines the shape of the label tensor as a scalar.
    # - Ensures that TensorFlow treats the label as a single integer value, suitable for classification tasks.
    # - Maintains consistency between serialized and deserialized label shapes.
    # - Prevents shape-related errors during loss calculation and metric evaluation.
    # Learn more about tensor shapes: https://www.tensorflow.org/guide/tensor

    return feature, label
    # Explanation:
    # - Returns a tuple containing the deserialized and reshaped feature tensor and label tensor.
    # - These tensors are ready to be used as inputs and targets in the training and evaluation pipelines.
    # - Facilitates seamless integration with TensorFlow's data processing and model training workflows.
    # Reference: https://www.tensorflow.org/guide/data

# Function to load tfrecord files
def load_tfrecords(tfrecord_file):
    # Create a dataset from the TFRecord file using TensorFlow's TFRecordDataset.
    # TFRecordDataset is an efficient data loader for large datasets stored in TFRecord format.
    # It allows for fast data streaming and is optimized for performance in TensorFlow.
    # Learn more: https://www.tensorflow.org/api_docs/python/tf/data/TFRecordDataset
    tfrecord_dataset = tf.data.TFRecordDataset(tfrecord_file)
    # Parse the serialized data in the TFRecord file by mapping each serialized example to a parsed example.
    # The 'parse_example' function should define how to decode the serialized data into usable tensors.
    # This step is crucial for converting raw data into a format suitable for model training.
    # Learn more about parsing TFRecords: https://www.tensorflow.org/tutorials/load_data/tfrecord
    return tfrecord_dataset.map(parse_example)

# Define a function to generate labels for original images
def generate_label(image, label):
    """
    Generates a label for the original (non-augmented) image.

    Parameters:
    image (tf.Tensor): The input image tensor.
    label (int): The original label associated with the image.

    Returns:
    tuple: A tuple containing the image and a constant label tensor.
    """
    # Return the original image and a label tensor with value [1]
    # The label '1' indicates that the image is an original, non-augmented instance.
    # This labeling scheme is useful for distinguishing between original and augmented images,
    # which is particularly beneficial in contrastive learning tasks where the model learns to differentiate
    # between similar and dissimilar pairs.
    # Learn more about labeling strategies: https://keras.io/api/preprocessing/image/
    return image, tf.constant(1, dtype=tf.int32)

# Define a function to generate augmented images and their labels
def generate_augmented(image, label):
    """
    Generates an augmented version of the input image and assigns it a label.

    Parameters:
    image (tf.Tensor): The input image tensor.
    label (int): The original label associated with the image.

    Returns:
    tuple: A tuple containing the augmented image and a constant label tensor.
    """
    # Apply augmentation on the GPU to leverage hardware acceleration for faster processing.
    # Data augmentation can be computationally intensive; performing it on the GPU speeds up the process.
    # 'fun_augment' is a predefined Keras Sequential model that applies augmentation transformations.
    # Learn more about TensorFlow device placement: https://www.tensorflow.org/guide/device_placement
    with tf.device('/GPU:0'):
        aug_image = fun_augment(image)
    # Explanation:
    # - Data augmentation increases the diversity of the training data by applying random transformations
    #   such as flips and rotations. This helps the model generalize better by learning invariant features.
    # - The 'fun_augment' model applies these transformations to the input image, creating a new augmented version.
    # - Performing augmentation on the GPU ensures that the training process remains efficient and avoids bottlenecks.
    # Learn more about data augmentation: https://www.tensorflow.org/tutorials/images/data_augmentation

    # Return the augmented image and a label tensor with value [0]
    # The label '0' indicates that the image is an augmented instance.
    # This labeling helps in distinguishing between original and augmented data during training,
    # facilitating tasks like contrastive learning where the model learns to differentiate between
    # similar (augmented) and dissimilar (original) pairs.
    # Learn more about data augmentation labeling: https://keras.io/api/preprocessing/image/
    return aug_image, tf.constant(0, dtype=tf.int32)

# Define a sequential model for data augmentation using TensorFlow's Keras API
fun_augment = tf.keras.Sequential([
    # Randomly flip the input images both horizontally and vertically
    # This introduces variability in the dataset by mirroring images along different axes,
    # which helps the model learn rotational and reflectional invariance.
    # Learn more about RandomFlip: https://keras.io/api/layers/preprocessing_layers/image_preprocessing/random_flip/
    tf.keras.layers.RandomFlip("horizontal_and_vertical"),

    # Randomly rotate the input images by a factor of 0.2 (20% of 2π radians)
    # This applies random rotations to images within a specified range, further increasing data diversity.
    # Rotations help the model become invariant to the orientation of objects in images.
    # Learn more about RandomRotation: https://keras.io/api/layers/preprocessing_layers/image_preprocessing/random_rotation/
    tf.keras.layers.RandomRotation(0.2),
])
# Explanation:
# - `fun_augment` is a Keras Sequential model that applies a series of data augmentation transformations.
# - `RandomFlip` randomly flips images horizontally and vertically, introducing variability in the dataset.
# - `RandomRotation` randomly rotates images by up to 20% of a full rotation (approximately ±72 degrees),
#   helping the model generalize better by learning invariant features.
# - These augmentations are crucial for increasing the robustness and generalization capability of the model.
# Learn more about data augmentation in TensorFlow: https://www.tensorflow.org/tutorials/images/data_augmentation
```

```python
#@title Create Models FSP and PRX
# ==============================
# Model Initialization
# ==============================

# Load the base model (DenseNet121 in this example) with ImageNet weights
model_base = tf.keras.applications.DenseNet121(
    include_top=False,              # Exclude the top classification layer
    weights='imagenet',             # Initialize with weights pre-trained on ImageNet
    input_shape=(res, res, 3)       # Define input shape: (height, width, channels)
)
# Explanation:
# - `DenseNet121` is a convolutional neural network architecture known for its efficiency and accuracy.
# - `include_top=False` excludes the final fully connected layers, allowing customization for specific tasks.
# - `weights='imagenet'` leverages transfer learning by initializing the model with weights trained on the ImageNet dataset.
# - `input_shape=(res, res, 3)` specifies the input dimensions, where `res` is the resolution (height and width) of the input images,
#   and `3` corresponds to the three color channels (RGB).
# Learn more about DenseNet121: https://keras.io/api/applications/densenet/

# We clone model_base to create separate instances for FSP and PRX models.
# P.S. We will clone the model_dws after the pretext task using model_prx.
model_base_fsp = tf.keras.models.clone_model(model_base)
model_base_prx = tf.keras.models.clone_model(model_base)
# Explanation:
# - `tf.keras.models.clone_model` creates a new instance of the model with the same architecture as `model_base`.
# - `model_base_fsp` and `model_base_prx` are separate models that will be used for different tasks (FSP and PRX respectively).
# - Cloning ensures that both models have identical architectures initially, allowing for fair comparisons and independent training.
# Learn more about clone_model: https://www.tensorflow.org/api_docs/python/tf/keras/models/clone_model

# We set the parameters of model_base_fsp and model_base_prx to be the same as the
# randomly generated parameters of model_base to have both models' initial
# weights (i.e., start-point) the same for a fair comparison.
model_base_fsp.set_weights(model_base.get_weights())
model_base_prx.set_weights(model_base.get_weights())
# Explanation:
# - `model_base.get_weights()` retrieves the current weights of `model_base`.
# - `set_weights` assigns these weights to `model_base_fsp` and `model_base_prx`.
# - This ensures that both cloned models start with the same weights, providing a consistent baseline for training.
# - Having identical starting weights is crucial for fair comparisons between different models or training strategies.
# Learn more about set_weights: https://www.tensorflow.org/api_docs/python/tf/keras/Model#set_weights

# Print examples of weights from the 3rd layer of both cloned models to verify they are identical
print('Example of weights in the 3rd layer of model_base_fsp:',
      model_base_fsp.layers[2].weights[0][0][0][0][:2])
print('Example of weights in the 3rd layer of model_base_prx:',
      model_base_prx.layers[2].weights[0][0][0][0][:2])
# Explanation:
# - Accesses the weights of the 3rd layer (`layers[2]`) in both `model_base_fsp` and `model_base_prx`.
# - `weights[0]` refers to the kernel weights of the convolutional layer.
# - `[0][0][0][0][:2]` slices the first two weights for demonstration purposes.
# - Printing these values helps verify that the weights have been correctly cloned and are identical in both models.
# Learn more about accessing layer weights: https://keras.io/api/layers/base_layer/#get_weights-method

# ==============================
# Model Construction for FSP
# ==============================

# Define inputs for the fully supervised pretext (FSP) model
inputs_fsp = tf.keras.Input(shape=(res, res, num_channels))
# Explanation:
# - `tf.keras.Input` creates a placeholder tensor for the input data.
# - `shape=(res, res, num_channels)` specifies the shape of each input sample, where:
#   - `res`: Resolution of the input images (height and width).
#   - `num_channels`: Number of channels in the input data (e.g., 8 channels as defined in hyperparameters).
# Learn more about tf.keras.Input: https://keras.io/api/layers/core_layers/input/

# Apply a 2D convolutional layer to transform the input channels to 3 (RGB)
x_fsp = tf.keras.layers.Conv2D(
    3,                  # Number of filters (output channels)
    (1, 1),             # Kernel size: 1x1 convolution
    activation='relu'   # Activation function: ReLU (Rectified Linear Unit)
)(inputs_fsp)
# Explanation:
# - `Conv2D` applies convolutional filters to the input data.
# - `(1, 1)` kernel size means each output channel is a linear combination of the input channels.
# - `activation='relu'` introduces non-linearity, enabling the model to learn complex patterns.
# - This layer effectively reduces or maps the number of channels from `num_channels` to `3`, matching the expected input shape of the base model.
# Learn more about Conv2D: https://keras.io/api/layers/convolution_layers/convolution2d/

# Pass the transformed input through the cloned base model for FSP
x_fsp = model_base_fsp(x_fsp)
# Explanation:
# - `model_base_fsp(x_fsp)` feeds the output of the convolutional layer into `model_base_fsp`.
# - Since `include_top=False`, the base model outputs feature maps instead of classification probabilities.
# - This step leverages the pre-trained DenseNet121 to extract high-level features from the input data.
# Learn more about DenseNet121: https://keras.io/api/applications/densenet/#densenet121-function

# Flatten the feature maps into a 1D vector
x_fsp = tf.keras.layers.Flatten()(x_fsp)
# Explanation:
# - `Flatten` converts multi-dimensional feature maps into a single 1D vector.
# - This is necessary before passing the data into dense (fully connected) layers, which expect 1D input.
# Learn more about Flatten layer: https://keras.io/api/layers/core_layers/flatten/

# Apply batch normalization to stabilize and accelerate training
x_fsp = tf.keras.layers.BatchNormalization()(x_fsp)
# Explanation:
# - `BatchNormalization` normalizes the activations of the previous layer for each batch.
# - This helps in reducing internal covariate shift, leading to faster and more stable training.
# - It also acts as a regularizer, potentially reducing the need for other forms of regularization.
# Learn more about BatchNormalization: https://keras.io/api/layers/normalization_layers/batch_normalization/

# Add a dense (fully connected) layer with softmax activation for classification
outputs_fsp = tf.keras.layers.Dense(
    2,                  # Number of output units (classes)
    activation='softmax' # Activation function: Softmax for multi-class classification
)(x_fsp)
# Explanation:
# - `Dense` layer with `2` units implies binary classification (e.g., class 0 and class 1).
# - `activation='softmax'` converts the output into probability distributions over the classes.
# - This layer serves as the final classification layer for the FSP model.
# Learn more about Dense layers: https://keras.io/api/layers/core_layers/dense/

# Define the Fully Supervised Pretext (FSP) model by specifying its inputs and outputs
model_fsp = tf.keras.Model(inputs_fsp, outputs_fsp)
# Explanation:
# - `tf.keras.Model` creates a Keras model by specifying the input and output tensors.
# - `model_fsp` is now a complete model that can be compiled and trained on data.
# - This model is tailored for the Fully Supervised Pretext (FSP) task.
# Learn more about Keras Model: https://keras.io/api/models/model/

# ==============================
# Model Construction for PRX
# ==============================

# Define inputs for the pretext (PRX) model
inputs_prx = tf.keras.Input(shape=(res, res, num_channels))
# Explanation:
# - Similar to `inputs_fsp`, this creates a placeholder tensor for the input data for the PRX model.
# - Ensures consistency in input dimensions across different models.
# Learn more about tf.keras.Input: https://keras.io/api/layers/core_layers/input/

# Apply a 2D convolutional layer to transform the input channels to 3 (RGB)
x_fsp = tf.keras.layers.Conv2D(
    3,                  # Number of filters (output channels)
    (1, 1),             # Kernel size: 1x1 convolution
    activation='relu'   # Activation function: ReLU
)(inputs_prx)
# Explanation:
# - This convolutional layer serves the same purpose as in the FSP model, mapping input channels to 3.
# - Maintains consistency in data preprocessing across different tasks.
# Learn more about Conv2D: https://keras.io/api/layers/convolution_layers/convolution2d/

# Pass the transformed input through the cloned base model for PRX
x_prx = model_base_prx(x_fsp)
# Explanation:
# - `model_base_prx(x_fsp)` feeds the output of the convolutional layer into `model_base_prx`.
# - This utilizes the pre-trained DenseNet121 to extract features specific to the PRX task.
# - Allows the PRX model to learn representations based on the self-supervised contrastive learning objective.
# Learn more about DenseNet121: https://keras.io/api/applications/densenet/#densenet121-function

# Flatten the feature maps into a 1D vector
x_prx = tf.keras.layers.Flatten()(x_prx)
# Explanation:
# - Similar to the FSP model, `Flatten` converts multi-dimensional feature maps into a 1D vector.
# - Prepares the data for the subsequent dense layers.
# Learn more about Flatten layer: https://keras.io/api/layers/core_layers/flatten/

# Apply batch normalization to stabilize and accelerate training
x_prx = tf.keras.layers.BatchNormalization()(x_prx)
# Explanation:
# - `BatchNormalization` normalizes the activations of the previous layer, aiding in faster and more stable training.
# - Helps in reducing internal covariate shift and acts as a regularizer.
# Learn more about BatchNormalization: https://keras.io/api/layers/normalization_layers/batch_normalization/

# Add a dense (fully connected) layer with softmax activation for classification
outputs_prx = tf.keras.layers.Dense(
    2,                  # Number of output units (classes)
    activation='softmax' # Activation function: Softmax for multi-class classification
)(x_prx)
# Explanation:
# - `Dense` layer with `2` units and `softmax` activation functions as the final classification layer for the PRX model.
# - Facilitates binary classification between original and augmented data instances.
# Learn more about Dense layers: https://keras.io/api/layers/core_layers/dense/

# Define the Pretext (PRX) model by specifying its inputs and outputs
model_prx = tf.keras.Model(inputs_prx, outputs_prx)
# Explanation:
# - `tf.keras.Model` creates the PRX model by linking the input and output tensors.
# - `model_prx` is now a complete model ready for compilation and training.
# - This model is tailored for the Pretext (PRX) task, focusing on self-supervised contrastive learning.
# Learn more about Keras Model: https://keras.io/api/models/model/
```

```python
#@title Some Useful Functions

# Function to compile a model with specified learning rate
def fun_model_compile(model, learning_rate):
    """
    Compiles the given Keras model with the Adam optimizer,
    sparse categorical crossentropy loss, and accuracy metric.

    Parameters:
    model (tf.keras.Model): The Keras model to compile.
    learning_rate (float): The learning rate for the Adam optimizer.

    Returns:
    tf.keras.Model: The compiled Keras model.
    """
    # Compiles the model with specified optimizer, loss function, and metrics
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),  # Adam optimizer with given learning rate
        loss='sparse_categorical_crossentropy',                           # Loss function for integer labels
        metrics=['accuracy'],                                             # Metrics to track during training
        jit_compile=False                                                  # Disables Just-In-Time (JIT) compilation for debugging
    )
    # Explanation:
    # - **Optimizer:** Adam is chosen for its adaptive learning rate capabilities, which often lead to faster convergence.
    # - **Learning Rate:** Controls the step size during optimization. A higher learning rate can speed up training but may overshoot minima.
    # - **Loss Function:** 'sparse_categorical_crossentropy' is suitable for multi-class classification tasks where labels are integers.
    # - **Metrics:** Tracking 'accuracy' provides a straightforward measure of model performance.
    # - **JIT Compilation:** Disabled here to simplify debugging. Enabling it can optimize training performance but may complicate error tracing.
    # Learn more about Adam optimizer: https://keras.io/api/optimizers/adam/
    # Learn more about sparse_categorical_crossentropy: https://keras.io/api/losses/probabilistic_losses/#sparsecategoricalcrossentropy-function
    # Learn more about model.compile: https://keras.io/api/models/model_training_apis/#compile-method

    return model
    # Returns the compiled model ready for training

# Function to print the summary of the model
def fun_model_summary(model):
    """
    Prints a summary of the Keras model architecture.

    Parameters:
    model (tf.keras.Model): The Keras model to summarize.

    Returns:
    None
    """
    model.summary()  # Prints the summary of the model architecture, including layers, output shapes, and parameter counts
    # Explanation:
    # - `model.summary()` provides a concise overview of the model's architecture, which is essential for verifying layer configurations and parameter sizes.
    # - It displays each layer's type, output shape, and number of parameters, helping in debugging and ensuring the model is structured correctly.
    # Learn more about model.summary(): https://keras.io/api/models/model/#summary-method

    print('\n')  # Prints a newline for better readability in the console
    # Explanation:
    # - Adding a newline enhances the readability of the output, especially when multiple summaries or logs are printed sequentially.

# Function to define callbacks for training
def fun_model_callbacks():
    """
    Defines and returns a list of Keras callbacks to be used during model training.

    Returns:
    list: A list containing EarlyStopping and ReduceLROnPlateau callbacks.
    """
    # Early stopping callback to stop training when validation loss doesn't improve
    cb_early_stopping = tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',            # Metric to monitor for improvement
        patience=3,                    # Number of epochs with no improvement after which training will be stopped
        restore_best_weights=True      # Restores model weights from the epoch with the best value of the monitored metric
    )
    # Explanation:
    # - **EarlyStopping:** Prevents overfitting by halting training when the model's performance on the validation set stops improving.
    # - **Monitor:** 'val_loss' is a common choice as it reflects the model's generalization ability.
    # - **Patience:** Allows the model some epochs to improve before stopping, accommodating minor fluctuations.
    # - **Restore Best Weights:** Ensures that the model retains the best-performing weights, not the weights from the final epoch.
    # Learn more about EarlyStopping: https://keras.io/api/callbacks/early_stopping/

    # Reduce learning rate callback when validation loss plateaus
    cb_reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',            # Metric to monitor for improvement
        factor=0.5,                    # Factor by which the learning rate will be reduced
        patience=2,                    # Number of epochs with no improvement after which learning rate will be reduced
        min_lr=0.00001                 # Lower bound on the learning rate
    )
    # Explanation:
    # - **ReduceLROnPlateau:** Dynamically adjusts the learning rate when the model's performance stagnates.
    # - **Factor:** Determines how much the learning rate is reduced. Here, it halves the learning rate.
    # - **Patience:** After 2 epochs without improvement, the learning rate is reduced.
    # - **min_lr:** Ensures that the learning rate does not become too small, which could hinder further learning.
    # Learn more about ReduceLROnPlateau: https://keras.io/api/callbacks/reduce_lr_on_plateau/

    return [cb_early_stopping, cb_reduce_lr]
    # Returns a list of the defined callbacks, which can be passed to the `model.fit` method during training
    # Learn more about Keras callbacks: https://keras.io/api/callbacks/

# Function to train the model with given dataset, epochs, and callbacks
def fun_model_train(model, dataset_tr, dataset_vl, epochs, callbacks, steps_per_epoch_tr, steps_per_epoch_vl, validation_batch_size):
    """
    Trains the given Keras model using the provided training and validation datasets.

    Parameters:
    model (tf.keras.Model): The Keras model to train.
    dataset_tr (tf.data.Dataset): The training dataset.
    dataset_vl (tf.data.Dataset): The validation dataset.
    epochs (int): Number of epochs to train the model.
    callbacks (list): List of Keras callbacks to apply during training.
    steps_per_epoch_tr (int): Number of training steps per epoch.
    steps_per_epoch_vl (int): Number of validation steps per epoch.
    validation_batch_size (int): Batch size for validation data.

    Returns:
    tf.keras.Model: The trained Keras model.
    """
    # Initiates the training process using the fit method
    model.fit(
        dataset_tr,                              # Training data
        epochs=epochs,                           # Total number of epochs to train
        verbose=1,                               # Verbosity mode: 0 = silent, 1 = progress bar, 2 = one line per epoch
        shuffle=True,                            # Whether to shuffle the training data before each epoch
        validation_data=dataset_vl,              # Validation data to evaluate the model at the end of each epoch
        validation_batch_size=validation_batch_size,  # Batch size for validation data
        steps_per_epoch=steps_per_epoch_tr,      # Total number of steps (batches of samples) before declaring one epoch finished
        validation_steps=steps_per_epoch_vl,     # Total number of steps (batches of samples) to validate after each epoch
        callbacks=callbacks                      # List of callback functions to apply during training
    )
    # Explanation:
    # - **dataset_tr:** The training dataset provides input data and labels for the model to learn from.
    # - **epochs:** Determines how many times the entire training dataset is passed through the model.
    # - **verbose:** Controls the amount of logging during training. '1' provides a progress bar with metrics.
    # - **shuffle:** Shuffling the training data each epoch helps prevent the model from learning the order of the data, promoting better generalization.
    # - **validation_data:** Enables evaluation of the model on unseen data at the end of each epoch, providing insight into overfitting and generalization.
    # - **validation_batch_size:** Defines how many samples per validation step. Adjusting this can balance memory usage and speed.
    # - **steps_per_epoch_tr & steps_per_epoch_vl:** Define how many batches are processed in each epoch for training and validation, respectively.
    # - **callbacks:** Incorporates mechanisms like early stopping and learning rate adjustments to enhance training efficiency and performance.
    # Learn more about model.fit: https://keras.io/api/models/model_training_apis/#fit-method

    return model
    # Returns the trained model, which can be further evaluated or used for predictions

# Function to evaluate the model with a given dataset
def fun_model_evaluate(model, data, steps):
    """
    Evaluates the given Keras model on the provided dataset.

    Parameters:
    model (tf.keras.Model): The Keras model to evaluate.
    data (tf.data.Dataset): The dataset to evaluate the model on.
    steps (int): Number of steps (batches of samples) to evaluate.

    Returns:
    list: A list containing the evaluation metric(s), e.g., accuracy.
    """
    # Evaluates the model on the provided dataset
    _, acc = model.evaluate(
        data,           # Dataset to evaluate
        steps=steps,    # Number of steps (batches) to evaluate
        verbose=1        # Verbosity mode: 0 = silent, 1 = progress bar, 2 = one line per epoch, 3 = one line per batch (if supported)
    )
    # Explanation:
    # - `model.evaluate` computes the loss and specified metrics on the provided dataset.
    # - The underscore `_` captures the loss value, while `acc` captures the accuracy metric.
    # - **Verbosity 3:** Typically not standard; TensorFlow usually supports 0, 1, or 2. If 3 is unsupported, it defaults to a lower verbosity level.
    # - Ensures that the evaluation is performed over a specified number of steps to handle large datasets efficiently.
    # Learn more about model.evaluate: https://keras.io/api/models/model_training_apis/#evaluate-method

    return [acc]
    # Returns the accuracy as a list. This can be extended to include other metrics if needed.

# Function to compare the weights of base and trained models in FSP and PRX tasks
def fun_control(model_base_fsp, model_base_prx, model_fsp, model_prx):
    """
    Prints examples of weights from specific layers of the base and trained models
    for Fully Supervised Pretext (FSP) and Pretext (PRX) tasks to verify consistency.

    Parameters:
    model_base_fsp (tf.keras.Model): The base model for the FSP task.
    model_base_prx (tf.keras.Model): The base model for the PRX task.
    model_fsp (tf.keras.Model): The trained model for the FSP task.
    model_prx (tf.keras.Model): The trained model for the PRX task.

    Returns:
    None
    """
    print('\n')  # Prints a newline for better readability

    # Print examples of weights from the 3rd layer of the base FSP model
    print('Example of weights in the 3rd layer of model_base_fsp: ',
          model_base_fsp.layers[2].weights[0][0][0][0][:2])
    # Explanation:
    # - Accesses the weights of the 3rd layer (`layers[2]`) in `model_base_fsp`.
    # - `weights[0]` refers to the kernel weights of the layer (as Keras layers can have multiple weight tensors, e.g., kernel and bias).
    # - `[0][0][0][0][:2]` slices the first two weights from the first filter for demonstration.
    # - This provides a glimpse into the initial weights, useful for verifying that models have identical starting points.
    # Learn more about Keras layers and weights: https://keras.io/api/layers/

    # Print examples of weights from the 3rd layer of the base PRX model
    print('Example of weights in the 3rd layer of model_base_prx: ',
          model_base_prx.layers[2].weights[0][0][0][0][:2])
    # Explanation:
    # - Similar to the previous print statement, but for `model_base_prx`.
    # - Verifies that both base models (`fsp` and `prx`) have identical initial weights.
    # Learn more about Keras layers and weights: https://keras.io/api/layers/

    # Print examples of weights from the last layer of the trained FSP model
    print('Example of weights in the last layer of model_fsp     : ',
          model_fsp.layers[-1].weights[0][:1][0][:2])
    # Explanation:
    # - Accesses the weights of the last layer (`layers[-1]`) in `model_fsp`.
    # - `weights[0]` refers to the kernel weights of the Dense layer.
    # - `[:1][0][:2]` slices the first two weights from the first unit for demonstration.
    # - This helps in observing how the weights have been updated after training.
    # Learn more about Keras layers and weights: https://keras.io/api/layers/

    # Print examples of weights from the last layer of the trained PRX model
    print('Example of weights in the last layer of model_prx     : ',
          model_prx.layers[-1].weights[0][:1][0])
    # Explanation:
    # - Similar to the previous print statement, but for `model_prx`.
    # - Provides insight into how the PRX model's weights have evolved during training.
    # Learn more about Keras layers and weights: https://keras.io/api/layers/

    # Overall Explanation:
    # - Comparing weights between the base models and the trained models helps in verifying that training has occurred correctly.
    # - By examining specific weights, students can understand how training updates model parameters.
    # - This function aids in debugging and ensuring that different tasks (FSP and PRX) are influencing the models as intended.
    # Learn more about Keras models and weights: https://keras.io/api/models/model/#get_weights-method

```

```python
#@title Create DWM

'''
Construct and prepare model_dwm for transfer learning.
'''
# Explanation:
# - This section constructs the Downstream Model (DWM) using the pre-trained PRX model (`model_base_prx`).
# - The DWM model is designed to leverage the representations learned during the self-supervised contrastive learning phase (PRX).
# - It incorporates a similar dense layer structure as the Fully Supervised Pretext (FSP) model to ensure consistency for fair comparison.
# Learn more about downstream tasks in transfer learning: https://machinelearningmastery.com/what-is-transfer-learning/

# Constructing model_dwm utilizing the pre-trained model_base_prx, incorporating
# a dense layer similar to that in model_fsp.
inputs_dwm = tf.keras.Input(shape=(res, res, num_channels))
# Explanation:
# - `tf.keras.Input` creates a placeholder tensor for the input data.
# - `shape=(res, res, num_channels)` defines the input dimensions, where:
#   - `res`: Resolution of the input images (height and width).
#   - `num_channels`: Number of channels in the input data (e.g., 3 for RGB images).
# - This input layer serves as the entry point for the DWM model.
# Learn more about tf.keras.Input: https://keras.io/api/layers/core_layers/input/

x_dwm = tf.keras.layers.Conv2D(
    3,                  # Number of filters (output channels)
    (1, 1),             # Kernel size: 1x1 convolution
    activation='relu'   # Activation function: ReLU (Rectified Linear Unit)
)(inputs_dwm)
# Explanation:
# - `Conv2D` layer applies convolutional filters to the input data.
# - `(1, 1)` kernel size means each output channel is a linear combination of the input channels.
# - `activation='relu'` introduces non-linearity, enabling the model to learn complex patterns.
# - This layer effectively reduces or maps the number of channels from `num_channels` to `3`, matching the expected input shape of the pre-trained PRX model.
# Learn more about Conv2D: https://keras.io/api/layers/convolution_layers/convolution2d/

x_dwm = model_base_prx(x_dwm)  # Use the trained base model.
# Explanation:
# - Passes the output of the convolutional layer through the pre-trained PRX base model (`model_base_prx`).
# - `model_base_prx` extracts high-level features from the input data, leveraging the representations learned during the self-supervised learning phase.
# - By using the pre-trained model, DWM benefits from rich feature representations without additional training on unlabeled data.
# Learn more about using pre-trained models in Keras: https://keras.io/guides/transfer_learning/#using-pre-trained-models-as-layers-in-your-own-models

x_dwm = tf.keras.layers.Flatten()(x_dwm)
# Explanation:
# - `Flatten` layer converts multi-dimensional feature maps into a single 1D vector.
# - This is necessary before passing the data into dense (fully connected) layers, which expect 1D input.
# - Prepares the extracted features for classification.
# Learn more about Flatten layer: https://keras.io/api/layers/core_layers/flatten/

x_dwm = tf.keras.layers.BatchNormalization()(x_dwm)
# Explanation:
# - `BatchNormalization` layer normalizes the activations of the previous layer for each batch.
# - Helps in reducing internal covariate shift, leading to faster and more stable training.
# - Acts as a regularizer, potentially reducing the need for other forms of regularization.
# Learn more about BatchNormalization: https://keras.io/api/layers/normalization_layers/batch_normalization/

outputs_dwm = tf.keras.layers.Dense(
    2,                  # Number of output units (classes)
    activation='softmax' # Activation function: Softmax for multi-class classification
)(x_dwm)  # Set for 10 output classes.
# Explanation:
# - `Dense` layer with `2` units implies binary classification (e.g., class 0 and class 1).
# - `activation='softmax'` converts the output into probability distributions over the classes.
# - **Note:** The comment mentions "Set for 10 output classes," but the Dense layer is configured for 2 classes. Ensure consistency between the comment and the actual configuration.
# - This layer serves as the final classification layer for the DWM model.
# Learn more about Dense layers: https://keras.io/api/layers/core_layers/dense/

model_dwm = tf.keras.Model(inputs_dwm, outputs_dwm)
# Explanation:
# - `tf.keras.Model` creates a Keras model by specifying the input and output tensors.
# - `model_dwm` is now a complete model that can be compiled and trained on data.
# - This model is tailored for the Downstream (DWM) task, utilizing the features learned from the PRX pretext task.
# Learn more about Keras Model: https://keras.io/api/models/model/

# For equitable comparison between fsp and dwm, we align the parameters of dwm's
# final dense layer with outputs_fsp_initial_parameters.
# Additionally, we transfer and freeze the pre-trained weights of the preceding
# dense layer from prx into dwm's structure to leverage their learned
# representations.
# Set the initial parameters for fair comparison and incorporate pre-trained layers.
model_dwm.layers[-1].set_weights(outputs_fsp_initial_parameters)  # Match initial output layer parameters.
# Explanation:
# - Sets the weights of the final dense layer in `model_dwm` to match the initial parameters of the FSP model's output layer (`outputs_fsp_initial_parameters`).
# - Ensures that both FSP and DWM models start with the same output layer weights, providing a fair baseline for comparison.
# - Facilitates direct comparison of how training affects each model's performance.
# Learn more about setting layer weights: https://keras.io/api/models/model/#set_weights-method

model_dwm.layers[-2].set_weights(model_prx.layers[-2].weights)  # Use pre-trained intermediate layer.
# Explanation:
# - Transfers the weights of the second-to-last layer from the PRX model (`model_prx.layers[-2].weights`) to the corresponding layer in the DWM model (`model_dwm.layers[-2].weights`).
# - This leverages the learned representations from the PRX task, allowing DWM to build upon these features for the downstream classification task.
# - By freezing or appropriately configuring the trainability of these layers, DWM can utilize these pre-trained weights effectively.
# Learn more about weight transfer in Keras: https://keras.io/guides/transfer_learning/#transfer-learning-by-freezing-and-finetuning

# Overall Explanation:
# - The DWM model is constructed to utilize the feature representations learned during the self-supervised PRX task.
# - By aligning the output layer weights with those of the FSP model and transferring the intermediate layer weights from PRX, DWM is set up for a fair and effective downstream classification task.
# - This setup allows for a direct comparison between fully supervised learning (FSP) and self-supervised learning followed by fine-tuning (DWM), demonstrating the effectiveness of self-supervised approaches.
# Learn more about model comparison in machine learning: https://towardsdatascience.com/model-evaluation-metrics-82c49b6e0f24
```
