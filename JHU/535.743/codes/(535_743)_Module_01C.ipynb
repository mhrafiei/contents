## **1.  ML Data**

### **1.1. Data Mathematics**

 ML mimics how the human brain learns from data it receives through sensory inputs. It interprets the inputs, extracts features from the inputs, combines the different modalities of inputs (e.g., vision & voice), & generates a response. The data can be represented mathematically in arrays in a  ML model.

> In a regression problem, one data point (e.g., one animal) can be represented as $\mathbf{X}_{i} = [X_{i,1}, ..., X_{i,k},... X_{i,K}]$ where array  $\mathbf{X}_{i}$ represents the $i^{th}$ data point (e.g., second animal), & $X_{i,k}$ represents the $k^{th}$ attribute (e.g., animal's weight) associated with the problem (e.g., regression problem).

>A  ML model for this particular problem (e.g., regression) can be represented as a function, $f$, such $\bar{Y}_{i} = f(\mathbf{X}_{i})$ where $\bar{Y}_{i}$ is the estimation associated with $\mathbf{X}_{i}$ (e.g., animal jump height). In a regression problem, we can measure how far these two are from one another using metrics such as Squared Error (SE), $SE = ||Y_{i} - \bar{Y}_{i}||^{2}$. If we do this for all $N$ data points (e.g., all $N$ animals), we can compute the mean SE (MSE) of this  ML model:
\begin{equation}
\text{MSE} = \frac{1}{N}\sum_{i=1}^{N}{||Y_{i} - \bar{Y}_{i}||^{2}}
\end{equation}
 ML aims to modify the parameters (e.g., weights & biases) of $f$ so the MSE maximizes; this modification is done by minimizing the loss function, $l$. If the loss function is minimized, the  ML model, $f$, is called "trained" or "fitted" over $N$ data points (i.e., training data points). Once $f$ is trained, we can use it to estimate a new data point.


> In a classification problem, one data point (e.g., one cat) can be represented as $\mathbf{X}_{i,j} = [X_{i,j,1}, ..., X_{i,j,k},... X_{i,j,K}]$ where array  $\mathbf{X}_{i,j}$ represents the $i^{th}$ data point (e.g., second cat) in category $j$ (e.g., category of cats), & $X_{i,j,k}$ represents the $k^{th}$ attribute (e.g., animal's weight) associated with the problem (e.g., animal classification problem).

> A  ML model for this particular problem (e.g., animal classification problem) can be represented as a function, $f$, such $\bar{Y}_{i,j} = f(\mathbf{X}_{i,j})$ where $\bar{Y}_{i,j}$ is the estimation associated with $\mathbf{X}_{i,j}$ (e.g., animal category). In a classification problem, either $\bar{Y}_{i,j} = Y_{i,j}$, where we count one correct estimation, or not, where we count one wrong estimation. If we do this for all $N$ data points (e.g., all cats & dogs), we can compute the accuracy of this  ML model:
\begin{equation}
\text{Accuracy} = \frac{\text{Number of Correctly Estimated Data Points}}{N}
\end{equation}
 ML aims to modify the parameters (e.g., weights & biases) of $f$ so the accuracy maximizes; this modification is done by minimizing the loss function, $l$. If the loss function is minimized, the  ML model, $f$, is called "trained" or "fitted" over $N$ data points (i.e., training data points). Once $f$ is trained, we can use it to estimate the category of a new data point.

> Remember, these are just the regression & classification examples. We have a different but similar logic in other types of  ML models. For example, in Generative Adversarial Networks for image generation, we have loss functions to be minimized for the goal of maximizing the quality of the generated images & confusing the discriminator. Again, there are loss functions to be minimized & some measurements to show the quality of the  ML model. The same logic is used in Natural Language Processing (NLP), Object Detection (OD), Generative Pre-training Transformer (GPT), & any other  ML models out there.


### **1.2. Data Jargon**

> Training Data: This is the data on which the  ML model is trained. The training phase involves adjusting the model's parameters to minimize the error between the predicted & actual outcomes.
The model learns from this data, & this is the bulk of the dataset.

> Validation data: After the model has been trained on the training data, it is then tested on the validation data. This data is used to provide an unbiased evaluation of a model fit during the training phase. It is crucial for hyperparameter tuning & for providing an estimate of the model's performance while tuning the model's architectures, such as the number of hidden units in an NN, learning rate, etc. The key point is that the model does not learn from this data, but it is used to make decisions about which model & model parameters are best.

> Testing Data: This is new data that has not been seen by the model during the training or validation phases. It is used to provide an unbiased evaluation of the final model fit. Testing data is crucial to ensure that the model's predictions are consistent across new, unseen data & not just the data it was trained on. It's the final check on the model performance & is only used once a model is completely trained & validated.

> Production data points: the data point a  ML estimates in a production phase, usually for a client. We cannot directly evaluate the estimations, but we have some confidence in their accuracy based on the training, validation, & testing results.  

> Inputs & Outputs: Whether a training, validation, testing, or production data point, a "data point" includes "inputs" & might include "output/s." When we have outputs & need outputs to train our  ML model, it is called a supervised problem; otherwise, it is unsupervised.

### **1.3 Data Train, Validation, & Test Division**

> Sometimes, the training, validation, and/or testing data points are separated for us in a  ML problem. In cases where it is not, we can separate the data using a predefined testing ratio. For example, 20% of a repository can be selected & set aside as the testing data & the remaining as the training data. If we are interested in having validation data, we can set aside a portion, say 5%, of the current training data as the validation data. We can also ignore having testing data & treat the 20% testing as the validation data. However, the best practice is to have the training, validation, & testing data separately.

```python
#@title Example of Training, Testing, & Validation Split
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Load the Wine Quality dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
data = pd.read_csv(url, delimiter=';')

# Separate features & target variable
X = data.drop('quality', axis=1)
y = data['quality']

# Split the data into training & test sets (80% train, 20% test)
datain_tr, datain_te, dataou_tr, dataou_te = train_test_split(X,
                                                              y,
                                                              test_size=0.2,
                                                              random_state = np.random.randint(10000))

# Split the training data to create a validation set (95% train, 5% validation)
datain_tr, datain_vl, dataou_tr, dataou_vl = train_test_split(datain_tr,
                                                              dataou_tr,
                                                              test_size=0.05,
                                                              random_state = np.random.randint(10000))

# Print the shape of the datasets
print("datain_tr shape:", datain_tr.shape)
print("dataou_tr shape:", dataou_tr.shape)
print("datain_vl shape:", datain_vl.shape)
print("dataou_vl shape:", dataou_vl.shape)
print("datain_te shape:", datain_te.shape)
print("dataou_te shape:", dataou_te.shape)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
try:
  from google.colab import runtime
  import time
  time.sleep(5)
  runtime.unassign()
except ImportError:
  pass
```

## **2. ML Problems & Models**

### **2.1.  ML Problem Types**

#### 2.1.1. Supervised Problems

Supervised Learning is a type of  ML where the model is trained on a labeled dataset. This means that the data used for training includes both the input data (features) & the correct output (labels or targets). The goal of supervised Learning is for the model to learn a mapping from inputs to outputs so that it can predict the output for new, unseen inputs based on the patterns it learned during training. Here's a breakdown of the key aspects:

> * Labeled Data: The dataset consists of examples, each with a set of features & a corresponding label. For example, in a spam detection system, each email is labeled as "spam" or "not spam."
> * Training Process: During training, the algorithm iteratively makes predictions on the training data & is corrected by the true labels, allowing it to learn over time.
> * Types of Problems:
 * Classification: The output variable is a category, such as "spam" or "not spam" in email filtering.
 * Regression: The output variable is a continuous value, for example, predicting house prices based on features like size & location.
> * Generalization: The ultimate goal is for the model to generalize from the training data to new, unseen data. This means it should make accurate predictions on data it hasn't encountered during training.
> * Evaluation: Models are evaluated using new data (test set) to assess their performance. Common metrics are accuracy for classification problems & mean squared error for regression problems.

Supervised Learning has numerous applications in mechanical engineering:

> * Advanced Manufacturing:
>  * Predictive Maintenance of Manufacturing Equipment: Using  ML to predict when manufacturing equipment will need maintenance based on historical data of machine wear & failure patterns.
>  * Quality Control in Production Lines: Implementing supervised learning algorithms to automatically inspect & identify product defects on a production line, learning from a dataset of images labeled with defect types.

> * Biomechanical Engineering:
>  * Prosthetic Limb Control: Developing a supervised learning model to interpret neural signals or muscle movements for controlling prosthetic limbs, learning from data of intended limb movements & the corresponding neural or muscular signals.
>  * Gait Analysis & Correction: Using supervised Learning to analyze & predict abnormalities in human gait, trained on biomechanical data of normal & abnormal walking patterns.

> * Fluid Mechanics & Thermal Science:
>  * Optimization of HVAC Systems: Applying supervised Learning to optimize heating, ventilation, & air conditioning systems in buildings, using historical data on temperature, humidity, & energy usage.
>  * Aerodynamic Design of Vehicles: Utilizing supervised Learning to predict the aerodynamic properties of different vehicle designs, trained on simulation or wind tunnel data.

> * Hypersonic Technologies:
>  * Thermal Protection System Analysis: Implementing supervised learning models to predict the performance & failure of thermal protection systems in hypersonic vehicles, using data from high-speed flight tests & simulations.
>  * Hypersonic Flight Path Optimization: Using supervised Learning to optimize flight paths of hypersonic vehicles for efficiency & safety based on historical flight data & atmospheric conditions.

> *  Robotics, Dynamics, & Controls:
>  * Autonomous Robot Navigation: Training supervised learning algorithms to navigate robots in complex environments, using data collected from sensors in real-world or simulated environments.
>  * Precision Control of Robotic Arms: Developing models to precisely control robotic arms in tasks like assembly or surgery, trained on data from successful maneuvers & corrections.

> *  Solids/Mechanics of Materials:
>  * Material Fatigue Prediction: Using supervised Learning to predict when materials will fail under stress, trained on historical data of material stress tests.
>  * Structural Health Monitoring: Implementing  ML models to monitor the health of structures like bridges or buildings, using data from sensors that measure vibrations, strains, or other indicators of structural integrity.

#### 2.1.2. Unsupervised Problems

In Unsupervised Learning, the model is trained on a dataset that does not have any labels or target values. The goal here is to discover underlying patterns, structures, or features in the data without any guidance on what to look for. Here are the key characteristics & types of unsupervised learning problems:

> * Unlabeled Data: The training data consists only of input data without any corresponding output labels. The model has to make sense of the data on its own.

> * Discovering Patterns: Unsupervised learning algorithms try to find patterns, groupings, or correlations within the data. They do this by analyzing the structure of the data & identifying any interesting or unusual patterns.

> * Types of Problems:
 * Clustering: This involves grouping data points into clusters of similar items. For example, clustering customers based on purchasing behavior.
 * Dimensionality Reduction: This is about reducing the number of variables in data while retaining its essential features. An example is Principal Component Analysis (PCA).
 * Association: Discovering rules that describe large portions of your data, such as people who buy X often buy Y.
 * Anomaly Detection: Identifying rare items or events in data that differ significantly from the majority of the data. This is often used in fraud detection.
> * No Explicit Feedback: Unlike supervised Learning, there is no explicit feedback or "correct answer." The algorithm must discern structures & patterns without guidance.

Unsupervised Learning has numerous applications in mechanical engineering:

> * Advanced Manufacturing:
>   * Anomaly Detection in Manufacturing Processes: Utilizing unsupervised Learning to identify unusual patterns or anomalies in manufacturing data, which could indicate problems or inefficiencies in the process.
>   * Process Optimization: Implementing unsupervised learning algorithms to analyze vast datasets from manufacturing processes to uncover hidden patterns, correlations, & insights for process improvement.

> * Biomechanical Engineering:
>   * Muscle Movement Clustering: Using unsupervised Learning to cluster muscle movement patterns to understand different types of muscle activities & their implications in various health conditions or athletic performances.
>   * Biomechanical Data Exploration: Applying unsupervised learning techniques to explore & analyze large datasets of biomechanical data to find underlying patterns or structures without predefined labels.

> * Fluid Mechanics & Thermal Science:
>   * Flow Pattern Analysis: Implementing Unsupervised Learning to analyze fluid flow patterns in various scenarios, helping in the classification & understanding of complex flow dynamics.
>   * Thermal System Behavior Clustering: Using unsupervised learning algorithms to cluster thermal system behaviors, aiding in identifying optimal operating conditions or potential issues.

> * Hypersonic Technologies:
>   * Material Property Exploration: Utilizing unsupervised learning methods to analyze & categorize various materials used in hypersonic technologies based on their properties & performance characteristics.
>   * Flight Data Segmentation: Applying unsupervised Learning to segment & analyze large volumes of flight test data for hypersonic vehicles to uncover patterns & insights.

> * Robotics, Dynamics, & Controls:
>   * Sensor Data Clustering: Using unsupervised Learning to cluster & analyze data from various sensors on robots, helping to improve understanding & calibration of robotic systems.
>   * Dynamic System Characterization: Implementing Unsupervised Learning to analyze & characterize the dynamics of robotic systems, identifying unique patterns & behaviors.

> * Solids/Mechanics of Materials:
>   * Material Characterization: Applying unsupervised Learning to analyze & classify different materials based on their stress-strain behavior & other mechanical properties.
>   * Structural Health Analysis: Utilizing unsupervised learning techniques to analyze data from structural health monitoring systems, identifying patterns that might indicate potential structural issues.


### **2.2.  ML Major Problems**

 ML is a vast field with numerous applications, especially in engineering disciplines like Mechanical Engineering. Here's a breakdown of the different types of problems it tackles, along with their explanations & examples relevant to Mechanical Engineering.

#### 2.2.1. Estimations

> * Regression: Involves predicting a continuous output variable based on one or more input features. This is essential in predicting quantities like temperature, pressure, or material properties.
 * Examples: Predicting the fatigue life of a component based on its material properties; Estimating the thermal conductivity of a new alloy; Forecasting energy consumption in a manufacturing process.
> * Classification: Focuses on categorizing data into predefined classes. It's widely used in fault diagnosis & quality control.
 * Examples: Classifying the type of defect in a welded joint; Identifying the grade of a material based on its microstructure; Detecting abnormal vibrations in rotating machinery for maintenance.
> * Anomaly Detection: This involves identifying unusual patterns that deviate from the norm. It's crucial for predictive maintenance & safety.
 * Examples: Detecting cracks in structural components using sensor data; Identifying unusual wear patterns in bearings; Monitoring for unexpected thermal hotspots in engines.

#### 2.2.2. Recognitions

> * Object Detection: Identifying & locating objects within images or videos. Key in quality control & automation.
 * Examples: Detecting defects in casting or machining processes through image analysis; Identifying & sorting different parts on a production line; Monitoring assembly operations for quality assurance.
> * Pattern Recognition: Involves recognizing patterns in data, essential in interpreting complex datasets.
 * Examples: Recognizing wear patterns in machinery; Identifying patterns in material failure modes; Analyzing flow patterns in fluid dynamics studies.
> * Speech Recognition: Converting spoken words into text, useful in voice-controlled systems.
 * Examples: Voice commands for operating machinery; Speech-to-text for documenting engineering processes; Voice-controlled interfaces for CAD software.

#### 2.2.3. Generations

> * Image Generation: Generating new images or modifying existing ones, often used for design & simulation.
 * Examples: Creating virtual prototypes of new components; Simulating material deformation under stress; Visualizing fluid flow in complex geometries.
> * Text Generation: Generating human-like text. Useful in automated reporting & documentation.
 * Examples: Generating maintenance reports from sensor data; Creating product descriptions for technical catalogs; Automating the generation of simulation result summaries.
> * Signal Generation: Creating artificial signals or modifying real ones used in control systems & diagnostics.
 * Examples: Synthesizing sensor signals for testing control algorithms; Generating acoustic emission signals for material testing; Simulating vibration signals for fault diagnosis training.


#### 2.2.4. Accelerations

> * Algorithm Acceleration: Using ML to speed up computational algorithms crucial in simulation & optimization.
 * Examples: Accelerating finite element analysis for structural simulations; Speeding up computational fluid dynamics simulations; Enhancing optimization algorithms for design problems.
> * Data Processing Acceleration: Speeding up the analysis of large datasets.
 * Examples: Rapid processing of sensor data from manufacturing equipment; Accelerating the analysis of material properties from large databases; Quick interpretation of thermal imaging data.
> * Decision-Making Acceleration: Enhancing the speed of decision-making processes.
 * Examples: Real-time decision-making in automated manufacturing lines; Quick adaptation of control systems in response to changing conditions; Fast risk analysis in project management.

#### 2.2.5. Feature Extractions

> * Dimensionality Reduction: Reducing the number of variables under consideration to simplify models without losing key information.
 * Examples: Reducing sensor data dimensions for monitoring machine health; Simplifying complex material property data; Compressing CAD model data for faster processing.
> * Feature Identification: Automatically identifying the most relevant features in a dataset.
 * Examples: Identifying key factors affecting product quality; Pinpointing critical parameters in thermal management systems; Highlighting dominant frequencies in vibration analysis.
> * Feature Transformation: Transforming features into a more advantageous or interpretable format.
 * Examples: Transforming raw sensor data for better anomaly detection; Converting time-series data for easier pattern recognition in machine behavior; Transforming stress-strain data for better material property predictions.
