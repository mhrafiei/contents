## **1. Labeling Challenges of Large Repositories**

> The task of manually labeling an extensive repository of ten million animal images is a prime example of the burdens associated with manual data annotation. This process demands significant human effort, as each image must be individually examined and categorized by human annotators. Given the sheer volume of images, the task would require a substantial team of labelers working for an extended period, potentially leading to inconsistencies due to fatigue or subjective judgment. Moreover, the cost of such an undertaking is considerable, encompassing the wages of the labelers and the management of the labeling process to ensure quality and consistency across the dataset.

> When dealing with textual data, such as identifying sentiment in thousands of documents, manual labeling presents challenges of scale and subjectivity. Each document must be meticulously read, understood, and then classified as positive, negative, or neutral. This process is not only time-intensive but also prone to human error. Sentiments can be subtle and context-dependent, which increases the difficulty of maintaining high accuracy and consistency. The requirement for trained personnel who can interpret nuanced language adds to the expense and complexity of the project.

> As another example, in the field of materials science, characterizing the properties of materials through manual experiments is a demanding task. It involves procuring the necessary ingredients and laboratory equipment, which can be costly. Furthermore, there is a need to employ experts capable of conducting experiments, such as building and testing thousands of material samples, which can be a slow and meticulous process. The manual nature of this work limits the number of experiments that can be conducted within a given timeframe, potentially delaying the advancement of research and increasing the overall cost of material development.

> In each of these scenarios, the burden of manual labeling is highlighted by the extensive resources required, including time, financial outlay, and skilled labor. These factors combined make manual labeling a significant bottleneck in data processing and analysis.



## **2. Labeling Challenges in Mechanical Engineering**

Here, we review a few examples of labeling challenges in different focus areas in Mechanical Engineering.

> **Advanced Manufacturing**
- **Example 1**: In quality control using visual inspection systems, accurately labeling manufacturing defects requires expert knowledge to distinguish between minor anomalies and actual defects, leading to a significant time investment and potential for inconsistency.
- **Example 2**: For tool wear classification from acoustic emission signals, collecting and labeling the dataset involves continuous monitoring of the machining process, which can be time-consuming and requires the correlation of signal patterns with specific wear types through expert analysis.

> **Biomechanical Engineering**
- **Example 1**: Labeling gait data for identifying abnormalities involves detailed analysis by healthcare professionals who can accurately categorize gait patterns. This process is labor-intensive and requires a high level of expertise to ensure accuracy.
- **Example 2**: In predicting the risk of osteoporosis from bone density scans, the labeling process must consider a range of factors including age, gender, and medical history, making it complex and burdensome to compile a comprehensive and accurately labeled dataset.

> **Fluid Mechanics and Thermal Science**
- **Example 1**: Classifying flow regimes in pipe systems based on velocity profiles necessitates extensive experimental setups to capture a wide range of flow conditions, followed by expert analysis to label each regime correctly, which is both costly and time-consuming.
- **Example 2**: Labeling thermal patterns for efficient cooling in electronics requires thermal imaging under various operational scenarios, demanding extensive testing and expert interpretation to categorize cooling efficiency accurately.

> **Hypersonic Technologies**
- **Example 1**: Labeling thermal protection system materials for different vehicle sections involves rigorous testing at hypersonic conditions to understand material behavior, a process that is exceedingly expensive and technically challenging.
- **Example 2**: Identifying and labeling potential areas of aerodynamic heating on hypersonic vehicles using infrared imaging requires simulation of flight conditions in specialized facilities, making the labeling process highly resource-intensive.

> **Robotics, Dynamics, and Controls**
- **Example 1**: For gesture recognition in human-robot interaction, the labeling process involves capturing a wide variety of human gestures across different users, requiring significant effort in data collection and labeling to account for human variability.
- **Example 2**: Labeling obstacles for autonomous robot navigation demands diverse environmental setups and object types to train robust models, entailing extensive data collection and precise annotation efforts to accurately represent real-world conditions.

> **Solids/Mechanics of Materials**
- **Example 1**: Classifying material failure modes under different stress conditions requires destructive testing and expert post-analysis to label the failure accurately, which is not only costly but also limits the amount of data that can be generated.
- **Example 2**: Labeling stress concentration types in structural components from strain data involves complex simulations or experimental setups to induce and measure stress concentrations, followed by detailed analysis to categorize them, a process that is both labor-intensive and technically demanding.


## **3. Supervised Learning Labeling with Limited Labeled Data**

A supervised annotator model can be developed with pre-trained parameters. In this lecture, we will explore an example of a supervised classification model using a portion of the CIFAR-10 dataset. The CIFAR-10 dataset is segmented into 50,000 images for training, in which only 5% is assumed to be labeled, & 10,000 for testing.

```python
#@title Example of Supervised Learning with Limited Labeled Data - EfficientNetB1
'''
Runtime: GPU $$$

tf.keras.datasets.cifar10:               https://www.tensorflow.org/api_docs/python/tf/keras/datasets/cifar10
tf.data.Dataset:                         https://www.tensorflow.org/api_docs/python/tf/data/Dataset
tf.keras.utils.to_categorical:           https://www.tensorflow.org/api_docs/python/tf/keras/utils/to_categorical
tf.keras.applications.MobileNet:         https://www.tensorflow.org/api_docs/python/tf/keras/applications/DenseNet121
tf.keras.layers.GlobalAveragePooling2D:  https://www.tensorflow.org/api_docs/python/tf/keras/layers/GlobalAveragePooling2D
tf.keras.layers.Dense:                   https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dense
tf.keras.layers.Dropout:                 https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dropout
tf.keras.Model:                          https://www.tensorflow.org/api_docs/python/tf/keras/Model
tf.keras.callbacks.EarlyStopping:        https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/EarlyStopping
tf.keras.callbacks.ReduceLROnPlateau:    https://www.tensorflow.org/api_docs/python/tf/keras/callbacks/ReduceLROnPlateau
tf.keras.optimizers.Adam:                https://www.tensorflow.org/api_docs/python/tf/keras/optimizers/Adam
'''

import tensorflow as tf

# Load CIFAR-10 dataset
(datain_tr, dataou_tr), (datain_vl, dataou_vl) = tf.keras.datasets.cifar10.load_data()

# Randomly select 5% of training data
rate_labeled = 0.05
num_labeled  = int(datain_tr.shape[0] * rate_labeled)
print("Number of labeled training data: {}".format(num_labeled))
index_tr  = tf.experimental.numpy.random.randint(0,
                                                 datain_tr.shape[0],
                                                 num_labeled)
datain_tr = datain_tr[index_tr,:,:,:]
dataou_tr = dataou_tr[index_tr,:]

# Print the shaped
print('Shape of datain_tr: {}'.format(datain_tr.shape))
print('Shape of datain_te: {}'.format(datain_vl.shape))
print('Shape of dataou_tr: {}'.format(dataou_tr.shape))
print('Shape of dataou_te: {}'.format(dataou_vl.shape))

# Normalize the images
datain_tr = datain_tr.astype('float32') / 255.0
datain_vl = datain_vl.astype('float32') / 255.0

# Convert labels to one-hot encoding
dataou_tr = tf.keras.utils.to_categorical(dataou_tr, 10)
dataou_vl = tf.keras.utils.to_categorical(dataou_vl, 10)

# Create TensorFlow datasets
dataset_tr = tf.data.Dataset.from_tensor_slices((datain_tr, dataou_tr)).batch(32).prefetch(tf.data.experimental.AUTOTUNE)
dataset_vl = tf.data.Dataset.from_tensor_slices((datain_vl, dataou_vl)).batch(32).prefetch(tf.data.experimental.AUTOTUNE)

# Load the DenseNet121 model pre-trained on ImageNet, excluding the top layer
base_model = tf.keras.applications.DenseNet121(weights='imagenet', include_top=False, input_shape=(32, 32, 3))

# Freeze the base model
base_model.trainable = False

# Create the model architecture
inputs = tf.keras.Input(shape=(32, 32, 3))
x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.2)(x)
outputs = tf.keras.layers.Dense(10, activation='softmax')(x)
model = tf.keras.Model(inputs, outputs)

# Compile the model
model.compile(optimizer=tf.keras.optimizers.Adam(),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=0.00001)

# Transfer Learning
history = model.fit(dataset_tr,
                    epochs=20,
                    validation_data=dataset_vl,
                    callbacks=[early_stopping, reduce_lr])

# Fine-tuning
model.compile(optimizer=tf.keras.optimizers.Adam(1e-5),
              loss='categorical_crossentropy',
              metrics=['accuracy'])
history_fine = model.fit(dataset_tr,
                         epochs=10,
                         validation_data=dataset_vl,
                         callbacks=[early_stopping, reduce_lr])

# Evaluate the model
test_loss, test_acc = model.evaluate(dataset_vl)
print(f'\nTest accuracy: {test_acc}, Test loss: {test_loss}')

# Model Summary
model.summary()

"""
The code illustrates an approach to supervised learning using a
significantly limited subset of labeled data from the CIFAR-10 dataset,
leveraging only 5% of the available training samples. This method simulates a
scenario common in many real-world applications where labeled data is scarce or
expensive to obtain. By randomly selecting a small fraction of the training data
for model training, it demonstrates a practical yet challenging situation faced
by machine learning practitioners. The use of TensorFlow and Keras for data
preprocessing, model construction, and training is evident, incorporating
advanced techniques such as transfer learning with a DenseNet121 base model
pre-trained on ImageNet, and fine-tuning to adapt the model to the specific task
with the minimal labeled dataset.

The results, may not be optimal due to the limited amount of labeled
data used for training. This situation underlines the significance of labeled
data in achieving high model performance and the challenges of working with
limited resources. Techniques like transfer learning and fine-tuning are
strategic attempts to mitigate these limitations by leveraging pre-learned
features from a model trained on a large, diverse dataset. Despite these
efforts, the constraint of limited labeled data likely results in a model that,
while potentially useful, may not fully capture the complexity or variability of
the dataset, leading to lower accuracy or generalizability when evaluated on
unseen data.

Remember, in all these practices, we did not touch the remaining, assumed to be
unlabeled, training input data.
"""

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```
