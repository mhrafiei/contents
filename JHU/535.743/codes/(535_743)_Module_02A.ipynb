## **1. Scikit-Learn Model Evaluations**

Scikit-Learn offers a variety of tools & functions for evaluating ML models. These tools are essential for assessing the performance of models across different metrics & scenarios. Here is a list of major model evaluation tools/functions in Scikit-Learn in this section.

### **1.1. Classification Metrics**

Major Classification Metrics in Scikit-Learn are:

* [Accuracy Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html) measures the accuracy of classification models.

* [Precision Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_score.html) is a measure of the accuracy of the positive predictions made by a classification model. It is defined as the ratio of true positive predictions to the total number of positive predictions (including both true positives & false positives). Precision is particularly important in scenarios where the cost of a false positive is high. For instance, in email spam detection, a high precision means that fewer legitimate emails are incorrectly classified as spam, thus avoiding the inconvenience of missing important emails. Precision is crucial when the consequences of false positives are more significant than false negatives.

* [Recall Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html), also known as sensitivity, measures the proportion of actual positives that are correctly identified by the model. It is calculated as the ratio of true positives to the sum of true positives & false negatives. Recall is especially important in situations where missing a positive instance is costly. For example, in medical diagnostics for a serious disease, a high recall implies that the model identifies most of the sick patients correctly, which is crucial as missing a sick patient could be fatal. In such cases, recall is prioritized, even if it leads to more false positives.

* [F1 Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html) is the harmonic mean of precision & recall. It is a single metric that balances both precision & recall, providing a convenient way to express the overall performance of a classifier when dealing with imbalanced datasets or when both false positives & false negatives carry significant costs. The F1 score is particularly useful when you need to find a balance between precision & recall, & there is an uneven class distribution (i.e., a significant imbalance between the positive & negative instances). It’s a crucial metric for scenarios where both types of errors (false positives & false negatives) are equally important.

* [The Receiver Operating Characteristic (ROC) Area Under Curve (AUC) Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_auc_score.html) computes the area under the ROC curve, a graphical representation of a classifier's performance. This score provides an aggregate measure of the model's performance across all possible classification thresholds, essentially capturing the trade-off between the true positive rate (TPR) & the false positive rate (FPR). A higher AUC indicates better model performance, with a perfect model scoring 1. The ROC AUC score is particularly valuable for evaluating classifiers on imbalanced datasets & for comparing different models, as it considers both the positive (true positives) & negative (true negatives) classes, providing a more comprehensive view of the model's performance beyond mere accuracy.

* [The Confusion Matrix](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.confusion_matrix.html) is a fundamental tool for understanding the performance of classification models. It is a table that visualizes the performance of an algorithm by comparing the actual & predicted classifications. The matrix outlines four outcomes: true positives (TP), true negatives (TN), false positives (FP), & false negatives (FN). This visualization makes it easier to see if the model is confusing two classes, which is particularly important in binary classification. It provides insight into not only the errors being made by a classifier but also the types of errors, helping in the fine-tuning of models, especially in domain-specific scenarios where certain types of errors may have different costs or implications.

* [The Classification Report](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.classification_report.html) in Scikit-Learn provides a quick overview of the precision, recall, F1 score, & support (the number of true instances for each label) for each class in a classification problem. This report is invaluable for getting a concise summary of the overall effectiveness of a classifier. By including multiple key metrics in one report, it allows for easy comparison between different classifiers or models. The classification report is particularly useful when dealing with multi-class problems, as it breaks down the performance metric for each class, giving insights into the model's strengths & weaknesses in classifying different classes.



```python
#@title Example of Classification Metrics
'''
Runtime: CPU
accuracy_score:        https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html
recall_score:          https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html
precision_score:       https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_score.html
f1_score:              https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html
roc_auc_score:         https://scikit-learn.org/stable/modules/generated/sklearn.metrics.roc_auc_score.html
confusion_matrix:      https://scikit-learn.org/stable/modules/generated/sklearn.metrics.confusion_matrix.html
classification_report: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.classification_report.html
make_pipeline:         https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.make_pipeline.html
StandardScaler:        https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html
'''

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, recall_score, precision_score, f1_score,
                             roc_auc_score, confusion_matrix, classification_report)
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Data Generation
np.random.seed(np.random.randint(1000))
datain, dataou = make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=np.random.randint(1000))

# Splitting Data
datain_tr, datain_te, dataou_tr, dataou_te = train_test_split(datain, dataou, test_size=0.2, random_state=np.random.randint(1000))

# Model
mdl = make_pipeline(StandardScaler(), LogisticRegression(random_state=np.random.randint(1000)))
mdl.fit(datain_tr, dataou_tr)

'''
Using mdl = make_pipeline(StandardScaler(), LogisticRegression(random_state=np.random.randint(1000)))
instead of mdl = LogisticRegression(random_state=np.random.randint(1000)) incorporates a crucial step
in the ML workflow: feature scaling. Let's break down the reasons for this:

- Feature Scaling with StandardScaler: StandardScaler standardizes features by removing the mean and
scaling to unit variance. This is particularly important for models like Logistic Regression that are
sensitive to the scale of input features. In Logistic Regression, features with large values can
disproportionately influence the prediction outcome, leading to a model that is biased towards these
features. Standard scaling ensures that all features contribute equally to the decision process, leading
to a more balanced & fair model.

- Pipeline Integration: make_pipeline is a utility from Scikit-Learn that helps to sequentially apply a list
of transformations & a final estimator. The primary advantage of using a pipeline is that it encapsulates
all preprocessing steps & the model into a single object. This is beneficial because:

 -- It ensures that the same preprocessing steps are applied to both training & testing data, avoiding
inconsistencies & potential errors.
 -- It simplifies code & makes it more readable & maintainable.
 -- When using cross-validation or grid searches for hyperparameter tuning, the pipeline ensures that
preprocessing steps like scaling are applied within each fold of the data, preventing data leakage from the
test fold into the training process.

- Random State in Logistic Regression: The use of random_state=np.random.randint(1000) in Logistic Regression
ensures reproducibility of results. Logistic Regression can involve random processes like data shuffling, and
setting a random state guarantees that you'll get the same results every time you run the model with the same
data & parameters.
'''
# Predictions
predictions = mdl.predict(datain_te)
probabilities = mdl.predict_proba(datain_te)[:, 1]

# Evaluation Metrics
accuracy = accuracy_score(dataou_te, predictions)
recall = recall_score(dataou_te, predictions)
precision = precision_score(dataou_te, predictions)
f1 = f1_score(dataou_te, predictions)
roc_auc = roc_auc_score(dataou_te, probabilities)
conf_matrix = confusion_matrix(dataou_te, predictions)
class_report = classification_report(dataou_te, predictions)

# Printing Metrics
print("Accuracy:", accuracy)
print("Recall:", recall)
print("Precision:", precision)
print("F1 Score:", f1)
print("ROC AUC Score:", roc_auc)
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", class_report)

# Confusion Matrix Visualization
plt.figure(figsize=(3, 3))
plt.imshow(conf_matrix, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.colorbar()
tick_marks = np.arange(2)
plt.xticks(tick_marks, ['Class 0', 'Class 1'])
plt.yticks(tick_marks, ['Class 0', 'Class 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
thresh = conf_matrix.max() / 2.
for i, j in np.ndindex(conf_matrix.shape):
    plt.text(j, i, conf_matrix[i, j],
             horizontalalignment="center",
             color="white" if conf_matrix[i, j] > thresh else "black")
plt.show()

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

### **1.2. Regression Metrics**

Major Regression Metrics in Scikit-Learn are:

* [The Mean Squared Error](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html) (MSE) is a widely used measure for evaluating the performance of regression models. It calculates the average of the squares of the differences between the actual & predicted values. By squaring the errors, MSE gives more weight to larger errors, making it particularly sensitive to outliers in the dataset. This sensitivity can be both an advantage & a disadvantage, depending on the context. MSE is beneficial when large errors are particularly undesirable, but it can be misleading if the dataset contains many outliers. A lower MSE indicates a better fit of the model to the data, with a value of zero representing a perfect fit.

* [The Mean Absolute Error](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html) (MAE) measures the average magnitude of errors in a set of predictions, without considering their direction. It is calculated as the average of the absolute differences between the predicted values & the observed values. Unlike the Mean Squared Error, MAE gives a linear score to the errors, which means that all errors are weighted equally in the average. This makes MAE a more robust metric against outliers compared to MSE. In practice, MAE can be more intuitive & easier to interpret, as it provides an estimate of the average error magnitude in the same units as the data.

* [The R2 Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html) ($R^2$) or the coefficient of determination, is a statistical measure that represents the proportion of the variance for the dependent variable that's explained by the independent variables in a regression model. It provides an indication of the goodness of fit of the model & is typically between 0 & 1. A score of 1 indicates that the model perfectly explains the variability of the target data around its mean, while a score of 0 means the model explains none of the variability. $R^2$ is particularly useful for comparing the performance of different regression models on the same dataset.

* [Explained Variance Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.explained_variance_score.html) is similar to the $R^2$ score in that it measures the proportion of variance in the dataset that is explained by the model. However, unlike the $R^2$ score, the explained variance score does not consider the mean error (the difference between the average of the predictions & the average of the actual values). This makes it less sensitive to shifts in the prediction scale & can sometimes provide a more accurate measure of the model's performance, particularly in cases where the mean error is significant. A higher explained variance score indicates a model that more accurately captures the variability in the dataset.


```python
#@title Example of Regression Metrics
'''
Runtime: CPU
mean_squared_error:       https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html
mean_absolute_error:      https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html
r2_score:                 https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html
explained_variance_score: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.explained_variance_score.html
'''
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, explained_variance_score

# Data Generation
np.random.seed(np.random.randint(1000))
datain, dataou = make_regression(n_samples=100, n_features=1, noise=10, random_state=np.random.randint(1000))

# Splitting Data
datain_tr, datain_te, dataou_tr, dataou_te = train_test_split(datain, dataou, test_size=0.2, random_state=np.random.randint(1000))

# Model
mdl = LinearRegression()
mdl.fit(datain_tr, dataou_tr)

# Predictions
predictions = mdl.predict(datain_te)

# Evaluation Metrics
mse = mean_squared_error(dataou_te, predictions)
mae = mean_absolute_error(dataou_te, predictions)
r2 = r2_score(dataou_te, predictions)
explained_var = explained_variance_score(dataou_te, predictions)

# Printing Metrics
print("Mean Squared Error:", mse)
print("Mean Absolute Error:", mae)
print("R2 Score:", r2)
print("Explained Variance Score:", explained_var)

# Plotting
plt.figure(figsize = [3,3])
plt.scatter(datain_te, dataou_te, color='blue', marker='o', s=10,  label='Actual')
plt.scatter(datain_te, predictions, color='red', marker='s', s=10, label='Predicted')

plt.title("Linear Regression Model")
plt.xlabel("Feature")
plt.ylabel("Target")
plt.legend()

# Turning on the grid
plt.grid(True, which='both')
plt.show()

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

### **1.3. Clustering Metrics**

Major Clustering Metrics in Scikit-Learn are:

* [The Silhouette Score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.silhouette_score.html) is a metric used to calculate the effectiveness of a clustering algorithm. It measures how similar each point in one cluster is to points in the neighboring clusters, thus providing a way to assess the separation distance between the resulting clusters. This score is calculated for each sample & can range from -1 to 1. A high silhouette score indicates that the object is well matched to its own cluster & poorly matched to neighboring clusters. If many points have a high value, the clustering configuration is appropriate. If many points have a low or negative value, the clustering configuration may have too many or too few clusters. The Silhouette Score is particularly useful when the ground truth about the data is unknown, & it helps in determining an optimal number of clusters.

* [The Adjusted Rand Score (AIR)](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.adjusted_rand_score.html) is a measure of the similarity between two data clustering assignments, considering all pairs of samples & counting pairs that are assigned in the same or different clusters in the predicted & true clusters. The ARI is adjusted for chance, meaning that it will take a value close to 0 for random labeling independently of the number of clusters & samples. Its range is from -1 to 1. A score close to 1 indicates that the clusters in the two assignments largely agree, whereas a score close to -1 indicates a disagreement between the clusters. The Adjusted Rand Score is particularly useful for measuring the performance of a clustering algorithm when the true cluster assignments are known, making it a good tool for validating clustering algorithms in a supervised learning context.


```python
#@title Example of Clustering Metrics
'''
Runtime: CPU
silhouette_score:    https://scikit-learn.org/stable/modules/generated/sklearn.metrics.silhouette_score.html
adjusted_rand_score: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.adjusted_rand_score.html
make_blobs:          https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_blobs.html
'''

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score

# Data Generation
np.random.seed(np.random.randint(1000))
datain, dataou = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

# K-Means Model
mdl = KMeans(n_clusters=4, random_state=np.random.randint(1000))
mdl.fit(datain)

# Predictions
cluster_labels = mdl.predict(datain)

# Silhouette Score
silhouette_avg = silhouette_score(datain, cluster_labels)
print("Silhouette Score: ", silhouette_avg)

# Adjusted Rand Score
rand_score = adjusted_rand_score(dataou, cluster_labels)
print("Adjusted Rand Score: ", rand_score)

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

### **1.4. Scorer Objects**

Major Scorer Objects in Scikit-Learn is [Make Scorer](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.make_scorer.html); it is a utility function that allows users to create a custom scorer object compatible with model evaluation & hyperparameter tuning tools, such as cross-validation routines (cross_val_score, cross_validate) & grid search (GridSearchCV, RandomizedSearchCV). This function is particularly useful when you need to assess a model based on a specific metric that is not directly available in Scikit-Learn's extensive list of built-in scoring metrics. make_scorer takes a performance metric or loss function, along with any necessary additional parameters, & converts it into a scorer object that can be passed to the aforementioned evaluation & optimization procedures. This flexibility allows for more tailored & nuanced model evaluation, enabling the application of domain-specific performance metrics or customized evaluation criteria that better align with specific project goals or business objectives. Using make_scorer thus extends the versatility of Scikit-Learn's model evaluation & selection capabilities, allowing for a more personalized & precise approach to ML model development.

```python
#@title Example of Scorer Objects
'''
Runtime: CPU
make_scorer:         https://scikit-learn.org/stable/modules/generated/sklearn.metrics.make_scorer.html
make_classification: https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_classification.html
LogisticRegression:  https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html
'''
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import make_scorer, accuracy_score

# Custom scoring function (example: square root of accuracy)
def fun_custom_accuracy(y_true, y_pred):
    return accuracy_score(y_true, y_pred) ** (0.5)

# Create a scorer from the custom scoring function
custom_scorer = make_scorer(fun_custom_accuracy)

# Data Generation with 2 features for visualization
np.random.seed(np.random.randint(1000))
datain, dataou = make_classification(n_samples=1000, n_features=2, n_informative=2, n_redundant=0, n_classes=2, random_state=np.random.randint(1000))

# Splitting Data
datain_tr, datain_te, dataou_tr, dataou_te = train_test_split(datain, dataou, test_size=0.2, random_state=np.random.randint(1000))

# Model
mdl = LogisticRegression(random_state=np.random.randint(1000))

# Cross-validation using custom scorer
scores = cross_val_score(mdl, datain_tr, dataou_tr, scoring=custom_scorer, cv=5)

# Printing the custom scores
print("Custom Accuracy scores for each fold:", scores)
print("Mean Custom Accuracy:", np.mean(scores))

# Fit the model & plot decision boundary
mdl.fit(datain_tr, dataou_tr)

# Plotting decision boundary
x_min, x_max = datain_tr[:, 0].min() - 1, datain_tr[:, 0].max() + 1
y_min, y_max = datain_tr[:, 1].min() - 1, datain_tr[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))

Z = mdl.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.4)
scatter = plt.scatter(datain_tr[:, 0], datain_tr[:, 1], c=dataou_tr, s=20, edgecolor='k')
plt.title('Logistic Regression with Custom Scorer')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(*scatter.legend_elements(), title="Classes")
plt.show()

'''
Plot Explanations:
The contour plot (plt.contourf) displays the decision boundary created by the
Logistic Regression model. The decision boundary separates the feature space
into two regions, each corresponding to one of the two classes (0 or 1) that the
model is predicting. The boundary line is where the model estimates a 50%
probability of belonging to either class. On one side of this line, the model
predicts class 0, & on the other side, it predicts class 1
'''

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

### **1.5. Model Inspection**

Major Model Inspection tools in Scikit-Learn are:

* [The Permutation Importance](https://scikit-learn.org/stable/modules/generated/sklearn.inspection.permutation_importance.html) is a technique used in feature evaluation to determine the importance of individual features for a model's predictive performance. It works by randomly shuffling a feature's values in the dataset & observing the effect on the model's accuracy or performance. The rationale is that shuffling an important feature will significantly decrease model performance, as the model relies heavily on that feature's original order for making accurate predictions. In contrast, shuffling a less important feature will have a minimal impact. This method is model-agnostic, meaning it can be used with any model, & it provides a straightforward & intuitive way to assess feature importance. Permutation importance is particularly useful when trying to understand the contribution of each feature to the model's decision-making process & when identifying features that might not be contributing much & could potentially be removed to simplify the model.

* [The Partial Dependence Plot (PDP)](https://scikit-learn.org/stable/modules/partial_dependence.html) is a visualization tool used in ML to show the effect of a single or a pair of features on the predicted outcome of a model, particularly in gradient boosting models. PDPs illustrate the dependence between the target response & a set of ‘target’ features, marginalizing over the values of all other features (the ‘complement’ features). They can show whether the relationship between the target & a feature is linear, monotonic, or more complex. For example, in a PDP, one might see that the probability of a certain prediction increases sharply up to a certain value of a feature & then levels off, indicating a saturation effect. Partial Dependence Plots are valuable for gaining a deeper understanding of the model's behavior & can be used for both regression & classification problems. They are particularly useful for interpreting complex models like ensemble methods & can help in identifying interaction effects between features.

```python
#@title Example of Model Inspection
'''
Runtime: CPU
https://scikit-learn.org/stable/modules/generated/sklearn.inspection.permutation_importance.html
https://scikit-learn.org/stable/modules/partial_dependence.html
'''

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.inspection import permutation_importance
from sklearn.inspection import PartialDependenceDisplay

# Data Generation
np.random.seed(np.random.randint(1000))
datain, dataou = make_classification(n_samples=1000, n_features=5, n_classes=2, random_state=np.random.randint(1000))

# Splitting Data
datain_tr, datain_te, dataou_tr, dataou_te = train_test_split(datain, dataou, test_size=0.2, random_state=np.random.randint(1000))

# Model
mdl = RandomForestClassifier(random_state=np.random.randint(1000))
mdl.fit(datain_tr, dataou_tr)

# Permutation Importance
perm_importance = permutation_importance(mdl, datain_te, dataou_te, n_repeats=30, random_state=np.random.randint(1000))

# Plotting Permutation Importance
sorted_idx = perm_importance.importances_mean.argsort()
plt.barh(range(len(sorted_idx)), perm_importance.importances_mean[sorted_idx])
plt.xlabel("Permutation Importance")
plt.ylabel("Feature Index")
plt.title("Permutation Importance of Each Feature")
plt.show()
print("\n")

# Partial Dependence Plot
features = [0, 1]  # Indices of features to plot
fig, ax = plt.subplots(figsize=(12, 4))
PartialDependenceDisplay.from_estimator(mdl, datain_tr, features, ax=ax)
plt.suptitle("Partial Dependence Plot")
plt.subplots_adjust(top=0.9)  # Adjust title position
plt.show()

'''
Hint: The PDP shows the dependence of the model's prediction on the selected
features (in this case, features with indices 0 & 1). For each feature, the
plot illustrates how changes in the value of that feature, while keeping all
other features constant, affect the predicted outcome. Essentially, it shows
the marginal effect of a feature on the predicted outcome. In this
two-dimensional plot, the x-axis represents the range of values for the feature,
and the y-axis shows the average predicted target outcome. The plotted line
(or curve) reveals how the predicted outcome varies with the feature value. If
the line is horizontal, the feature does not have a significant effect on the
predictions. A sloped line indicates a positive or negative relationship between
the feature & the target. The PDP helps in understanding the behavior of the
model & how it responds to changes in input features.
'''

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

#### 2.1.1. Linear Models

Major [Linear](https://scikit-learn.org/stable/modules/linear_model.html) models in Scikit-Learn are:

* [Linear Regression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html): Used for predicting continuous values.

The Scikit-Learn library provides a comprehensive set of tools for unsupervised learning, which is another commonly used branches of ML. In this section, we review major unsupervised learning models available in Scikit-Learn.

#### 2.2.1. Clustering Algorithms

Major Clustering Algorithms in Scikit-Learn are:

* [K-Means](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html): A method to partition data into K distinct clusters based on distance to the centroid of a cluster.
* [Hierarchical Clustering (Agglomerative Clustering)](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.AgglomerativeClustering.html): Builds nested clusters by merging or splitting them successively.
* [Density-Based Spatial Clustering of Applications with Noise (DBSCAN)](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.DBSCAN.html): Clusters points that are closely packed together, marking as outliers points that lie alone in low-density regions.
* [Spectral Clustering](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.SpectralClustering.html): Uses the eigenvalues of a similarity matrix to reduce dimensionality before clustering in fewer dimensions.
* [Mean Shift](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.MeanShift.html): Aims to discover “blobs” in a smooth density of samples using a sliding-window-based search.
* [Affinity Propagation](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.AffinityPropagation.html): Creates clusters by sending messages between pairs of samples until convergence.

```python
#@title Example of Agglomerative Clustering
'''
Runtime: CPU
KMeans:              https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html
Agglomerative:       https://scikit-learn.org/stable/modules/generated/sklearn.cluster.AgglomerativeClustering.html
DBSCAN:              https://scikit-learn.org/stable/modules/generated/sklearn.cluster.DBSCAN.html
Spectral:            https://scikit-learn.org/stable/modules/generated/sklearn.cluster.SpectralClustering.html
MeanShift:           https://scikit-learn.org/stable/modules/generated/sklearn.cluster.MeanShift.html
AffinityPropagation: https://scikit-learn.org/stable/modules/generated/sklearn.cluster.AffinityPropagation.html
make_moons:          https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_moons.html
'''
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import make_moons

# Data Generation
np.random.seed(np.random.randint(1000))
datain, _ = make_moons(n_samples=200, noise=0.05, random_state=0)

# Agglomerative Clustering Model
mdl = AgglomerativeClustering(n_clusters=2)
mdl.fit(datain)

# Plotting
plt.figure(figsize = [5,3])
plt.scatter(datain[mdl.labels_ == 0][:, 0], datain[mdl.labels_ == 0][:, 1], color='blue', label='Cluster 1')
plt.scatter(datain[mdl.labels_ == 1][:, 0], datain[mdl.labels_ == 1][:, 1], color='green', label='Cluster 2')
plt.title("Hierarchical Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

#### 2.2.2. Dimensionality Reduction

Major Dimensionality Reduction models in Scikit-Learn are:

* [Principal Component Analysis (PCA)](https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.PCA.html): Linear dimensionality reduction using Singular Value Decomposition of the data & retaining most of the variance.
* [t-Distributed Stochastic Neighbor Embedding (t-SNE)](https://scikit-learn.org/stable/modules/generated/sklearn.manifold.TSNE.html): A tool to visualize high-dimensional data by reducing it to two or three dimensions.
* [Linear Discriminant Analysis (LDA)](https://scikit-learn.org/stable/modules/generated/sklearn.discriminant_analysis.LinearDiscriminantAnalysis.html): A classifier with a linear decision boundary, generated by fitting class conditional densities to the data & using Bayes’ rule.
* [Non-negative Matrix Factorization (NMF)](https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.NMF.html): Factorizes a matrix into one matrix with non-negative elements & a reduced rank.
* [Isomap](https://scikit-learn.org/stable/modules/generated/sklearn.manifold.Isomap.html): A non-linear dimensionality reduction method based on the Isometric Mapping algorithm.

```python
#@title Example of t-SNE
'''
Runtime: CPU
PCA:         https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.PCA.html
TSNE:        https://scikit-learn.org/stable/modules/generated/sklearn.manifold.TSNE.html
LDA:         https://scikit-learn.org/stable/modules/generated/sklearn.discriminant_analysis.LinearDiscriminantAnalysis.html
NMF:         https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.NMF.html
Isomap:      https://scikit-learn.org/stable/modules/generated/sklearn.manifold.Isomap.html
load_digits: https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_digits.html
'''

import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.datasets import load_digits

# Data Loading
digits = load_digits()
datain, dataou = digits.data, digits.target
print("Shape of Original Data Inputs:", datain.shape)
print("\n")

# t-SNE Model
np.random.seed(np.random.randint(1000))
mdl = TSNE(n_components=2, random_state=np.random.randint(1000))
datain_tr = mdl.fit_transform(datain)

# Plotting
plt.figure(figsize=(10, 4))
scatter = plt.scatter(datain_tr[:, 0], datain_tr[:, 1], c=dataou, cmap='viridis', alpha=0.6)
plt.legend(*scatter.legend_elements(), title="Digits")
plt.title("t-SNE - Dimensionality Reduction")
plt.xlabel("t-SNE Feature 1")
plt.ylabel("t-SNE Feature 2")
plt.show()

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```

#### 2.2.3. Anomaly Detection

Major Anomaly Detection models in Scikit-Learn are:

* [One-Class SVM](https://scikit-learn.org/stable/modules/generated/sklearn.svm.OneClassSVM.html): Unsupervised Outlier Detection.
* [Isolation Forest](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.IsolationForest.html): An ensemble method for anomaly detection that isolates anomalies instead of profiling normal data points.

```python
#@title Example of One-Class SVM:
'''
Runtime: CPU
One-Class SVM:    https://scikit-learn.org/stable/modules/generated/sklearn.svm.OneClassSVM.html
Isolation Forest: https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.IsolationForest.html
make_blobs:       https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_blobs.html
'''
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import OneClassSVM
from sklearn.datasets import make_blobs

# Data Generation
np.random.seed(np.random.randint(1000))
datain_tr, _ = make_blobs(n_samples=200, centers=1, cluster_std=0.3, random_state=0)
datain_te = np.random.uniform(low=-6, high=6, size=(20, 2))

# One-Class SVM Model
mdl = OneClassSVM(kernel='rbf', gamma=0.5, nu=0.1)
mdl.fit(datain_tr)

# Predictions
preds_tr = mdl.predict(datain_tr)
preds_te = mdl.predict(datain_te)

# Plotting
plt.scatter(datain_tr[:, 0], datain_tr[:, 1], c='blue', label='Training data')
plt.scatter(datain_te[preds_te == -1][:, 0], datain_te[preds_te == -1][:, 1], c='red', label='Outliers')
plt.title("One-Class SVM Outlier Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()

# Auto runtime disconnection to save CPU/GPU/TPU allocations
from google.colab import runtime
import time
time.sleep(5)
runtime.unassign()
```
