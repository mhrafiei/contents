## **1. Beobachtungen Merkmale Zielwerte**

### **1.1. Zeilen als Stichproben**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_01.jpg?v=1787629421" width="250">



>* Jede Zeile beschreibt einen konkreten Fall
>* Alle Spaltenwerte gehören zu dieser Einheit

>* Zeilen zeigen nur einen Datenausschnitt.
>* Auswahl beeinflusst sinnvolle Aussagen.

>* Jede Zeile braucht eine vergleichbare Einheit
>* Wiederholte Beobachtungen können abhängig sein



```
#@title Python-Code - Zeilen als Stichproben

# Dieses Beispiel zeigt Zeilen als einzelne Stichproben.
# Jede Zeile beschreibt genau eine Wohnung.
# Wir trennen Merkmale und Zielwert sichtbar.

import pandas as pd
import matplotlib.pyplot as plt

# Der kleine Datensatz bleibt vollständig im Arbeitsspeicher.
housing_data = pd.DataFrame(
    {
        "area_m2": [45, 62, 80, 55],
        "rooms": [2, 3, 4, 2],
        "rent_eur": [720, 980, 1350, 810],
    }
)

# Diese Prüfung bestätigt die erwartete Tabellenform.
expected_shape = (4, 3)
if housing_data.shape != expected_shape:
    raise ValueError("Die Tabelle hat nicht die erwartete Form.")

# Jede Zeile ist eine Beobachtung derselben Art.
print("Beobachtungen: 4 Wohnungen, also 4 Zeilen")
print("Merkmale: area_m2 und rooms")
print("Zielwert: rent_eur")

# X enthält nur beschreibende Merkmale.
X = housing_data[["area_m2", "rooms"]]
y = housing_data["rent_eur"]

# Die Formen zeigen Zeilen und Spalten getrennt.
print(f"Form von X: {X.shape} = Zeilen mal Merkmale")
print(f"Form von y: {y.shape} = ein Zielwert pro Zeile")

# Eine einzelne Zeile bleibt eine zusammengehörige Stichprobe.
sample_index = 1
sample = housing_data.loc[sample_index]
print(f"Zeile {sample_index}: {sample['area_m2']} m², {sample['rooms']} Zimmer")
print(f"Zielwert derselben Zeile: {sample['rent_eur']} Euro Miete")

# Das Diagramm markiert jede Zeile als Punkt.
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(housing_data["area_m2"], housing_data["rent_eur"], s=80)

# Die Beschriftungen verbinden Punkte mit Zeilennummern.
for row_index, row in housing_data.iterrows():
    ax.annotate(f"Zeile {row_index}", (row["area_m2"], row["rent_eur"]))

# Achsen und Titel nennen Einheiten und Bedeutung.
ax.set_title("Jede Zeile ist eine Wohnungs-Stichprobe")
ax.set_xlabel("Wohnfläche in Quadratmetern")
ax.set_ylabel("Miete in Euro")
plt.show()

```

### **1.2. Merkmale als Spalten**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_02.jpg?v=1787629423" width="250">



>* Spalten beschreiben gemeinsame Merkmale aller Beobachtungen
>* Gleiche Merkmale machen Muster analysierbar

>* Merkmalsspalten eindeutig benennen und erklären
>* Einheiten, Wertebereiche und Bedeutungen dokumentieren

>* Merkmale haben unterschiedliche Datentypen.
>* Jede Spalte enthält ein klares Merkmal.



```
#@title Python-Code - Merkmale als Spalten

# Dieses Beispiel zeigt Merkmale als Tabellenspalten.
# Jede Spalte beschreibt dieselben Beobachtungen vergleichbar.
# Am Ende entstehen Eingabematrix und Zielvektor.

import pandas as pd

# Wir bauen einen kleinen Datensatz direkt im Speicher.
data = pd.DataFrame(
    {
        "area_m2": [45, 62, 80, 55],
        "rooms": [2, 3, 4, 2],
        "building_year": [1998, 2010, 1985, 2020],
        "rent_eur": [720, 980, 1150, 890],
    }
)

# Diese Spalten sind Merkmale, die Wohnungen beschreiben.
feature_columns = ["area_m2", "rooms", "building_year"]
target_column = "rent_eur"

# X enthält nur die Merkmalsspalten.
X = data[feature_columns]
y = data[target_column]

# Eine einfache Prüfung macht die Tabellenform sichtbar.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen haben.")

# Die Ausgabe bleibt klein und zeigt die wichtigsten Begriffe.
print("Datensatz: 4 Beobachtungen, 3 Merkmale, 1 Zielwert")
print("Merkmalsspalten:", list(X.columns))
print("Zielspalte:", target_column)
print("Form von X:", X.shape)
print("Form von y:", y.shape)
print("Erste Beobachtung als Merkmale:", X.iloc[0].to_dict())
print("Passender Zielwert in Euro:", int(y.iloc[0]))

```

### **1.3. Zielwerte und Labels**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_03.jpg?v=1787629424" width="250">



>* Zielwerte zeigen das Analyseergebnis.
>* Sie machen die Fragestellung konkret.

>* Labels sind numerisch oder kategorial
>* Bedeutungen müssen klar dokumentiert sein

>* Zielwerte auf Vollständigkeit und Gültigkeit prüfen
>* Entstehung, Bedeutung und mögliche Verzerrungen dokumentieren



```
#@title Python-Code - Zielwerte und Labels

# Dieses Beispiel trennt Merkmale und Zielwerte.
# Labels machen die Analysefrage im Datensatz sichtbar.
# Die Ausgabe zeigt Spaltenrollen und Array-Formen.

import pandas as pd

# Wir bauen einen kleinen tabellarischen Datensatz.
data = pd.DataFrame(
    {
        "area_m2": [45, 80, 120, 60],
        "rooms": [2, 3, 5, 2],
        "year_built": [1998, 2010, 1985, 2020],
        "price_eur": [210000, 390000, 520000, 310000],
    }
)

# Diese Spalten beschreiben beobachtete Eigenschaften.
feature_columns = ["area_m2", "rooms", "year_built"]
target_column = "price_eur"

# X enthält Merkmale, y enthält den Zielwert.
X = data[feature_columns]
y = data[target_column]

# Eine einfache Prüfung schützt vor vertauschten Rollen.
if len(X) != len(y):
    raise ValueError("X und y müssen gleich viele Beobachtungen haben.")

# Die Ausgabe bleibt kurz und zeigt die wichtigsten Rollen.
print("Beobachtungen:", len(data))
print("Merkmale:", list(X.columns))
print("Zielwert:", target_column)
print("Form von X:", X.shape)
print("Form von y:", y.shape)
print("Erste Beobachtung als Merkmale:", X.iloc[0].to_dict())
print("Passendes Label in Euro:", int(y.iloc[0]))

```

## **2. X und y**

### **2.1. Eingabematrix erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_01.jpg?v=1787629432" width="250">



>* Zeilen sind Beobachtungen, Spalten sind Merkmale
>* Zielwerte gehören nicht in die Eingabematrix

>* Nur erklärende Merkmale für X auswählen
>* Spaltenreihenfolge in allen Zeilen beibehalten

>* Merkmale passend vorbereiten und prüfen
>* Matrixform: Zeilen Beobachtungen, Spalten Merkmale



```
#@title Python-Code - Eingabematrix erstellen

# Wir erstellen eine Eingabematrix aus Tabellendaten.
# Zielwerte bleiben bewusst außerhalb der Eingabematrix.
# Am Ende sehen wir X, y und Formen.

import pandas as pd
import numpy as np

# Dieser kleine Datensatz enthält Merkmale und einen Zielwert.
data = pd.DataFrame(
    {
        "age_years": [18, 19, 20, 21],
        "study_hours": [6.0, 8.5, 4.0, 10.0],
        "attendance_percent": [90, 95, 70, 98],
        "final_grade": [2.3, 1.7, 3.0, 1.3],
    }
)

# Diese Spalten beschreiben die Beobachtungen.
feature_columns = ["age_years", "study_hours", "attendance_percent"]

# Diese Spalte ist der vorherzusagende Zielwert.
target_column = "final_grade"

# X enthält nur die ausgewählten Eingabemerkmale.
X = data[feature_columns].to_numpy(dtype=float)

# y enthält nur den Zielwert jeder Beobachtung.
y = data[target_column].to_numpy(dtype=float)

# Eine einfache Prüfung schützt vor vertauschten Zeilen.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Die Ausgabe zeigt die Trennung von Eingaben und Zielwerten.
print("Ausgewählte Merkmale:", feature_columns)
print("Zielspalte:", target_column)
print("X-Form:", X.shape)
print("y-Form:", y.shape)
print("Erste Zeile von X:", np.round(X[0], 1).tolist())
print("Erster Wert von y:", round(float(y[0]), 1))

```

### **2.2. Zielwerte als Vektor**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_02.jpg?v=1787629436" width="250">



>* Zielvektor enthält die vorherzusagende Antwort
>* Eine Antwort gehört zu jeder Datenzeile

>* Zielwerte müssen zur Beobachtungsreihenfolge passen
>* Verschiebungen erzeugen falsche Modellzusammenhänge

>* Zielwerte können numerisch oder kategorial sein
>* Klare Bedeutung und Format erleichtern Modellierung



```
#@title Python-Code - Zielwerte als Vektor

# Dieses Beispiel trennt Zielwerte von Eingabemerkmalen.
# Der Zielvektor enthält eine Antwort pro Beobachtung.
# Die Ausgabe zeigt Formen und passende Zuordnung.

import numpy as np
import pandas as pd

# Wir bauen einen kleinen Datensatz direkt im Speicher.
data = pd.DataFrame(
    {
        "area_m2": [45, 60, 75, 90],
        "rooms": [2, 2, 3, 4],
        "price_eur": [210000, 285000, 340000, 420000],
    }
)

# X enthält nur Merkmale, y enthält nur Zielwerte.
X = data[["area_m2", "rooms"]].to_numpy()
y = data["price_eur"].to_numpy()

# Diese Prüfung schützt vor verschobenen Zielwerten.
if X.shape[0] != y.shape[0]:
    raise ValueError("Jede Beobachtung braucht genau einen Zielwert.")

# Wir zeigen die wichtigsten Formen und die erste Zuordnung.
print("Datensatz: 4 Wohnungen, Zielwert ist der Preis in Euro.")
print(f"Form von X: {X.shape} = Beobachtungen mal Merkmale.")
print(f"Form von y: {y.shape} = ein Zielwert pro Beobachtung.")
print(f"Erste Beobachtung in X: {X[0].tolist()}.")
print(f"Passender erster Zielwert in y: {int(y[0])} Euro.")

# Ein Vektor bleibt eindimensional, auch bei mehreren Beobachtungen.
print(f"Datentyp von y: {y.dtype}.")
print(f"Wertebereich von y: {int(np.min(y))} bis {int(np.max(y))} Euro.")
print("Merke: Die Reihenfolge in X und y muss gleich bleiben.")

```

### **2.3. Batchformen verstehen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_03.jpg?v=1787629434" width="250">



>* Ein Batch bündelt mehrere Beobachtungen.
>* Seine Form zeigt Beobachtungen und Merkmale.

>* Zeilen sind Beobachtungen, Spalten sind Merkmale
>* Gleiche Spaltenbedeutung sichert vergleichbare Daten

>* Große Daten werden in Batches verarbeitet
>* Gleiche Struktur sichert passende X- und y-Daten



```
#@title Python-Code - Batchformen verstehen

# Dieses Beispiel zeigt Batchformen in kleinen Tabellen.
# X enthält Merkmale, y enthält Zielwerte.
# Die Ausgabe vergleicht einzelne und mehrere Beobachtungen.

import numpy as np
import pandas as pd

# Wir bauen einen kleinen Datensatz direkt im Speicher.
data = pd.DataFrame(
    {
        "age_years": [25, 41, 36, 52],
        "income_eur": [32000, 54000, 47000, 68000],
        "previous_purchases": [2, 7, 4, 9],
        "will_buy": [0, 1, 0, 1],
    }
)

# Diese Spalten sind Merkmale für die Eingabematrix X.
feature_columns = ["age_years", "income_eur", "previous_purchases"]
X = data[feature_columns].to_numpy()

y = data["will_buy"].to_numpy()

# Eine Beobachtung bleibt zweidimensional, wenn sie ein Batch ist.
single_observation_batch = X[0:1]
first_three_batch = X[0:3]

# Diese Prüfung schützt vor vertauschten oder fehlenden Zielwerten.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

print("Merkmale:", feature_columns)
print("X-Form:", X.shape, "= Beobachtungen x Merkmale")
print("y-Form:", y.shape, "= ein Zielwert pro Beobachtung")
print("Einzelner Batch:", single_observation_batch.shape)
print("Dreier-Batch:", first_three_batch.shape)
print("Erste Beobachtung in X:", single_observation_batch[0].tolist())
print("Passender Zielwert in y:", int(y[0]))

```

## **3. Datentypen dokumentieren**

### **3.1. Numerisch und kategorial**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_01.jpg?v=1787629426" width="250">



>* Numerische Werte erlauben sinnvolle Berechnungen
>* Kategorien beschreiben Bedeutungen, nicht Rechengrößen

>* Nominale und ordinale Kategorien unterscheiden
>* Codierungen und zulässige Werte dokumentieren

>* Wertebereiche, Einheiten und Kategorien klar dokumentieren
>* Speicherformen und Messfehler als Risiken erkennen



```
#@title Python-Code - Numerisch und kategorial

# Dieses Beispiel dokumentiert Datentypen in einer kleinen Tabelle.
# Numerische und kategoriale Merkmale werden bewusst unterschieden.
# Am Ende entsteht ein kompaktes Datenwörterbuch.

import pandas as pd

# Wir bauen einen kleinen Datensatz direkt im Arbeitsspeicher.
data = pd.DataFrame(
    {
        "age_years": [22, 35, 41, 29],
        "city": ["Berlin", "Köln", "Berlin", "Hamburg"],
        "satisfaction_level": ["mittel", "hoch", "niedrig", "hoch"],
        "monthly_spending_eur": [120.50, 89.90, 210.00, 55.25],
    }
)

# Diese Liste beschreibt die fachliche Bedeutung jeder Spalte.
documentation = pd.DataFrame(
    {
        "column": ["age_years", "city", "satisfaction_level", "monthly_spending_eur"],
        "kind": ["numerisch", "kategorial nominal", "kategorial ordinal", "numerisch"],
        "unit_or_values": ["Jahre", "Berlin/Köln/Hamburg", "niedrig < mittel < hoch", "Euro"],
        "shape_role": ["X-Spalte", "X-Spalte", "X-Spalte", "y-Zielwert"],
    }
)

# X enthält Eingabemerkmale, y enthält den Zielwert.
X = data[["age_years", "city", "satisfaction_level"]]
y = data["monthly_spending_eur"]

# Wir prüfen die wichtigsten Formen für die Dokumentation.
if X.shape != (4, 3):
    raise ValueError("X sollte 4 Zeilen und 3 Spalten haben.")

# Auch der Zielvektor muss zur Zeilenzahl passen.
if len(y) != len(X):
    raise ValueError("y muss genauso viele Werte wie X haben.")

# Die Ausgabe bleibt kurz und zeigt nur die Dokumentation.
print("Datenwörterbuch für numerische und kategoriale Merkmale:")
print(documentation.to_string(index=False))
print("Formen: X = " + str(X.shape) + ", y = " + str(y.shape))

```

### **3.2. Zielwerte richtig einordnen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_02.jpg?v=1787629430" width="250">



>* Zielwerte nach Typ klar einordnen
>* Einordnung bestimmt Analyse und Interpretation

>* Darstellung und Bedeutung getrennt prüfen
>* Codes, Reihenfolgen und fehlende Werte dokumentieren

>* Wertebereich, Einheit und Skala festhalten
>* Entstehungszeitpunkt verhindert falsche Eingaben



```
#@title Python-Code - Zielwerte richtig einordnen

# Dieses Beispiel ordnet Zielwerte bewusst fachlich ein.
# Technischer Datentyp und Bedeutung werden getrennt dokumentiert.
# Am Ende entsteht eine kompakte Zielwertdokumentation.

import pandas as pd

# Ein kleiner Datensatz zeigt drei typische Zielwertarten.
data = pd.DataFrame(
    {
        "case_id": [1, 2, 3, 4, 5],
        "price_eur": [240000, 315000, 198000, 410000, 275000],
        "risk_code": [1, 2, 3, 2, 1],
        "churned": [0, 1, 0, 0, 1],
    }
)

# Diese Metadaten beschreiben die fachliche Bedeutung der Zielspalten.
target_metadata = pd.DataFrame(
    {
        "target": ["price_eur", "risk_code", "churned"],
        "technical_dtype": ["int64", "int64", "int64"],
        "meaning": ["numerisch", "ordinal codiert", "binär kategorial"],
        "unit_or_values": ["Euro", "1=niedrig, 2=mittel, 3=hoch", "0=nein, 1=ja"],
    }
)

# Die Form zeigt, wie viele Beobachtungen und Zielspalten vorliegen.
target_columns = ["price_eur", "risk_code", "churned"]
y = data[target_columns]

# Wertebereiche helfen, unplausible oder unbekannte Zielwerte zu erkennen.
ranges = y.agg(["min", "max"])

print("Zielmatrix y: 5 Beobachtungen, 3 dokumentierte Zielspalten")
print(target_metadata.to_string(index=False))
print("Wertebereiche der Zielspalten:")
print(ranges.to_string())
print("Merksatz: Zahlen können Codes sein, nicht automatisch Messwerte.")

```

### **3.3. Datenwörterbuch erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_03.jpg?v=1787629428" width="250">



>* Datenwörterbücher erklären Spalten, Werte und Nutzung
>* Sie verhindern Missverständnisse und sichern Reproduzierbarkeit

>* Merkmale mit Typ, Einheit und Bereich beschreiben
>* Zielwerte eindeutig definieren und Missverständnisse vermeiden

>* Spaltenrollen und Array-Formen klar dokumentieren
>* Umformungsfehler und Zielwert-Lecks vermeiden



```
#@title Python-Code - Datenwörterbuch erstellen

# Dieses Beispiel erstellt ein kleines Datenwörterbuch.
# Es dokumentiert Typen, Einheiten und Array-Formen.
# Die Ausgabe zeigt prüfbare Metadaten kompakt.

import pandas as pd

# Jede Zeile beschreibt eine Wohnung als Beobachtung.
data = pd.DataFrame(
    {
        "area_m2": [45, 72, 88, 120],
        "rooms": [2, 3, 4, 5],
        "district": ["Zentrum", "Nord", "Nord", "Süd"],
        "price_eur": [210000, 330000, 390000, 520000],
    }
)

# Diese Spalten bilden die Eingabematrix X.
feature_columns = ["area_m2", "rooms", "district"]
target_column = "price_eur"

# X enthält Merkmale, y enthält den Zielwert.
X = data[feature_columns]
y = data[target_column]

# Ein Datenwörterbuch beschreibt jede Spalte fachlich.
data_dictionary = pd.DataFrame(
    [
        ["area_m2", "Merkmal", "int64", "45 bis 120", "m²", "Wohnfläche"],
        ["rooms", "Merkmal", "int64", "2 bis 5", "Anzahl", "Zimmerzahl"],
        ["district", "Merkmal", "object", "Nord, Süd, Zentrum", "keine", "Lagekategorie"],
        ["price_eur", "Zielwert", "int64", "210000 bis 520000", "Euro", "Verkaufspreis"],
    ],
    columns=["Spalte", "Rolle", "Datentyp", "Wertebereich", "Einheit", "Bedeutung"],
)

# Eine einfache Prüfung schützt vor falschen Formen.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Die Ausgabe bleibt kurz und zeigt die wichtigsten Metadaten.
print("Datenwörterbuch:")
print(data_dictionary[["Spalte", "Rolle", "Datentyp", "Einheit"]].to_string(index=False))
print(f"Form von X: {X.shape[0]} Beobachtungen, {X.shape[1]} Merkmale")
print(f"Form von y: {y.shape[0]} Zielwerte")

```
