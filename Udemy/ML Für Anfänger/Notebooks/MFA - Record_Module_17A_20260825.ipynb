### **1.1. Version prüfen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_01_01.jpg?v=1787660689" width="250">



>* PyTorch-Version vor Tensorarbeit prüfen
>* Versionswissen hilft bei Fehlersuche

>* PyTorch-Versionen für reproduzierbare Ergebnisse vergleichen
>* Kleine Änderungen beeinflussen Tensorarbeit und Geräteverwaltung

>* Arbeitsumgebung vor Tensorübungen bewusst prüfen
>* CPU als verlässliche Grundlage dokumentieren



```
#@title Python-Code - Version prüfen

# Dieses Beispiel prüft die installierte PyTorch-Version.
# Es zeigt Tensoren, Datentypen und CPU-Geräte.
# Die Ausgabe dokumentiert die aktuelle Lernumgebung.

import torch

# Die Versionsnummer hilft beim Vergleichen von Lernumgebungen.
torch_version = torch.__version__

# Ein kleiner Tensor reicht für die erste Umgebungskontrolle.
sample_tensor = torch.tensor([[1, 2, 3], [4, 5, 6]])

# Diese Prüfung bestätigt die erwartete Tensorform.
if sample_tensor.shape != torch.Size([2, 3]):
    raise ValueError("Die Tensorform ist unerwartet.")

# Die Ausgabe bleibt kurz und konzentriert.
print(f"PyTorch-Version: {torch_version}")
print(f"Tensorform: {tuple(sample_tensor.shape)}")
print(f"Datentyp: {sample_tensor.dtype}")
print(f"Gerät: {sample_tensor.device}")
print("Versionsprüfung abgeschlossen.")

```

### **1.2. Formen und Datentypen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_01_02.jpg?v=1787660691" width="250">



>* Tensoren beschreiben Daten mit mehreren Dimensionen
>* Formen zeigen passende Größen und Fehlerquellen

>* Datentypen bestimmen Werte, Speicher und Operationen
>* Form und Datentyp immer bewusst prüfen

>* Form und Datentyp gemeinsam prüfen
>* CPU erleichtert Verstehen und Fehlervermeidung



```
#@title Python-Code - Formen und Datentypen

# Dieses Beispiel untersucht Tensorformen und Datentypen.
# Es zeigt typische Tensoren auf der CPU.
# Die Ausgabe vergleicht Form, Typ und Gerät.

import torch

# Kleine Beispieldaten bleiben übersichtlich und nachvollziehbar.
scalar = torch.tensor(7.5)
vector = torch.tensor([1, 2, 3, 4])

# Eine Tabelle wird als zweidimensionaler Tensor gespeichert.
table = torch.tensor([[1.0, 2.5, 3.0], [4.0, 5.5, 6.0]])

# Labels sind oft Ganzzahlen für Klassen oder Indizes.
labels = torch.tensor([0, 1, 1, 0], dtype=torch.long)

# Masken verwenden boolesche Werte für gültige Positionen.
mask = torch.tensor([True, False, True, True])

# Diese Prüfung macht eine erwartete Tabellenform explizit.
expected_shape = (2, 3)
shape_is_correct = tuple(table.shape) == expected_shape

print("Tensor | Form | Datentyp | Gerät")
print(f"Skalar | {tuple(scalar.shape)} | {scalar.dtype} | {scalar.device}")
print(f"Vektor | {tuple(vector.shape)} | {vector.dtype} | {vector.device}")
print(f"Tabelle | {tuple(table.shape)} | {table.dtype} | {table.device}")
print(f"Labels | {tuple(labels.shape)} | {labels.dtype} | {labels.device}")
print(f"Maske | {tuple(mask.shape)} | {mask.dtype} | {mask.device}")
print(f"Formprüfung Tabelle == {expected_shape}: {shape_is_correct}")
print("Merke: Form beschreibt die Anordnung, dtype die gespeicherten Werte.")

```

### **1.3. NumPy Konvertierung**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_01_03.jpg?v=1787660693" width="250">



>* NumPy bereitet Daten für PyTorch-Modelle vor
>* Tensoren lassen sich für Analysen zurückwandeln

>* Geteilter Speicher spart Kopien, verändert aber beide
>* Bewusst zwischen Teilen und Kopieren entscheiden

>* NumPy-Konvertierung erzeugt zunächst CPU-Tensoren
>* Form und Datentyp vor Operationen prüfen



```
#@title Python-Code - NumPy Konvertierung

# Dieses Beispiel zeigt NumPy und PyTorch gemeinsam.
# Wir untersuchen Form, Datentyp und Gerät.
# Änderungen zeigen gemeinsam genutzten Speicher.

import numpy as np
import torch

# Ein kleines NumPy-Array bleibt übersichtlich.
numpy_array = np.array([[1.0, 2.0], [3.0, 4.0]], dtype=np.float32)

# from_numpy erstellt einen Tensor auf der CPU.
tensor_from_numpy = torch.from_numpy(numpy_array)

# Diese Prüfung macht die erwartete Form explizit.
if tensor_from_numpy.shape != (2, 2):
    raise ValueError("Die Form sollte 2 mal 2 sein.")

# Wir zeigen die wichtigsten Eigenschaften beider Objekte.
print(f"NumPy: shape={numpy_array.shape}, dtype={numpy_array.dtype}")
print(f"Tensor: shape={tuple(tensor_from_numpy.shape)}, dtype={tensor_from_numpy.dtype}")
print(f"Tensor-Gerät: {tensor_from_numpy.device}")

# Eine Änderung im Array erscheint auch im Tensor.
numpy_array[0, 0] = 99.0
print(f"Nach NumPy-Änderung: Tensor[0,0]={tensor_from_numpy[0, 0].item():.1f}")

# clone trennt den Speicher bewusst vom NumPy-Array.
independent_tensor = torch.from_numpy(numpy_array).clone()

# Diese spätere Änderung betrifft nur den geteilten Tensor.
numpy_array[0, 1] = -5.0
print(f"Geteilter Tensor[0,1]={tensor_from_numpy[0, 1].item():.1f}")
print(f"Kopierter Tensor[0,1]={independent_tensor[0, 1].item():.1f}")

# Zurück nach NumPy ist für CPU-Tensoren direkt möglich.
back_to_numpy = tensor_from_numpy.numpy()
print(f"Zurück nach NumPy: shape={back_to_numpy.shape}, dtype={back_to_numpy.dtype}")

```

## **2. Autograd Grundlagen**

### **2.1. Indexieren und Broadcasting**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_02_01.jpg?v=1787660697" width="250">



>* Indexieren wählt gezielt Tensorbereiche aus
>* Ausgewählte Werte beeinflussen spätere Gradienten

>* Broadcasting nutzt kompatible Tensorformen automatisch.
>* Autograd summiert Gradienten gebroadcasteter Werte.

>* Nur genutzte Tensorwerte erhalten Gradienten
>* Broadcasting sammelt Beiträge über viele Positionen



```
#@title Python-Code - Indexieren und Broadcasting

# Dieses Beispiel verbindet Indexieren mit Autograd.
# Broadcasting zeigt geteilte Parameter in Berechnungen.
# Die Gradienten zeigen verwendete Tensoranteile.

import torch

# Wir erzwingen CPU und gut lesbare Zahlen.
torch.set_printoptions(precision=2, sci_mode=False)

# Vier Messwerte bilden einen kleinen lernbaren Tensor.
values = torch.tensor([1.0, 2.0, 3.0, 4.0], requires_grad=True)

# Ein einzelner Bias wird später auf mehrere Werte gebroadcastet.
bias = torch.tensor(0.5, requires_grad=True)

# Nur drei Positionen gehen in die Berechnung ein.
selected = values[[0, 2, 3]]

# Der skalare Bias wird automatisch auf alle drei Werte angewendet.
shifted = selected + bias

# Ein einfacher Verlust fasst die ausgewählten Ergebnisse zusammen.
loss = (shifted ** 2).mean()

# Backward berechnet Gradienten für alle beteiligten Ursprungstensoren.
loss.backward()

print("Ausgewählte Werte:", selected.detach().tolist())
print("Ergebnis nach Broadcasting:", shifted.detach().round(decimals=2).tolist())
print("Loss:", round(loss.item(), 2))
print("Gradienten von values:", values.grad.round(decimals=2).tolist())
print("Gradient von bias:", round(bias.grad.item(), 2))
print("Nicht ausgewählter Wert hat Gradient:", round(values.grad[1].item(), 2))

```

### **2.2. Gradienten aktivieren**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_02_02.jpg?v=1787660699" width="250">



>* requires_grad aktiviert PyTorchs Rechengraphen
>* Meist nur Modellparameter verfolgen

>* backward berechnet Einflüsse auf verfolgte Tensoren
>* Gradienten zeigen sinnvolle Parameteranpassungen

>* Gradienten nur bei Bedarf aktivieren
>* Gesammelte Gradienten bewusst zurücksetzen



```
#@title Python-Code - Gradienten aktivieren

# Dieses Beispiel aktiviert Gradienten für einen Tensor.
# PyTorch merkt sich dadurch einfache Rechenschritte.
# Nach backward erscheint die berechnete Ableitung.

import torch

# Ein einzelner Wert wird als lernbarer Tensor markiert.
x = torch.tensor(3.0, requires_grad=True)

# Diese Rechnung erzeugt einen kleinen Rechengraphen.
y = x * x + 2 * x + 1

# backward berechnet die Ableitung von y nach x.
y.backward()

# Die erwartete Ableitung lautet 2*x plus 2.
expected_gradient = 2 * x.detach() + 2

# Eine kurze Prüfung macht die Annahme sichtbar.
if x.grad is None:
    raise RuntimeError("Kein Gradient wurde berechnet.")

print(f"Tensorwert x: {x.item():.1f}")
print(f"Berechneter Wert y: {y.item():.1f}")
print(f"Gradient in x.grad: {x.grad.item():.1f}")
print(f"Erwarteter Gradient: {expected_gradient.item():.1f}")
print("requires_grad hat die automatische Ableitung aktiviert.")

```

### **2.3. Gradienten prüfen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_02_03.jpg?v=1787660695" width="250">



>* PyTorch speichert Gradienten im Rechengraphen
>* Gradienten zeigen Empfindlichkeit gegenüber Eingabeänderungen

>* Gradienten erscheinen meist nur bei Blatt-Tensoren
>* Rückwärtsdurchläufe addieren Gradienten ohne Zurücksetzen

>* Gradienten zeigen Lernprobleme und Instabilität
>* Regelmäßige Prüfung macht Autograd verständlicher



```
#@title Python-Code - Gradienten prüfen

# Dieses Beispiel prüft gespeicherte Gradienten.
# Autograd berechnet Ableitungen nach backward.
# Die Ausgabe zeigt Werte und Akkumulation.

import torch

# Ein Blatt-Tensor speichert Gradienten nach backward.
x = torch.tensor(3.0, requires_grad=True)

# Diese Rechnung hat die bekannte Ableitung 2x plus 2.
y = x ** 2 + 2 * x + 1

# Vor backward ist noch kein Gradient gespeichert.
print(f"Vor backward: x.grad = {x.grad}")

# backward berechnet den Gradienten von y nach x.
y.backward()

# Bei x gleich 3 erwarten wir 2 mal 3 plus 2.
expected_gradient = 2 * x.detach() + 2

# detach trennt Werte für eine sichere Anzeige.
print(f"Tensorwert x: {x.item():.1f}")
print(f"Funktionswert y: {y.item():.1f}")
print(f"Gespeicherter Gradient: {x.grad.item():.1f}")
print(f"Erwarteter Gradient: {expected_gradient.item():.1f}")

# Ein zweiter Rückwärtslauf addiert den neuen Gradienten.
y_again = x ** 2 + 2 * x + 1

# Ohne Zurücksetzen wird der Gradient akkumuliert.
y_again.backward()
print(f"Nach zweitem backward: x.grad = {x.grad.item():.1f}")

# Für einen frischen Schritt setzt man Gradienten zurück.
x.grad.zero_()
print(f"Nach zero_: x.grad = {x.grad.item():.1f}")

```

## **3. Dataset und DataLoader**

### **3.1. TensorDataset erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_03_01.jpg?v=1787660687" width="250">



>* TensorDataset bündelt Eingaben und Zielwerte.
>* Jedes Beispiel bleibt korrekt zugeordnet.

>* Erste Dimension muss gleich lang sein
>* Weitere Dimensionen dürfen unterschiedlich sein

>* Stabile Schnittstelle für vorbereitete Tensor-Daten
>* Einfacher Einstieg, bei komplexen Daten begrenzt



```
#@title Python-Code - TensorDataset erstellen

# Dieses Beispiel erstellt ein TensorDataset.
# Eingaben und Labels bleiben korrekt gepaart.
# Die Ausgabe zeigt einzelne Datensatz-Elemente.

import torch
from torch.utils.data import TensorDataset

# Kleine Beispieldaten machen die Zuordnung gut sichtbar.
features = torch.tensor(
    [[1.2, 0.7], [2.1, 1.9], [0.3, 3.4], [1.8, 2.2]],
    dtype=torch.float32,
)

# Jedes Label gehört zur Zeile mit gleichem Index.
labels = torch.tensor([0, 1, 0, 1], dtype=torch.long)

# Die erste Dimension muss bei beiden Tensoren übereinstimmen.
if features.shape[0] != labels.shape[0]:
    raise ValueError("Eingaben und Labels brauchen gleich viele Beispiele.")

# TensorDataset verbindet die Tensoren ohne Kopieren der Bedeutung.
dataset = TensorDataset(features, labels)

# Ein einzelnes Element enthält Eingabe und passendes Label.
first_features, first_label = dataset[0]

print(f"Anzahl Beispiele im Dataset: {len(dataset)}")
print(f"Form der Eingaben: {tuple(features.shape)}")
print(f"Form der Labels: {tuple(labels.shape)}")
print(f"Erstes Beispiel: {first_features.tolist()} -> Label {first_label.item()}")
print(f"Drittes Beispiel: {dataset[2][0].tolist()} -> Label {dataset[2][1].item()}")

```

### **3.2. Daten effizient laden**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_03_02.jpg?v=1787660686" width="250">



>* DataLoader liefert Daten in handlichen Batches
>* Er organisiert Zugriff und schont Speicher

>* Trainingsdaten mischen für vielfältigere Batches
>* Validierung und Tests reproduzierbar laden

>* Datenfluss nutzt Rechenressourcen besser
>* DataLoader hält Trainingsschleifen übersichtlich



```
#@title Python-Code - Daten effizient laden

# Dieses Beispiel zeigt effizientes Laden mit DataLoader.
# Batches ordnen Daten für Training und Bewertung.
# Die Ausgabe vergleicht Train, Validierung und Test.

import torch
from torch.utils.data import DataLoader
from torch.utils.data import TensorDataset

# Kleine Beispieldaten bleiben vollständig im Arbeitsspeicher.
features = torch.arange(40, dtype=torch.float32).reshape(20, 2)
targets = (features[:, 0] > 18).long()

# Feste Aufteilungen simulieren Train, Validierung und Test.
train_dataset = TensorDataset(features[:12], targets[:12])
valid_dataset = TensorDataset(features[12:16], targets[12:16])
test_dataset = TensorDataset(features[16:], targets[16:])

# Diese Prüfung macht die erwarteten Größen sichtbar.
expected_sizes = (12, 4, 4)
actual_sizes = (len(train_dataset), len(valid_dataset), len(test_dataset))

# Bei falschen Größen wäre die Aufteilung missverständlich.
if actual_sizes != expected_sizes:
    raise ValueError("Die Dataset-Aufteilung hat unerwartete Größen.")

# Ein Generator macht das Mischen reproduzierbar.
generator = torch.Generator()
generator.manual_seed(42)

# Trainingsdaten werden gemischt, Bewertungsdaten bleiben geordnet.
train_loader = DataLoader(
    train_dataset, batch_size=4, shuffle=True, generator=generator
)

# Validierungsdaten werden ohne Mischen geladen.
valid_loader = DataLoader(valid_dataset, batch_size=4, shuffle=False)

# Testdaten werden ebenfalls ohne Mischen geladen.
test_loader = DataLoader(test_dataset, batch_size=4, shuffle=False)

# Wir betrachten jeweils nur den ersten Batch.
train_batch_features, train_batch_targets = next(iter(train_loader))
valid_batch_features, valid_batch_targets = next(iter(valid_loader))

test_batch_features, test_batch_targets = next(iter(test_loader))

# Die Ausgabe zeigt Batchgrößen und Reihenfolge.
print("Train-Dataset: 12 Beispiele, Batchgröße 4, shuffle=True")
print("Valid-Dataset: 4 Beispiele, Batchgröße 4, shuffle=False")
print("Test-Dataset: 4 Beispiele, Batchgröße 4, shuffle=False")
print(f"Train-Batch-Form: {tuple(train_batch_features.shape)}")
print(f"Valid-Batch-Form: {tuple(valid_batch_features.shape)}")
print(f"Test-Batch-Form: {tuple(test_batch_features.shape)}")
print(f"Train-Ziele im ersten Batch: {train_batch_targets.tolist()}")
print(f"Valid-Ziele im ersten Batch: {valid_batch_targets.tolist()}")
print(f"Test-Ziele im ersten Batch: {test_batch_targets.tolist()}")

```

### **3.3. Datenaufteilung mit Loadern**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_17/Lecture_A/image_03_03.jpg?v=1787660684" width="250">



>* Training lernt, Validierung steuert Entwicklung
>* Testdaten prüfen ehrlich neue Beispiele

>* Eigene Loader für jede Datenphase
>* Training mischen, Auswertung reproduzierbar halten

>* Datenleckage durch fachlich sinnvolle Aufteilung vermeiden
>* DataLoader bündeln getrennte Daten wiederholbar in Batches



```
#@title Python-Code - Datenaufteilung mit Loadern

# Wir teilen Tensoren in drei Datenbereiche.
# DataLoader liefern passende Batches pro Bereich.
# Die Ausgabe zeigt Größen und Batchformen.

import torch
from torch.utils.data import DataLoader
from torch.utils.data import TensorDataset
from torch.utils.data import random_split

# Ein Generator macht die zufällige Aufteilung reproduzierbar.
generator = torch.Generator().manual_seed(42)

# Kleine Beispieldaten simulieren Merkmale und Klassenlabels.
features = torch.arange(60, dtype=torch.float32).reshape(20, 3)
labels = (features[:, 0] > 27).long()

# TensorDataset verbindet jedes Merkmal mit seinem Label.
full_dataset = TensorDataset(features, labels)

# Die drei Längen müssen zusammen alle Beispiele abdecken.
split_lengths = [12, 4, 4]

if sum(split_lengths) != len(full_dataset):
    raise ValueError("Die Aufteilung passt nicht zur Datensatzgröße.")

# random_split erzeugt getrennte Dataset-Objekte.
train_dataset, val_dataset, test_dataset = random_split(
    full_dataset, split_lengths, generator=generator
)

# Training wird gemischt, Auswertung bleibt geordnet.
train_loader = DataLoader(train_dataset, batch_size=5, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=4, shuffle=False)

# Ein erster Batch zeigt die Form der geladenen Daten.
train_batch_features, train_batch_labels = next(iter(train_loader))

print(f"Gesamtdatensatz: {len(full_dataset)} Beispiele")
print(f"Training: {len(train_dataset)} Beispiele, {len(train_loader)} Batches")
print(f"Validierung: {len(val_dataset)} Beispiele, {len(val_loader)} Batch")
print(f"Test: {len(test_dataset)} Beispiele, {len(test_loader)} Batch")
print(f"Erster Trainingsbatch: X{tuple(train_batch_features.shape)}, y{tuple(train_batch_labels.shape)}")
print(f"Labels im ersten Trainingsbatch: {train_batch_labels.tolist()}")

```
