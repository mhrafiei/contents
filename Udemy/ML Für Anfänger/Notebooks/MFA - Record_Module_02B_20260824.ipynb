## **1. Überwachtes Lernen**

### **1.1. Bekannte Zielwerte**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_01_01.jpg?v=1787628644" width="250">



>* Trainingsdaten enthalten Merkmale und bekannte Ergebnisse
>* Modelle lernen Muster für neue Fälle

>* Trainingsdaten enthalten richtige Antworten als Orientierung
>* Bekannte Zielwerte kennzeichnen überwachtes Lernen

>* Bekannte Ergebnisse sprechen für überwachtes Lernen
>* Zielwerte können Kategorien oder Zahlen sein



```
#@title Python-Code - Bekannte Zielwerte

# Dieses Beispiel zeigt bekannte Zielwerte im überwachten Lernen.
# Wir trennen Merkmale klar vom bekannten Ergebnis.
# Am Ende erkennt man Klassifikation und Regression.

import pandas as pd

# Kleine Alltagsdaten zeigen Merkmale und Zielwerte nebeneinander.
housing_data = pd.DataFrame(
    {
        "area_m2": [45, 60, 80, 100],
        "rooms": [2, 2, 3, 4],
        "rent_eur": [720, 900, 1250, 1600],
    }
)

# Der Zielwert ist hier eine Zahl: der Mietpreis.
housing_features = housing_data[["area_m2", "rooms"]]
housing_target = housing_data["rent_eur"]

# Ein zweites Beispiel nutzt Kategorien als Zielwerte.
email_data = pd.DataFrame(
    {
        "links": [0, 3, 1, 5],
        "urgent_words": [0, 4, 1, 6],
        "label": ["kein Spam", "Spam", "kein Spam", "Spam"],
    }
)

# Der Zielwert ist hier eine Kategorie: Spam oder nicht.
email_features = email_data[["links", "urgent_words"]]
email_target = email_data["label"]

# Eine einfache Prüfung macht die Lernart sichtbar.
if len(housing_features) != len(housing_target):
    raise ValueError("Merkmale und Zielwerte müssen gleich viele Zeilen haben.")

if len(email_features) != len(email_target):
    raise ValueError("Merkmale und Zielwerte müssen gleich viele Zeilen haben.")

print("Überwachtes Lernen: Zu jedem Beispiel gibt es einen bekannten Zielwert.")
print(f"Wohnungen: {len(housing_data)} Beispiele, Zielwert = rent_eur.")
print("rent_eur ist numerisch, daher passt Regression.")
print(f"E-Mails: {len(email_data)} Beispiele, Zielwert = label.")
print("label ist eine Kategorie, daher passt Klassifikation.")
print("Ohne diese Zielwerte wäre es kein typisches überwachtes Lernen.")

```

### **1.2. Kategorien vorhersagen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_01_02.jpg?v=1787628648" width="250">



>* Modell lernt aus gelabelten Beispielen
>* Neue Fälle werden Kategorien zugeordnet

>* Klassifikation ordnet Daten festen Klassen zu
>* Gesucht ist eine Kategorie, kein Zahlenwert

>* Trainingsdaten brauchen bekannte richtige Kategorien
>* Gelabelte Beispiele müssen vielfältig und zuverlässig sein



```
#@title Python-Code - Kategorien vorhersagen

# Dieses Beispiel zeigt überwachte Klassifikation.
# Bekannte Kategorien steuern das Training.
# Neue Blumen erhalten vorhergesagte Kategorien.

import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Wir laden kleine, gelabelte Beispieldaten.
iris = load_iris()
features = iris.data[:, [2, 3]]
labels = iris.target

# Diese Prüfung macht die Datenannahme sichtbar.
if features.shape[0] != labels.shape[0]:
    raise ValueError("Jede Blume braucht genau eine bekannte Kategorie.")

# Training und Test bleiben sauber getrennt.
X_train, X_test, y_train, y_test = train_test_split(
    features, labels, test_size=0.3, stratify=labels, random_state=42
)

# Das Modell lernt aus Merkmalen und bekannten Kategorien.
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)

# Danach ordnet es neue Beispiele Kategorien zu.
predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print(f"scikit-learn-Version: {sklearn.__version__}")
print("Lernart: überwachtes Lernen, weil Zielkategorien bekannt sind.")
print(f"Testgenauigkeit: {accuracy:.2f}")

# Drei Vorhersagen zeigen die Idee der Klassifikation.
for index in range(3):
    true_name = iris.target_names[y_test[index]]
    predicted_name = iris.target_names[predictions[index]]
    print(f"Beispiel {index + 1}: echt {true_name}, vorhergesagt {predicted_name}")

# Die Grafik zeigt Klassen und neue Zuordnungen.
fig, ax = plt.subplots(figsize=(7, 5))
scatter = ax.scatter(
    X_train[:, 0], X_train[:, 1], c=y_train, cmap="viridis", label="Training"
)

ax.scatter(
    X_test[:12, 0], X_test[:12, 1], c=predictions[:12], cmap="viridis",
    marker="x", s=90, label="Vorhersage"
)

ax.set_title("Kategorien aus gelabelten Beispielen vorhersagen")
ax.set_xlabel("Blütenblattlänge in cm")
ax.set_ylabel("Blütenblattbreite in cm")
ax.legend(loc="lower right")
plt.show()

```

### **1.3. Werte vorhersagen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_01_03.jpg?v=1787628646" width="250">



>* Überwachtes Lernen kann Zahlenwerte vorhersagen
>* Bekannte Beispiele zeigen Muster und Zielwerte

>* Wertvorhersagen unterstützen mengenabhängige Entscheidungen
>* Modelle lernen aus bekannten früheren Ergebnissen

>* Zahlenvorhersagen von Kategorien klar unterscheiden
>* Datenqualität und passende Anwendung sorgfältig prüfen



```
#@title Python-Code - Werte vorhersagen

# Dieses Beispiel zeigt überwachte Wertvorhersage.
# Regression lernt aus bekannten Zahlenwerten.
# Am Ende schätzen wir neue Mietpreise.

import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.linear_model import LinearRegression

# Kleine Trainingsdaten enthalten Wohnfläche und bekannte Miete.
area_sqm = np.array([35, 45, 55, 65, 75, 85, 95, 105]).reshape(-1, 1)
rent_eur = np.array([520, 610, 700, 790, 880, 970, 1060, 1150])

# Diese Prüfung verhindert unpassende Eingabeformen.
if area_sqm.shape[0] != rent_eur.shape[0]:
    raise ValueError("Jede Wohnung braucht genau einen bekannten Mietwert.")

# Ein Regressionsmodell lernt die Beziehung aus Beispielen.
model = LinearRegression()
model.fit(area_sqm, rent_eur)

# Neue Wohnungen haben Merkmale, aber noch keinen Zielwert.
new_area_sqm = np.array([50, 80, 110]).reshape(-1, 1)
predicted_rent_eur = model.predict(new_area_sqm)

print(f"scikit-learn-Version: {sklearn.__version__}")
print("Lernart: überwachtes Lernen, weil Zielwerte bekannt sind.")
print("Aufgabentyp: Regression, weil eine Zahl vorhergesagt wird.")

for area, rent in zip(new_area_sqm.ravel(), predicted_rent_eur):
    print(f"{int(area)} m² -> geschätzte Miete: {round(rent)} Euro")

# Die Grafik zeigt bekannte Werte und die gelernte Gerade.
plot_area_sqm = np.linspace(30, 115, 100).reshape(-1, 1)
plot_rent_eur = model.predict(plot_area_sqm)

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(area_sqm.ravel(), rent_eur, label="bekannte Trainingswerte")
ax.plot(plot_area_sqm.ravel(), plot_rent_eur, label="gelernte Vorhersage")
ax.scatter(new_area_sqm.ravel(), predicted_rent_eur, label="neue Schätzungen")

ax.set_title("Regression: Mietpreis als Zahlenwert vorhersagen")
ax.set_xlabel("Wohnfläche in Quadratmetern")
ax.set_ylabel("Miete in Euro")
ax.legend()
plt.show()

```

## **2. Lernen ohne Zielwerte**

### **2.1. Gruppen in Daten**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_02_01.jpg?v=1787628650" width="250">



>* Clustering entdeckt Gruppen ohne vorgegebene Antworten
>* Sinnvolle Gruppen hängen von Merkmalen ab

>* Clustering entdeckt verborgene Muster in Daten
>* Gefundene Gruppen brauchen fachliche Deutung

>* Clustering hat keine bekannten Zielwerte
>* Gefundene Gruppen müssen fachlich sinnvoll sein



```
#@title Python-Code - Gruppen in Daten

# Dieses Beispiel zeigt Gruppen ohne Zielwerte.
# Clustering findet ähnliche Punkte automatisch zusammen.
# Die Grafik zeigt entdeckte Daten-Gruppen.

import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Wir erzeugen einfache zweidimensionale Beispieldaten.
features, hidden_groups = make_blobs(
    n_samples=90,
    centers=3,
    cluster_std=0.75,
    random_state=42,
)

# Diese Prüfung macht die erwartete Datenform sichtbar.
if features.shape != (90, 2):
    raise ValueError("Die Beispieldaten sollten 90 Zeilen und 2 Spalten haben.")

# K-Means sucht drei Gruppen ohne bekannte Zielwerte.
model = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = model.fit_predict(features)

# Wir zählen, wie viele Punkte in jeder Gruppe liegen.
cluster_counts = np.bincount(cluster_labels)
print(f"scikit-learn-Version: {sklearn.__version__}")
print("Lernart: unüberwachtes Lernen, weil keine Zielwerte genutzt werden.")
print(f"Gefundene Gruppen: {len(cluster_counts)}")
print(f"Punkte pro Gruppe: {cluster_counts.tolist()}")

# Die Farben zeigen die vom Algorithmus gefundenen Gruppen.
fig, ax = plt.subplots(figsize=(7, 5))
scatter = ax.scatter(
    features[:, 0],
    features[:, 1],
    c=cluster_labels,
    cmap="viridis",
)

# Die roten Kreuze markieren die berechneten Gruppenmitten.
ax.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    c="red",
    marker="X",
    s=160,
    label="Gruppenmitten",
)

ax.set_title("Clustering: Gruppen in Daten ohne Zielwerte")
ax.set_xlabel("Merkmal 1")
ax.set_ylabel("Merkmal 2")
ax.legend()
plt.show()

```

### **2.3. Ungewöhnliche Muster erkennen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_02_03.jpg?v=1787628653" width="250">



>* Anomalien weichen vom normalen Muster ab
>* Auffälliges Verhalten braucht genauere Prüfung

>* Seltene Fälle oft ohne bekannte Beispiele
>* Unerwartete Abweichungen in vielen Bereichen erkennen

>* Clustering gruppiert, Anomalieerkennung findet Abweichungen
>* Kontext entscheidet, Frühwarnung statt Endurteil



```
#@title Python-Code - Ungewöhnliche Muster erkennen

# Dieses Beispiel zeigt Anomalieerkennung ohne Zielwerte.
# Normale Punkte bilden ein typisches Muster.
# Auffällige Punkte werden als Hinweise markiert.

import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.ensemble import IsolationForest

# Wir erzeugen normale Messwerte und wenige ungewöhnliche Fälle.
rng = np.random.default_rng(42)
normal_data = rng.normal(loc=0.0, scale=1.0, size=(120, 2))

anomaly_data = np.array([[4.0, 4.2], [4.5, -3.8], [-4.2, 3.7]])
measurements = np.vstack((normal_data, anomaly_data))

# Diese Prüfung schützt vor unerwarteten Datenformen.
if measurements.shape != (123, 2):
    raise ValueError("Die Beispieldaten haben nicht die erwartete Form.")

# IsolationForest lernt, was im Datensatz typisch wirkt.
model = IsolationForest(contamination=0.03, random_state=42)
model.fit(measurements)

predictions = model.predict(measurements)
anomaly_mask = predictions == -1

# Wir zählen, wie viele Punkte als auffällig gelten.
anomaly_count = int(np.sum(anomaly_mask))
normal_count = int(len(measurements) - anomaly_count)

print(f"scikit-learn-Version: {sklearn.__version__}")
print(f"Als normal markiert: {normal_count} Punkte")
print(f"Als auffällig markiert: {anomaly_count} Punkte")
print("Wichtig: Auffällig bedeutet Prüfhinweis, nicht automatisch Fehler.")

# Die Grafik zeigt normale und auffällige Punkte gemeinsam.
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(measurements[~anomaly_mask, 0], measurements[~anomaly_mask, 1], label="typisch")

ax.scatter(
    measurements[anomaly_mask, 0],
    measurements[anomaly_mask, 1],
    label="auffällig",
    marker="x",
    s=90,
)

ax.set_title("Anomalieerkennung ohne Zielwerte")
ax.set_xlabel("Merkmal 1")
ax.set_ylabel("Merkmal 2")
ax.legend()
plt.show()

```

### **3.3. Aufgaben passend zuordnen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_02/Lecture_B/image_03_03.jpg?v=1787628659" width="250">



>* Problemstruktur vor Verfahren genau prüfen
>* Antwortform bestimmt die passende Lernart

>* Bekannte Zielgröße spricht für überwachtes Lernen
>* Ohne Zielgröße Muster oder Abweichungen finden

>* Zweck und Fehlfolgen mitbedenken
>* Ziel, Daten und Ausgabe begründen



```
#@title Python-Code - Aufgaben passend zuordnen

# Dieses Beispiel ordnet Alltagsaufgaben passenden Lernarten zu.
# Die Zielgröße entscheidet oft über die Lernart.
# Am Ende entsteht eine begründete Zuordnung.

import pandas as pd

# Jede Aufgabe beschreibt Daten, Ziel und bekannte Antworten.
tasks = pd.DataFrame(
    {
        "Aufgabe": ["Spam erkennen", "Miete schätzen", "Musikgruppen finden", "Betrug entdecken"],
        "Gewünschte Ausgabe": ["Kategorie", "Zahl", "Gruppen", "Auffälligkeit"],
        "Bekannte Antworten": ["ja", "ja", "nein", "nein"],
    }
)

# Diese einfachen Regeln bilden die fachliche Begründung nach.
learning_types = []
for output, known in zip(tasks["Gewünschte Ausgabe"], tasks["Bekannte Antworten"]):
    if known == "ja" and output == "Kategorie":
        learning_types.append("Klassifikation")
    elif known == "ja" and output == "Zahl":
        learning_types.append("Regression")
    elif known == "nein" and output == "Gruppen":
        learning_types.append("Clustering")
    else:
        learning_types.append("Anomalieerkennung")

# Die neue Spalte zeigt die passende Lernart.
tasks["Passende Lernart"] = learning_types

# Eine kurze Prüfung macht die Zuordnung nachvollziehbar.
if len(tasks) != len(learning_types):
    raise ValueError("Die Anzahl der Zuordnungen passt nicht.")

print("Aufgaben passend zuordnen:")
print(tasks.to_string(index=False))
print("Merksatz: Ausgabeform und bekannte Antworten begründen die Lernart.")

```
