## **1. Zeitreihen erzeugen**

### **1.1. Zeitachse und Abtastrate**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_01_01.jpg?v=1787634185" width="250">



>* Zeitpunkte machen Messwerte zur Zeitreihe
>* Beobachtungsdauer und Messzeitpunkte vorher festlegen

>* Abtastrate bestimmt Messhäufigkeit pro Zeiteinheit
>* Wahl hängt von Signal und Aufwand ab

>* Dichte Zeitpunkte zeigen schnelle Signaländerungen
>* Gleichmäßige Zeitachsen ermöglichen sinnvolle Auswertung



```
#@title Python-Code - Zeitachse und Abtastrate

# Wir erzeugen eine einfache Zeitachse.
# Die Abtastrate bestimmt die Messpunkte.
# Der Plot zeigt dichte und grobe Abtastung.

import numpy as np
import matplotlib.pyplot as plt

# Diese Werte legen Dauer und Abtastraten fest.
duration_seconds = 2.0
low_rate_hz = 8
high_rate_hz = 80

# Die Anzahl der Messpunkte folgt aus Dauer mal Abtastrate.
low_count = int(duration_seconds * low_rate_hz)
high_count = int(duration_seconds * high_rate_hz)

# Gleichmäßige Zeitachsen enthalten den Start, aber nicht das Ende.
low_time = np.arange(low_count) / low_rate_hz
high_time = np.arange(high_count) / high_rate_hz

# Ein Sinussignal macht den Unterschied der Abtastung sichtbar.
frequency_hz = 2.0
low_signal = np.sin(2 * np.pi * frequency_hz * low_time)
high_signal = np.sin(2 * np.pi * frequency_hz * high_time)

# Eine einfache Prüfung schützt vor unpassenden Zeitachsen.
if len(low_time) != len(low_signal):
    raise ValueError("Zeitachse und Signal müssen gleich lang sein.")

# Die Zeitabstände zeigen die Bedeutung der Abtastrate.
low_step = 1 / low_rate_hz
high_step = 1 / high_rate_hz

print(f"Dauer: {duration_seconds:.1f} s")
print(f"Niedrige Abtastrate: {low_rate_hz} Hz, Abstand {low_step:.3f} s")
print(f"Hohe Abtastrate: {high_rate_hz} Hz, Abstand {high_step:.3f} s")
print(f"Messpunkte: {low_count} grob und {high_count} dicht")

# Beide Kurven zeigen dasselbe Signal mit anderer zeitlicher Auflösung.
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(high_time, high_signal, label="80 Hz: dichte Zeitachse")
ax.plot(low_time, low_signal, "o--", label="8 Hz: grobe Zeitachse")

# Beschriftungen machen Einheiten und Aussage des Diagramms klar.
ax.set_title("Zeitachse und Abtastrate bei einem einfachen Signal")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Signalwert")
ax.legend()

plt.show()

```

### **1.2. Sinus mit Rauschen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_01_02.jpg?v=1787634187" width="250">



>* Sinus beschreibt regelmäßige periodische Schwingungen
>* Amplitude, Frequenz und Phase modellieren Messsignale

>* Rauschen macht reale Signale unruhig
>* Verrauschte Sinusse zeigen nötige Verarbeitung

>* Signalstruktur und Rauschen gemeinsam erkennen
>* Darstellung zeigt Ausreißer und Messunsicherheit



```
#@title Python-Code - Sinus mit Rauschen

# Wir erzeugen ein einfaches verrauschtes Sinussignal.
# Die Zeitachse ordnet jedem Messwert einen Zeitpunkt zu.
# Die Grafik zeigt ideale und realistische Messwerte.

import numpy as np
import matplotlib.pyplot as plt

# Diese Werte steuern Dauer und Abtastrate.
sampling_rate_hz = 100
signal_duration_s = 2

# Die Zeitachse enthält gleichmäßig verteilte Messzeitpunkte.
time_s = np.arange(0, signal_duration_s, 1 / sampling_rate_hz)
expected_points = sampling_rate_hz * signal_duration_s

# Diese Prüfung macht die erwartete Signalgröße sichtbar.
if len(time_s) != expected_points:
    raise ValueError("Die Zeitachse hat nicht die erwartete Länge.")

# Amplitude und Frequenz bestimmen die ideale Sinuskurve.
amplitude = 1.0
frequency_hz = 3.0

# Ein fester Zufallsgenerator macht das Rauschen reproduzierbar.
rng = np.random.default_rng(42)
noise_strength = 0.25

# Das ideale Signal wird mit zufälligem Rauschen überlagert.
clean_signal = amplitude * np.sin(2 * np.pi * frequency_hz * time_s)
noise = rng.normal(0, noise_strength, size=len(time_s))
noisy_signal = clean_signal + noise

# Kurze Kennzahlen helfen beim Einordnen der Messreihe.
print(f"Messpunkte: {len(time_s)}")
print(f"Dauer: {signal_duration_s} s, Abtastrate: {sampling_rate_hz} Hz")
print(f"Frequenz: {frequency_hz} Hz, Rauschstärke: {noise_strength}")
print(f"Erster verrauschter Messwert: {noisy_signal[0]:.3f}")

# Eine einzelne Grafik vergleicht Modell und verrauschte Messung.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(time_s, clean_signal, label="Idealer Sinus", linewidth=2)
ax.plot(time_s, noisy_signal, label="Sinus mit Rauschen", alpha=0.75)

# Beschriftungen machen Zeit und Signalwert eindeutig interpretierbar.
ax.set_title("Sinussignal mit zufälligem Rauschen")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Signalwert")
ax.legend()

plt.show()

```

### **1.3. CSV Messreihe**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_01_03.jpg?v=1787634188" width="250">



>* CSV speichert Zeitpunkte und Messwerte.
>* Reale Daten enthalten oft Fehler und Trends.

>* Zeitachse aus Spalte oder Abtastrate bestimmen
>* Messwerte nur im Zeitkontext richtig deuten

>* Zeitdiagramme zeigen Trends, Rauschen und Ausreißer
>* Plausibilität vor weiterer Analyse prüfen



```
#@title Python-Code - CSV Messreihe

# Wir simulieren eine kleine CSV-Messreihe.
# Zeit, Trend und Rauschen werden sichtbar.
# Die Grafik zeigt Rohwerte und Trend.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Eine feste Zufallsquelle macht das Beispiel wiederholbar.
rng = np.random.default_rng(42)

# Die Zeitachse beschreibt Messpunkte im Abstand von 0,1 Sekunden.
sample_count = 80
sample_interval_s = 0.1

# Aus Messnummern entsteht eine echte Zeitachse.
sample_index = np.arange(sample_count)
time_s = sample_index * sample_interval_s

# Das Signal kombiniert Grundwert, Trend, Schwingung und Rauschen.
trend = 0.08 * time_s
wave = 0.35 * np.sin(2 * np.pi * 0.7 * time_s)
noise = rng.normal(loc=0.0, scale=0.08, size=sample_count)

# Diese Tabelle entspricht einer eingelesenen CSV-Messreihe.
measurement = 20.0 + trend + wave + noise
csv_like_data = pd.DataFrame(
    {"time_s": time_s, "temperature_c": measurement}
)

# Eine einfache Prüfung schützt vor falsch aufgebauten Messreihen.
expected_columns = ["time_s", "temperature_c"]
if list(csv_like_data.columns) != expected_columns:
    raise ValueError("Die Messreihe hat nicht die erwarteten Spalten.")

# Ein gleitender Mittelwert macht den langsamen Verlauf besser sichtbar.
window_size = 7
csv_like_data["smooth_c"] = (
    csv_like_data["temperature_c"].rolling(window_size, center=True).mean()
)

print("Erste fünf Zeilen der simulierten CSV-Messreihe:")
print(csv_like_data.head().round(3).to_string(index=False))
print(f"Messpunkte: {len(csv_like_data)}, Messintervall: {sample_interval_s} s")

# Die Grafik zeigt Messwerte und geglätteten Verlauf gemeinsam.
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(csv_like_data["time_s"], csv_like_data["temperature_c"], label="Messwert")
ax.plot(csv_like_data["time_s"], csv_like_data["smooth_c"], label="Gleitender Mittelwert")

ax.set_title("CSV-Messreihe mit Zeitachse, Trend und Rauschen")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Temperatur in Grad Celsius")
ax.legend()

plt.show()

```

## **2. Fenster und Merkmale**

### **2.1. Gleitende Fenster**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_02_01.jpg?v=1787634177" width="250">



>* Signale abschnittsweise mit gleitenden Fenstern analysieren
>* Lokale Veränderungen und Muster besser erkennen

>* Fensterlänge steuert Tempo, Stabilität und Rauschen
>* Schrittweite bestimmt Auflösung und Rechenaufwand

>* Fenster liefern lokale Signalmerkmale.
>* Merkmale machen Messreihen tabellarisch auswertbar.



```
#@title Python-Code - Gleitende Fenster

# Dieses Beispiel zeigt gleitende Fenster.
# Lokale Merkmale beschreiben Signalabschnitte.
# Die Grafik macht Fensterwerte sichtbar.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Wir erzeugen ein kleines künstliches Sensorsignal.
rng = np.random.default_rng(42)
time_seconds = np.arange(0, 60, 1)
trend = 0.03 * time_seconds

# Eine Schwingung und Rauschen bilden Messwerte.
wave = np.sin(2 * np.pi * time_seconds / 15)
noise = rng.normal(0, 0.18, size=time_seconds.size)
signal = 20 + trend + wave + noise

# Diese Prüfung schützt vor unpassenden Fenstergrößen.
window_size = 10
step_size = 5
if window_size > signal.size:
    raise ValueError("Das Fenster ist länger als das Signal.")

# Für jedes Fenster berechnen wir einfache Merkmale.
rows = []
for start in range(0, signal.size - window_size + 1, step_size):
    end = start + window_size
    window = signal[start:end]
    rows.append(
        {
            "start_s": int(time_seconds[start]),
            "mean": float(np.mean(window)),
            "std": float(np.std(window)),
            "range": float(np.max(window) - np.min(window)),
        }
    )

# Die Merkmalstabelle fasst lokale Signalabschnitte zusammen.
features = pd.DataFrame(rows)
print("Erste Fenstermerkmale, gerundet:")
print(features.head(4).round(2).to_string(index=False))

# Die Grafik vergleicht Rohsignal und lokale Mittelwerte.
window_centers = features["start_s"].to_numpy() + window_size / 2
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(time_seconds, signal, label="Rohsignal", color="steelblue")

# Die Punkte zeigen den Mittelwert jedes Fensters.
ax.plot(
    window_centers,
    features["mean"],
    "o-",
    label="Fenster-Mittelwert",
    color="darkorange",
)

# Achsen und Titel machen die Darstellung lesbar.
ax.set_title("Gleitende Fenster: lokale Mittelwerte im Signal")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Messwert in Einheiten")
ax.legend()
plt.show()

```

### **2.2. Lokale Kennzahlen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_02_02.jpg?v=1787634173" width="250">



>* Kurze Fenster zeigen lokale Signalmerkmale.
>* Kennzahlen machen Rohdaten interpretierbar.

>* Kennzahl passend zur Signaleigenschaft wählen
>* Mehrere Kennzahlen zeigen verschiedene Auffälligkeiten

>* Fensterlänge passend zum Signal wählen
>* Kennzahlen als Modellierungsentscheidung verstehen



```
#@title Python-Code - Lokale Kennzahlen

# Dieses Beispiel berechnet lokale Kennzahlen aus Signalfenstern.
# Fenster zeigen Niveau, Streuung und lokale Extremwerte.
# Die Grafik verbindet Rohsignal und Fenstermerkmale.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Wir erzeugen ein kleines künstliches Temperatursignal.
rng = np.random.default_rng(42)
time_minutes = np.arange(0, 120)
trend = 20 + 0.03 * time_minutes
wave = 1.2 * np.sin(2 * np.pi * time_minutes / 40)
noise = rng.normal(0, 0.35, size=time_minutes.size)
signal = trend + wave + noise

# Ein kurzer Ausreißer macht robuste Kennzahlen anschaulich.
signal[72] = signal[72] + 4.0
window_size = 20
step_size = 10

# Diese Prüfung verhindert unpassende Fenstergrößen.
if window_size > signal.size:
    raise ValueError("Die Fensterlänge muss kleiner als das Signal sein.")

# Wir sammeln Kennzahlen für überlappende Fenster.
rows = []
for start in range(0, signal.size - window_size + 1, step_size):
    end = start + window_size
    window = signal[start:end]
    center = time_minutes[start:end].mean()
    rows.append(
        {
            "Mitte_min": center,
            "Mittelwert_C": window.mean(),
            "Median_C": np.median(window),
            "Spannweite_C": window.max() - window.min(),
            "Std_C": window.std(ddof=0),
        }
    )

# Die Tabelle enthält pro Fenster eine kompakte Beschreibung.
features = pd.DataFrame(rows)
preview = features.head(5).round(2)
print("Erste fünf Fenstermerkmale:")
print(preview.to_string(index=False))

# Die Grafik zeigt Rohsignal und lokale Mittelwerte gemeinsam.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(time_minutes, signal, color="lightgray", label="Signal")
ax.plot(
    features["Mitte_min"],
    features["Mittelwert_C"],
    marker="o",
    color="tab:blue",
    label="Lokaler Mittelwert",
)
ax.set_title("Lokale Kennzahlen in gleitenden Fenstern")
ax.set_xlabel("Zeit in Minuten")
ax.set_ylabel("Temperatur in °C")
ax.legend()
plt.show()

```

### **2.3. Änderungen erkennen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_02_03.jpg?v=1787634175" width="250">



>* Signalverlauf zeigt mehr als Einzelwerte
>* Änderungsmerkmale machen wichtige Ereignisse sichtbar

>* Werte vergleichen zeigt Ruhe oder Aktivität
>* Zeitabstand und Messkontext richtig berücksichtigen

>* Fenster machen Änderungserkennung robuster gegen Rauschen
>* Sie markieren Ereignisse und bedeutsame Übergänge



```
#@title Python-Code - Änderungen erkennen

# Dieses Beispiel erkennt Änderungen in einem Signal.
# Differenzen zeigen schnelle Anstiege und Abfälle.
# Markierte Punkte machen Ereignisse im Verlauf sichtbar.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Wir erzeugen ein kleines Signal mit Trend und Ereignis.
rng = np.random.default_rng(42)
time_seconds = np.arange(0, 60, 1)

# Das Signal enthält Rauschen und einen plötzlichen Sprung.
trend = 0.03 * time_seconds
noise = rng.normal(0, 0.15, size=time_seconds.size)
signal = 2.0 + trend + noise

# Ein kurzer Abschnitt steigt absichtlich deutlich schneller.
signal[30:36] = signal[30:36] + np.linspace(0.0, 3.0, 6)
signal[36:] = signal[36:] + 3.0

# Die erste Differenz misst Änderung pro Sekunde.
change_per_second = np.diff(signal, prepend=signal[0])
absolute_change = np.abs(change_per_second)

# Ein Schwellenwert markiert ungewöhnlich starke Änderungen.
threshold = 0.8
change_points = absolute_change > threshold

# Eine kleine Tabelle zeigt die wichtigsten Änderungsmerkmale.
feature_table = pd.DataFrame(
    {
        "Zeit_s": time_seconds,
        "Signal": signal,
        "Aenderung": change_per_second,
        "Stark": change_points,
    }
)

# Wir geben nur die stärksten Änderungen aus.
strong_rows = feature_table.loc[change_points].head(5).copy()
strong_rows["Signal"] = strong_rows["Signal"].round(2)
strong_rows["Aenderung"] = strong_rows["Aenderung"].round(2)

print("Starke Änderungen werden über Differenzen erkannt.")
print("Schwellenwert:", threshold, "Einheiten pro Sekunde")
print(strong_rows.to_string(index=False))

# Die Grafik zeigt Signal und erkannte Änderungspunkte.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(time_seconds, signal, label="Signal", color="steelblue")
ax.scatter(time_seconds[change_points], signal[change_points], color="crimson", label="starke Änderung")

ax.set_title("Änderungen in einer Messreihe erkennen")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Messwert")
ax.legend()

plt.show()

```

## **3. Frequenzen und Merkmalstabellen**

### **3.1. Fehlpunkte und Ausreißer**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_03_01.jpg?v=1787634181" width="250">



>* Fehlpunkte fehlen, Ausreißer weichen stark ab
>* Kontextprüfung schützt Merkmale vor Verzerrung

>* Fehlende Werte kontextabhängig ersetzen oder markieren
>* Qualitätsangaben machen Merkmale nachvollziehbar

>* Ausreißer erst fachlich prüfen, nicht löschen
>* Robuste Merkmale und Ausreißerinformationen nutzen



```
#@title Python-Code - Fehlpunkte und Ausreißer

# Wir prüfen Signalabschnitte auf fehlende Werte.
# Ausreißer werden markiert und robust ersetzt.
# Eine kleine Merkmalstabelle zeigt Qualitätsinformationen.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ein reproduzierbares Signal enthält Trend, Rauschen und Fehler.
rng = np.random.default_rng(42)
time_seconds = np.arange(0, 60, 1)
clean_signal = 20 + 0.03 * time_seconds + np.sin(time_seconds / 6)
measured_signal = clean_signal + rng.normal(0, 0.15, size=time_seconds.size)

# Einige Messwerte fehlen, ein Wert ist ein Ausreißer.
measured_signal[[12, 13, 38]] = np.nan
measured_signal[27] = 28.0
signal = pd.Series(measured_signal, index=time_seconds)

# Die Zeitachse muss zur Signalreihe passen.
if len(signal) != len(time_seconds):
    raise ValueError("Zeitachse und Signal haben unterschiedliche Längen.")

# Fehlpunkte werden linear aus Nachbarwerten geschätzt.
filled_signal = signal.interpolate(method="linear")

# Ausreißer erkennen wir über robuste Median-Abweichungen.
median_value = filled_signal.median()
absolute_deviation = (filled_signal - median_value).abs()
mad_value = absolute_deviation.median()

# Eine kleine Sicherheitsprüfung verhindert Division durch null.
if mad_value == 0:
    raise ValueError("Die robuste Streuung ist null.")

robust_z = 0.6745 * (filled_signal - median_value) / mad_value
outlier_mask = robust_z.abs() > 3.5
cleaned_signal = filled_signal.mask(outlier_mask).interpolate(method="linear")

# Signalabschnitte werden zu einer kompakten Merkmalstabelle.
segment_id = time_seconds // 15
feature_table = pd.DataFrame({"segment": segment_id})
feature_table["value"] = cleaned_signal.to_numpy()
feature_table["missing"] = signal.isna().to_numpy()
feature_table["outlier"] = outlier_mask.to_numpy()

# Pro Abschnitt speichern wir Kennwerte und Qualitätsmerkmale.
features = feature_table.groupby("segment").agg(
    mean_value=("value", "mean"),
    max_value=("value", "max"),
    missing_count=("missing", "sum"),
    outlier_count=("outlier", "sum"),
)

# Gerundete Werte sind für Anfänger leichter lesbar.
features = features.round(2)
print("Merkmalstabelle pro 15-Sekunden-Abschnitt:")
print(features.head(4).to_string())

# Die Grafik zeigt Rohsignal, Bereinigung und markierte Probleme.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(time_seconds, signal, "o-", label="Rohsignal", alpha=0.55)
ax.plot(time_seconds, cleaned_signal, "-", label="bereinigtes Signal")
ax.scatter(time_seconds[outlier_mask], filled_signal[outlier_mask], color="red", label="Ausreißer")

# Achsenbeschriftungen nennen die verwendeten Einheiten.
ax.set_title("Fehlpunkte und Ausreißer vor der Merkmalstabelle")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Temperatur in Grad Celsius")
ax.legend()
plt.show()

```

### **3.2. Frequenzen im Signal**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_03_02.jpg?v=1787634183" width="250">



>* Frequenzen zeigen wiederkehrende Signalmuster.
>* Merkmalstabellen erfassen rhythmische Abschnittseigenschaften.

>* Signalabschnitte zeigen lokale Frequenzmuster.
>* Merkmale machen Abschnitte vergleichbar und auswertbar.

>* Frequenzen immer im Messkontext deuten
>* Merkmalstabellen verbinden mehrere Signalperspektiven



```
#@title Python-Code - Frequenzen im Signal

# Dieses Beispiel zeigt Frequenzmerkmale aus Signalabschnitten.
# Eine Fourier-Analyse findet dominante Schwingungen.
# Am Ende entsteht eine kleine Merkmalstabelle.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Wir erzeugen ein kurzes Sensorsignal mit zwei Abschnitten.
sampling_rate_hz = 100
duration_seconds = 4
time = np.arange(0, duration_seconds, 1 / sampling_rate_hz)

# Der erste Abschnitt schwingt langsam, der zweite schneller.
first_part = np.sin(2 * np.pi * 3 * time[:200])
second_part = 0.8 * np.sin(2 * np.pi * 12 * time[200:])
signal = np.concatenate([first_part, second_part])

# Ein kleines Rauschen macht das Signal realistischer.
rng = np.random.default_rng(42)
noise = rng.normal(0, 0.15, size=signal.size)
signal = signal + noise

# Wir prüfen, ob die Abschnittslänge zum Signal passt.
window_size = 100
if signal.size % window_size != 0:
    raise ValueError("Die Abschnittslänge passt nicht zum Signal.")

# Für jeden Abschnitt berechnen wir einfache Frequenzmerkmale.
rows = []
for start in range(0, signal.size, window_size):
    segment = signal[start:start + window_size]
    frequencies = np.fft.rfftfreq(window_size, d=1 / sampling_rate_hz)
    spectrum = np.abs(np.fft.rfft(segment))
    spectrum[0] = 0
    dominant_index = int(np.argmax(spectrum))
    dominant_frequency = frequencies[dominant_index]
    total_energy = float(np.sum(spectrum ** 2))
    high_energy = float(np.sum(spectrum[frequencies >= 10] ** 2))
    high_share = high_energy / total_energy if total_energy > 0 else 0
    rows.append([start // window_size, dominant_frequency, high_share])

# Die Merkmalstabelle fasst jeden Abschnitt kompakt zusammen.
feature_table = pd.DataFrame(
    rows,
    columns=["Abschnitt", "dominante_Frequenz_Hz", "Anteil_hohe_Frequenzen"],
)

# Gerundete Werte sind für den ersten Blick leichter lesbar.
display_table = feature_table.round(2)
print("Merkmalstabelle aus vier Signalabschnitten:")
print(display_table.to_string(index=False))

# Die Grafik zeigt das Signal und die Abschnittsgrenzen.
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(time, signal, label="Signal")
ax.set_title("Signal mit langsamem und schnellem Frequenzanteil")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Messwert")

# Abschnittsgrenzen helfen beim Bezug zur Tabelle.
for boundary in [1, 2, 3]:
    ax.axvline(boundary, color="gray", linestyle="--", linewidth=1)

ax.legend()
plt.tight_layout()
plt.show()

```

### **3.3. Signal Mini Projekt**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_05/Lecture_B/image_03_03.jpg?v=1787634179" width="250">



>* Signalabschnitte werden zu Merkmalen zusammengefasst
>* Tabellen erleichtern Vergleich, Visualisierung und Analyse

>* Abschnitte einheitlich bilden, Merkmale klar zuordnen
>* Fehlwerte und Ausreißer fachlich begründet behandeln

>* Merkmalstabellen machen Signalabschnitte vergleichbar
>* Passende Merkmale unterstützen begründete Datenentscheidungen



```
#@title Python-Code - Signal Mini Projekt

# Dieses Mini Projekt erzeugt ein künstliches Sensorsignal.
# Signalabschnitte werden in Merkmale übersetzt.
# Am Ende entsteht eine kleine Merkmalstabelle.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Eine feste Zufallsquelle macht das Beispiel reproduzierbar.
rng = np.random.default_rng(42)

# Wir erzeugen eine Zeitachse mit Messwerten pro Sekunde.
sampling_rate_hz = 50
seconds = 12
sample_count = sampling_rate_hz * seconds

# Das Signal enthält Trend, Schwingung und zufälliges Rauschen.
time_s = np.arange(sample_count) / sampling_rate_hz
trend = 0.03 * time_s
wave = np.sin(2 * np.pi * 2.0 * time_s)

# Ein zweiter Anteil macht spätere Abschnitte etwas unruhiger.
extra_wave = 0.4 * np.sin(2 * np.pi * 6.0 * time_s)
noise = rng.normal(0, 0.18, sample_count)
signal = trend + wave + extra_wave + noise

# Wir prüfen, ob die Fenster sauber in das Signal passen.
window_size = 100
window_count = sample_count // window_size
usable_count = window_count * window_size

# Die Messreihe wird in gleich lange Abschnitte zerlegt.
trimmed_time = time_s[:usable_count]
trimmed_signal = signal[:usable_count]
windows = trimmed_signal.reshape(window_count, window_size)

# Für jedes Fenster berechnen wir kompakte Signalmerkmale.
rows = []
for window_index in range(window_count):
    window = windows[window_index]
    changes = np.diff(window)
    spectrum = np.fft.rfft(window - np.mean(window))
    frequencies = np.fft.rfftfreq(window_size, d=1 / sampling_rate_hz)

    magnitudes = np.abs(spectrum)
    magnitudes[0] = 0
    dominant_frequency = frequencies[np.argmax(magnitudes)]
    start_s = window_index * window_size / sampling_rate_hz

    rows.append(
        {
            "start_s": start_s,
            "mean": np.mean(window),
            "std": np.std(window),
            "mean_abs_change": np.mean(np.abs(changes)),
            "dominant_hz": dominant_frequency,
        }
    )

# Die Liste der Merkmale wird zu einer tabellarischen Darstellung.
feature_table = pd.DataFrame(rows)
rounded_table = feature_table.round(3)

# Die Ausgabe zeigt nur wenige Zeilen für gute Lesbarkeit.
print("Merkmalstabelle: eine Zeile beschreibt ein Signalfenster.")
print(rounded_table.head(5).to_string(index=False))
print(f"Fensterlänge: {window_size} Messpunkte = {window_size / sampling_rate_hz:.1f} s")

# Die Grafik verbindet Rohsignal und Fenstergrenzen.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(trimmed_time, trimmed_signal, label="Signal")

# Vertikale Linien markieren die gebildeten Abschnitte.
for boundary in range(1, window_count):
    boundary_time = boundary * window_size / sampling_rate_hz
    ax.axvline(boundary_time, color="gray", alpha=0.25)

ax.set_title("Signalabschnitte als Grundlage der Merkmalstabelle")
ax.set_xlabel("Zeit in Sekunden")
ax.set_ylabel("Messwert")
ax.legend()
plt.show()

```
