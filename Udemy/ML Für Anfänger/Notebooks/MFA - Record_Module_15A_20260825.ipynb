>* TensorFlow-Version vor Übungen prüfen

>* Softwaredetails beeinflussen Berechnungen und Leistung

```
#@title Python-Code - TensorFlow Version prüfen

# Dieses Beispiel prüft die installierte TensorFlow-Version.
# Es verbindet Versionswissen mit ersten Tensor-Eigenschaften.
# Die Ausgabe zeigt Version, Form und Datentyp.

import tensorflow as tf

# Die Versionsnummer beschreibt die geladene TensorFlow-Bibliothek.
tensorflow_version = tf.__version__

# Ein kleiner Tensor reicht für eine sichere Prüfung.
sample_tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

# Diese Prüfung macht eine unerwartete Umgebung früh sichtbar.
if not tensorflow_version.startswith("2."):
    raise RuntimeError("Dieses Beispiel erwartet TensorFlow 2.x.")

# Form und Datentyp sind zentrale Eigenschaften jedes Tensors.
tensor_shape = sample_tensor.shape
tensor_dtype = sample_tensor.dtype.name

print(f"TensorFlow-Version: {tensorflow_version}")
print(f"Beispiel-Tensor: Rang {len(tensor_shape)}")
print(f"Form: {tuple(tensor_shape)}")
print(f"Datentyp: {tensor_dtype}")

```

### **1.2. Formen und Datentypen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_01_02.jpg?v=1787655778" width="250">



>* Tensorformen zeigen Anordnung und Dimensionen.
>* Passende Formen verhindern Modell- und Datenfehler.

>* Datentypen bestimmen Werteart und Genauigkeit
>* Passende Typen vermeiden Speicher- und Rechenprobleme

>* Form und Datentyp zeigen Datenstruktur.
>* Prüfen verhindert Fehler und Verbindungsprobleme.



```
#@title Python-Code - Formen und Datentypen

# Dieses Beispiel untersucht Tensorformen und Datentypen.
# Tensoren speichern Werte in geordneten Dimensionen.
# Die Ausgabe zeigt Form, Rang und dtype.

import tensorflow as tf

# Ein Skalar enthält genau einen einzelnen Wert.
scalar = tf.constant(7)

# Ein Vektor enthält eine geordnete Reihe von Werten.
vector = tf.constant([1.5, 2.5, 3.5], dtype=tf.float32)

# Eine Matrix ordnet Werte in Zeilen und Spalten.
matrix = tf.constant([[1, 2, 3], [4, 5, 6]], dtype=tf.int32)

# Ein Bildstapel hat Batch, Höhe, Breite und Kanäle.
image_batch = tf.zeros((2, 4, 4, 3), dtype=tf.float32)

# Diese Prüfung macht eine erwartete Form explizit.
tf.debugging.assert_shapes([(image_batch, (2, 4, 4, 3))])

print("Skalar: shape=(), Rang=0, dtype=" + scalar.dtype.name)
print("Vektor: shape=" + str(vector.shape) + ", Rang=" + str(vector.ndim))
print("Matrix: shape=" + str(matrix.shape) + ", dtype=" + matrix.dtype.name)
print("Bildstapel: shape=" + str(image_batch.shape))
print("Bildstapel-Rang: " + str(image_batch.ndim))
print("Erster Bildwert: " + str(float(image_batch[0, 0, 0, 0].numpy())))
print("Formprüfung bestanden: Batch, Höhe, Breite und Kanäle passen.")

```

## **2. Operationen und Gradienten**

### **2.1. Broadcasting verstehen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_02_01.jpg?v=1787655785" width="250">



>* Broadcasting kombiniert kompatible Tensorformen effizient.
>* Kleinere Tensoren werden ohne Kopien wiederverwendet.

>* Formen von rechts nach links vergleichen
>* Farbwerte automatisch auf alle Pixel anwenden

>* Tensorformen bestimmen die Bedeutung der Daten
>* Achsen prüfen verhindert plausible Fehlberechnungen



```
#@title Python-Code - Broadcasting verstehen

# Dieses Beispiel zeigt Broadcasting mit TensorFlow-Tensoren.
# Formen entscheiden, welche Werte automatisch wiederholt werden.
# Das Ergebnis zeigt normalisierte Messwerte und Gradienten.

import tensorflow as tf

# Beobachtungen haben drei Zeilen und zwei Merkmale.
measurements = tf.constant(
    [[10.0, 100.0], [12.0, 110.0], [14.0, 130.0]]
)

# Ein Mittelwert pro Merkmal passt zur letzten Achse.
feature_means = tf.constant([12.0, 110.0])

# Diese Prüfung macht die beabsichtigte Bedeutung sichtbar.
if measurements.shape[-1] != feature_means.shape[0]:
    raise ValueError("Die Merkmalsanzahl passt nicht zum Mittelwertvektor.")

# TensorFlow broadcastet den Vektor über alle Beobachtungen.
centered = measurements - feature_means

# GradientTape zeigt, dass Broadcasting auch ableitbar bleibt.
scale = tf.Variable([1.0, 0.1])

with tf.GradientTape() as tape:
    scaled = centered * scale
    loss = tf.reduce_mean(tf.square(scaled))

gradient = tape.gradient(loss, scale)

print("Form Messwerte:", measurements.shape)
print("Form Mittelwerte:", feature_means.shape)
print("Form Ergebnis:", centered.shape)
print("Zentrierte Werte:", centered.numpy().round(1).tolist())
print("Loss:", round(float(loss.numpy()), 3))
print("Gradient nach scale:", gradient.numpy().round(3).tolist())

```

### **2.2. Gradienten automatisch berechnen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_02_02.jpg?v=1787655787" width="250">



>* Gradienten zeigen nötige Änderungen beim Lernen
>* GradientTape zeichnet Schritte auf und leitet rückwärts ab

>* GradientTape berechnet Fehlergradienten für Modellgewichte
>* Optimierer verbessern Gewichte schrittweise automatisch

>* GradientTape verfolgt nur relevante Operationen.
>* Unterscheide Daten, Konstanten und trainierbare Variablen.



```
#@title Python-Code - Gradienten automatisch berechnen

# Dieses Beispiel zeigt automatische Gradienten mit TensorFlow.
# GradientTape verfolgt Operationen für eine einfache Funktion.
# Die Ausgabe vergleicht berechnete und erwartete Steigungen.

import tensorflow as tf

# Eine Variable wird automatisch vom GradientTape beobachtet.
x = tf.Variable(3.0)

# Innerhalb des Tape-Bereichs entsteht der Zielwert.
with tf.GradientTape() as tape:
    y = x * x + 2.0 * x + 1.0

# TensorFlow berechnet die Ableitung von y nach x.
gradient = tape.gradient(y, x)

# Für diese Funktion lautet die Ableitung zwei mal x plus zwei.
expected_gradient = 2.0 * x + 2.0

# Eine kleine Prüfung macht die Annahme sichtbar.
if x.shape != ():
    raise ValueError("x sollte ein einzelner Skalar sein.")

print("Funktion: y = x² + 2x + 1")
print(f"Eingesetzter Wert x: {x.numpy():.1f}")
print(f"Berechneter Funktionswert y: {y.numpy():.1f}")
print(f"Gradient laut TensorFlow: {gradient.numpy():.1f}")
print(f"Erwarteter Gradient: {expected_gradient.numpy():.1f}")
print("GradientTape zeigt, wie stark y auf x reagiert.")

```

### **2.3. Manueller Gradientenvergleich**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_02_03.jpg?v=1787655789" width="250">



>* GradientTape-Ergebnisse manuell auf Plausibilität prüfen
>* Kleine Änderungen zeigen lokale Empfindlichkeit

>* Manuelle Prüfung erkennt Modell- und Skalierungsfehler
>* GradientTape erfasst nur beobachtete TensorFlow-Schritte

>* Nur Diagnose, abhängig von sinnvoller Schrittgröße
>* Richtung verstehen, Training und Fehler besser deuten



```
#@title Python-Code - Manueller Gradientenvergleich

# Wir vergleichen automatische und numerische Gradienten.
# GradientTape liefert die lokale Steigung automatisch.
# Die Näherung sollte fast gleich ausfallen.

import tensorflow as tf
import matplotlib.pyplot as plt

# Diese kleinen Tensoren bilden ein einfaches Regressionsbeispiel.
x_values = tf.constant([1.0, 2.0, 3.0], dtype=tf.float32)
y_true = tf.constant([2.0, 4.0, 6.0], dtype=tf.float32)

# Das Gewicht ist die Variable, deren Gradient gesucht wird.
weight = tf.Variable(1.5, dtype=tf.float32)
epsilon = tf.constant(0.001, dtype=tf.float32)

# Diese Funktion berechnet den mittleren quadratischen Fehler.
def loss_for_weight(test_weight):
    predictions = test_weight * x_values
    errors = predictions - y_true
    return tf.reduce_mean(tf.square(errors))

# GradientTape verfolgt die TensorFlow-Operationen automatisch.
with tf.GradientTape() as tape:
    loss_value = loss_for_weight(weight)

auto_gradient = tape.gradient(loss_value, weight)

# Die zentrale Differenz nähert dieselbe Steigung numerisch an.
loss_plus = loss_for_weight(weight + epsilon)
loss_minus = loss_for_weight(weight - epsilon)
manual_gradient = (loss_plus - loss_minus) / (2.0 * epsilon)

# Eine einfache Prüfung schützt vor unerwarteten Formen.
if auto_gradient.shape != weight.shape:
    raise ValueError("Der Gradient hat nicht die erwartete Form.")

# Die Ausgabe zeigt Richtung und Größenordnung beider Gradienten.
print(f"Gewicht: {weight.numpy():.3f}")
print(f"Verlust am Gewicht: {loss_value.numpy():.4f}")
print(f"GradientTape-Gradient: {auto_gradient.numpy():.4f}")
print(f"Manuelle Näherung: {manual_gradient.numpy():.4f}")
print(f"Absoluter Unterschied: {abs(auto_gradient.numpy() - manual_gradient.numpy()):.6f}")

# Die Kurve macht die lokale Steigung am Gewicht sichtbar.
weight_grid = tf.linspace(1.0, 2.5, 80)
loss_grid = [loss_for_weight(w).numpy() for w in weight_grid]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(weight_grid.numpy(), loss_grid, label="Verlustkurve")
ax.scatter([weight.numpy()], [loss_value.numpy()], color="red", label="aktuelles Gewicht")
ax.set_title("Manueller Gradientenvergleich")
ax.set_xlabel("Gewicht")
ax.set_ylabel("Mittlerer quadratischer Fehler")
ax.legend()
plt.show()

```

## **3. tf data Grundlagen**

### **3.1. Dataset erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_03_01.jpg?v=1787655780" width="250">



>* Datasets bündeln Daten für Modelle strukturiert
>* TensorFlow steuert Reihenfolge und Verarbeitung

>* Daten aus Listen, Arrays oder Tensoren nutzen
>* Formen, Typen und Label-Zuordnung früh prüfen

>* Datasets strukturieren Daten für realistische Workflows
>* Pipelines ermöglichen Batching, Shuffling und Mapping



```
#@title Python-Code - Dataset erstellen

# Dieses Beispiel erstellt ein kleines TensorFlow Dataset.
# Es zeigt Elemente, Formen und Datentypen.
# Am Ende sehen wir gebündelte Trainingsdaten.

import numpy as np
import tensorflow as tf

# Kleine Arrays stehen hier für Eingaben und Zielwerte.
features = np.array([[18.0, 1.0], [22.0, 0.0], [35.0, 1.0], [46.0, 0.0]])
labels = np.array([0, 0, 1, 1])

# Diese Prüfung verhindert falsch gekoppelte Beispiele.
if len(features) != len(labels):
    raise ValueError("Eingaben und Labels brauchen gleich viele Zeilen.")

# from_tensor_slices verbindet jede Zeile mit dem passenden Label.
dataset = tf.data.Dataset.from_tensor_slices((features, labels))

# Wir betrachten zuerst ein einzelnes Element des Datasets.
first_features, first_label = next(iter(dataset))
print("Ein Element:", first_features.numpy().tolist(), "Label:", int(first_label))

# element_spec beschreibt Struktur, Form und Datentyp.
print("Struktur:", dataset.element_spec)

# Shuffling mischt die Reihenfolge reproduzierbar.
shuffled_dataset = dataset.shuffle(buffer_size=4, seed=42)

# Batching fasst mehrere Beispiele zu einem Trainingsschritt zusammen.
batched_dataset = shuffled_dataset.batch(2)

# Mapping kann Werte direkt in der Pipeline vorbereiten.
prepared_dataset = batched_dataset.map(lambda x, y: (x / 50.0, y))

# Wir zeigen die Form eines vorbereiteten Mini-Batches.
batch_features, batch_labels = next(iter(prepared_dataset))
print("Batch-Form:", batch_features.shape, "Label-Form:", batch_labels.shape)

# Kleine gerundete Werte bleiben gut lesbar.
print("Erster vorbereiteter Batch:", np.round(batch_features.numpy(), 2).tolist())
print("Batch-Labels:", batch_labels.numpy().tolist())

```

### **3.2. Batching und Shuffling**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_03_02.jpg?v=1787655782" width="250">



>* Batches bündeln mehrere Beispiele für parallele Verarbeitung
>* Gleiche Struktur ermöglicht gemeinsame Tensoren

>* Shuffling mischt Daten zufällig vor dem Training
>* Batches werden vielfältiger und weniger sortierungsabhängig

>* Erst mischen, dann meist besser batchen
>* Puffer und Batchgröße passend wählen



```
#@title Python-Code - Batching und Shuffling

# Dieses Beispiel zeigt Batching und Shuffling.
# Kleine Zahlen stehen für einzelne Trainingsbeispiele.
# Die Ausgabe vergleicht geordnete und gemischte Batches.

import tensorflow as tf

# Wir erzeugen zwölf einfache Beispiele mit passenden Labels.
features = tf.range(12)
labels = features * 10

# Jedes Dataset-Element besteht aus Merkmal und Label.
base_dataset = tf.data.Dataset.from_tensor_slices((features, labels))

# Ohne Shuffling bleiben benachbarte Beispiele im selben Batch.
ordered_dataset = base_dataset.batch(4)

# Mit Seed wird die zufällige Reihenfolge reproduzierbar.
shuffled_dataset = base_dataset.shuffle(buffer_size=12, seed=42).batch(4)

# Diese Prüfung macht die erwartete Batchgröße sichtbar.
first_batch_features, first_batch_labels = next(iter(ordered_dataset))
if first_batch_features.shape[0] != 4:
    raise ValueError("Die Batchgröße sollte vier Beispiele enthalten.")

print("Ausgangsdaten: 12 Beispiele, Batchgröße 4")
print("Geordnete Batches:")

for batch_number, (batch_features, batch_labels) in enumerate(ordered_dataset, start=1):
    feature_list = batch_features.numpy().tolist()
    label_list = batch_labels.numpy().tolist()
    print(f"Batch {batch_number}: x={feature_list}, y={label_list}")

print("Gemischte Batches mit shuffle(buffer_size=12, seed=42):")

for batch_number, (batch_features, batch_labels) in enumerate(shuffled_dataset, start=1):
    feature_list = batch_features.numpy().tolist()
    label_list = batch_labels.numpy().tolist()
    print(f"Batch {batch_number}: x={feature_list}, y={label_list}")

```

### **3.3. Mapping Übung**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_15/Lecture_A/image_03_03.jpg?v=1787655783" width="250">



>* Mapping verändert jedes Dataset-Element systematisch.
>* Vorverarbeitung bleibt direkt in der Pipeline.

>* Kleines Dataset mit klarer Transformation nutzen
>* Konsistente Formen und Datentypen sicherstellen

>* Mapping mit Shuffling und Batching kombinieren
>* Reihenfolge, Form und Datentypen bewusst prüfen



```
#@title Python-Code - Mapping Übung

# Diese Übung zeigt Mapping in tf.data.
# Jedes Element wird gleich transformiert.
# Am Ende sehen wir saubere Batches.

import tensorflow as tf
import matplotlib.pyplot as plt

# Kleine Rohdaten machen die Pipeline gut nachvollziehbar.
features = tf.constant([0, 1, 2, 3, 4, 5], dtype=tf.float32)
labels = tf.constant([0, 0, 1, 1, 1, 0], dtype=tf.int32)

# Das Dataset enthält Paare aus Eingabe und Label.
dataset = tf.data.Dataset.from_tensor_slices((features, labels))

# Diese Funktion wird auf jedes einzelne Element angewendet.
def scale_and_expand(feature, label):
    scaled_feature = feature / 5.0
    expanded_feature = tf.stack([scaled_feature, scaled_feature * scaled_feature])
    return expanded_feature, label

# Mapping verändert jedes Beispiel vor Shuffle und Batch.
mapped_dataset = dataset.map(scale_and_expand)
shuffled_dataset = mapped_dataset.shuffle(buffer_size=6, seed=42)
batched_dataset = shuffled_dataset.batch(3)

# Wir prüfen Form und Datentyp des ersten Batches.
first_features, first_labels = next(iter(batched_dataset))
print("Form der Batch-Features:", first_features.shape)
print("Datentyp der Batch-Features:", first_features.dtype.name)
print("Labels im ersten Batch:", first_labels.numpy().tolist())

# Die Werte zeigen die Wirkung der Mapping-Funktion.
rounded_features = tf.round(first_features * 100) / 100
print("Gemappte Features im ersten Batch:", rounded_features.numpy().tolist())

# Die Grafik vergleicht Rohwerte und skalierte Werte.
raw_values = features.numpy()
scaled_values = (features / 5.0).numpy()

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(raw_values, scaled_values, marker="o", label="feature / 5")
ax.set_title("Mapping skaliert jedes Dataset-Element")
ax.set_xlabel("Rohwert")
ax.set_ylabel("Skalierter Wert")
ax.legend()
plt.show()

```
