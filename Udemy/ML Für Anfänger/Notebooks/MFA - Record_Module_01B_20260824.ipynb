### **1.1. Arrays erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_01_01.jpg?v=1787626112" width="250">



>* NumPy-Arrays speichern gleichartige Daten effizient
>* Form und Datentyp erleichtern Berechnungen

>* Array-Form muss zur Bedeutung passen
>* Startwerte gezielt für Berechnungen wählen

>* Datentyp beeinflusst Speicher, Genauigkeit und Geschwindigkeit
>* Array-Eigenschaften vor Analysen immer prüfen



```
#@title Python-Code - Arrays erstellen

# Dieses Beispiel erstellt kleine NumPy-Arrays.
# Formen und Datentypen werden sichtbar geprüft.
# Die Ausgabe zeigt passende Array-Eigenschaften.

import numpy as np

# Eine Python-Liste wird in ein eindimensionales Array umgewandelt.
temperatures = np.array([18.5, 20.0, 21.5, 19.0], dtype=np.float32)

# Eine verschachtelte Liste erzeugt ein zweidimensionales Array.
sales_table = np.array([[120, 135, 150], [90, 110, 105]], dtype=np.int16)

# Ein Startarray mit Nullen reserviert Platz für spätere Werte.
empty_scores = np.zeros((2, 3), dtype=np.float32)

# Diese Prüfung schützt vor einer unerwarteten Tabellenform.
if sales_table.shape != (2, 3):
    raise ValueError("Die Verkaufstabelle sollte zwei Zeilen und drei Spalten haben.")

print("Temperaturen:", temperatures.tolist())
print("Temperaturen: Form", temperatures.shape, "Dimensionen", temperatures.ndim)
print("Temperaturen: Größe", temperatures.size, "Datentyp", temperatures.dtype)
print("Verkaufstabelle: Form", sales_table.shape, "Datentyp", sales_table.dtype)
print("Startarray mit Nullen:", empty_scores.tolist())

```

### **1.2. Indexierung und Masken**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_01_02.jpg?v=1787626110" width="250">



>* Indizes wählen Elemente in mehrdimensionalen Arrays.
>* Slices wählen zusammenhängende Datenbereiche gezielt aus.

>* Masken filtern Array-Werte nach Bedingungen
>* Große Datenmengen gezielt auswählen und bereinigen

>* Arrayform und Masken müssen zusammenpassen
>* Indexierung nach Position, Maskierung nach Bedingung



```
#@title Python-Code - Indexierung und Masken

# Dieses Beispiel zeigt Indexierung und Masken.
# Wir untersuchen ein kleines Temperatur-Array.
# Am Ende sehen wir gefilterte Werte.

import numpy as np

# Die Matrix enthält Tage als Zeilen.
temperatures = np.array(
    [[18, 21, 24], [20, 26, 31], [17, 29, 33], [19, 22, 28]]
)

# Diese Prüfung macht die erwartete Form sichtbar.
expected_shape = (4, 3)
if temperatures.shape != expected_shape:
    raise ValueError("Die Temperaturmatrix hat eine unerwartete Form.")

# Einzelne Positionen werden mit Zeile und Spalte gewählt.
second_day_noon = temperatures[1, 1]
first_two_days = temperatures[:2, :]

# Eine Maske markiert alle Werte über dem Schwellenwert.
hot_mask = temperatures > 30
hot_values = temperatures[hot_mask]

# Die Ausgabe bleibt klein und gut lesbar.
print("Form:", temperatures.shape)
print("Datentyp:", temperatures.dtype)
print("Tag 2, Mittag:", int(second_day_noon), "°C")
print("Erste zwei Tage:", first_two_days.tolist())
print("Maske über 30 °C:", hot_mask.astype(int).tolist())
print("Gefilterte Werte:", hot_values.tolist(), "°C")

```

### **1.3. Vektorisierte Berechnungen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_01_03.jpg?v=1787626114" width="250">



>* Operationen wirken direkt auf ganze Arrays
>* Kürzerer, schnellerer Code durch NumPy

>* Gleiche Daten gemeinsam umrechnen oder anpassen
>* Passende Arrays werden elementweise verrechnet

>* Broadcasting nutzt kompatible Array-Formen automatisch.
>* Form und Datentyp bewusst prüfen.



```
#@title Python-Code - Vektorisierte Berechnungen

# Dieses Beispiel zeigt vektorisierte NumPy-Berechnungen.
# Arrays werden elementweise ohne Schleifen verarbeitet.
# Formen und Datentypen werden sichtbar geprüft.

import numpy as np

# Kleine Messwerte bleiben übersichtlich und realistisch.
temperatures_celsius = np.array([18.0, 20.5, 22.0, 19.5], dtype=np.float64)

# Ein einzelner Wert wird auf alle Elemente angewendet.
temperatures_fahrenheit = temperatures_celsius * 9 / 5 + 32

# Zwei gleich geformte Arrays werden elementweise kombiniert.
corrections_celsius = np.array([0.2, -0.1, 0.0, 0.3], dtype=np.float64)
corrected_celsius = temperatures_celsius + corrections_celsius

# Diese Prüfung macht die erwartete Form ausdrücklich sichtbar.
if temperatures_celsius.shape != corrections_celsius.shape:
    raise ValueError("Die Arrays müssen dieselbe Form haben.")

# Broadcasting nutzt einen Faktor pro Spalte.
sales_euro = np.array([[120, 150, 90], [100, 180, 110]], dtype=np.float64)
branch_factors = np.array([1.10, 0.95, 1.05], dtype=np.float64)
adjusted_sales = sales_euro * branch_factors

# Kurze Ausgaben zeigen Ergebnis, Form und Datentyp.
print("Temperaturen in °C:", temperatures_celsius)
print("Umgerechnet in °F:", np.round(temperatures_fahrenheit, 1))
print("Korrigierte °C:", np.round(corrected_celsius, 1))
print("Umsatz-Form:", sales_euro.shape)
print("Faktor-Form:", branch_factors.shape)
print("Angepasste Umsätze:", np.round(adjusted_sales, 1))
print("Datentyp der Umsätze:", adjusted_sales.dtype)

```

## **2. pandas Tabellen**

### **2.1. Series und DataFrames**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_02_01.jpg?v=1787626106" width="250">



>* Series sind beschriftete einzelne Datenspalten
>* DataFrames organisieren Tabellen zum Analysieren

>* Series: ein Merkmal über viele Beobachtungen
>* DataFrames: mehrere Merkmale derselben Fälle

>* Spaltennamen und Indizes machen Daten verständlich
>* Auswählen, filtern und gruppieren wird einfacher



```
#@title Python-Code - Series und DataFrames

# Dieses Beispiel zeigt Series und DataFrames.
# Wir wählen, filtern und gruppieren Tabellendaten.
# Die Ausgabe zeigt typische pandas-Ergebnisse.

import pandas as pd

# Eine Series ist eine beschriftete einzelne Spalte.
sales_series = pd.Series([120, 80, 150, 90], name="sales_eur")

# Ein DataFrame bündelt mehrere beschriftete Spalten.
orders = pd.DataFrame(
    {
        "product": ["Buch", "Stift", "Buch", "Heft"],
        "region": ["Nord", "Süd", "Nord", "Süd"],
        "sales_eur": [120, 80, 150, 90],
    }
)

# Diese Prüfung macht die erwartete Tabellenform sichtbar.
if orders.shape != (4, 3):
    raise ValueError("Die Beispieltabelle sollte vier Zeilen haben.")

# Eine einzelne Spalte aus dem DataFrame ist wieder eine Series.
selected_sales = orders["sales_eur"]

# Eine Bedingung filtert nur passende Zeilen heraus.
high_sales = orders.loc[orders["sales_eur"] >= 100, ["product", "sales_eur"]]

# Gruppieren fasst Zeilen nach einer gemeinsamen Eigenschaft zusammen.
region_summary = orders.groupby("region", as_index=False)["sales_eur"].sum()

print("Series-Werte:", selected_sales.tolist())
print("DataFrame-Form:", orders.shape)
print("Gefilterte Zeilen:", high_sales.to_dict("records"))
print("Umsatz je Region:", region_summary.to_dict("records"))

```

### **2.2. Daten laden und filtern**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_02_02.jpg?v=1787626104" width="250">



>* Datenquellen als DataFrame laden
>* Spalten, Datentypen und Auffälligkeiten prüfen

>* Relevante Daten mit Bedingungen gezielt auswählen
>* Filter fachlich prüfen und Verzerrungen vermeiden

>* Nur relevante Spalten für klare Analysen wählen
>* Datenprobleme früh erkennen und Qualität prüfen



```
#@title Python-Code - Daten laden und filtern

# Wir laden eine kleine Tabelle direkt im Speicher.
# Danach filtern wir Zeilen mit klaren Bedingungen.
# Am Ende sehen wir einen fokussierten Datenausschnitt.

import pandas as pd

# Diese Beispieldaten stehen für eine kleine Studierendentabelle.
data = {
    "student": ["Aylin", "Ben", "Clara", "David", "Elif", "Finn"],
    "program": ["Informatik", "BWL", "Informatik", "Design", "BWL", "Informatik"],
    "semester": [1, 3, 2, 4, 1, 5],
    "grade": [1.7, 2.3, 1.3, 3.0, 2.0, 1.0],
}

# Aus dem Dictionary entsteht ein pandas DataFrame.
students = pd.DataFrame(data)

# Eine einfache Prüfung schützt vor unerwarteten Spaltennamen.
required_columns = {"student", "program", "semester", "grade"}
if set(students.columns) != required_columns:
    raise ValueError("Die Tabelle hat nicht die erwarteten Spalten.")

# Wir wählen nur Zeilen, die zur Fragestellung passen.
filtered_students = students[
    (students["program"] == "Informatik") & (students["grade"] <= 1.7)
]

# Danach reduzieren wir die Tabelle auf wichtige Spalten.
selected_columns = filtered_students[["student", "semester", "grade"]]

# Die Ausgabe bleibt klein und gut lesbar.
print("Gefilterte Informatik-Studierende mit Note 1,7 oder besser:")
print(selected_columns.to_string(index=False))

# Eine kurze Kennzahl zeigt die Größe des Ausschnitts.
print("Anzahl gefilterter Zeilen:", len(selected_columns))

```

### **2.3. Gruppen zusammenfassen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_02_03.jpg?v=1787626108" width="250">



>* Gruppen zeigen Verhalten innerhalb wiederholter Kategorien
>* Kennzahlen verdichten Teilmengen zu erkennbaren Mustern

>* Gruppieren: Merkmal wählen, Zeilen bündeln, berechnen
>* Kennzahlen zeigen neue Muster in Daten

>* Kennzahlen kritisch prüfen, Ausreißer beachten
>* Saubere Gruppen ermöglichen bessere Vergleiche



```
#@title Python-Code - Gruppen zusammenfassen

# Wir fassen Bestellungen nach Produktgruppen zusammen.
# pandas groupby berechnet Kennzahlen pro Kategorie.
# Das Ergebnis zeigt kompakte Gruppenvergleiche.

import pandas as pd
import matplotlib.pyplot as plt

# Diese kleine Tabelle steht für einzelne Bestellungen.
sales = pd.DataFrame(
    {
        "group": ["Buch", "Buch", "Spiel", "Spiel", "Spiel", "Stift"],
        "revenue_eur": [18, 22, 35, 40, 120, 3],
        "returned": [False, False, True, False, True, False],
    }
)

# Wir prüfen eine einfache Annahme zur Tabellengröße.
if len(sales) == 0:
    raise ValueError("Die Beispieltabelle darf nicht leer sein.")

# groupby teilt die Tabelle nach Produktgruppe auf.
summary = sales.groupby("group", as_index=False).agg(
    orders=("revenue_eur", "count"),
    total_revenue_eur=("revenue_eur", "sum"),
    average_revenue_eur=("revenue_eur", "mean"),
    returns=("returned", "sum"),
)

# Sortieren macht den Vergleich der Gruppen leichter.
summary = summary.sort_values("total_revenue_eur", ascending=False)
summary["average_revenue_eur"] = summary["average_revenue_eur"].round(1)

# Die Ausgabe bleibt klein und zeigt nur Gruppenergebnisse.
print("Zusammenfassung pro Produktgruppe:")
print(summary.to_string(index=False))

# Das Balkendiagramm zeigt die wichtigste Kennzahl visuell.
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(summary["group"], summary["total_revenue_eur"], color="steelblue")
ax.set_title("Gesamtumsatz pro Produktgruppe")
ax.set_xlabel("Produktgruppe")

# Die y-Achse enthält messbare Euro-Werte.
ax.set_ylabel("Gesamtumsatz in Euro")
plt.show()

```

## **3. Tabellen prüfen**

### **3.1. Fehlende Werte erkennen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_03_01.jpg?v=1787626103" width="250">



>* Fehlende Werte früh erkennen
>* Lücken können Kennzahlen deutlich verzerren

>* Muster fehlender Werte gezielt untersuchen
>* Systematische Lücken können Analysen verzerren

>* Fehlende Werte fachlich begründet behandeln
>* Entscheidungen transparent machen und Auswirkungen prüfen



```
#@title Python-Code - Fehlende Werte erkennen

# Wir erkennen fehlende Werte in einer kleinen Tabelle.
# Pandas markiert Lücken und berechnet einfache Prüfzahlen.
# Am Ende sehen wir problematische Spalten sofort.

import pandas as pd
import matplotlib.pyplot as plt

# Diese Beispieltabelle enthält absichtlich einige fehlende Werte.
data = {
    "student": ["Ada", "Ben", "Cem", "Dina", "Eli"],
    "points": [88, None, 74, 91, None],
    "hours": [5.0, 3.5, None, 6.0, 4.0],
    "group": ["A", "A", "B", None, "B"],
}

# Aus dem Wörterbuch entsteht eine übersichtliche pandas-Tabelle.
students = pd.DataFrame(data)

# Diese Prüfung schützt vor unerwartet leeren Beispieldaten.
if students.shape != (5, 4):
    raise ValueError("Die Beispieltabelle hat nicht die erwartete Form.")

# isna erkennt fehlende Werte, sum zählt sie spaltenweise.
missing_count = students.isna().sum()
missing_percent = (missing_count / len(students) * 100).round(0)

# Eine kleine Prüftabelle fasst Anzahl und Anteil zusammen.
missing_summary = pd.DataFrame(
    {"fehlend": missing_count, "anteil_prozent": missing_percent}
)

# Nur Spalten mit mindestens einer Lücke werden angezeigt.
problem_columns = missing_summary[missing_summary["fehlend"] > 0]

print("Erste Prüfung: fehlende Werte pro Spalte")
print(problem_columns.to_string())
print("Spalte mit den meisten Lücken:", missing_count.idxmax())

# Das Balkendiagramm macht das Muster der Lücken sichtbar.
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(missing_count.index, missing_count.values, color="tomato")
ax.set_title("Fehlende Werte pro Spalte")

ax.set_xlabel("Spalte")
ax.set_ylabel("Anzahl fehlender Werte")
ax.set_ylim(0, max(missing_count.values) + 1)
plt.show()

```

### **3.2. Datentypen sicher prüfen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_03_02.jpg?v=1787626101" width="250">



>* Datentypen bestimmen sinnvolle Auswertungen
>* Falsche Typen führen zu Analysefehlern

>* Importe können Datentypen unerwartet verfälschen
>* Technischen Typ mit fachlicher Bedeutung abgleichen

>* Datentypen vor Kennzahlen zuverlässig prüfen
>* Datenprobleme früh erkennen und vermeiden



```
#@title Python-Code - Datentypen sicher prüfen

# Wir prüfen Datentypen in einer kleinen Tabelle.
# Fachliche Erwartungen werden mit technischen Typen verglichen.
# Problemspalten werden sichtbar und gezielt korrigiert.

import pandas as pd

# Diese Tabelle enthält typische Importprobleme.
data = pd.DataFrame(
    {
        "order_id": ["001", "002", "003", "004"],
        "price_eur": ["12.50", "8,90", "fehlt", "15.00"],
        "order_date": ["2024-01-05", "05.01.2024", "2024-01-07", "unbekannt"],
        "region": ["Nord", " nord ", "Süd", "Sued"],
    }
)

# Wir prüfen zuerst die technischen Datentypen.
dtype_summary = data.dtypes.astype(str)
print("Technische Datentypen:", dtype_summary.to_dict())

# Fachlich erwarten wir Text, Zahl, Datum und Kategorie.
expected_types = {
    "order_id": "Text",
    "price_eur": "Zahl",
    "order_date": "Datum",
    "region": "Kategorie",
}
print("Fachliche Erwartung:", expected_types)

# Zahlen werden vorsichtig umgewandelt, ungültige Werte werden NaN.
clean_price_text = data["price_eur"].str.replace(",", ".", regex=False)
price_numeric = pd.to_numeric(clean_price_text, errors="coerce")

# Datumswerte werden vorsichtig umgewandelt, ungültige Werte werden NaT.
date_values = pd.to_datetime(data["order_date"], errors="coerce", dayfirst=True)

# Kategorien werden vereinheitlicht, bevor Varianten gezählt werden.
region_clean = data["region"].str.strip().str.lower()
region_clean = region_clean.replace({"sued": "süd"})

# Diese Kennzahlen zeigen einfache Datenprobleme.
invalid_prices = int(price_numeric.isna().sum())
invalid_dates = int(date_values.isna().sum())
region_count = int(region_clean.nunique())

print("Ungültige Preise:", invalid_prices)
print("Ungültige Datumswerte:", invalid_dates)
print("Regionen nach Bereinigung:", region_count)

# Nach der Umwandlung sind Kennzahlen sinnvoller interpretierbar.
mean_price = round(float(price_numeric.mean()), 2)
print("Durchschnittspreis gültiger Werte:", mean_price, "EUR")

# Eine kurze Prüfung bestätigt die erwartete Tabellenform.
expected_columns = 4
print("Spaltenprüfung bestanden:", data.shape[1] == expected_columns)

```

### **3.3. Kurze Abschlussanalyse**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_01/Lecture_B/image_03_03.jpg?v=1787626099" width="250">



>* Wichtigste Beobachtungen verantwortungsvoll zusammenfassen
>* Klare Befunde von offenen Fragen trennen

>* Kennzahlen machen Tabellen verständlich und prüfbar
>* Datenprobleme können Entscheidungen stark verzerren

>* Verwendbarkeit der Tabelle klar einschätzen
>* Datenprobleme transparent benennen und berücksichtigen



```
#@title Python-Code - Kurze Abschlussanalyse

# Diese Analyse fasst eine geprüfte Tabelle zusammen.
# Kennzahlen zeigen Muster und mögliche Datenprobleme.
# Am Ende entsteht eine kurze fachliche Einschätzung.

import pandas as pd

# Eine kleine Verkaufstabelle enthält absichtlich typische Datenprobleme.
sales_data = pd.DataFrame(
    {
        "branch": ["Nord", "Nord", "Süd", "Süd", "West", "West", "West"],
        "region": ["A", "A", "B", None, "C", "C", None],
        "revenue_eur": [1200, 1350, 980, 1020, 5000, 1100, 1150],
        "orders": [24, 27, 20, 21, 80, 22, 23],
    }
)

# Diese Prüfung verhindert eine Analyse mit leerer Tabelle.
if len(sales_data) == 0:
    raise ValueError("Die Tabelle enthält keine Zeilen.")

# Fehlende Werte zeigen, welche Spalten vorsichtig genutzt werden sollten.
missing_region_count = sales_data["region"].isna().sum()
missing_region_share = missing_region_count / len(sales_data)

# Kennzahlen verdichten die Umsatzspalte zu wenigen verständlichen Werten.
mean_revenue = sales_data["revenue_eur"].mean()
median_revenue = sales_data["revenue_eur"].median()
max_revenue = sales_data["revenue_eur"].max()

# Ein einfacher Ausreißerhinweis vergleicht Maximum und Median.
outlier_limit = 3 * median_revenue
has_high_outlier = max_revenue > outlier_limit

# Gruppierung zeigt, ob einzelne Filialen besonders auffallen.
branch_summary = sales_data.groupby("branch", sort=True)["revenue_eur"].mean()
top_branch = branch_summary.idxmax()
top_branch_mean = branch_summary.max()

# Die Abschlussanalyse trennt klare Befunde von offenen Fragen.
print("Kurze Abschlussanalyse der Verkaufstabelle:")
print(f"Zeilen: {len(sales_data)}, Spalten: {sales_data.shape[1]}")
print(f"Mittlerer Umsatz: {mean_revenue:.0f} EUR")
print(f"Median-Umsatz: {median_revenue:.0f} EUR")
print(f"Höchster Umsatz: {max_revenue:.0f} EUR")
print(f"Fehlende Regionen: {missing_region_count} ({missing_region_share:.0%})")
print(f"Auffälligste Filiale nach Mittelwert: {top_branch} ({top_branch_mean:.0f} EUR)")

# Die letzte Einschätzung benennt die Verwendbarkeit der Daten.
if has_high_outlier or missing_region_count > 0:
    print("Einschätzung: nutzbar, aber Ausreißer und fehlende Regionen prüfen.")
else:
    print("Einschätzung: für einfache Auswertungen direkt gut nutzbar.")

```
