## **1. Daten laden**

### **1.1. Listen und Dictionaries**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_01_01.jpg?v=1787630300" width="250">



>* Listen sammeln kleine Daten geordnet im Speicher

>* Einheitliche Namen und Formate sichern Analysequalität

>* Kleine Datensätze zeigen Datenstrukturen sofort
>* Frühe Experimente decken Datenprobleme auf



```
#@title Python-Code - Listen und Dictionaries

# Wir laden kleine Daten direkt aus Python-Strukturen.
# Dictionaries beschreiben Beobachtungen mit klaren Spaltennamen.
# Am Ende entsteht eine übersichtliche Tabelle.

import pandas as pd

# Jede Dictionary-Zeile beschreibt ein Wohnungsangebot.
listings = [
    {"city": "Berlin", "rooms": 2, "area_m2": 54, "rent_eur": 980},
    {"city": "Hamburg", "rooms": 3, "area_m2": 72, "rent_eur": 1250},
    {"city": "Leipzig", "rooms": 1, "area_m2": 38, "rent_eur": 520},
]

# Pandas macht daraus eine tabellarische Datenstruktur.
listings_table = pd.DataFrame(listings)

# Eine einfache Prüfung schützt vor unerwarteten Spalten.
expected_columns = ["city", "rooms", "area_m2", "rent_eur"]
actual_columns = list(listings_table.columns)
columns_are_expected = actual_columns == expected_columns

# Diese Ausgabe zeigt Struktur und erste Daten kompakt.
print("Geladene Zeilen:", len(listings_table))
print("Spalten passen:", columns_are_expected)
print("Spaltennamen:", actual_columns)
print("Datentypen:", listings_table.dtypes.astype(str).to_dict())
print("Vorschau:")
print(listings_table.to_string(index=False))

```

### **1.2. CSV aus StringIO**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_01_02.jpg?v=1787630296" width="250">



>* CSV-Text direkt im Arbeitsspeicher laden
>* StringIO erleichtert Üben ohne externe Dateien

>* CSV-Struktur: Kopfzeile, Datenzeilen und Trennzeichen
>* StringIO zeigt typische Fehlerquellen kontrolliert

>* CSV-Laden bedeutet Daten richtig interpretieren
>* Frühe Entscheidungen beeinflussen spätere Prüfungen



```
#@title Python-Code - CSV aus StringIO

# Wir laden CSV-Text direkt aus dem Arbeitsspeicher.
# Pandas interpretiert den Text wie eine Datei.
# Am Ende sehen wir eine kleine Tabelle.

import pandas as pd

# Diese Zeichenkette enthält den Inhalt einer kleinen CSV-Datei.
csv_text = """customer_id;age;city;spend_eur
1;34;Berlin;120.50
2;28;Hamburg;89.90
3;;München;45.00
4;41;Köln;210.30
"""

# StringIO ist über pandas erreichbar und benötigt keinen Extra-Import.
csv_buffer = pd.io.common.StringIO(csv_text)

# Das Semikolon wird hier bewusst als Trennzeichen angegeben.
data = pd.read_csv(csv_buffer, sep=";")

# Eine einfache Prüfung bestätigt die erwartete Tabellenform.
expected_shape = (4, 4)
shape_is_valid = data.shape == expected_shape

# Fehlende Werte werden nach dem Laden als Tabelle erkennbar.
missing_values = int(data.isna().sum().sum())

print("Geladene Zeilen und Spalten:", data.shape)
print("Erwartete Form erfüllt:", shape_is_valid)
print("Fehlende Werte insgesamt:", missing_values)
print("Vorschau der geladenen CSV-Tabelle:")
print(data.head(4).to_string(index=False))

```

### **1.3. JSON als Tabelle**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_01_03.jpg?v=1787630298" width="250">



>* Strukturierte Objekte in Tabellen umwandeln
>* Zeilen und Spalten erleichtern Analysen

>* Beobachtungsebene vor der Tabellenbildung klären
>* Verschachtelte Daten fachlich sinnvoll strukturieren

>* Tabellen zeigen Inhalt und Qualität schnell
>* Rohdaten werden lesbar und zuverlässig nutzbar



```
#@title Python-Code - JSON als Tabelle

# Dieses Beispiel wandelt JSON-ähnliche Daten in Tabellen um.
# Verschachtelte Felder werden als eigene Spalten sichtbar.
# Am Ende sehen wir eine kompakte Prüfübersicht.

import pandas as pd

# Kleine JSON-ähnliche Daten stehen direkt im Arbeitsspeicher.
orders = [
    {"order_id": 101, "customer": {"city": "Berlin"}, "total_eur": 39.9},
    {"order_id": 102, "customer": {"city": "Hamburg"}, "total_eur": 24.5},
    {"order_id": 103, "customer": {"city": "Berlin"}, "total_eur": 58.0},
]

# pandas normalisiert verschachtelte Wörterbücher zu Tabellenspalten.
orders_table = pd.json_normalize(orders)

# Eine einfache Prüfung schützt vor unerwarteten Strukturen.
expected_columns = ["order_id", "customer.city", "total_eur"]

# Fehlende Spalten würden die spätere Analyse verfälschen.
missing_columns = []
for column in expected_columns:
    if column not in orders_table.columns:
        missing_columns.append(column)

# Die Ausgabe bleibt klein und zeigt nur zentrale Informationen.
print("Tabelle aus JSON-ähnlichen Daten:")
print(orders_table.to_string(index=False))
print("Zeilen und Spalten:", orders_table.shape)
print("Fehlende erwartete Spalten:", missing_columns)
print("Datentypen:", orders_table.dtypes.astype(str).to_dict())

```

## **2. Quellen dokumentieren**

### **2.1. Mehrere Dateien**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_02_01.jpg?v=1787630302" width="250">



>* Jede Datei als eigene Quelle erfassen
>* Herkunft, Zeitraum und Zweck dokumentieren

>* Dateiunterschiede können Qualitätsprüfungen verfälschen
>* Auffälligkeiten immer nach Herkunft prüfen

>* Dateiherkunft und Annahmen nachvollziehbar festhalten
>* Inkonsistenzen früh erkennen und Modellverzerrungen vermeiden



```
#@title Python-Code - Mehrere Dateien

# Dieses Beispiel prüft Daten aus mehreren Dateien.
# Die Herkunft jeder Zeile bleibt sichtbar.
# Ein kurzer Bericht zeigt typische Qualitätsprobleme.

import pandas as pd

# Drei kleine Tabellen stehen für drei Quelldateien.
file_tables = {
    "filiale_januar.csv": pd.DataFrame({
        "order_id": [101, 102, 103],
        "customer_id": ["C001", "C002", "C003"],
        "amount_eur": [29.9, 15.0, 120.0],
        "items": [1, 2, 4],
    }),
    "filiale_februar.csv": pd.DataFrame({
        "order_id": [103, 104, 105],
        "customer_id": ["C003", "C004", None],
        "amount_eur": [120.0, -5.0, 42.5],
        "items": [4, 1, 2],
    }),
    "filiale_maerz.csv": pd.DataFrame({
        "order_id": [106, 107],
        "customer_id": ["C006", "C007"],
        "amount_eur": [18.0, 9999.0],
        "item_count": [1, 3],
    }),
}

# Erwartete Spalten und Wertebereiche werden vorher festgelegt.
expected_columns = ["order_id", "customer_id", "amount_eur", "items"]
expected_types = {"order_id": "int64", "amount_eur": "float64", "items": "int64"}
valid_ranges = {"amount_eur": (0, 1000), "items": (1, 20)}

# Jede Tabelle bekommt eine Spalte mit ihrer Herkunft.
combined_parts = []
for file_name, table in file_tables.items():
    table_with_source = table.copy()
    table_with_source["source_file"] = file_name
    combined_parts.append(table_with_source)

# Beim Zusammenführen bleiben fehlende Spalten als NaN sichtbar.
combined = pd.concat(combined_parts, ignore_index=True, sort=False)

# Der Bericht sammelt pro Datei kompakte Qualitätskennzahlen.
report_rows = []
for file_name, table in file_tables.items():
    missing_columns = sorted(set(expected_columns) - set(table.columns))
    extra_columns = sorted(set(table.columns) - set(expected_columns))
    missing_values = int(table.isna().sum().sum())
    duplicate_orders = int(table.duplicated(subset=["order_id"]).sum())

    range_problems = 0
    for column, limits in valid_ranges.items():
        if column in table.columns:
            low, high = limits
            range_problems += int((~table[column].between(low, high)).sum())

    report_rows.append({
        "Datei": file_name,
        "fehlende Spalten": len(missing_columns),
        "Zusatzspalten": len(extra_columns),
        "fehlende Werte": missing_values,
        "Wertebereich": range_problems,
    })

# Duplikate über alle Dateien zeigen wiederholte Bestellungen.
cross_file_duplicates = int(combined.duplicated(subset=["order_id"]).sum())
report = pd.DataFrame(report_rows)

# Eine einfache Annahme schützt vor unerwartet leeren Quellen.
if combined.shape[0] == 0:
    raise ValueError("Der zusammengeführte Datensatz ist leer.")

print("Validierungsbericht für mehrere Dateien:")
print(report.to_string(index=False))
print(f"Zeilen gesamt: {combined.shape[0]}")
print(f"Duplikate über Dateien hinweg: {cross_file_duplicates}")
print("Herkunftsspalte vorhanden: source_file")

```

### **2.3. Offline Ersatz**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_02_03.jpg?v=1787630304" width="250">



>* Lokale Ersatzdaten sichern reproduzierbare Qualitätsprüfungen
>* Alle prüfen dieselben bekannten Datenprobleme

>* Relevante Struktur und typische Fehler abbilden
>* Vertrauliche Daten anonymisieren oder synthetisch ersetzen

>* Ersatzdaten klar dokumentieren und einordnen
>* Version, Zweck und Prüfbarkeit transparent machen



```
#@title Python-Code - Offline Ersatz

# Wir prüfen einen lokalen Offline Ersatz.
# Qualitätsregeln machen Datenprobleme sichtbar.
# Am Ende entsteht ein kompakter Bericht.

import pandas as pd

# Diese kleine Tabelle ersetzt eine externe Quelle.
data = {
    "customer_id": [101, 102, 102, 104, 105],
    "region": ["Nord", "Süd", "Süd", None, "West"],
    "revenue_eur": [120.0, 85.5, 85.5, -10.0, 9999.0],
    "signup_year": [2021, 2020, 2020, 2019, 2035],
}

# Aus den Listen entsteht ein reproduzierbarer DataFrame.
df = pd.DataFrame(data)

# Erwartete Spalten und Typen dokumentieren die Quelle.
expected_columns = ["customer_id", "region", "revenue_eur", "signup_year"]
expected_types = {"customer_id": "int64", "revenue_eur": "float64"}

# Jede Prüfung liefert eine kurze, verständliche Kennzahl.
missing_columns = sorted(set(expected_columns) - set(df.columns))
duplicate_ids = int(df["customer_id"].duplicated().sum())
missing_values = int(df.isna().sum().sum())

# Wertebereiche zeigen fachlich unplausible Einträge.
invalid_revenue = int(((df["revenue_eur"] < 0) | (df["revenue_eur"] > 5000)).sum())
invalid_year = int(((df["signup_year"] < 2018) | (df["signup_year"] > 2024)).sum())

# Datentypen werden mit der dokumentierten Erwartung verglichen.
type_problems = []
for column, expected_type in expected_types.items():
    actual_type = str(df[column].dtype)
    if actual_type != expected_type:
        type_problems.append(column)

# Der Bericht fasst die wichtigsten Qualitätsrisiken zusammen.
report = pd.DataFrame(
    {
        "Prüfung": ["Spalten", "Datentypen", "Fehlende Werte", "Duplikate", "Wertebereiche"],
        "Problemzahl": [len(missing_columns), len(type_problems), missing_values, duplicate_ids, invalid_revenue + invalid_year],
    }
)

print("Offline Ersatz: synthetische Kundendaten, 5 Zeilen.")
print(report.to_string(index=False))
print("Fazit: Der Ersatz enthält absichtlich typische Qualitätsprobleme.")

```

## **3. Kompakter Validierungsbericht**

### **3.1. Spalten und Typen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_03_01.jpg?v=1787630308" width="250">



>* Spalten und Datentypen zuerst erfassen
>* Typen mit fachlicher Bedeutung abgleichen

>* Merkmalstypen gezielt unterscheiden und behandeln
>* Fehlklassifikationen im Bericht sichtbar machen

>* Typfehler früh erkennen und vermeiden
>* Bereinigungsschritte für Modellierung festhalten



```
#@title Python-Code - Spalten und Typen

# Dieser Bericht prüft Spalten und Datentypen.
# Fachliche Rollen werden neben technischen Typen sichtbar.
# Das Ergebnis zeigt konkrete Hinweise zur ML-Vorbereitung.

import pandas as pd

# Ein kleiner Datensatz enthält absichtlich typische Typ-Probleme.
data = {
    "customer_id": [101, 102, 103, 103],
    "age": [34, 28, None, 45],
    "price_eur": ["19.99", "29.50", "kostenlos", "41.00"],
    "signup_date": ["2024-01-05", "05.02.2024", "2024-03-10", "unbekannt"],
    "region": ["Nord", "Süd", "Nord", "West"],
}

df = pd.DataFrame(data)

# Erwartete Rollen helfen beim fachlichen Vergleich.
expected_roles = {
    "customer_id": "Kategorie",
    "age": "Numerisch",
    "price_eur": "Numerisch",
    "signup_date": "Zeit",
    "region": "Kategorie",
}

# Diese Funktion übersetzt pandas-Typen in einfache Rollen.
def detected_role(series):
    if pd.api.types.is_numeric_dtype(series):
        return "Numerisch"
    if pd.api.types.is_datetime64_any_dtype(series):
        return "Zeit"
    return "Text oder Kategorie"

# Für jede Spalte entsteht eine kompakte Berichtszeile.
report_rows = []
for column in df.columns:
    role = detected_role(df[column])
    expected = expected_roles[column]
    status = "prüfen" if role != expected else "ok"
    report_rows.append([column, str(df[column].dtype), expected, role, status])

report = pd.DataFrame(
    report_rows,
    columns=["Spalte", "pandas-Typ", "Erwartet", "Erkannt", "Status"],
)

# Zwei gezielte Umwandlungen zeigen typische Validierungsschritte.
price_numeric = pd.to_numeric(df["price_eur"], errors="coerce")
signup_dates = pd.to_datetime(df["signup_date"], errors="coerce", dayfirst=True)

# Kurze Kennzahlen machen die Typ-Probleme messbar.
price_invalid = int(price_numeric.isna().sum() - df["price_eur"].isna().sum())
date_invalid = int(signup_dates.isna().sum() - df["signup_date"].isna().sum())
duplicate_rows = int(df.duplicated().sum())

print("Kompakter Bericht zu Spalten und Typen:")
print(report.to_string(index=False))
print(f"Ungültige Preiswerte nach Umwandlung: {price_invalid}")
print(f"Ungültige Datumswerte nach Umwandlung: {date_invalid}")
print(f"Vollständige Duplikatzeilen: {duplicate_rows}")
print("Fazit: Spalten mit Status 'prüfen' vor dem Modellieren bereinigen.")

```

### **3.2. Fehlwerte und Wertebereiche**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_03_02.jpg?v=1787630311" width="250">



>* Fehlwerte verstehen, nicht vorschnell verwerfen
>* Häufigkeit, Spalten und Muster dokumentieren

>* Plausible Wertebereiche und Ausreißer erkennen
>* Kategorien, Schreibweisen und seltene Werte prüfen

>* Fehlwerte und Wertebereiche gemeinsam prüfen
>* Risiken erkennen und nächste Schritte planen



```
#@title Python-Code - Fehlwerte und Wertebereiche

# Wir prüfen Fehlwerte und plausible Wertebereiche.
# Ein kleiner Bericht zeigt wichtige Datenrisiken.
# Die Ausgabe unterstützt die ML-Vorbereitung.

import pandas as pd

# Dieser Beispieldatensatz enthält typische Qualitätsprobleme.
data = pd.DataFrame(
    {
        "age_years": [34, 52, None, -3, 120, 41],
        "income_eur": [42000, None, 38000, 51000, 999999, 45000],
        "country": ["DE", "DE", "AT", None, "deutschland", "CH"],
    }
)

# Diese Regeln beschreiben einfache fachliche Erwartungsbereiche.
range_rules = {
    "age_years": (0, 100),
    "income_eur": (0, 200000),
}

# Fehlwerte werden pro Spalte gezählt und prozentual eingeordnet.
missing_count = data.isna().sum()
missing_percent = (missing_count / len(data) * 100).round(1)

# Numerische Werte außerhalb der Regeln werden gezählt.
out_of_range = {}
for column, limits in range_rules.items():
    lower_limit, upper_limit = limits
    valid_values = data[column].dropna()
    outside = (valid_values < lower_limit) | (valid_values > upper_limit)
    out_of_range[column] = int(outside.sum())

# Der Bericht verbindet Fehlwerte und Wertebereichsprobleme kompakt.
report = pd.DataFrame(
    {
        "fehlwerte": missing_count,
        "fehlwerte_prozent": missing_percent,
        "ausserhalb_bereich": pd.Series(out_of_range),
    }
).fillna(0)

# Für Anfängerinnen und Anfänger zeigen wir nur zentrale Kennzahlen.
report["ausserhalb_bereich"] = report["ausserhalb_bereich"].astype(int)
print("Kompakter Validierungsbericht:")
print(report.to_string())

# Diese Zeile macht die nächste Entscheidung sichtbar.
problem_columns = report.query("fehlwerte > 0 or ausserhalb_bereich > 0").index.tolist()
print("Zu prüfen vor dem Training:", ", ".join(problem_columns))

```

### **3.3. Kompakter Bericht**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_B/image_03_03.jpg?v=1787630309" width="250">



>* Datenqualität schnell und verständlich zusammenfassen
>* Probleme für ML früh sichtbar machen

>* Datenprobleme können Modellleistung früh verzerren
>* Befunde fachlich prüfen, nicht vorschnell bereinigen

>* Risiken priorisieren und nächste Schritte ableiten
>* Unsicherheiten dokumentieren für reproduzierbare Entscheidungen



```
#@title Python-Code - Kompakter Bericht

# Wir erstellen einen kompakten Validierungsbericht.
# Der Bericht prüft typische Datenqualitätsrisiken.
# Am Ende entsteht eine klare ML-Einschätzung.

import pandas as pd

# Dieser kleine Datensatz enthält absichtlich mehrere Qualitätsprobleme.
data = pd.DataFrame(
    {
        "order_id": [101, 102, 103, 103, 105, 106],
        "region": ["Nord", "Süd", "Süd", "Süd", None, "West"],
        "delivery_days": [2, 5, -1, -1, 40, 3],
        "target_delayed": [0, 1, 0, 0, 1, None],
    }
)

# Diese Regeln beschreiben einfache Erwartungen vor dem Modelltraining.
expected_ranges = {
    "delivery_days": (0, 30),
    "target_delayed": (0, 1),
}

# Grunddaten zeigen, wie groß der Datensatz ist.
row_count = len(data)
column_count = len(data.columns)

# Fehlende Werte werden pro Spalte gezählt.
missing_counts = data.isna().sum()
missing_columns = missing_counts[missing_counts > 0]

# Duplikate können Trainingsdaten verzerren.
duplicate_count = data.duplicated().sum()

# Wertebereiche werden gegen einfache Fachregeln geprüft.
range_issues = []
for column, limits in expected_ranges.items():
    low, high = limits
    outside = data[column].dropna().between(low, high).eq(False).sum()
    if outside > 0:
        range_issues.append(f"{column}: {outside} auffällige Werte")

# Datentypen helfen beim Erkennen falscher Spalteninterpretationen.
type_summary = data.dtypes.astype(str).to_dict()

# Eine einfache Ampel priorisiert die nächsten Schritte.
risk_score = len(missing_columns) + duplicate_count + len(range_issues)
if risk_score == 0:
    decision = "bereit für erste ML-Vorbereitung"
elif risk_score <= 3:
    decision = "erst gezielt prüfen und bereinigen"
else:
    decision = "vor Modelltraining gründlich klären"

print("Kompakter Validierungsbericht")
print(f"Größe: {row_count} Zeilen, {column_count} Spalten")
print(f"Datentypen: {type_summary}")
print(f"Fehlende Werte: {missing_columns.to_dict()}")
print(f"Doppelte Zeilen: {duplicate_count}")
print(f"Wertebereich-Probleme: {range_issues}")
print(f"Einschätzung: {decision}")

```
