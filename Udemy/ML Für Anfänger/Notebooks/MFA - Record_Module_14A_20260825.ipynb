## **1. Neuronen und Schichten**

### **1.1. Das künstliche Neuron**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_01_01.jpg?v=1787653406" width="250">



>* Neuron gewichtet Eingaben und erzeugt Ausgaben
>* Es lernt wichtige Muster aus Daten

>* Gewichte bestimmen den Einfluss jeder Eingabe
>* Bias verschiebt das entstehende Neuronensignal

>* NumPy berechnet Neuronen effizient mit Vektoren
>* Viele Neuronen erkennen komplexe Muster



```
#@title Python-Code - Das künstliche Neuron

# Dieses Beispiel berechnet ein einzelnes künstliches Neuron.
# Eingaben, Gewichte und Bias ergeben ein Signal.
# Die Sigmoid-Aktivierung macht daraus eine verständliche Ausgabe.

import numpy as np
import matplotlib.pyplot as plt

# Kleine Vektoren machen die Rechnung gut nachvollziehbar.
inputs = np.array([0.8, 0.4, 0.9])
weights = np.array([1.2, -0.7, 0.5])
bias = -0.3

# Diese Prüfung verhindert unpassende Vektorlängen.
if inputs.shape != weights.shape:
    raise ValueError("Eingaben und Gewichte müssen gleich lang sein.")

# Das Neuron bildet zuerst die gewichtete Summe.
weighted_parts = inputs * weights
linear_signal = np.dot(inputs, weights) + bias

# Die Sigmoid-Funktion begrenzt die Ausgabe zwischen null und eins.
activation = 1 / (1 + np.exp(-linear_signal))

print("Eingaben:", np.round(inputs, 2))
print("Gewichte:", np.round(weights, 2))
print("Gewichtete Beiträge:", np.round(weighted_parts, 2))
print("Bias:", round(bias, 2))
print("Lineares Signal:", round(linear_signal, 3))
print("Neuron-Ausgabe nach Sigmoid:", round(activation, 3))

# Die Balken zeigen, welche Merkmale stark beitragen.
feature_names = ["Wetter", "Entfernung", "Motivation"]
colors = ["tab:green" if value >= 0 else "tab:red" for value in weighted_parts]

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(feature_names, weighted_parts, color=colors)
ax.axhline(0, color="black", linewidth=1)
ax.set_title("Beiträge der Eingaben zu einem künstlichen Neuron")

ax.set_xlabel("Eingabemerkmal")
ax.set_ylabel("Eingabe mal Gewicht")
plt.show()

```

### **1.2. Dense Schichten**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_01_02.jpg?v=1787653408" width="250">



>* Mehrere Neuronen verarbeiten dieselben Eingaben parallel
>* Jedes Neuron ist mit allen Merkmalen verbunden

>* Dense-Schichten als effiziente Matrixoperationen berechnen
>* Batches, Gewichte und Bias formen Ausgaben

>* Lernen verborgene Beziehungen zwischen Merkmalen
>* Viele Parameter brauchen gute Trainingsdaten



```
#@title Python-Code - Dense Schichten

# Diese Demo berechnet eine kleine Dense-Schicht.
# Gewichte verbinden jedes Merkmal mit jedem Neuron.
# Die Ausgabe zeigt Formen und Aktivierungen.

import numpy as np

# Drei Beispiele mit jeweils zwei Eingabemerkmalen.
inputs = np.array([[1.0, 2.0], [0.5, -1.0], [3.0, 0.0]])

# Jede Spalte enthält die Gewichte eines Ausgabeneurons.
weights = np.array([[0.2, -0.4, 0.1], [0.7, 0.3, -0.5]])

# Jeder Bias gehört zu genau einem Ausgabeneuron.
biases = np.array([0.1, -0.2, 0.05])

# Die Matrixmultiplikation berechnet alle Neuronen gleichzeitig.
linear_output = inputs @ weights + biases

# ReLU lässt positive Werte stehen und setzt negative auf null.
activated_output = np.maximum(0.0, linear_output)

# Diese Prüfung macht die erwarteten Formen explizit.
if linear_output.shape != (3, 3):
    raise ValueError("Die Ausgabeform sollte drei Beispiele und drei Neuronen haben.")

print("Eingabeform:", inputs.shape)
print("Gewichtsform:", weights.shape)
print("Ausgabeform:", linear_output.shape)
print("Lineare Ausgabe, erstes Beispiel:", np.round(linear_output[0], 2))
print("Nach ReLU, erstes Beispiel:", np.round(activated_output[0], 2))

```

### **1.3. Aktivierungsfunktionen verstehen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_01_03.jpg?v=1787653410" width="250">



>* Aktivierungen machen lineare Neuronen flexibel
>* Nichtlinearität ermöglicht komplexe Mustererkennung

>* Aktivierungen steuern Reaktionen auf Eingangssignale
>* Ihre Form beeinflusst Lernen und Gradienten

>* Aktivierungen verarbeiten ganze Schichten effizient.
>* Sie ermöglichen abstrakte, nichtlineare Muster.



```
#@title Python-Code - Aktivierungsfunktionen verstehen

# Dieses Beispiel vergleicht wichtige Aktivierungsfunktionen.
# NumPy berechnet die Funktionen elementweise.
# Der Plot zeigt unterschiedliche Reaktionsregeln.

import numpy as np
import matplotlib.pyplot as plt

# Diese Eingabewerte stehen für Neuronensignale vor der Aktivierung.
x_values = np.linspace(-4.0, 4.0, 200)

# ReLU lässt positive Werte durch und blockiert negative Werte.
relu_values = np.maximum(0.0, x_values)

# Sigmoid drückt Werte in den Bereich zwischen null und eins.
sigmoid_values = 1.0 / (1.0 + np.exp(-x_values))

# Tanh drückt Werte symmetrisch in den Bereich minus eins bis eins.
tanh_values = np.tanh(x_values)

# Eine kleine Prüfung macht die erwarteten Formen sichtbar.
if x_values.shape != relu_values.shape:
    raise ValueError("Die Aktivierung muss die Array-Form erhalten.")

print("Eingabeform:", x_values.shape)
print("ReLU bei -2, 0, 2:", np.round(np.maximum(0.0, [-2, 0, 2]), 2))
print("Sigmoid bei -2, 0, 2:", np.round(1.0 / (1.0 + np.exp([2, 0, -2])), 2))
print("Tanh bei -2, 0, 2:", np.round(np.tanh([-2, 0, 2]), 2))

# Ein gemeinsamer Plot zeigt die verschiedenen Reaktionsregeln.
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_values, relu_values, label="ReLU")
ax.plot(x_values, sigmoid_values, label="Sigmoid")
ax.plot(x_values, tanh_values, label="Tanh")

ax.set_title("Aktivierungsfunktionen im Vergleich")
ax.set_xlabel("Rohwert vor der Aktivierung")
ax.set_ylabel("Ausgabe nach der Aktivierung")
ax.legend()

plt.show()

```

## **2. Vorwärtsrechnung**

### **2.1. Stabile Softmax Berechnung**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_02_01.jpg?v=1787653414" width="250">



>* Softmax macht Rohwerte zu Klassenwahrscheinlichkeiten
>* Relative Abstände bestimmen die Wahrscheinlichkeitsverteilung

>* Extreme Werte können Softmax numerisch überfordern
>* Stabile Berechnung erhält sinnvolle Wahrscheinlichkeiten

>* Maximalwert abziehen, Abstände bleiben erhalten
>* Stabilere Wahrscheinlichkeiten für zuverlässige Vorwärtsrechnung



```
#@title Python-Code - Stabile Softmax Berechnung

# Wir vergleichen direkte und stabile Softmax-Berechnung.
# Große Rohwerte zeigen das numerische Problem.
# Die stabile Variante liefert gültige Wahrscheinlichkeiten.

import numpy as np
import matplotlib.pyplot as plt

# Diese Rohwerte könnten aus einer letzten Dense-Schicht stammen.
logits = np.array([[1000.0, 1001.0, 1002.0]])
class_names = ["Katze", "Hund", "Fahrrad"]

# Die direkte Berechnung kann bei großen Zahlen überlaufen.
direct_exp = np.exp(logits)
direct_softmax = direct_exp / np.sum(direct_exp, axis=1, keepdims=True)

# Die stabile Berechnung verschiebt jede Zeile um ihr Maximum.
shifted_logits = logits - np.max(logits, axis=1, keepdims=True)
stable_exp = np.exp(shifted_logits)
stable_softmax = stable_exp / np.sum(stable_exp, axis=1, keepdims=True)

# Diese Prüfung bestätigt die erwartete Form der Ausgabe.
if stable_softmax.shape != logits.shape:
    raise ValueError("Die Softmax-Ausgabe muss die gleiche Form haben.")

# Die Summe jeder Softmax-Zeile sollte eins ergeben.
probability_sum = np.sum(stable_softmax, axis=1)[0]
print("Rohwerte:", logits[0].astype(int).tolist())
print("Direkte Softmax enthält gültige Zahlen:", bool(np.all(np.isfinite(direct_softmax))))
print("Verschobene Rohwerte:", shifted_logits[0].astype(int).tolist())
print("Stabile Softmax:", np.round(stable_softmax[0], 4).tolist())
print("Summe der Wahrscheinlichkeiten:", round(float(probability_sum), 4))

# Das Balkendiagramm zeigt die stabile Wahrscheinlichkeitsverteilung.
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(class_names, stable_softmax[0], color=["#4C78A8", "#F58518", "#54A24B"])
ax.set_title("Stabile Softmax aus großen Rohwerten")
ax.set_xlabel("Klasse")
ax.set_ylabel("Wahrscheinlichkeit")
ax.set_ylim(0, 1)
plt.show()

```

### **2.2. Zwei Schichten**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_02_02.jpg?v=1787653416" width="250">



>* Erste Schicht bildet nützliche Zwischenmerkmale
>* Zweite Schicht erzeugt daraus Vorhersagen

>* Dense-Schichten kombinieren alle vorherigen Ausgaben
>* Aktivierungen ermöglichen komplexe nichtlineare Muster

>* Dimensionen und Stapel sorgfältig verfolgen
>* Feste Parameter ergeben eindeutige Vorhersagen



```
#@title Python-Code - Zwei Schichten

# Dieses Beispiel zeigt zwei Dense-Schichten.
# Softmax macht Ausgaben zu Klassenwahrscheinlichkeiten.
# Der Verlust bewertet die richtige Klasse.

import numpy as np

# Kleine Eingaben stehen für drei Datenbeispiele.
inputs = np.array([[1.0, 0.5], [0.2, 1.0], [1.2, 0.1]])
labels = np.array([0, 1, 0])

# Die erste Schicht erzeugt drei Zwischenmerkmale.
weights_1 = np.array([[0.8, -0.4, 0.3], [0.2, 0.9, -0.5]])
bias_1 = np.array([0.1, -0.2, 0.0])

# Die zweite Schicht erzeugt zwei Klassenwerte.
weights_2 = np.array([[0.7, -0.6], [-0.3, 0.8], [0.5, 0.2]])
bias_2 = np.array([0.0, 0.1])

# Diese Prüfung verhindert unpassende Matrixformen.
if inputs.shape[1] != weights_1.shape[0]:
    raise ValueError("Eingaben passen nicht zur ersten Schicht.")

# ReLU lässt positive Zwischenwerte unverändert.
hidden_linear = inputs @ weights_1 + bias_1
hidden_activation = np.maximum(0.0, hidden_linear)

# Die zweite Schicht berechnet rohe Klassenwerte.
logits = hidden_activation @ weights_2 + bias_2
shifted_logits = logits - np.max(logits, axis=1, keepdims=True)

# Softmax wandelt Klassenwerte in Wahrscheinlichkeiten um.
exp_scores = np.exp(shifted_logits)
probabilities = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

# Kreuzentropie misst die Wahrscheinlichkeit der richtigen Klasse.
correct_probabilities = probabilities[np.arange(labels.size), labels]
loss = -np.mean(np.log(correct_probabilities))

# Die Ausgabe zeigt Formen und ein konkretes Ergebnis.
print("Eingabeform:", inputs.shape)
print("Nach Schicht 1:", hidden_activation.shape)
print("Nach Schicht 2:", logits.shape)
print("Softmax Beispiel 1:", np.round(probabilities[0], 3))
print("Mittlerer Kreuzentropie-Verlust:", round(float(loss), 3))

```

### **2.3. Verluste berechnen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_02_03.jpg?v=1787653412" width="250">



>* Verlust misst Abweichung von der Zielantwort
>* Er steuert spätere Verbesserungen des Netzes

>* Softmax-Verlust bewertet die richtige Klassenwahrscheinlichkeit
>* Starke Fehlzuordnung erhöht den gemittelten Verlust

>* Einzelverluste werden zum Stapelmittelwert zusammengefasst
>* Durchschnittsverlust zeigt nicht alle Problemfälle



```
#@title Python-Code - Verluste berechnen

# Dieses Beispiel berechnet Klassifikationsverluste mit Softmax.
# Wir betrachten richtige Klassen und ihre Wahrscheinlichkeiten.
# Am Ende vergleichen wir Einzelverluste und Mittelwert.

import numpy as np
import matplotlib.pyplot as plt

# Diese Rohwerte könnten aus der letzten Netzschicht stammen.
logits = np.array(
    [[2.0, 1.0, 0.1], [0.2, 2.5, 0.3], [1.8, 0.4, 2.2]]
)

# Jede Zeile gehört zu einem Beispiel im Stapel.
true_classes = np.array([0, 1, 0])

# Die Formprüfung verhindert unpassende Klassenindizes.
if logits.shape[0] != true_classes.size:
    raise ValueError("Jedes Beispiel braucht genau eine richtige Klasse.")

# Softmax wird stabiler durch Abziehen des Zeilenmaximums.
shifted_logits = logits - np.max(logits, axis=1, keepdims=True)
exp_scores = np.exp(shifted_logits)
probabilities = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)

# Wir lesen die Wahrscheinlichkeit der richtigen Klasse aus.
row_indices = np.arange(true_classes.size)
correct_probabilities = probabilities[row_indices, true_classes]

# Ein kleines Epsilon schützt den Logarithmus vor null.
epsilon = 1e-12
sample_losses = -np.log(correct_probabilities + epsilon)
mean_loss = np.mean(sample_losses)

# Kurze Ausgaben zeigen den Zusammenhang direkt.
print("Wahrscheinlichkeiten der richtigen Klassen:", np.round(correct_probabilities, 3))
print("Einzelverluste:", np.round(sample_losses, 3))
print("Mittlerer Verlust:", round(float(mean_loss), 3))

# Das Balkendiagramm macht hohe und niedrige Verluste sichtbar.
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(["Beispiel 1", "Beispiel 2", "Beispiel 3"], sample_losses)
ax.axhline(mean_loss, color="red", linestyle="--", label="Mittelwert")
ax.set_title("Kreuzentropie-Verlust pro Beispiel")

ax.set_xlabel("Beispiel im Stapel")
ax.set_ylabel("Verlust")
ax.legend()
plt.tight_layout()

plt.show()

```

## **3. Backpropagation mit NumPy**

### **3.1. Kettenregel anwenden**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_03_01.jpg?v=1787653418" width="250">



>* Kettenregel verfolgt Fehlereinfluss rückwärts.
>* Dense-Gradienten entstehen aus Zwischenergebnissen und Fehlersignal.

>* Fehlersignal zeigt Empfindlichkeit der Dense-Schicht
>* Kettenregel bestimmt sinnvolle Parameteränderungen

>* Array-Formen und Mini-Batch-Achsen beachten
>* Gradienten zeigen lokale Empfindlichkeiten



```
#@title Python-Code - Kettenregel anwenden

# Wir berechnen Dense-Gradienten mit der Kettenregel.
# Ein Zahlenbeispiel zeigt lokale Ableitungen.
# Numerische Prüfung bestätigt die Formeln.

import numpy as np

# Kleine Batch-Daten machen alle Formen gut sichtbar.
inputs = np.array([[1.0, -2.0], [0.5, 1.5], [-1.0, 2.0]])
weights = np.array([[0.2, -0.4], [0.7, 0.1]])
bias = np.array([0.05, -0.2])

# Zielwerte definieren einen einfachen quadratischen Verlust.
targets = np.array([[0.0, 1.0], [1.0, 0.0], [-1.0, 0.5]])
batch_size = inputs.shape[0]

# Vorwärtsrechnung einer Dense-Schicht.
outputs = inputs @ weights + bias
loss = np.mean((outputs - targets) ** 2)

# Das zurückkommende Signal ist die lokale Verlustableitung.
grad_outputs = 2.0 * (outputs - targets) / outputs.size
grad_weights = inputs.T @ grad_outputs
grad_bias = np.sum(grad_outputs, axis=0)
grad_inputs = grad_outputs @ weights.T

# Eine kleine Störung prüft einen einzelnen Gewichtsgradienten.
epsilon = 1e-5
weights_plus = weights.copy()
weights_minus = weights.copy()

# Wir verändern nur Gewicht Zeile null, Spalte eins.
weights_plus[0, 1] += epsilon
weights_minus[0, 1] -= epsilon

# Der Differenzenquotient nähert denselben Gradienten an.
loss_plus = np.mean((inputs @ weights_plus + bias - targets) ** 2)
loss_minus = np.mean((inputs @ weights_minus + bias - targets) ** 2)
numeric_gradient = (loss_plus - loss_minus) / (2.0 * epsilon)

# Die Ausgabe vergleicht analytische und numerische Werte.
print("Verlust:", round(float(loss), 4))
print("Form grad_weights:", grad_weights.shape)
print("Form grad_bias:", grad_bias.shape)
print("Form grad_inputs:", grad_inputs.shape)
print("Analytischer Gradient W[0,1]:", round(float(grad_weights[0, 1]), 6))
print("Numerischer Gradient W[0,1]:", round(float(numeric_gradient), 6))
print("Abweichung:", round(float(abs(grad_weights[0, 1] - numeric_gradient)), 10))

```

### **3.2. Gradienten numerisch prüfen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_14/Lecture_A/image_03_02.jpg?v=1787653419" width="250">



>* Numerische Prüfung vergleicht geschätzte und berechnete Gradienten
>* Sie findet typische NumPy-Implementierungsfehler

>* Einzelne Parameter minimal verändern und Verlust vergleichen
>* Nützlich für kleine Dense-Schicht-Tests

>* Kleine Abweichungen sind normal, große verdächtig
>* Früh prüfen, später wegen Langsamkeit vermeiden



```
#@title Python-Code - Gradienten numerisch prüfen

# Wir prüfen Gradienten einer kleinen Dense-Schicht.
# Numerische Steigungen vergleichen wir mit Backpropagation.
# Kleine Abweichungen bestätigen die Ableitungen.

import numpy as np
import matplotlib.pyplot as plt

# Feste Werte machen das Beispiel vollständig reproduzierbar.
inputs = np.array([[0.2, -0.4], [0.7, 0.1], [-0.3, 0.5]])
targets = np.array([[0.1, -0.2], [0.4, 0.0], [-0.1, 0.3]])

# Diese Parameter gehören zu einer Dense-Schicht.
weights = np.array([[0.3, -0.2], [0.5, 0.1]])
biases = np.array([[0.05, -0.03]])

# Eine kleine Prüfung verhindert unklare Formfehler.
if inputs.shape[1] != weights.shape[0]:
    raise ValueError("Eingaben und Gewichte passen nicht zusammen.")

# Der Verlust ist der mittlere quadratische Fehler.
def loss_for(current_weights, current_biases):
    predictions = inputs @ current_weights + current_biases
    squared_errors = (predictions - targets) ** 2
    return np.mean(squared_errors)

# Vorwärtsrechnung und Rückwärtsrechnung liefern analytische Gradienten.
predictions = inputs @ weights + biases
loss = loss_for(weights, biases)
loss_gradient = 2 * (predictions - targets) / predictions.size

# Die Dense-Schicht nutzt Matrixmultiplikation rückwärts.
analytic_weight_gradients = inputs.T @ loss_gradient
analytic_bias_gradients = np.sum(loss_gradient, axis=0, keepdims=True)

# Zentrale Differenzen schätzen einzelne Gradienten unabhängig.
epsilon = 1e-5
numeric_weight_gradients = np.zeros_like(weights)
numeric_bias_gradients = np.zeros_like(biases)

# Jedes Gewicht wird einzeln minimal erhöht und verringert.
for row in range(weights.shape[0]):
    for col in range(weights.shape[1]):
        changed_weights = weights.copy()
        changed_weights[row, col] += epsilon
        loss_plus = loss_for(changed_weights, biases)
        changed_weights[row, col] -= 2 * epsilon
        loss_minus = loss_for(changed_weights, biases)
        numeric_weight_gradients[row, col] = (loss_plus - loss_minus) / (2 * epsilon)

# Auch jeder Bias wird einzeln numerisch geprüft.
for col in range(biases.shape[1]):
    changed_biases = biases.copy()
    changed_biases[0, col] += epsilon
    loss_plus = loss_for(weights, changed_biases)
    changed_biases[0, col] -= 2 * epsilon
    loss_minus = loss_for(weights, changed_biases)
    numeric_bias_gradients[0, col] = (loss_plus - loss_minus) / (2 * epsilon)

# Die größte Abweichung zeigt die Qualität der Prüfung.
weight_difference = np.max(np.abs(analytic_weight_gradients - numeric_weight_gradients))
bias_difference = np.max(np.abs(analytic_bias_gradients - numeric_bias_gradients))

print(f"Verlust vor der Prüfung: {loss:.6f}")
print(f"Maximale Gewichts-Abweichung: {weight_difference:.10f}")
print(f"Maximale Bias-Abweichung: {bias_difference:.10f}")
print("Sehr kleine Abweichungen sprechen für korrekte Backpropagation.")

# Das Diagramm vergleicht alle geprüften Parameter direkt.
analytic_values = np.concatenate(
    [analytic_weight_gradients.ravel(), analytic_bias_gradients.ravel()]
)
numeric_values = np.concatenate(
    [numeric_weight_gradients.ravel(), numeric_bias_gradients.ravel()]
)

# Punkte nahe der Diagonale bedeuten gute Übereinstimmung.
fig, ax = plt.subplots(figsize=(5, 4))
ax.scatter(analytic_values, numeric_values, color="tab:blue", label="Parameter")

# Die Diagonale markiert perfekte Übereinstimmung.
line_min = min(analytic_values.min(), numeric_values.min())
line_max = max(analytic_values.max(), numeric_values.max())
ax.plot([line_min, line_max], [line_min, line_max], color="tab:orange", label="ideal")

ax.set_title("Analytische und numerische Gradienten")
ax.set_xlabel("Analytischer Gradient")
ax.set_ylabel("Numerischer Gradient")
ax.legend()
plt.show()

```

### **3.3. Ein Trainingsschritt**

>* Vorwärts rechnen, Verlust messen, rückwärts ableiten

>* Gewichts- und Bias-Gradienten aus Fehlersignalen

>* Parameter mit passender Lernrate aktualisieren
>* Verlust, Gradienten und Änderungen regelmäßig prüfen



```
#@title Python-Code - Ein Trainingsschritt

# Dieser Trainingsschritt zeigt Gradienten einer Dense-Schicht.
# Vorwärtsrechnung und Rückwärtsrechnung werden direkt verbunden.
# Die Ausgabe vergleicht analytische und numerische Gradienten.

import numpy as np

# Kleine feste Daten machen alle Rechnungen nachvollziehbar.
inputs = np.array([[1.0, -2.0], [0.5, 1.0], [-1.5, 2.0]])
targets = np.array([[0.2], [0.8], [-0.4]])

# Die Dense-Schicht besitzt Gewichte und einen Bias.
weights = np.array([[0.10], [-0.20]])
bias = np.array([[0.05]])
learning_rate = 0.1

# Diese Prüfung verhindert unpassende Matrixformen.
if inputs.shape[1] != weights.shape[0]:
    raise ValueError("Eingaben und Gewichte passen nicht zusammen.")

# Vorwärtsrechnung: lineare Dense-Ausgabe und mittlerer quadratischer Fehler.
predictions = inputs @ weights + bias
errors = predictions - targets
loss = np.mean(errors ** 2)

# Rückwärtsrechnung: Gradient des Verlusts nach der Dense-Ausgabe.
batch_size = inputs.shape[0]
grad_output = 2.0 * errors / batch_size

# Dense-Gradienten entstehen aus Eingaben und rückwärts kommendem Signal.
grad_weights = inputs.T @ grad_output
grad_bias = np.sum(grad_output, axis=0, keepdims=True)
grad_inputs = grad_output @ weights.T

# Ein numerischer Gradient prüft ein einzelnes Gewicht unabhängig.
epsilon = 1e-5
weights_plus = weights.copy()
weights_minus = weights.copy()

# Wir verändern nur das erste Gewicht minimal nach oben und unten.
weights_plus[0, 0] += epsilon
weights_minus[0, 0] -= epsilon

# Aus den zwei Verlusten entsteht die finite Differenz.
loss_plus = np.mean((inputs @ weights_plus + bias - targets) ** 2)
loss_minus = np.mean((inputs @ weights_minus + bias - targets) ** 2)
numeric_gradient = (loss_plus - loss_minus) / (2.0 * epsilon)

# Die Parameter werden einmal in Gegenrichtung der Gradienten aktualisiert.
new_weights = weights - learning_rate * grad_weights
new_bias = bias - learning_rate * grad_bias
new_loss = np.mean((inputs @ new_weights + new_bias - targets) ** 2)

print(f"Startverlust: {loss:.4f}")
print(f"Verlust nach einem Schritt: {new_loss:.4f}")
print(f"Analytischer Gradient w[0,0]: {grad_weights[0, 0]:.4f}")
print(f"Numerischer Gradient w[0,0]: {numeric_gradient:.4f}")
print(f"Bias-Gradient: {grad_bias[0, 0]:.4f}")
print(f"Gradient zu den Eingaben hat Form: {grad_inputs.shape}")

```
