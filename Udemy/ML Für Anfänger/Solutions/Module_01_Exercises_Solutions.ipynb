## Einrichtung und gemeinsame Datenbasis

Die folgende Zelle erzeugt einen kleinen Verkaufsdatensatz direkt im Arbeitsspeicher. Damit sind keine externen Dateien erforderlich.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

# Kleine Array-Daten für die ersten Aufgaben.
temperaturen_c = np.array([18.5, 20.0, 21.5, 24.0, 27.5, 29.0, 26.0, 23.5], dtype=float)
netto_preise = np.array([12.50, 8.90, 19.90, 5.50], dtype=float)

# Ein reproduzierbarer Verkaufsdatensatz mit numerischen und kategorialen Spalten.
verkauf = pd.DataFrame(
    {
        "Tag": pd.date_range("2026-07-01", periods=14, freq="D"),
        "Region": ["Nord", "Süd"] * 7,
        "Produkt": ["Sensor", "Ventil", "Sensor", "Pumpe", "Ventil", "Pumpe", "Sensor"] * 2,
        "Stückzahl": [18, 12, 24, 10, 16, 15, 29, 14, 21, 11, 19, 17, 32, 13],
        "Preis_EUR": [42, 55, 42, 80, 55, 80, 42, 55, 42, 80, 55, 80, 42, 55],
        "Temperatur_C": [21, 22, 24, 25, 27, 28, 30, 29, 27, 25, 23, 22, 24, 26],
        "Bewertung": [4.5, 4.1, 4.7, np.nan, 4.0, 4.3, 4.8, 3.9, 4.4, 4.2, np.nan, 4.6, 4.9, 4.1],
    }
)
verkauf["Umsatz_EUR"] = verkauf["Stückzahl"] * verkauf["Preis_EUR"]

print("Einrichtung abgeschlossen.")
print(f"Array-Form: {temperaturen_c.shape}")
print(f"Verkaufsdaten: {verkauf.shape[0]} Zeilen und {verkauf.shape[1]} Spalten")
display(verkauf.head())
```

### Aufgabe 1: NumPy-Arrays untersuchen und maskieren

Bearbeiten Sie die folgenden Teilaufgaben mit `temperaturen_c`:

1. Geben Sie Form, Anzahl der Dimensionen und Datentyp aus.
2. Wählen Sie die ersten drei sowie die letzten zwei Messwerte mit Slicing aus.
3. Erzeugen Sie eine boolesche Maske für Temperaturen von mindestens 25 °C.
4. Speichern Sie die gefilterten Werte in `warme_tage` und berechnen Sie deren Anzahl und Mittelwert.
5. Prüfen Sie mit einer geeigneten Bedingung, ob alle Werte im plausiblen Bereich von -30 bis 55 °C liegen.

```python
# Das Array wurde in der Einrichtungszelle erzeugt.
print("Temperaturen:", temperaturen_c)

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# shape beschreibt die Länge jeder Achse. Das Array ist eindimensional.
print("Form:", temperaturen_c.shape)
print("Dimensionen:", temperaturen_c.ndim)
print("Datentyp:", temperaturen_c.dtype)

# Slices schließen die obere Grenze aus. [:3] liefert daher die Positionen 0, 1 und 2.
erste_drei = temperaturen_c[:3]
letzte_zwei = temperaturen_c[-2:]
print("Erste drei Werte:", erste_drei)
print("Letzte zwei Werte:", letzte_zwei)

# Der Vergleich wird elementweise ausgeführt und erzeugt ein boolesches Array.
maske_warm = temperaturen_c >= 25.0
warme_tage = temperaturen_c[maske_warm]
print("Maske:", maske_warm)
print("Warme Tage:", warme_tage)
print("Anzahl warmer Tage:", warme_tage.size)
print("Mittelwert warmer Tage:", warme_tage.mean())

# np.all() ist nur dann True, wenn jede einzelne Teilbedingung erfüllt ist.
plausibel = np.all((temperaturen_c >= -30.0) & (temperaturen_c <= 55.0))
print("Alle Werte plausibel:", plausibel)
```

> **Musterantwort und Interpretation**
>
> Slicing wählt Positionen oder zusammenhängende Positionsbereiche, zum Beispiel die ersten drei Elemente. Eine boolesche Maske wählt dagegen alle Elemente aus, die eine inhaltliche Bedingung erfüllen, unabhängig von ihrer Position.

### Aufgabe 2: Vektorisierte Berechnungen und Broadcasting

Berechnen Sie ohne Python-Schleife:

1. Die Bruttopreise bei 19 % Mehrwertsteuer.
2. Einen Preisnachlass von 10 % auf die Bruttopreise.
3. Eine Matrix `tagespreise`, in der die vier rabattierten Preise mit den Tagesfaktoren `[1.00, 1.05, 0.95]` kombiniert werden. Die erwartete Form ist `(3, 4)`.
4. Prüfen Sie die Form und runden Sie die Ausgabe auf zwei Dezimalstellen.

```python
mehrwertsteuer = 0.19
rabatt = 0.10
tagesfaktoren = np.array([1.00, 1.05, 0.95], dtype=float)

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# NumPy wendet die Multiplikation direkt auf alle Preise an.
brutto_preise = netto_preise * (1.0 + mehrwertsteuer)

# Auch der Rabatt wird vektorisiert auf das gesamte Array angewendet.
rabattierte_preise = brutto_preise * (1.0 - rabatt)

# Durch [:, np.newaxis] erhalten die Tagesfaktoren die Form (3, 1).
# Diese Form ist mit (4,) kompatibel. NumPy broadcastet die Werte zu einer 3x4-Matrix.
tagespreise = tagesfaktoren[:, np.newaxis] * rabattierte_preise

print("Bruttopreise:", np.round(brutto_preise, 2))
print("Rabattierte Preise:", np.round(rabattierte_preise, 2))
print("Form der Tagespreismatrix:", tagespreise.shape)
print(np.round(tagespreise, 2))

# Eine explizite Prüfung macht Formfehler sofort sichtbar.
assert tagespreise.shape == (3, 4)
```

> **Musterantwort und Interpretation**
>
> Broadcasting drückt die gemeinsame Rechenregel kompakt aus und vermeidet wiederholten Code. Dadurch bleibt die Lösung leichter prüfbar und kann ohne strukturelle Änderung auf mehr Preise oder Tagesfaktoren erweitert werden.

### Aufgabe 3: DataFrame auswählen, filtern und erweitern

Arbeiten Sie mit dem DataFrame `verkauf`:

1. Wählen Sie nur `Tag`, `Region`, `Produkt`, `Stückzahl` und `Umsatz_EUR` aus.
2. Filtern Sie alle Zeilen aus der Region **Nord** mit mindestens 20 verkauften Einheiten.
3. Erzeugen Sie eine neue Spalte `Umsatz_pro_Stück` und kontrollieren Sie, ob sie dem Preis entspricht.
4. Sortieren Sie die gefilterte Tabelle absteigend nach `Umsatz_EUR`.
5. Geben Sie nur die drei umsatzstärksten Zeilen aus.

```python
# Arbeiten Sie auf einer Kopie, damit spätere Aufgaben dieselbe Ausgangstabelle verwenden.
verkauf_arbeit = verkauf.copy()

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# Eine Liste von Spaltennamen erzeugt einen neuen DataFrame mit genau diesen Spalten.
auswahl = verkauf_arbeit[["Tag", "Region", "Produkt", "Stückzahl", "Umsatz_EUR"]]
print("Ausgewählte Spalten:")
display(auswahl.head())

# Beide Bedingungen müssen für eine Zeile wahr sein. Jede Bedingung steht in Klammern.
nord_gross = verkauf_arbeit[
    (verkauf_arbeit["Region"] == "Nord") & (verkauf_arbeit["Stückzahl"] >= 20)
].copy()

# Die Division wirkt zeilenweise auf die beiden Series.
nord_gross["Umsatz_pro_Stück"] = nord_gross["Umsatz_EUR"] / nord_gross["Stückzahl"]

# sort_values() ordnet die vollständigen Zeilen anhand einer Spalte.
top_drei = nord_gross.sort_values("Umsatz_EUR", ascending=False).head(3)
display(top_drei)

# Die neue Spalte sollte bis auf numerische Rundung mit Preis_EUR übereinstimmen.
assert np.allclose(nord_gross["Umsatz_pro_Stück"], nord_gross["Preis_EUR"])
```

### Aufgabe 4: Fehlende Werte und Gruppenzusammenfassungen

1. Zählen Sie fehlende Werte je Spalte.
2. Ersetzen Sie fehlende Bewertungen durch den Median der jeweiligen Region. Speichern Sie das Ergebnis in `Bewertung_bereinigt`, ohne die ursprüngliche Spalte zu überschreiben.
3. Erstellen Sie eine Gruppentabelle nach `Region` und `Produkt` mit:
   - Gesamtstückzahl,
   - Gesamtumsatz,
   - mittlerer bereinigter Bewertung.
4. Sortieren Sie die Gruppentabelle nach Gesamtumsatz absteigend.

```python
verkauf_gruppen = verkauf.copy()

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# isna() markiert fehlende Werte; sum() zählt True-Werte spaltenweise.
fehlwerte = verkauf_gruppen.isna().sum()
print("Fehlende Werte je Spalte:")
print(fehlwerte)

# transform("median") liefert für jede Zeile den Median ihrer Region.
# Dadurch kann fillna() nur die fehlenden Bewertungen mit dem passenden Gruppenwert ersetzen.
regionaler_median = verkauf_gruppen.groupby("Region")["Bewertung"].transform("median")
verkauf_gruppen["Bewertung_bereinigt"] = verkauf_gruppen["Bewertung"].fillna(regionaler_median)

# Benannte Aggregationen erzeugen direkt verständliche Ergebnisnamen.
gruppenbericht = (
    verkauf_gruppen.groupby(["Region", "Produkt"], as_index=False)
    .agg(
        Gesamtstückzahl=("Stückzahl", "sum"),
        Gesamtumsatz_EUR=("Umsatz_EUR", "sum"),
        Mittlere_Bewertung=("Bewertung_bereinigt", "mean"),
    )
    .sort_values("Gesamtumsatz_EUR", ascending=False)
)

gruppenbericht["Mittlere_Bewertung"] = gruppenbericht["Mittlere_Bewertung"].round(2)
display(gruppenbericht)

# Nach der Ersetzung dürfen in der neuen Spalte keine Fehlwerte verbleiben.
assert verkauf_gruppen["Bewertung_bereinigt"].isna().sum() == 0
```

> **Musterantwort und Interpretation**
>
> Ein regionaler Median berücksichtigt, dass sich typische Bewertungen zwischen Regionen unterscheiden können. Er ist außerdem weniger empfindlich gegenüber einzelnen extremen Bewertungen als ein Mittelwert. Die Ersetzung muss trotzdem dokumentiert werden, weil sie keine echte Beobachtung ist.

### Aufgabe 5: Passende Diagramme erstellen

Erstellen Sie eine Abbildung mit vier Diagrammen:

1. **Linie:** Umsatz pro Tag.
2. **Balken:** Gesamtumsatz je Produkt.
3. **Histogramm:** Verteilung der Stückzahlen.
4. **Streudiagramm:** Temperatur gegen Stückzahl, wobei die beiden Regionen unterscheidbar sind.

Beschriften Sie Titel und Achsen. Verwenden Sie für das Streudiagramm eine Legende.

```python
plot_daten = verkauf.copy()

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# Für den täglichen Verlauf werden Umsätze pro Datum zusammengefasst.
tagesumsatz = plot_daten.groupby("Tag", as_index=False)["Umsatz_EUR"].sum()

# Für den Kategorienvergleich aggregieren wir nach Produkt.
produktumsatz = (
    plot_daten.groupby("Produkt", as_index=False)["Umsatz_EUR"]
    .sum()
    .sort_values("Umsatz_EUR", ascending=False)
)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Linie für eine geordnete Zeitachse.
axes[0, 0].plot(tagesumsatz["Tag"], tagesumsatz["Umsatz_EUR"], marker="o")
axes[0, 0].set_title("Täglicher Umsatz")
axes[0, 0].set_xlabel("Tag")
axes[0, 0].set_ylabel("Umsatz in EUR")
axes[0, 0].tick_params(axis="x", rotation=45)

# 2. Balken für den Vergleich diskreter Kategorien.
axes[0, 1].bar(produktumsatz["Produkt"], produktumsatz["Umsatz_EUR"])
axes[0, 1].set_title("Gesamtumsatz nach Produkt")
axes[0, 1].set_xlabel("Produkt")
axes[0, 1].set_ylabel("Umsatz in EUR")

# 3. Histogramm für die Form einer numerischen Verteilung.
axes[1, 0].hist(plot_daten["Stückzahl"], bins=6, edgecolor="black")
axes[1, 0].set_title("Verteilung der Stückzahlen")
axes[1, 0].set_xlabel("Stückzahl")
axes[1, 0].set_ylabel("Häufigkeit")

# 4. Streudiagramm für den Zusammenhang zweier numerischer Merkmale.
for region, teilmenge in plot_daten.groupby("Region"):
    axes[1, 1].scatter(
        teilmenge["Temperatur_C"],
        teilmenge["Stückzahl"],
        label=region,
        alpha=0.8,
    )
axes[1, 1].set_title("Temperatur und Stückzahl")
axes[1, 1].set_xlabel("Temperatur in °C")
axes[1, 1].set_ylabel("Stückzahl")
axes[1, 1].legend(title="Region")

plt.tight_layout()
plt.show()
```

> **Musterantwort und Interpretation**
>
> Die Linie bewahrt die zeitliche Reihenfolge und macht Veränderungen sichtbar. Balken vergleichen klar abgegrenzte Produkte. Das Histogramm zeigt Häufungen und Streuung einer numerischen Variable. Das Streudiagramm zeigt, ob Temperatur und Stückzahl gemeinsam variieren, und die Gruppierung macht mögliche regionale Unterschiede sichtbar.

### Aufgabe 6: Integrationsaufgabe: kompakte Verkaufsanalyse

Erstellen Sie einen kleinen Analysebericht, der mindestens Folgendes enthält:

1. Gesamtumsatz und mittlere Stückzahl.
2. Umsatzstärkste Region und umsatzstärkstes Produkt.
3. Korrelation zwischen Temperatur und Stückzahl.
4. Die drei Tage mit dem höchsten Umsatz.
5. Ein geeignetes Diagramm, das Ihre wichtigste Beobachtung unterstützt.
6. Eine kurze Interpretation mit mindestens einer Einschränkung der Analyse.

```python
analyse_daten = verkauf.copy()

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# Grundkennzahlen beantworten die ersten Fragen direkt.
gesamtumsatz = analyse_daten["Umsatz_EUR"].sum()
mittlere_stueckzahl = analyse_daten["Stückzahl"].mean()

# idxmax() liefert den Index des größten aggregierten Werts.
umsatz_region = analyse_daten.groupby("Region")["Umsatz_EUR"].sum()
umsatz_produkt = analyse_daten.groupby("Produkt")["Umsatz_EUR"].sum()
beste_region = umsatz_region.idxmax()
bestes_produkt = umsatz_produkt.idxmax()

# corr() berechnet hier die lineare Pearson-Korrelation.
korrelation = analyse_daten[["Temperatur_C", "Stückzahl"]].corr().loc[
    "Temperatur_C", "Stückzahl"
]

# Die drei stärksten Tage werden nach Umsatz sortiert.
top_tage = analyse_daten.nlargest(3, "Umsatz_EUR")[["Tag", "Region", "Produkt", "Umsatz_EUR"]]

print(f"Gesamtumsatz: {gesamtumsatz:,.2f} EUR")
print(f"Mittlere Stückzahl: {mittlere_stueckzahl:.2f}")
print(f"Umsatzstärkste Region: {beste_region}")
print(f"Umsatzstärkstes Produkt: {bestes_produkt}")
print(f"Korrelation Temperatur/Stückzahl: {korrelation:.3f}")
print("Drei umsatzstärkste Tage:")
display(top_tage)

# Das Diagramm vergleicht die Regionsergebnisse, weil dies eine zentrale Managementfrage ist.
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(umsatz_region.index, umsatz_region.values)
ax.set_title("Gesamtumsatz nach Region")
ax.set_xlabel("Region")
ax.set_ylabel("Umsatz in EUR")
plt.show()
```

> **Musterantwort und Interpretation**
>
> Die aggregierten Werte zeigen, welche Region und welches Produkt im betrachteten Zeitraum den größten Umsatz liefern. Eine Korrelation zwischen Temperatur und Stückzahl beschreibt nur einen linearen Zusammenhang und beweist keine Ursache. Der Datensatz umfasst lediglich vierzehn Tage und enthält unterschiedliche Produkte und Preise, sodass saisonale oder längerfristige Aussagen nicht belastbar sind. Für eine Entscheidung wären ein längerer Zeitraum, weitere Einflussgrößen und eine Prüfung getrennt nach Produkt sinnvoll.
