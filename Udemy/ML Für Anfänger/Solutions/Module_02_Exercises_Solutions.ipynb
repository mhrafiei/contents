## Einrichtung und gemeinsame Datenbasis

Die Setup-Zelle erzeugt kleine synthetische Daten für überwachte und unüberwachte Beispiele. Die Modelle dienen nur zur Veranschaulichung der Lernarten, nicht zur umfassenden Optimierung.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

# Daten für eine kleine Klassifikationsaufgabe: zwei Messwerte und eine bekannte Kategorie.
X_klassifikation = np.array(
    [[1.0, 1.2], [1.3, 0.8], [1.8, 1.6], [3.0, 3.2], [3.4, 2.8], [3.8, 3.7],
     [1.1, 1.7], [3.2, 3.8], [1.6, 1.0], [3.7, 3.0]],
    dtype=float,
)
y_klassifikation = np.array([0, 0, 0, 1, 1, 1, 0, 1, 0, 1], dtype=int)

# Daten für eine kleine Regression: Fläche und Energieverbrauch.
flaeche = np.array([45, 55, 65, 80, 95, 110, 125, 145], dtype=float).reshape(-1, 1)
verbrauch = np.array([72, 84, 91, 112, 128, 151, 169, 192], dtype=float)

# Daten ohne Labels für Gruppen- und Anomaliebeispiele.
cluster_a = rng.normal(loc=[-2.0, -1.5], scale=0.45, size=(35, 2))
cluster_b = rng.normal(loc=[2.0, 2.2], scale=0.50, size=(35, 2))
X_unbeschriftet = np.vstack([cluster_a, cluster_b, [[6.5, -5.5], [-6.0, 5.8]]])

print("Einrichtung abgeschlossen.")
print("Klassifikationsdaten:", X_klassifikation.shape)
print("Regressionsdaten:", flaeche.shape)
print("Unbeschriftete Daten:", X_unbeschriftet.shape)
```

> **Musterantwort und Interpretation**
>
> Vorteile sind die vollständige Nachvollziehbarkeit und die zuverlässige Wiederholung derselben Entscheidung. Grenzen entstehen, wenn wichtige Sonderfälle fehlen oder Regeln häufig geändert werden müssen. Außerdem kann die Regel keine unscharfen Muster aus vielen Merkmalen lernen.

### Aufgabe 4: Überwachtes Lernen an zwei Mini-Beispielen

1. Teilen Sie die Klassifikationsdaten reproduzierbar in Training und Test auf.
2. Trainieren Sie einen `KNeighborsClassifier` mit `n_neighbors=3` und berechnen Sie die Genauigkeit.
3. Trainieren Sie eine lineare Regression für Fläche und Verbrauch. Verwenden Sie die letzten zwei Beobachtungen als Testdaten und berechnen Sie den MAE.
4. Geben Sie für beide Beispiele aus, was Merkmale, Zielwerte und neue Vorhersagen bedeuten.

```python
# Die Daten X_klassifikation, y_klassifikation, flaeche und verbrauch sind vorhanden.

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# Klassifikation: Die Zielwerte sind Kategorien 0 und 1.
X_train, X_test, y_train, y_test = train_test_split(
    X_klassifikation,
    y_klassifikation,
    test_size=0.3,
    random_state=RANDOM_SEED,
    stratify=y_klassifikation,
)

klassifikator = KNeighborsClassifier(n_neighbors=3)
klassifikator.fit(X_train, y_train)
klassen_vorhersage = klassifikator.predict(X_test)
klassifikationsgenauigkeit = accuracy_score(y_test, klassen_vorhersage)

print("Klassifikation")
print("Merkmale: zwei Messwerte je Objekt")
print("Zielwert: bekannte Kategorie 0 oder 1")
print("Vorhersagen:", klassen_vorhersage)
print(f"Genauigkeit: {klassifikationsgenauigkeit:.3f}")

# Regression: Die Zielwerte sind kontinuierliche Verbrauchswerte.
X_reg_train, X_reg_test = flaeche[:-2], flaeche[-2:]
y_reg_train, y_reg_test = verbrauch[:-2], verbrauch[-2:]

regressor = LinearRegression()
regressor.fit(X_reg_train, y_reg_train)
verbrauch_vorhersage = regressor.predict(X_reg_test)
regressions_mae = mean_absolute_error(y_reg_test, verbrauch_vorhersage)

print("\nRegression")
print("Merkmal: Gebäudefläche in Quadratmetern")
print("Zielwert: Energieverbrauch als Zahl")
print("Vorhersagen:", np.round(verbrauch_vorhersage, 1))
print(f"MAE: {regressions_mae:.2f} kWh")
```

> **Musterantwort und Interpretation**
>
> Genauigkeit zählt den Anteil korrekt vorhergesagter Kategorien und passt daher zu einer Klassifikationsaufgabe. Der MAE misst dagegen durchschnittliche numerische Abstände in der Einheit des Regressionsziels. Beide Kennzahlen beantworten unterschiedliche Fragen.

### Aufgabe 5: Unüberwachtes Lernen und Anomalien erkunden

1. Wenden Sie `KMeans` mit zwei Clustern auf `X_unbeschriftet` an.
2. Markieren Sie zusätzlich mögliche Anomalien mit `IsolationForest(contamination=0.03)`.
3. Visualisieren Sie die Punkte nach Clusterzuordnung und kennzeichnen Sie Anomalien besonders.
4. Erklären Sie, warum Cluster-IDs keine vorgegebenen fachlichen Klassen sind.

```python
X_unbeobachtet = X_unbeschriftet.copy()

# ============================================================
# MUSTERLÖSUNG
# ============================================================

# KMeans erhält keine Zielwerte. Es gruppiert nur anhand geometrischer Nähe.
kmeans = KMeans(n_clusters=2, random_state=RANDOM_SEED, n_init=10)
cluster_ids = kmeans.fit_predict(X_unbeobachtet)

# IsolationForest liefert 1 für gewöhnliche Punkte und -1 für auffällige Punkte.
anomalie_modell = IsolationForest(
    contamination=0.03,
    random_state=RANDOM_SEED,
)
anomalie_labels = anomalie_modell.fit_predict(X_unbeobachtet)
ist_anomalie = anomalie_labels == -1

print("Clustergrößen:", np.bincount(cluster_ids))
print("Anzahl markierter Anomalien:", int(ist_anomalie.sum()))

fig, ax = plt.subplots(figsize=(7, 5))
streuung = ax.scatter(
    X_unbeobachtet[:, 0],
    X_unbeobachtet[:, 1],
    c=cluster_ids,
    alpha=0.75,
)
ax.scatter(
    X_unbeobachtet[ist_anomalie, 0],
    X_unbeobachtet[ist_anomalie, 1],
    marker="x",
    s=120,
    linewidths=2,
    label="mögliche Anomalie",
)
ax.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="P",
    s=160,
    label="Clusterzentren",
)
ax.set_title("Clustering und Anomaliehinweise")
ax.set_xlabel("Merkmal 1")
ax.set_ylabel("Merkmal 2")
ax.legend()
plt.show()
```

> **Musterantwort und Interpretation**
>
> Die gefundenen Gruppen müssen anhand des Anwendungskontexts geprüft und beschrieben werden. Die numerischen Cluster-IDs 0 und 1 besitzen von sich aus keine Bedeutung, können sich zwischen Läufen vertauschen und garantieren weder nützliche noch faire Segmente.
