```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: NumPy-Arrays gezielt untersuchen
#
# Ziel dieser Codezelle:
# Erstellen Sie aus den unten vorgegebenen Messwerten ein zweidimensionales NumPy-
# Array. 1. Legen Sie den Datentyp ausdrücklich als float64 fest. 2. Geben Sie Form,
# Anzahl der Dimensionen und Datentyp aus. 3. Wählen Sie...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

measurements = [
    [20.5, 21.0, 23.5, 22.0],
    [19.5, 24.0, 25.5, 21.5],
    [18.0, 20.0, 22.5, 23.0],
]

# Das Array wird ausdrücklich als float64 angelegt. Dadurch ist der
# Datentyp nicht von einer automatischen Interpretation abhängig.
measurement_array = np.array(measurements, dtype=np.float64)

# shape beschreibt die Länge jeder Dimension. ndim gibt die Anzahl
# der Dimensionen an, und dtype zeigt den gespeicherten Zahlentyp.
print("Form:", measurement_array.shape)
print("Dimensionen:", measurement_array.ndim)
print("Datentyp:", measurement_array.dtype)

# Index 1 ist die zweite Zeile, weil Python bei null zu zählen beginnt.
second_row = measurement_array[1, :]

# Index 2 ist die dritte Spalte. Der Doppelpunkt wählt alle Zeilen.
third_column = measurement_array[:, 2]

# Der Vergleich erzeugt zunächst eine boolesche Maske gleicher Form.
above_limit_mask = measurement_array > 22.0

# Mit der Maske werden nur die zutreffenden Werte als 1-D-Array gewählt.
values_above_limit = measurement_array[above_limit_mask]

# axis=1 fasst jede Zeile zusammen. Es entstehen drei Mittelwerte.
row_means = measurement_array.mean(axis=1)

# axis=0 fasst jede Spalte zusammen. Es entstehen vier Mittelwerte.
column_means = measurement_array.mean(axis=0)

print("Zweite Zeile:", second_row)
print("Dritte Spalte:", third_column)
print("Maske > 22.0:\n", above_limit_mask)
print("Gefilterte Werte:", values_above_limit)
print("Mittelwert je Zeile:", row_means)
print("Mittelwert je Spalte:", column_means)
```

### Reflexion zu Aufgabe 1

Das Array hat die Form `(3, 4)`: drei Messreihen und vier Messpunkte pro Reihe. Bei `axis=1` wird innerhalb jeder Zeile zusammengefasst, während `axis=0` die Werte derselben Spalte über alle Zeilen kombiniert. Eine häufige Fehlerquelle ist die Verwechslung dieser Achsen oder die Annahme, dass eine boolesche Maske die ursprüngliche zweidimensionale Form beibehält. Bei direkter Maskierung entsteht hier ein eindimensionaler Vektor der ausgewählten Werte.

## Aufgabe 2: Tabellen filtern, berechnen und gruppieren

    Arbeiten Sie mit der Tabelle `orders`.

1. Erzeugen Sie eine neue Spalte `revenue_eur = units * unit_price_eur`.
2. Wählen Sie nur Bestellungen aus der Region `Nord` mit mindestens 7 Einheiten.
3. Erstellen Sie eine Zusammenfassung je Kategorie mit Gesamtstückzahl, Gesamtumsatz und mittlerer Zufriedenheit.
4. Sortieren Sie die Zusammenfassung absteigend nach Gesamtumsatz.
5. Geben Sie die umsatzstärkste Kategorie aus.

> **Hinweis:** Kontrollieren Sie vor der Gruppierung, ob `revenue_eur` numerisch ist.

```python
orders_work = orders.copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Tabellen filtern, berechnen und gruppieren
#
# Ziel dieser Codezelle:
# Arbeiten Sie mit der Tabelle orders. 1. Erzeugen Sie eine neue Spalte revenueeur =
# units unitpriceeur. 2. Wählen Sie nur Bestellungen aus der Region Nord mit
# mindestens 7 Einheiten. 3. Erstellen Sie eine Zusammenfassu...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

# Eine Kopie verhindert, dass die gemeinsame Ausgangstabelle unbemerkt
# für spätere Aufgaben verändert wird.
orders_work = orders.copy()

# Die Multiplikation wird elementweise zwischen zwei gleich langen
# Series ausgeführt. Das Ergebnis wird als neue Spalte gespeichert.
orders_work["revenue_eur"] = (
    orders_work["units"] * orders_work["unit_price_eur"]
)

# Jede Teilbedingung steht in Klammern. Das kaufmännische Und `&`
# kombiniert die beiden booleschen Series zeilenweise.
north_large_orders = orders_work[
    (orders_work["region"] == "Nord")
    & (orders_work["units"] >= 7)
].copy()

# groupby teilt die Zeilen nach Kategorie auf. agg berechnet mehrere
# klar benannte Kennzahlen in einem Schritt.
category_summary = (
    orders_work.groupby("category", as_index=False)
    .agg(
        total_units=("units", "sum"),
        total_revenue_eur=("revenue_eur", "sum"),
        mean_satisfaction=("satisfaction", "mean"),
    )
    .sort_values("total_revenue_eur", ascending=False)
    .reset_index(drop=True)
)

# iloc[0] ist nach der Sortierung die erste und damit umsatzstärkste Zeile.
top_category = category_summary.loc[0, "category"]

print("Gefilterte Nord-Bestellungen:")
print(north_large_orders.to_string(index=False))
print("\nZusammenfassung nach Kategorie:")
print(category_summary.round(2).to_string(index=False))
print("\nUmsatzstärkste Kategorie:", top_category)
```

### Reflexion zu Aufgabe 2

Filtern und Gruppieren beantworten unterschiedliche Fragen. Der Filter betrachtet einzelne passende Bestellungen. Die Gruppierung verdichtet dagegen alle Bestellungen einer Kategorie. Die Sortierung muss nach der Aggregation erfolgen, weil erst dann der Gesamtumsatz je Kategorie vorliegt. Eine häufige Fehlerquelle ist die Verwendung von `and` statt `&` bei pandas-Bedingungen.

## Aufgabe 3: Datenqualität systematisch prüfen

    Untersuchen Sie `orders_raw`, das absichtlich einen fehlenden Wert, einen doppelten Datensatz und eine als Text gespeicherte Datumsspalte enthält.

1. Zeigen Sie Form, Spaltennamen und Datentypen.

5. Erstellen Sie einen kompakten Qualitätsbericht als DataFrame mit Spalte, Datentyp, Anzahl Fehlwerte und Anzahl eindeutiger Werte.

```python
quality_data = orders_raw.copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Datenqualität systematisch prüfen
#
# Ziel dieser Codezelle:
# Untersuchen Sie ordersraw, das absichtlich einen fehlenden Wert, einen doppelten
# Datensatz und eine als Text gespeicherte Datumsspalte enthält. 1. Zeigen Sie Form,
# Spaltennamen und Datentypen. 2. Zählen Sie fehlende W...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

# Wir arbeiten auf einer Kopie, damit die absichtlich fehlerhafte
# Ausgangstabelle für Kontrollzwecke unverändert bleibt.
quality_data = orders_raw.copy()

print("Form vor Bereinigung:", quality_data.shape)
print("Spalten:", quality_data.columns.tolist())
print("\nDatentypen vor Konvertierung:")
print(quality_data.dtypes)

# isna().sum() zählt für jede Spalte die fehlenden Einträge.
missing_by_column = quality_data.isna().sum()

# duplicated() markiert ab der zweiten identischen Zeile ein Duplikat.
duplicate_count = int(quality_data.duplicated().sum())

# errors='coerce' würde unlesbare Werte in NaT umwandeln. Danach muss
# geprüft werden, ob durch die Konvertierung neue Fehlwerte entstanden.
quality_data["date"] = pd.to_datetime(
    quality_data["date"], errors="coerce"
)

# drop_duplicates entfernt vollständige doppelte Zeilen. reset_index
# erzeugt anschließend einen sauberen fortlaufenden Index.
cleaned_orders = (
    quality_data.drop_duplicates()
    .reset_index(drop=True)
)

# Der Bericht wird zeilenweise aus den Spalten der bereinigten Tabelle
# aufgebaut. astype(str) liefert gut lesbare Typbezeichnungen.
quality_report = pd.DataFrame(
    {
        "column": cleaned_orders.columns,
        "dtype": cleaned_orders.dtypes.astype(str).values,
        "missing": cleaned_orders.isna().sum().values,
        "unique": cleaned_orders.nunique(dropna=True).values,
    }
)

print("\nFehlwerte vor Bereinigung:")
print(missing_by_column.to_string())
print("Vollständige Duplikate:", duplicate_count)
print("Form nach Entfernen der Duplikate:", cleaned_orders.shape)
print("\nQualitätsbericht:")
print(quality_report.to_string(index=False))
```

### Reflexion zu Aufgabe 3

Ein Qualitätsbericht beschreibt Probleme, löst sie aber nicht automatisch. Der fehlende Zufriedenheitswert bleibt deshalb bewusst sichtbar. Seine Behandlung hängt von der späteren Fragestellung ab. Die Datumsumwandlung ist erfolgreich, wenn die Spalte den Typ `datetime64` besitzt und keine unerwarteten `NaT`-Werte entstanden sind. Duplikate sollten nur entfernt werden, wenn identische Zeilen tatsächlich dieselbe Beobachtung darstellen.

## Aufgabe 4: Passende Diagramme erstellen

    Visualisieren Sie vier unterschiedliche Aspekte der Verkaufstabelle.

1. Liniendiagramm des täglichen Gesamtumsatzes.
2. Horizontales Balkendiagramm der verkauften Einheiten je Kategorie.

4. Streudiagramm von Stückzahl und Umsatz, wobei die Regionen unterscheidbar sind.

Jedes Diagramm benötigt einen aussagekräftigen Titel, Achsenbeschriftungen und, wo sinnvoll, eine Legende oder ein Raster. Erstellen Sie jedes Diagramm als eigene Abbildung.

```python
plot_data = orders.copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Passende Diagramme erstellen
#
# Ziel dieser Codezelle:
# Visualisieren Sie vier unterschiedliche Aspekte der Verkaufstabelle. 1.
# Liniendiagramm des täglichen Gesamtumsatzes. 2. Horizontales Balkendiagramm der
# verkauften Einheiten je Kategorie. 3. Histogramm der Umsätze einz...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

plot_data = orders.copy()

# Der Umsatz wird einmal berechnet und anschließend in mehreren
# Diagrammen wiederverwendet.
plot_data["revenue_eur"] = (
    plot_data["units"] * plot_data["unit_price_eur"]
)

# 1) Für einen zeitlichen Verlauf werden die Umsätze zunächst je Datum
# zusammengefasst und chronologisch sortiert.
daily_revenue = (
    plot_data.groupby("date", as_index=False)["revenue_eur"]
    .sum()
    .sort_values("date")
)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(
    daily_revenue["date"],
    daily_revenue["revenue_eur"],
    marker="o",
    linewidth=2,
)
ax.set_title("Täglicher Gesamtumsatz")
ax.set_xlabel("Datum")
ax.set_ylabel("Umsatz in Euro")
ax.grid(True, alpha=0.3)
fig.autofmt_xdate()
plt.tight_layout()
plt.show()

# 2) Ein Balkendiagramm eignet sich für den Vergleich diskreter
# Kategorien. Die Sortierung erhöht die Lesbarkeit.
units_by_category = (
    plot_data.groupby("category")["units"]
    .sum()
    .sort_values()
)
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(units_by_category.index, units_by_category.values)
ax.set_title("Verkaufte Einheiten je Kategorie")
ax.set_xlabel("Einheiten")
ax.set_ylabel("Kategorie")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()

# 3) Ein Histogramm zeigt die Verteilung numerischer Einzelwerte.
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(plot_data["revenue_eur"], bins=6, edgecolor="white")
ax.set_title("Verteilung der Bestellumsätze")
ax.set_xlabel("Umsatz je Bestellung in Euro")
ax.set_ylabel("Anzahl der Bestellungen")
plt.tight_layout()
plt.show()

# 4) Für den Gruppenvergleich wird jede Region separat gezeichnet.
# Dadurch erhält jede Gruppe automatisch einen Legendeneintrag.
fig, ax = plt.subplots(figsize=(7, 4))
for region, region_data in plot_data.groupby("region"):
    ax.scatter(
        region_data["units"],
        region_data["revenue_eur"],
        label=region,
        s=65,
        alpha=0.8,
    )
ax.set_title("Stückzahl und Umsatz nach Region")
ax.set_xlabel("Verkaufte Einheiten")
ax.set_ylabel("Bestellumsatz in Euro")
ax.legend(title="Region")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Reflexion zu Aufgabe 4

## Aufgabe 5: Integrationsaufgabe: Kurze Verkaufsanalyse

    Beantworten Sie die Frage: **Welche Kombination aus Region und Kategorie trägt am meisten zum Umsatz bei, und wie ist ihre mittlere Kundenzufriedenheit einzuordnen?**

1. Berechnen Sie den Umsatz je Bestellung.
2. Gruppieren Sie nach Region und Kategorie und berechnen Sie Anzahl Bestellungen, Einheiten, Umsatz und mittlere Zufriedenheit.
3. Sortieren Sie nach Umsatz und zeigen Sie die fünf stärksten Kombinationen.
4. Erstellen Sie ein Balkendiagramm der Umsätze dieser fünf Kombinationen.
5. Formulieren Sie anschließend eine vorsichtige Interpretation. Nennen Sie mindestens eine Einschränkung des kleinen Datensatzes.

> **Hinweis:** Trennen Sie Beobachtung, mögliche Erklärung und nachgewiesene Ursache sprachlich voneinander.

```python
analysis_data = orders.copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Integrationsaufgabe: Kurze Verkaufsanalyse
#
# Ziel dieser Codezelle:
# Beantworten Sie die Frage: Welche Kombination aus Region und Kategorie trägt am
# meisten zum Umsatz bei, und wie ist ihre mittlere Kundenzufriedenheit einzuordnen?
# 1. Berechnen Sie den Umsatz je Bestellung. 2. Gruppier...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

analysis_data = orders.copy()

# Der Zeilenumsatz verbindet Stückzahl und Stückpreis. Beide Spalten
# wurden vorher als numerisch geprüft.
analysis_data["revenue_eur"] = (
    analysis_data["units"] * analysis_data["unit_price_eur"]
)

# Die Kombination aus Region und Kategorie bildet die fachliche
# Vergleichseinheit dieser Analyse.
regional_category_summary = (
    analysis_data.groupby(["region", "category"], as_index=False)
    .agg(
        order_count=("category", "size"),
        total_units=("units", "sum"),
        total_revenue_eur=("revenue_eur", "sum"),
        mean_satisfaction=("satisfaction", "mean"),
    )
    .sort_values("total_revenue_eur", ascending=False)
    .reset_index(drop=True)
)

# Da der Datensatz nur neun mögliche Kombinationen besitzt, sind die
# fünf größten für eine kompakte Darstellung ausreichend.
top_five = regional_category_summary.head(5).copy()
top_five["region_category"] = (
    top_five["region"] + " | " + top_five["category"]
)

print(top_five.round(2).to_string(index=False))

# Die Reihenfolge wird für ein horizontales Balkendiagramm umgedreht,
# damit der größte Balken oben erscheint.
chart_data = top_five.sort_values("total_revenue_eur")
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.barh(
    chart_data["region_category"],
    chart_data["total_revenue_eur"],
)
ax.set_title("Umsatzstärkste Regions-Kategorie-Kombinationen")
ax.set_xlabel("Gesamtumsatz in Euro")
ax.set_ylabel("Region | Kategorie")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()

# Die führende Zeile wird für eine kurze textuelle Zusammenfassung
# gezielt ausgelesen.
leader = top_five.iloc[0]
print(
    f"Führend ist {leader['region']} / {leader['category']} mit "
    f"{leader['total_revenue_eur']:.2f} Euro Umsatz und einer "
    f"mittleren Zufriedenheit von {leader['mean_satisfaction']:.2f}."
)
```

### Reflexion zu Aufgabe 5
