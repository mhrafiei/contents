## **1. Beobachtungen Merkmale Zielwerte**

### **1.1. Zeilen als Stichproben**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_01.jpg?v=1787537070" width="250">



>* Jede Zeile beschreibt eine Beobachtung
>* Die Zeilenbedeutung bestimmt die Interpretation

>* Zeilen zeigen nur eine ausgewählte Stichprobe.
>* Auswahl begrenzt mögliche Aussagen aus Daten.

>* Zeilenbeziehungen früh erkennen und dokumentieren
>* Dubletten, Lücken und Einheiten prüfen



```
#@title Python-Code - Zeilen als Stichproben

# Dieses Beispiel zeigt Zeilen als einzelne Stichproben.
# Jede Zeile beschreibt genau eine beobachtete Wohnung.
# Die Ausgabe trennt Beobachtungen, Merkmale und Zielwert.

import pandas as pd

# Wir bauen eine kleine Tabelle direkt im Speicher.
housing_data = pd.DataFrame(
    {
        "area_m2": [45, 62, 80, 95],
        "rooms": [2, 3, 3, 4],
        "distance_km": [1.2, 3.5, 5.0, 2.1],
        "rent_eur": [720, 890, 1050, 1320],
    }
)

# Diese Prüfung macht die erwartete Tabellenform sichtbar.
expected_shape = (4, 4)
if housing_data.shape != expected_shape:
    raise ValueError("Die Beispieltabelle hat nicht die erwartete Form.")

# Jede Tabellenzeile ist eine Beobachtung oder Stichprobe.
row_count = len(housing_data)
feature_names = ["area_m2", "rooms", "distance_km"]
target_name = "rent_eur"

# Merkmale beschreiben die Beobachtung, der Zielwert ist gesucht.
X = housing_data[feature_names]
y = housing_data[target_name]

# Wir betrachten eine einzelne Zeile als konkreten Fall.
sample_index = 1
sample_features = X.iloc[sample_index]
sample_target = y.iloc[sample_index]

# Die Ausgabe bleibt kurz und zeigt die Rollen der Spalten.
print(f"Anzahl der Zeilen, also Stichproben: {row_count}")
print(f"Merkmale in X: {list(X.columns)}")
print(f"Zielwert in y: {target_name}")
print(f"Form von X: {X.shape}")
print(f"Form von y: {y.shape}")
print(f"Beispielzeile {sample_index}: {sample_features.to_dict()}")
print(f"Zielwert derselben Zeile: {sample_target} Euro")

```

### **1.2. Merkmale als Spalten**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_02.jpg?v=1787537069" width="250">



>* Merkmale stehen als Spalten in Tabellen
>* Jede Spalte bleibt inhaltlich einheitlich

>* Spalten beantworten Fragen zu Beobachtungen
>* Datentyp, Einheit und Bedeutung klären

>* Spaltennamen und Merkmale kritisch prüfen
>* Merkmale klar von Zielwerten trennen



```
#@title Python-Code - Merkmale als Spalten

# Dieses Beispiel zeigt Merkmale als Tabellenspalten.
# Jede Zeile bleibt eine einzelne Beobachtung.
# Am Ende entstehen X und y.

import pandas as pd

# Wir bauen einen kleinen Datensatz direkt im Speicher.
data = pd.DataFrame(
    {
        "age_years": [29, 41, 35, 52],
        "height_cm": [168, 181, 174, 166],
        "visits_last_year": [2, 5, 1, 7],
        "risk_group": ["niedrig", "hoch", "niedrig", "hoch"],
    }
)

# Diese Spalte ist der Zielwert, nicht ein Merkmal.
data["needs_follow_up"] = [0, 1, 0, 1]

# Merkmale sind die beschreibenden Spalten für jede Beobachtung.
feature_columns = ["age_years", "height_cm", "visits_last_year", "risk_group"]
target_column = "needs_follow_up"

# X enthält nur Merkmale, y enthält nur den Zielwert.
X = data[feature_columns]
y = data[target_column]

# Eine einfache Prüfung schützt vor vertauschten Formen.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen haben.")

# Die Ausgabe zeigt Spaltenrollen und Array-Formen.
print("Datensatz: 4 Beobachtungen, 4 Merkmale, 1 Zielwert")
print("Merkmalspalten:", ", ".join(feature_columns))
print("Zielspalte:", target_column)
print("Form von X:", X.shape)
print("Form von y:", y.shape)
print("Datentypen:", X.dtypes.astype(str).to_dict())
print("Erste Beobachtung als Merkmale:", X.iloc[0].to_dict())

```

### **1.3. Zielwerte und Labels**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_01_03.jpg?v=1787537072" width="250">



>* Zielwerte sind vorherzusagende Ergebnisse
>* Die Fragestellung bestimmt ihre Rolle

>* Labels beschreiben Klassen oder Kategorien.
>* Herkunft und Zuverlässigkeit kritisch prüfen.

>* Zielwerte können numerisch, kategorial oder mehrfach sein
>* Zielwert und Merkmale klar dokumentieren



```
#@title Python-Code - Zielwerte und Labels

# Dieses Beispiel trennt Merkmale und Zielwerte.
# Labels zeigen die gesuchte Kategorie jeder Beobachtung.
# Die Ausgabe zeigt X, y und Formen.

import pandas as pd

# Ein kleiner Datensatz macht die Rollen sichtbar.
data = pd.DataFrame(
    {
        "area_m2": [45, 80, 120, 60],
        "rooms": [2, 3, 5, 2],
        "year_built": [1998, 2010, 1985, 2020],
        "price_label": ["günstig", "mittel", "hoch", "mittel"],
    }
)

# Merkmale beschreiben jede Beobachtung.
feature_columns = ["area_m2", "rooms", "year_built"]
X = data[feature_columns]

# Der Zielwert ist die gesuchte Antwortspalte.
target_column = "price_label"
y = data[target_column]

# Eine einfache Prüfung verhindert vertauschte Längen.
if len(X) != len(y):
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Die Ausgabe bleibt klein und zeigt die wichtigsten Rollen.
print("Beobachtungen:", len(data))
print("Merkmale:", list(X.columns))
print("Zielwert:", target_column)
print("X-Form:", X.shape)
print("y-Form:", y.shape)
print("Erste Labels:", list(y.head(4)))

```

## **2. X und y**

### **2.1. Eingabematrix erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_01.jpg?v=1787537083" width="250">



>* Zeilen sind Beobachtungen, Spalten sind Merkmale
>* Zielwerte bleiben getrennt von Eingaben

>* Nur sinnvolle Merkmale auswählen und ordnen
>* Einheiten, Kategorien und fehlende Werte prüfen

>* Eingabematrizen sorgfältig prüfen und dokumentieren
>* Gute Struktur erleichtert spätere Interpretationen



```
#@title Python-Code - Eingabematrix erstellen

# Wir erstellen eine Eingabematrix aus Tabellendaten.
# Merkmale werden vom Zielwert sauber getrennt.
# Am Ende prüfen wir Formen und Inhalte.

import pandas as pd

# Dieser kleine Datensatz beschreibt vier Wohnungen.
data = pd.DataFrame(
    {
        "area_m2": [45, 70, 85, 120],
        "rooms": [2, 3, 3, 5],
        "distance_km": [1.2, 4.5, 3.0, 8.0],
        "price_eur": [210000, 320000, 390000, 520000],
    }
)

# Diese Spalten sind erklärende Merkmale für das Modell.
feature_columns = ["area_m2", "rooms", "distance_km"]

# Der Zielwert wird getrennt gespeichert.
target_column = "price_eur"

# X enthält nur Merkmale, y enthält nur Zielwerte.
X = data[feature_columns].to_numpy()
y = data[target_column].to_numpy()

# Eine einfache Prüfung schützt vor vertauschten Zeilenzahlen.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Die Ausgabe zeigt Struktur statt den ganzen Datensatz.
print("Merkmalsspalten:", feature_columns)
print("Zielfeld:", target_column)
print("Form von X:", X.shape)
print("Form von y:", y.shape)
print("Erste Beobachtung in X:", X[0].tolist())
print("Passender Zielwert in y:", int(y[0]))

```

### **2.2. Zielwerte als Vektor**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_02.jpg?v=1787537079" width="250">



>* Zielwerte sind die gewünschten Ergebnisse.
>* Ihre Reihenfolge passt zu den Beobachtungen.

>* Ein Zielwert pro Beobachtung
>* Reihenfolge muss exakt erhalten bleiben

>* Zielwerte können Zahlen oder Kategorien sein
>* Reihenfolge verbindet Ausgaben mit Beobachtungen



```
#@title Python-Code - Zielwerte als Vektor

# Dieses Beispiel trennt Zielwerte aus einer Tabelle.
# Der Zielvektor y bleibt zeilenweise passend.
# Die Ausgabe zeigt Werte, Form und Bedeutung.

import pandas as pd

# Eine kleine Tabelle enthält Merkmale und einen Zielwert.
data = pd.DataFrame(
    {
        "area_m2": [45, 60, 80, 100],
        "rooms": [2, 2, 3, 4],
        "price_eur": [210000, 285000, 390000, 520000],
    }
)

# X enthält nur beschreibende Merkmale.
X = data[["area_m2", "rooms"]].to_numpy()

# y enthält genau einen Zielwert pro Beobachtung.
y = data["price_eur"].to_numpy()

# Diese Prüfung schützt die zeilenweise Zuordnung.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Kurze Ausgaben zeigen die wichtigste Struktur.
print("Tabelle: 4 Wohnungen, Zielwert = price_eur.")
print(f"X-Form: {X.shape} für Beobachtungen und Merkmale.")
print(f"y-Form: {y.shape} als eindimensionaler Zielvektor.")
print(f"y-Werte in Euro: {y.tolist()}.")
print(f"Erste Beobachtung passt zu Zielwert: {X[0].tolist()} -> {y[0]} Euro.")

```

### **2.3. Batchformen verstehen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_02_03.jpg?v=1787537081" width="250">



>* Batchform zeigt Beobachtungen und Merkmale gemeinsam
>* Dimensionen prüfen korrekte Datenstruktur

>* Batchform zeigt Beobachtungen und Merkmale klar
>* Vertauschte Dimensionen führen zu falschen Ergebnissen

>* Batches ermöglichen Verarbeitung in kleineren Datengruppen
>* Klare Formen sichern vergleichbare Datenvorbereitung



```
#@title Python-Code - Batchformen verstehen

# Dieses Beispiel zeigt Batchformen mit Wohnungsdaten.
# X enthält Merkmale, y enthält Zielwerte.
# Die Ausgabe vergleicht korrekte und vertauschte Formen.

import numpy as np

# Jede Zeile beschreibt eine Wohnung mit drei Merkmalen.
X = np.array(
    [[55, 2, 4.5], [80, 3, 2.0], [42, 1, 6.0], [95, 4, 1.5], [68, 3, 3.2]]
)

# Der Zielvektor enthält einen Preis pro Wohnung.
y = np.array([210000, 340000, 180000, 420000, 290000])

# Diese Prüfung verbindet Beobachtungen in X mit Zielwerten in y.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen enthalten.")

# Eine vertauschte Matrix zeigt einen typischen Formfehler.
X_swapped = X.T

print("Korrekte Batchform X: 5 Wohnungen x 3 Merkmale ->", X.shape)
print("Zielvektor y: ein Preis pro Wohnung ->", y.shape)
print("Erste Wohnung: Fläche, Zimmer, Entfernung ->", X[0].tolist())
print("Erster Zielwert: Preis in Euro ->", int(y[0]))
print("Vertauschte Form X.T: 3 Zeilen x 5 Spalten ->", X_swapped.shape)
print("Merksatz: X.shape[0] muss zu y.shape[0] passen.")

```

## **3. Datentypen dokumentieren**

### **3.1. Numerisch oder kategorial**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_01.jpg?v=1787537076" width="250">



>* Numerisch bedeutet sinnvoll berechenbare Größen
>* Kategorial beschreibt Gruppen, auch bei Ziffern

>* Numerische Werte: kontinuierlich oder gezählt
>* Kategorien: nominal oder geordnet

>* Merkmale korrekt einordnen, Fehlinterpretationen vermeiden
>* Werte, Einheiten und Kodierungen dokumentieren



```
#@title Python-Code - Numerisch oder kategorial

# Dieses Beispiel dokumentiert Datentypen in Tabellendaten.
# Numerische und kategoriale Merkmale werden fachlich unterschieden.
# Am Ende entsteht eine kompakte Datendokumentation.

import pandas as pd

# Ein kleiner Datensatz enthält verschiedene fachliche Merkmalstypen.
data = pd.DataFrame(
    {
        "age_years": [22, 35, 41, 29],
        "postal_code": [10115, 80331, 50667, 20095],
        "device_type": ["Laptop", "Smartphone", "Tablet", "Laptop"],
        "satisfaction_level": ["niedrig", "mittel", "hoch", "mittel"],
    }
)

# Die Zielvariable wird getrennt vom Eingabebereich dokumentiert.
target = pd.Series([0, 1, 1, 0], name="buys_course")

# Technische Datentypen allein reichen für die Bedeutung nicht aus.
technical_types = data.dtypes.astype(str).to_dict()

# Fachliche Typen beschreiben, wie Werte interpretiert werden sollen.
documentation = pd.DataFrame(
    {
        "technical_dtype": [technical_types[column] for column in data.columns],
        "meaning_type": ["numerisch", "kategorial", "kategorial", "ordinal"],
        "unit_or_order": ["Jahre", "keine Einheit", "keine Einheit", "niedrig < mittel < hoch"],
        "value_range": ["22 bis 41", "erlaubte Codes", "3 Kategorien", "3 geordnete Stufen"],
    },
    index=data.columns,
)

# X enthält Merkmale, y enthält den Zielwert.
X = data.copy()
y = target.copy()

# Eine einfache Prüfung macht die Array-Formen nachvollziehbar.
if X.shape[0] != y.shape[0]:
    raise ValueError("X und y müssen gleich viele Beobachtungen haben.")

# Die Ausgabe zeigt Dokumentation und Formen kompakt zusammen.
print("Datendokumentation für vier Merkmale:")
print(documentation)
print(f"Form von X: {X.shape[0]} Beobachtungen, {X.shape[1]} Merkmale")
print(f"Form von y: {y.shape[0]} Zielwerte")
print("Hinweis: postal_code sieht numerisch aus, ist aber kategorial.")

```

### **3.2. Zielwerte richtig einordnen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_02.jpg?v=1787537074" width="250">



>* Zielwerte beschreiben das Analyseergebnis
>* Typ und Bedeutung sorgfältig dokumentieren

>* Gespeicherte Zahlen fachlich richtig deuten
>* Binäre, mehrklassige und numerische Ziele unterscheiden

>* Datentyp, Bereich, Einheit und Sonderfälle festhalten
>* Fehlende Werte und Zielvektorform klar beschreiben



```
#@title Python-Code - Zielwerte richtig einordnen

# Dieses Beispiel ordnet Zielwerte bewusst ein.
# Datentypen, Wertebereiche und Formen werden dokumentiert.
# Die Ausgabe zeigt eine kompakte Zielwertbeschreibung.

import pandas as pd

# Ein kleiner Datensatz enthält Merkmale und einen Zielwert.
data = pd.DataFrame(
    {
        "study_hours": [2.0, 4.5, 1.0, 6.0, 3.5],
        "practice_tests": [1, 3, 0, 4, 2],
        "passed": ["nein", "ja", "nein", "ja", "ja"],
    }
)

# X enthält Eingabemerkmale, y enthält den Zielwert.
X = data[["study_hours", "practice_tests"]]
y = data["passed"]

# Diese Angaben dokumentieren die fachliche Bedeutung.
target_documentation = pd.DataFrame(
    {
        "Angabe": ["Rolle", "Typ", "Werte", "Einheit", "Form"],
        "Beschreibung": [
            "Zielwert: Prüfung bestanden",
            "kategorial, binär",
            "ja oder nein",
            "keine Einheit",
            str(y.shape),
        ],
    }
)

# Eine einfache Prüfung schützt vor unerwarteten Zielwerten.
allowed_values = {"ja", "nein"}
observed_values = set(y.unique())

# Die Form von X und y muss zusammenpassen.
shape_is_valid = len(X) == len(y)
values_are_valid = observed_values <= allowed_values

print("Zielwerte richtig einordnen:")
print(target_documentation.to_string(index=False))
print("X-Form:", X.shape)
print("y-Form:", y.shape)
print("Prüfung gültiger Werte:", values_are_valid)
print("Prüfung passender Zeilenanzahl:", shape_is_valid)

```

### **3.3. Datenwörterbuch erstellen**

<img src="https://cdn.jsdelivr.net/gh/mhrafiei/contents@main/Udemy/ML Für Anfänger/Module_03/Lecture_A/image_03_03.jpg?v=1787537078" width="250">



>* Datenwörterbuch erklärt Spalten, Werte und Nutzung
>* Klare Beschreibungen verhindern Missverständnisse im Team

>* Datentyp, Wertebereich und Einheit festhalten
>* Kategorien und fehlende Werte eindeutig erklären

>* Eingabemerkmale, Zielwert und Array-Formen festhalten
>* Fehler vermeiden und Analysen nachvollziehbar machen



```
#@title Python-Code - Datenwörterbuch erstellen

# Dieses Beispiel erstellt ein kleines Datenwörterbuch.
# Es dokumentiert Typen, Bereiche, Einheiten und Formen.
# Am Ende siehst du kompakte Prüfergebnisse.

import numpy as np
import pandas as pd

# Wir bauen einen kleinen tabellarischen Datensatz im Speicher.
data = pd.DataFrame(
    {
        "study_hours": [2.0, 3.5, 1.0, 4.0, 5.5],
        "attendance_percent": [80, 90, 60, 95, 100],
        "passed": [0, 1, 0, 1, 1],
    }
)

# X enthält Eingabemerkmale, y enthält den Zielwert.
feature_columns = ["study_hours", "attendance_percent"]
target_column = "passed"

X = data[feature_columns].to_numpy()
y = data[target_column].to_numpy()

# Ein Datenwörterbuch beschreibt jede Spalte fachlich und technisch.
data_dictionary = pd.DataFrame(
    {
        "spalte": ["study_hours", "attendance_percent", "passed"],
        "rolle": ["Merkmal", "Merkmal", "Zielwert"],
        "datentyp": ["float", "int", "int"],
        "einheit": ["Stunden", "Prozent", "0=nein, 1=ja"],
        "wertebereich": ["0.0 bis 24.0", "0 bis 100", "0 oder 1"],
    }
)

# Diese Prüfungen machen erwartete Array-Formen ausdrücklich sichtbar.
expected_x_shape = (len(data), len(feature_columns))
expected_y_shape = (len(data),)

if X.shape != expected_x_shape:
    raise ValueError("X hat nicht die erwartete Form.")

if y.shape != expected_y_shape:
    raise ValueError("y hat nicht die erwartete Form.")

# Die Ausgabe bleibt kurz und zeigt nur die wichtigsten Dokumentationspunkte.
print("Datenwörterbuch:")
print(data_dictionary.to_string(index=False))
print(f"Form von X: {X.shape} = Beobachtungen x Merkmale")
print(f"Form von y: {y.shape} = ein Zielwert pro Beobachtung")
print("Prüfung: Datentypen, Wertebereiche, Einheiten und Formen sind dokumentiert.")

```
