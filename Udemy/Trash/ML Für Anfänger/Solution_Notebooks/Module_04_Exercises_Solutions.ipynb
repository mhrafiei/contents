## Aufgabe 2: Datentypen, Texte, Duplikate und Fehlwerte bereinigen

    Erstellen Sie eine bereinigte Kopie.

1. Entfernen Sie vollständige Duplikate.
2. Wandeln Sie `monthly_income` robust in eine numerische Euro-Spalte um.
3. Vereinheitlichen Sie `city` und `channel` durch Entfernen von Leerzeichen und konsistente Schreibweise.
4. Konvertieren Sie `signup_date` nach Datum. Nicht lesbare Werte sollen fehlen.
5. Ersetzen Sie fehlendes `age` durch den Median plausibler Alterswerte und fehlendes Einkommen durch den Median der jeweiligen Stadt, mit globalem Median als Fallback.

> **Hinweis:** Bewahren Sie Prüfspalten auf, damit spätere Nutzer erkennen können, welche Werte verändert wurden.

```python
cleaned = customer_raw.copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Datentypen, Texte, Duplikate und Fehlwerte bereinigen
#
# Ziel dieser Codezelle:
# Erstellen Sie eine bereinigte Kopie. 1. Entfernen Sie vollständige Duplikate. 2.
# Wandeln Sie monthlyincome robust in eine numerische Euro-Spalte um. 3.
# Vereinheitlichen Sie city und channel durch Entfernen von Leerzei...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

cleaned = customer_raw.copy()

# Vollständige Duplikate werden nur einmal behalten. Die ursprüngliche
# Tabelle bleibt für die Dokumentation unverändert.
cleaned = cleaned.drop_duplicates().reset_index(drop=True)

# Reguläre Ausdrücke entfernen Punkte, Leerzeichen und das Eurozeichen.
# Danach kann pandas die verbleibenden Ziffern numerisch interpretieren.
cleaned["monthly_income_eur"] = pd.to_numeric(
    cleaned["monthly_income"]
    .astype("string")
    .str.replace(r"[^0-9,.-]", "", regex=True)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce",
)

# strip entfernt Randabstände. casefold ist für sprachunabhängige
# Kleinschreibung robuster als ein einfacher lower-Aufruf.
cleaned["city"] = cleaned["city"].astype("string").str.strip().str.casefold()
cleaned["channel"] = cleaned["channel"].astype("string").str.strip().str.casefold()

# dayfirst=True ermöglicht das Format 05.02.2026. errors='coerce'
# macht das absichtlich fehlerhafte Datum als NaT sichtbar.
cleaned["signup_date"] = pd.to_datetime(
    cleaned["signup_date"],
    errors="coerce",
    dayfirst=True,
    format="mixed",
)

# Unplausible Alterswerte werden markiert und anschließend als fehlend
# behandelt, damit sie den Imputationsmedian nicht verzerren.
cleaned["age_implausible"] = ~cleaned["age"].between(16, 100)
cleaned.loc[cleaned["age_implausible"], "age"] = np.nan

plausible_age_median = cleaned["age"].median()
cleaned["age"] = cleaned["age"].fillna(plausible_age_median)

# transform('median') liefert für jede Zeile den Median ihrer Stadt.
city_income_median = cleaned.groupby("city")["monthly_income_eur"].transform(
    "median"
)
global_income_median = cleaned["monthly_income_eur"].median()
cleaned["monthly_income_eur"] = (
    cleaned["monthly_income_eur"]
    .fillna(city_income_median)
    .fillna(global_income_median)
)

print(cleaned.to_string(index=False))
print("\nFehlwerte nach den gezielten Schritten:")
print(cleaned.isna().sum().to_string())
```

### Reflexion zu Aufgabe 2

Die Bereinigung ist nachvollziehbar, weil problematische Werte zunächst markiert und erst danach behandelt werden. Ein globaler Median ist robuster gegen Ausreißer als der Mittelwert. Der Stadtmedian kann lokale Unterschiede bewahren, ist bei sehr kleinen Gruppen aber unsicher. Das unlesbare Datum bleibt als `NaT` erhalten, weil keine belastbare Information für eine Ersetzung vorliegt.

## Aufgabe 3: Ausreißer, Skalierung, Binning und Log-Transformation

    Verwenden Sie eine bereinigte Tabelle und führen Sie folgende Schritte durch:

1. Markieren Sie Ausreißer in `spend_eur` mit der IQR-Regel.
2. Erzeugen Sie eine manuelle Min-Max-Skalierung für `visits`.
3. Erzeugen Sie eine manuelle Standardisierung für `monthly_income_eur`.
4. Bilden Sie `log_spend = log1p(spend_eur)`.
5. Teilen Sie `age` mit sinnvollen Grenzen in Altersgruppen ein.
6. Vergleichen Sie die Verteilung von `spend_eur` und `log_spend` in getrennten Histogrammen.

> **Hinweis:** Speichern Sie Skalierungsparameter, wenn dieselbe Transformation später auf neue Daten angewendet werden soll.

```python
# Nutzen Sie Ihre bereinigte Tabelle aus Aufgabe 2 oder erstellen Sie hier eine Kopie.
transformed = cleaned.copy() if "cleaned" in globals() else customer_raw.drop_duplicates().copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Ausreißer, Skalierung, Binning und Log-Transformation
#
# Ziel dieser Codezelle:
# Verwenden Sie eine bereinigte Tabelle und führen Sie folgende Schritte durch: 1.
# Markieren Sie Ausreißer in spendeur mit der IQR-Regel. 2. Erzeugen Sie eine
# manuelle Min-Max-Skalierung für visits. 3. Erzeugen Sie eine...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

# Für eine unabhängige Ausführung wird Aufgabe 2 in kompakter Form
# wiederholt, falls `cleaned` noch nicht vorhanden ist.
if "cleaned" in globals():
    transformed = cleaned.copy()
else:
    transformed = customer_raw.drop_duplicates().copy()
    transformed["monthly_income_eur"] = pd.to_numeric(
        transformed["monthly_income"]
        .astype("string")
        .str.replace(r"[^0-9]", "", regex=True),
        errors="coerce",
    )
    transformed["city"] = transformed["city"].astype("string").str.strip().str.casefold()
    transformed.loc[~transformed["age"].between(16, 100), "age"] = np.nan
    transformed["age"] = transformed["age"].fillna(transformed["age"].median())
    transformed["monthly_income_eur"] = transformed["monthly_income_eur"].fillna(
        transformed["monthly_income_eur"].median()
    )

# Die IQR-Regel verwendet die mittleren 50 Prozent der Verteilung.
q1 = transformed["spend_eur"].quantile(0.25)
q3 = transformed["spend_eur"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
transformed["spend_outlier"] = ~transformed["spend_eur"].between(
    lower_bound, upper_bound
)

# Min-Max-Skalierung bildet das beobachtete Minimum auf 0 und das
# beobachtete Maximum auf 1 ab.
visits_min = transformed["visits"].min()
visits_max = transformed["visits"].max()
transformed["visits_minmax"] = (
    transformed["visits"] - visits_min
) / (visits_max - visits_min)

# Standardisierung zentriert um den Mittelwert und teilt durch die
# Standardabweichung. ddof=0 passt zur üblichen ML-Definition.
income_mean = transformed["monthly_income_eur"].mean()
income_std = transformed["monthly_income_eur"].std(ddof=0)
transformed["income_zscore"] = (
    transformed["monthly_income_eur"] - income_mean
) / income_std

# log1p ist für den Wert null definiert und komprimiert große Werte.
transformed["log_spend"] = np.log1p(transformed["spend_eur"])

# Die Grenzen sind fachlich interpretierbar und werden ausdrücklich
# dokumentiert. right=False macht die Intervalle links geschlossen.
transformed["age_group"] = pd.cut(
    transformed["age"],
    bins=[16, 30, 45, 60, 101],
    labels=["16-29", "30-44", "45-59", "60-100"],
    right=False,
)

print(
    transformed[
        [
            "customer_id", "spend_eur", "spend_outlier", "visits_minmax",
            "income_zscore", "log_spend", "age_group",
        ]
    ].round(3).to_string(index=False)
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(transformed["spend_eur"], bins=7, edgecolor="white")
ax.set_title("Originale Ausgaben")
ax.set_xlabel("Euro")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(transformed["log_spend"], bins=7, edgecolor="white")
ax.set_title("Log-transformierte Ausgaben")
ax.set_xlabel("log(1 + Ausgaben)")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.show()
```

### Reflexion zu Aufgabe 3

Die IQR-Markierung ist ein statistischer Hinweis, kein automatischer Löschbefehl. Min-Max-Skalierung ist empfindlich gegenüber Extremwerten, während Standardisierung ebenfalls durch starke Ausreißer beeinflusst werden kann. Die Log-Transformation reduziert die visuelle Dominanz großer positiver Werte. Binning vereinfacht ein kontinuierliches Merkmal, verliert dabei aber Detailinformation und hängt von den gewählten Grenzen ab.

    Erweitern Sie die bereinigte Tabelle um:

1. Jahr, Monat und Wochentag aus `signup_date`,

```python
feature_data = cleaned.copy() if "cleaned" in globals() else customer_raw.drop_duplicates().copy()

# ============================================================
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Kategorien, Datumsmerkmale und Interaktionen erzeugen
#
# Ziel dieser Codezelle:
# Erweitern Sie die bereinigte Tabelle um: 1. Jahr, Monat und Wochentag aus
# signupdate, 2. One-Hot-Spalten für city und channel, 3. spendpervisit mit sicherer
# Behandlung einer möglichen Division durch null, 4. eine Inte...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

if "cleaned" in globals():
    feature_data = cleaned.copy()
else:
    # Kompakter Fallback für eine unabhängige Ausführung.
    feature_data = customer_raw.drop_duplicates().copy()
    feature_data["city"] = feature_data["city"].astype("string").str.strip().str.casefold()
    feature_data["channel"] = feature_data["channel"].astype("string").str.strip().str.casefold()
    feature_data["signup_date"] = pd.to_datetime(
        feature_data["signup_date"], errors="coerce", dayfirst=True, format="mixed"
    )
    feature_data["monthly_income_eur"] = pd.to_numeric(
        feature_data["monthly_income"].astype("string").str.replace(r"[^0-9]", "", regex=True),
        errors="coerce",
    ).fillna(3500)

# Datumsbestandteile werden nur aus gültigen Zeitstempeln gewonnen.
feature_data["signup_year"] = feature_data["signup_date"].dt.year
feature_data["signup_month"] = feature_data["signup_date"].dt.month
feature_data["signup_weekday"] = feature_data["signup_date"].dt.dayofweek

# get_dummies erzeugt nachvollziehbare Indikatorspalten. dtype=int
# vermeidet boolesche Ausgaben, wenn numerische Merkmale gewünscht sind.
encoded_categories = pd.get_dummies(
    feature_data[["city", "channel"]],
    prefix=["city", "channel"],
    dtype=int,
)
feature_data = pd.concat([feature_data, encoded_categories], axis=1)

# replace(0, np.nan) verhindert eine unendliche Division. Ein fehlender
# Wert signalisiert, dass die Kennzahl fachlich nicht definiert ist.
safe_visits = feature_data["visits"].replace(0, np.nan)
feature_data["spend_per_visit"] = feature_data["spend_eur"] / safe_visits

# Eine Interaktion kann ausdrücken, dass die gemeinsame Größe zweier
# Merkmale relevant sein könnte. Sie beweist noch keinen Zusammenhang.
feature_data["income_x_visits"] = (
    feature_data["monthly_income_eur"] * feature_data["visits"]
)

created_columns = [
    "signup_year",
    "signup_month",
    "signup_weekday",
    "spend_per_visit",
    "income_x_visits",
    *encoded_categories.columns.tolist(),
]
control_table = pd.DataFrame(
    {
        "column": created_columns,
        "dtype": feature_data[created_columns].dtypes.astype(str).values,
        "missing": feature_data[created_columns].isna().sum().values,
        "min": [
            feature_data[col].min(skipna=True)
            if pd.api.types.is_numeric_dtype(feature_data[col])
            else np.nan
            for col in created_columns
        ],
        "max": [
            feature_data[col].max(skipna=True)
            if pd.api.types.is_numeric_dtype(feature_data[col])
            else np.nan
            for col in created_columns
        ],
    }
)

print(control_table.to_string(index=False))
```

```python
# ============================================================
# KOMMENTIERTE MUSTERLÖSUNG: Integrationsaufgabe: Überprüfbare Vorverarbeitungsfunktion
#
# Ziel dieser Codezelle:
# Schreiben Sie eine Funktion preprocesscustomers(df), die die wesentlichen Schritte
# aus den vorherigen Aufgaben reproduzierbar ausführt und zwei Objekte zurückgibt:
# 1. die verarbeitete Tabelle, 2. einen Kontrollbericht...
#
# Die Lösung folgt bewusst einer gut prüfbaren Schrittfolge.
# Zwischenwerte und Ausgaben machen Formen, Annahmen und Ergebnisse sichtbar.
# Die fachliche Interpretation und typische Fehlerquellen stehen in der
# ausführlichen Markdown-Reflexion direkt unter dieser Codezelle.
# ============================================================

def preprocess_customers(df: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    # Bereinigt und transformiert Kundendaten ohne Seiteneffekte.

    # Eine tiefe Kopie schützt das übergebene Objekt vor Veränderungen.
    result = df.copy(deep=True)
    rows_before = len(result)
    duplicate_count = int(result.duplicated().sum())

    result = result.drop_duplicates().reset_index(drop=True)

    # Textbasierte Eurobeträge werden in Zahlen umgewandelt.
    result["monthly_income_eur"] = pd.to_numeric(
        result["monthly_income"]
        .astype("string")
        .str.replace(r"[^0-9,.-]", "", regex=True)
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False),
        errors="coerce",
    )

    # Kategorien werden mit denselben festen Regeln vereinheitlicht.
    for column in ["city", "channel"]:
        result[column] = (
            result[column].astype("string").str.strip().str.casefold()
        )

    result["signup_date"] = pd.to_datetime(
        result["signup_date"],
        errors="coerce",
        dayfirst=True,
        format="mixed",
    )

    # Unplausible Alterswerte werden nachvollziehbar markiert.
    result["age_implausible"] = ~result["age"].between(16, 100)
    result.loc[result["age_implausible"], "age"] = np.nan
    result["age"] = result["age"].fillna(result["age"].median())

    # Erst gruppenspezifischer, dann globaler Median als Fallback.
    city_median = result.groupby("city")["monthly_income_eur"].transform("median")
    result["monthly_income_eur"] = (
        result["monthly_income_eur"]
        .fillna(city_median)
        .fillna(result["monthly_income_eur"].median())
    )

    # IQR-Ausreißer werden markiert, aber nicht automatisch entfernt.
    q1 = result["spend_eur"].quantile(0.25)
    q3 = result["spend_eur"].quantile(0.75)
    iqr = q3 - q1
    result["spend_outlier"] = ~result["spend_eur"].between(
        q1 - 1.5 * iqr,
        q3 + 1.5 * iqr,
    )

    # Reproduzierbare Merkmale werden mit klaren Namen ergänzt.
    result["signup_month"] = result["signup_date"].dt.month
    result["spend_per_visit"] = result["spend_eur"] / result["visits"].replace(0, np.nan)
    result["log_spend"] = np.log1p(result["spend_eur"])

    categorical_features = pd.get_dummies(
        result[["city", "channel"]],
        prefix=["city", "channel"],
        dtype=int,
    )
    result = pd.concat([result, categorical_features], axis=1)

    report = {
        "rows_before": rows_before,
        "rows_after": len(result),
        "removed_duplicates": duplicate_count,
        "remaining_missing_values": int(result.isna().sum().sum()),
        "marked_age_issues": int(result["age_implausible"].sum()),
        "marked_spend_outliers": int(result["spend_outlier"].sum()),
        "created_feature_count": int(
            3 + categorical_features.shape[1]
        ),
    }

    return result, report

processed_customers, preprocessing_report = preprocess_customers(customer_raw)

print("Kontrollbericht:")
print(preprocessing_report)
print("\nVerarbeitete Stichprobe:")
print(processed_customers.head().to_string(index=False))

# Ausreißer werden als separate Gruppe gezeichnet. Das Diagramm zeigt
# einen Zusammenhang, kann aber keine Ursache nachweisen.
fig, ax = plt.subplots(figsize=(7, 5))
for is_outlier, group in processed_customers.groupby("spend_outlier"):
    ax.scatter(
        group["monthly_income_eur"],
        group["spend_eur"],
        label="IQR-Ausreißer" if is_outlier else "kein IQR-Ausreißer",
        s=70,
        alpha=0.8,
    )
ax.set_title("Einkommen und Ausgaben nach Ausreißermarkierung")
ax.set_xlabel("Monatliches Einkommen in Euro")
ax.set_ylabel("Ausgaben in Euro")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
