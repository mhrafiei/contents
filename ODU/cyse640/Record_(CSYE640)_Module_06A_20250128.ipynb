## **1. Introduction to Generative AI Models: Transformers and Diffusion**

### **1.1. Contrasting Mechanisms: Transformers vs. Diffusion Models**

<img src="https://storage.googleapis.com/mrafiei-odu/module_06/Lecture%20A/image_01_01.png" width="250">



>* Transformers use self-attention for language processing.
* Diffusion models refine noise for image generation.

>* Transformers excel in sequence-based generative tasks.
* Diffusion models generate diverse, high-fidelity outputs.

>* Transformers face computational cost and data challenges.
* Optimizing diffusion models for stability and efficiency.



```
#@title Python Code - Transformers vs. Diffusion Models

# This code demonstrates the basic usage of a Transformer model for text generation
# and a Diffusion model for image generation. The purpose is to illustrate the contrasting
# mechanisms of these two types of generative models. We will use Hugging Face's
# Transformers and Diffusions libraries for the text and image generation, respectively.

# Import necessary libraries
from transformers import pipeline
from torchvision import transforms
from torch import nn
from diffusers import StableDiffusionPipeline
import torch
import matplotlib.pyplot as plt
import numpy as np

# Part 1: Text Generation using a Transformer Model
# We will use a pre-trained GPT-2 model to generate text based on a given prompt.
# Initialize the text generation pipeline with GPT-2
text_generator = pipeline('text-generation', model='gpt2')

# Define a prompt for text generation
prompt = "Once upon a time in a land far, there was a prince who"

# Generate text
generated_text = text_generator(prompt, max_length=50, num_return_sequences=1)

# Print the generated text
print("Generated Text:\n")
print(generated_text[0]['generated_text'])
print("\n")

# Part 2: Image Generation using a Hugging Face Diffusion Model
# Load the Stable Diffusion pipeline
# (you can replace "runwayml/stable-diffusion-v1-5" with another model if desired)
model_id = "runwayml/stable-diffusion-v1-5"

# By default, Stable Diffusion requires a GPU for best performance.
# If you're running on CPU or in a limited environment, consider using a smaller model or
# enabling `torch_dtype=torch.float16` if GPU with half-precision is available.
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe.to("cuda" if torch.cuda.is_available() else "cpu")

# Define a prompt for image generation
prompt = "A fantasy landscape with towering crystal mountains and golden clouds"

# Generate the image
# If potential Not Safe for Work (NSFW) content was detected in one or more images,
# a black image will be returned instead.
with torch.autocast("cuda" if torch.cuda.is_available() else "cpu"):
    image = pipe(prompt, num_inference_steps=30, guidance_scale=7.5).images[0]

# Display the image
plt.figure(figsize=(3, 3))
plt.imshow(image)
plt.title("Stable Diffusion Output")
plt.axis("off")
plt.show()
```
