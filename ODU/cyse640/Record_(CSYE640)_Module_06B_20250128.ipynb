### **1.1. Adapting Pre-trained Models for Specific Domains**

<img src="https://storage.googleapis.com/mrafiei-odu/module_06/Lecture%20B/image_01_01.png" width="250">



>* Fine-tune models with domain-specific datasets.
* Enhance model performance for specialized fields.

>* Address domain shift with transfer learning.
* Use data augmentation for limited datasets.

>* Balance model complexity with computational efficiency.
* Use techniques for lightweight, efficient AI deployment.



```
#@title Python Code - Adapting Pre-trained Models for Specific Domains

# This Python code example demonstrates how to adapt a pre-trained model for a specific domain using transfer learning.
# We will use a pre-trained BERT model from the Hugging Face Transformers library and fine-tune it on a domain-specific dataset.
# The chosen dataset is the IMDb movie reviews dataset, which is suitable for sentiment analysis in the entertainment domain.
# The code will illustrate the process of loading a pre-trained model, preparing the dataset, and fine-tuning the model.

# Install required libraries (uncomment if needed; in Colab these are usually pre-installed)
!pip install -q datasets

# Import necessary libraries
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
from datasets import load_dataset
import torch

# Load the IMDb dataset from the Hugging Face datasets library
# This dataset is used for sentiment analysis, which is a common task in the entertainment domain
imdb_dataset = load_dataset('imdb')

# Load the pre-trained BERT tokenizer
# The tokenizer is responsible for converting text into tokens that the model can understand
model_name = 'bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(model_name)

# Tokenize the dataset
# This step involves converting the text data into a format suitable for the BERT model
# We will tokenize the text and truncate or pad it to a fixed length
max_length = 128
def tokenize_function(examples):
    return tokenizer(examples['text'], padding='max_length', truncation=True, max_length=max_length)

# Apply the tokenization function to the dataset
encoded_dataset = imdb_dataset.map(tokenize_function, batched=True)

# Load the pre-trained BERT model for sequence classification
# We will fine-tune this model on the IMDb dataset
model = BertForSequenceClassification.from_pretrained(model_name, num_labels=2)

# Define the training arguments
# These arguments specify the training configuration, such as the number of epochs and batch size
training_args = TrainingArguments(
    output_dir='./results',          # Output directory
    num_train_epochs=1,              # Number of training epochs
    per_device_train_batch_size=8,   # Batch size for training
    per_device_eval_batch_size=8,    # Batch size for evaluation
    warmup_steps=500,                # Number of warmup steps for learning rate scheduler
    weight_decay=0.01,               # Strength of weight decay
    logging_dir='./logs',            # Directory for storing logs
    logging_steps=10,
)

# Define the Trainer
# The Trainer class provides an easy-to-use interface for training and evaluating models
trainer = Trainer(
    model=model,                         # The pre-trained model
    args=training_args,                  # Training arguments
    train_dataset=encoded_dataset['train'],  # Training dataset
    eval_dataset=encoded_dataset['test']     # Evaluation dataset
)

# Fine-tune the model
# This step involves training the model on the domain-specific dataset
trainer.train()

# Evaluate the model
# After fine-tuning, we evaluate the model to see how well it performs on the test set
trainer.evaluate()
```

```
#@title Python Code - Crafting Effective Prompts for Enhanced Model Performance

# This code demonstrates how to craft effective prompts for enhancing the performance of a language model.
# We will use a simple example with a pre-trained language model from the Hugging Face Transformers library.
# The purpose is to show how different prompts can influence the model's output.
# We will use the GPT-2 model to generate text based on different prompts.
# Note: This code requires internet access to download the pre-trained model.

from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

# Load pre-trained model and tokenizer
model_name = "gpt2"
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = GPT2Tokenizer.from_pretrained(model_name)

# Function to generate text based on a given prompt
def generate_text(prompt, max_length=50):
    # Encode the prompt text
    input_ids = tokenizer.encode(prompt, return_tensors='pt')
    # Generate text
    output = model.generate(input_ids, max_length=max_length, num_return_sequences=1)
    # Decode the generated text
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text

# Example prompts
prompt_1 = "Summarize the main features of the new smartphone:"
prompt_2 = "List the benefits of using renewable energy sources:"
prompt_3 = "Explain the impact of climate change on polar bears:"

# Generate and print text for each prompt
print("Prompt 1 Output:")
print(generate_text(prompt_1))
print("\nPrompt 2 Output:")
print(generate_text(prompt_2))
print("\nPrompt 3 Output:")
print(generate_text(prompt_3))
```

### **1.5. Mitigating Memory Loss in Model Training**

<img src="https://storage.googleapis.com/mrafiei-odu/module_06/Lecture%20B/image_01_05.png" width="250">



>* Retain relevant information, discard unnecessary data.
* Use continual learning to prevent memory loss.

>* Incorporate memory-augmented neural networks (MANNs).
* Enhance storage and retrieval with external memory.

>* Human oversight enhances model memory retention.
* Regular audits ensure ethical and efficient learning.



```
#@title Python Code - Mitigating Memory Loss in Model Training

# This Python code example demonstrates the use of Elastic Weight Consolidation (EWC) to mitigate memory loss in model training.
# The code uses a simple neural network on the MNIST dataset to illustrate how EWC helps prevent catastrophic forgetting
# when the model is trained sequentially on different tasks. The example shows how EWC can be implemented using PyTorch.

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Define a simple neural network model
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(28 * 28, 256)
        self.fc2 = nn.Linear(256, 10)
    def forward(self, x):
        x = x.view(-1, 28 * 28)
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Function to calculate the EWC loss
# EWC loss helps in retaining the knowledge of previous tasks by penalizing changes to important weights
class EWC:
    def __init__(self, model, dataloader, importance=1000):
        self.model = model
        self.dataloader = dataloader
        self.importance = importance
        self.params = {n: p.clone().detach() for n, p in model.named_parameters()}
        self.fisher = self._compute_fisher_information()
    def _compute_fisher_information(self):
        fisher = {n: torch.zeros_like(p) for n, p in self.model.named_parameters()}
        self.model.eval()
        for data, target in self.dataloader:
            self.model.zero_grad()
            output = self.model(data)
            loss = nn.CrossEntropyLoss()(output, target)
            loss.backward()
            for n, p in self.model.named_parameters():
                fisher[n] += p.grad ** 2
        fisher = {n: p / len(self.dataloader) for n, p in fisher.items()}
        return fisher
    def penalty(self, model):
        loss = 0
        for n, p in model.named_parameters():
            _loss = self.fisher[n] * (p - self.params[n]) ** 2
            loss += _loss.sum()
        return self.importance * loss

# Load MNIST dataset
transform = transforms.Compose([transforms.ToTensor()])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)

# Initialize model, optimizer, and EWC
model = SimpleNN()
optimizer = optim.SGD(model.parameters(), lr=0.01)
ewc = EWC(model, train_loader)

# Training loop with EWC
for epoch in range(1, 3):  # Train for 2 epochs for demonstration
    model.train()
    for data, target in train_loader:
        optimizer.zero_grad()
        output = model(data)
        loss = nn.CrossEntropyLoss()(output, target)
        # Add EWC penalty to the loss
        loss += ewc.penalty(model)
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch}, Loss: {loss.item()}')
```
